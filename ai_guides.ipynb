- this is a college level programming/computer science intro class
- this book is inspired by https://introcs.cs.luc.edu/ 
- the problem with introcs.cs.luc.edu is that they are not structured by the weeks. Ifi you go to their [index page](https://introcs.cs.luc.edu/index.html) you see the book has many chapters and sections that can be better grouped.
- I want to create a better version of the luc book as a contribution to the open world. Andrew N. Harrington and George Thiruvathukal would love this idea.
- NN01 covers major and overall concepts of the chapter
- NN02~N are sections each with a focus
- each chapter and section should have pedagogically sound structure
- each section should have code examples when possible
- when possible, use ASCII diagrams to explain concepts. Use the arrows and line components as seen in 0101, they are probably UTF-8 symbols.
- each chapter should have a preview quiz, lab, and a review (homework) section.
- each chapter covers one week of work/lecture. 
- each ## and ### should start with their own markdown cell
- this book is not just about programming, it's about computer science, so the conceptual parts are important, not just syntax.
- consistency is key. 
- all NN00.ipynb should have # header, local TOC (only first and 2nd levels), and intro the the chapter according to the content of the chapter. The intro should be conceptually clear.
- chapter review questions should be coding mostly.
- lab questions should be a series of connected technical questions (around 5 questions for a 30 to 50 minutes lab time) using the output from the previous questions.
- the sections are for lectures and they should cover topics for 100 to 120 minutes of lecturing.
- There's a way to make covering 15 chapters possible: just keep the class going during midterm week, which is doable, except that we now have spring break, spring recession, and fall recession... sigh.
- The labs should eventually become a Solution with projects inside. A whole semester of work and everybody walks home with a nice project they can demonstrate in github and, even better, having a web component (i know mobile app is not possible).
- when reviewing each chapter, make sure that:
  - important/essential topics are covered
  - the sections (## and ###) are organized soundly conceptually and pedagogically.
- All preview, lab, review (HW) questions should be graded automatically. 
- Every content sections (other than preview, lab, and review) should have Footnotes at the end of the section. The syntax is ```{rubric} Footnotes\n```. The syntax for the anchors and the footnote entries is [^N].
