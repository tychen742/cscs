- this is a college level programming/computer science intro class
- this book is inspired by https://introcs.cs.luc.edu/ 
- the problem with introcs.cs.luc.edu is that they are not structured by the weeks. Ifi you go to their [index page](https://introcs.cs.luc.edu/index.html) you see the book has many chapters and sections that can be better grouped.
- I want to create a better version of the luc book as a contribution to the open world. Andrew N. Harrington and George Thiruvathukal would love this idea.
- NN01 covers major and overall concepts of the chapter
- NN02~N are sections each with a focus
- each chapter and section should have pedagogically sound structure
- each section should have code examples when possible
- when possible, use ASCII diagrams to explain concepts.
- each chapter should have a lab, a review.
- each chapter covers one week of work/lecture. 
- each ## and ### should start with their own markdown cell
- this book is not just about programming, it's about computer science, so the conceptual parts are important, not just syntax.
