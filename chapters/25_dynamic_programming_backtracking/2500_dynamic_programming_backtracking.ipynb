# Dynamic Programming and Backtracking

Repeated subproblems invite reuse; a tree of choices requires careful exploration and undo. This chapter connects memoization and tabulation to complete state definitions, optimization recurrences, reconstructed solutions, and backtracking. You will compare reusable-pack planning with one-use record selection and report resource-limited searches honestly.

- **Overlapping subproblems**: The same smaller problem appears repeatedly.
- **Optimal substructure**: A larger solution can be built from smaller optimal solutions.
- **Memoization**: A top-down technique that stores recursive results for reuse.
- **Tabulation**: A bottom-up technique that fills a table from small cases to large cases.
- **Backtracking**: A search pattern that makes a choice, explores it, then undoes the choice.

4. Search one-use signed records with position-aware states, safe suffix bounds, and failed-state memoization.
