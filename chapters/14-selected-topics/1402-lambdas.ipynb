# Lambda Expressions

A **lambda expression** is an anonymous function — a compact way to define behavior inline, without declaring a named method.

```csharp
(parameters) => expression
(parameters) => { statements; }
```


## Basic Syntax

Use `=>` (the *lambda arrow*). The left side lists parameters; the right side is the body.


````
// Single expression — no braces needed
var square = (int x) => x * x;
Console.WriteLine(square(5));  // 25

// Multi-statement body
var greet = (string name) =>
{
    string msg = $"Hello, {name}!";
    Console.WriteLine(msg);
};
greet("Alice");  // Hello, Alice!
````

## Func and Action

Lambdas are stored in delegate types:

- `Func<T, TResult>` — takes input, returns a value.
- `Action<T>` — takes input, returns nothing (`void`).
- `Predicate<T>` — takes input, returns `bool`.


````
Func<int, int> square   = x => x * x;
Func<int, int, int> add = (a, b) => a + b;
Action<string> print    = msg => Console.WriteLine(msg);
Predicate<int> isEven   = n => n % 2 == 0;

Console.WriteLine(square(6));       // 36
Console.WriteLine(add(3, 4));       // 7
print("hello");                     // hello
Console.WriteLine(isEven(8));       // True
````

## Lambdas with Collections

Lambdas shine when passed to methods like `Sort`, `Find`, or `RemoveAll`.


````
var names = new List<string> { "Charlie", "Alice", "Bob" };

// Sort alphabetically
names.Sort((a, b) => string.Compare(a, b));
Console.WriteLine(string.Join(", ", names));  // Alice, Bob, Charlie

// Find first short name
string? short_ = names.Find(n => n.Length <= 3);
Console.WriteLine(short_);  // Bob

// Remove names starting with 'A'
names.RemoveAll(n => n.StartsWith('A'));
Console.WriteLine(string.Join(", ", names));  // Bob, Charlie
````

## Capturing Variables (Closures)

A lambda can *capture* variables from the surrounding scope — this is called a **closure**.


````
int threshold = 10;

// Lambda captures 'threshold' from the enclosing scope
Predicate<int> isAbove = n => n > threshold;

Console.WriteLine(isAbove(15));  // True
Console.WriteLine(isAbove(5));   // False

threshold = 20;
Console.WriteLine(isAbove(15));  // False  (captured by reference!)
````

## Passing Lambdas as Parameters

Any method that accepts a `Func` or `Action` can receive a lambda directly.


````
static void ApplyTwice(Action<string> action, string value)
{
    action(value);
    action(value);
}

ApplyTwice(s => Console.WriteLine(s.ToUpper()), "hello");
// HELLO
// HELLO
````
