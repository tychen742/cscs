## True or False

**1.** A recursive method must have a base case to prevent infinite recursion.

```{dropdown} Answer
**True** — Without a base case the call stack will overflow.
```

**2.** Lambda expressions in C# always require an explicit `return` statement.

```{dropdown} Answer
**False** — Single-expression lambdas return implicitly: `x => x * 2`.
```

**3.** LINQ queries can only be used with in-memory collections, not databases.

```{dropdown} Answer
**False** — LINQ to Entities / EF Core translates LINQ queries to SQL.
```

**4.** `Func<int, bool>` represents a function that takes an `int` and returns a `bool`.

```{dropdown} Answer
**True** — The last type parameter of `Func<>` is always the return type.
```

**5.** LINQ's `Select` method filters elements; `Where` transforms them.

```{dropdown} Answer
**False** — `Where` filters; `Select` transforms (projects).
```

## Multiple Choice

**1.** What is the base case in a recursive method?

a) The first method call  
b) The condition that stops further recursion  
c) The return value type  
d) The method name

```{dropdown} Answer
**b)** The base case returns a result directly without another recursive call.
```

**2.** Which LINQ method filters elements based on a predicate?

a) `Select()`  
b) `OrderBy()`  
c) `Where()`  
d) `First()`

```{dropdown} Answer
**c) `Where()`** — `Where(x => condition)` keeps only elements satisfying the condition.
```

**3.** What does the lambda `x => x * 2` do?

a) Takes no parameters and returns 2  
b) Takes one parameter `x` and returns `x * 2`  
c) Takes two parameters  
d) Returns only 2 always

```{dropdown} Answer
**b)** `x =>` declares one parameter; the body `x * 2` is returned implicitly.
```

**4.** What is a **closure** in the context of lambda expressions?

a) A lambda that takes no parameters  
b) A lambda that captures variables from its enclosing scope  
c) A lambda with more than one expression  
d) A static method stored in a variable

```{dropdown} Answer
**b)** A closure captures (closes over) variables from the surrounding scope.
```

**5.** Which LINQ method produces a single aggregated value from a sequence?

a) `SelectMany()`  
b) `GroupBy()`  
c) `Sum()` / `Count()` / `Aggregate()`  
d) `Zip()`

```{dropdown} Answer
**c)** Aggregation methods reduce a sequence to a single value.
```
