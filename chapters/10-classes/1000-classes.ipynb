<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Class | A programmer-defined type that groups data and behavior. |
| 2 | Object | An instance of a class created while a program runs. |
| 3 | Instance | A particular object created from a class. |
| 4 | Field | A variable stored inside an object. |
| 5 | Property | A controlled public member used to read or change object state. |
| 6 | Constructor | A special method that initializes a new object. |
| 7 | Method | A behavior defined by a class. |
| 8 | Access modifier | A keyword such as `public` or `private` that controls visibility. |
| 9 | Encapsulation | Keeping object state protected behind public operations. |
| 10 | Reference type | A type where variables refer to objects rather than directly storing all object data. |
| 11 | Static member | A member that belongs to the class itself rather than one instance. |
| 12 | Instance member | A member that belongs to a particular object. |
| 13 | Operator overloading | Defining how an operator works for a programmer-defined type. |
