Arrays are fine if you know ahead of time how long your sequence of items is.
Then you create your array with that length, and you are all set.

If you want a variable sized container, you are likely to want a `List`.
List is a type of collection that is used when the number of elements is unknown.
As with arrays, you might want a collection of any particular type.
Unfortunately, you cannot use the simple notation of arrays to specify
the type of element in a `List`. Array syntax is
*built into* the language. Lists are handled in the *library* of types
provided by C# from .NET.

## Generics

C# collection types use *generics*: a type in angle brackets after a class name tells the compiler what element type the collection holds. The generic collection types live in the `System.Collections.Generic` namespace, which C# imports automatically.

The generic list syntax is:

```csharp
List<T>
```

To create a list, supply a type argument for the elements it will store:

```{index} List; basics
```

## Lists Basics


A `List<T>` starts with an element type and then supports the everyday sequence operations you expect: add an item, count the items, read an item by index, and loop through the items. The index rules are the same as arrays: the first item is at index `0`.


````csharp
using System;
using System.Collections.Generic;

List<int> numbers = new List<int>();
numbers.Add(10);
numbers.Add(20);
numbers.Add(30);

Console.WriteLine(numbers[0]);
Console.WriteLine(numbers.Count);

foreach (int number in numbers)
{
    Console.WriteLine(number);
}
````

A list can also be initialized with values. A `for` loop is useful when the index is part of the output or logic; a `foreach` loop is simpler when you only need each value.


````csharp
using System;
using System.Collections.Generic;

List<string> names = new List<string> { "Alice", "Bob", "Charlie" };

for (int i = 0; i < names.Count; i++)
{
    Console.WriteLine($"{i}: {names[i]}");
}
````

## Lists Constructors and Methods

We can explore `List` methods with standalone C# examples. When you want to display the contents of a `List`, use a loop or `string.Join(...)`; printing the `List` object itself displays its type name, not its items.


````csharp
List<string> words = new List<string>();
Console.WriteLine($"Count: {words.Count}");
Console.WriteLine($"Items: [{string.Join(", ", words)}]");
````

With the no-parameter `constructor` (the parentheses at the end of the variable declaration), the `List` is empty to start:


````csharp
List<string> words = new List<string>();
Console.WriteLine($"Count: {words.Count}");
Console.WriteLine($"Items: [{string.Join(", ", words)}]");
````

You can add elements, and keep count with the `Count` property
as the size changes:

````csharp
List<string> myNames = new List<string>();
myNames.Add("Alice");
myNames.Add("Bob");
myNames.Add("Charles");
````

You can reference and change elements by index, like with `arrays`:

````csharp
List<string> words = new List<string> { "Apple", "Banana", "Cherry" };
Console.WriteLine(words[0]);
Console.WriteLine(words[2]);
words[2] = "Coconut";
Console.WriteLine(string.Join(", ", words));
````

You can use `foreach` like with arrays or other sequences:

````csharp
List<string> myNames = new List<string> { "Alice", "Bob", "Charles" };

foreach (string name in myNames)
{
    Console.WriteLine(name);
}
````

Next, compare `Remove`, which finds the first matching element and removes it,
and `RemoveAt`, which removes the element at a specified index.
`Remove` returns whether the List has been changed:

````csharp
List<string> words = new List<string> { "Apple", "Banana", "Cherry" };
words.Remove("Apple");
Console.WriteLine(words);
words.Add("Avocado");
words.Add("Durian");
Console.WriteLine(words);
words.RemoveAt(3);
Console.WriteLine(words);
````

Removing does not leave a "hole" in the `List`: The list closes up,
so the index decreases for the elements after the removed one:

````csharp
List<string> words = new List<string> { "Banana", "Cherry", "Avocado" };
Console.WriteLine(words.Count);
````

You can check for membership in a `List` with `Contains`:

````csharp
List<string> words = new List<string>();
Console.WriteLine(words.Contains("Apple"));
Console.WriteLine(words.Contains("Banana"));
````

`Contains` only tells you whether a value is present. When you need its position, use `IndexOf`, which returns the index of the first match or `-1` if the value is not found:

````csharp
List<string> words = new List<string> { "Apple", "Banana", "Cherry" };
Console.WriteLine(words.IndexOf("Banana"));
Console.WriteLine(words.IndexOf("Mango"));
````

You can also remove all elements at once:

````csharp
List<string> words = new List<string>();
words.Clear();
Console.WriteLine(words);
````

## Common Pitfall: Modifying a List While Iterating

`foreach` keeps an internal position in the list as it loops. Adding or removing elements through that same list while a `foreach` is still running invalidates that position, and C# detects it: the loop throws an `InvalidOperationException` rather than silently skipping or duplicating elements.

````csharp
List<int> numbers = new List<int> { 1, 2, 3, 4, 5 };
try
{
    foreach (int n in numbers)
    {
        if (n % 2 == 0)
        {
            numbers.Remove(n);  // modifying the list mid-foreach
        }
    }
}
catch (InvalidOperationException ex)
{
    Console.WriteLine($"Error: {ex.Message}");
}
````

Loop over the indices backwards instead, so removing an element does not disturb the indices you still need to visit. (`List<T>` also has a `RemoveAll` method that removes every matching element in one call — it is explained once you have seen the lambda syntax it takes, in a later chapter.)

````csharp
List<int> numbers = new List<int> { 1, 2, 3, 4, 5 };
for (int i = numbers.Count - 1; i >= 0; i--)
{
    if (numbers[i] % 2 == 0)
    {
        numbers.RemoveAt(i);
    }
}
Console.WriteLine(string.Join(", ", numbers));
````

An aside on the `Remove` method: it both causes a side effect (changing the list) *and* returns a value (whether an element was actually removed). If a method returns a value, that value is normally used, as in an `if` condition. `Remove` shows this is not always required: if you just want the side effect — try to remove an element, whether or not it is present — there is no need to check the return value. This is a complete, fine statement on its own:

```csharp
someList.Remove(element);
```

Think carefully before *defining* a method that both has a side effect and returns a value. Most methods that return a value do not also have a side effect; when a method is used as an ordinary expression, it is easy to forget that it might also be doing something.

## Interactive List Example

As in contrast to `array`, `lists` are handy when you do not know how much
data there will be. A simple example would be reading in lines from the user
interactively:



````csharp
/// Return a List of lines entered by the user in response
/// to the prompt.  Lines in the List will be nonempty, since an
/// empty line terminates the input.
List<string> ReadLines(string prompt)
{
   List<string> lines = new List<string>();
   Console.WriteLine(prompt);
   Console.WriteLine("An empty line terminates input.");
   string line = Console.ReadLine();
   while (line.Length > 0) {
      lines.Add(line);
      line = Console.ReadLine();
   }
   return lines;
}
````
