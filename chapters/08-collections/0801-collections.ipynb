# Data Collections

```{contents}
:local:
:depth: 2
```


Arrays are useful when a program has a fixed-size sequence of values. Collections are built-in types that manage groups of related values when the program needs more flexibility: adding items, removing items, looking up items, or enforcing a particular access pattern.

C# collection types are classes. That means collection variables are reference variables: two variables can refer to the same collection object, and a change through either variable changes that one shared object. This is different from simple value types such as `int`, `double`, and `bool`, where each variable stores its own value.

This chapter focuses on the two collection types introductory C# programmers use most often: `List<T>` for ordered sequences and `Dictionary<TKey,TValue>` for key/value lookup.

## Collection Families


Many C# collection types include angle brackets, such as `List<string>` or `Dictionary<string,int>`. The part inside the angle brackets tells C# what kinds of values the collection works with. A `List<string>` stores strings. A `List<int>` stores integers. A `Dictionary<string,int>` uses string keys to look up integer values.

In documentation, you may also see placeholder names such as `List<T>` or `Dictionary<TKey,TValue>`. These are not types you write exactly as shown. They mean "fill in the real type here." For example, `T` might become `string`, and `TKey,TValue` might become `string,int`.

### Collections by Access Pattern

A collection choice usually starts with the way the program needs to retrieve data. Use this table as a map for the collection types you will see in this chapter and later chapters. The leftmost column groups collections by *how you get an element back out* — the same question [Microsoft's own C# documentation](https://learn.microsoft.com/en-us/dotnet/csharp/language-reference/builtin-types/collections) uses to organize collections.

| Access pattern | Need | Common type | Where it belongs |
|---|---|---|---|
| Indexed | Fixed-size sequential values | array, such as `int[]` or `string[]` | Chapter 7 |
| Indexed | Variable-size sequential values | `List<T>` | This chapter |
| Key/Value | Key/value lookup | `Dictionary<TKey,TValue>` | This chapter |
| Membership | Unique membership tests | `HashSet<T>` | Chapter 17 |
| Restricted Order | Last-in-first-out access | `Stack<T>` | Chapter 15 |
| Restricted Order | First-in-first-out access | `Queue<T>` | Chapter 15 |
| Restricted Order | Priority-based access | `PriorityQueue<TElement,TPriority>` | Chapter 17 |
| Structural | Hierarchical relationships | tree structures | Chapter 16 |
| Structural | Network relationships | graph structures | Chapter 18 |

Lists and dictionaries cover many everyday programming tasks. Stacks, queues, sets, trees, and graphs become more important when the access rule or performance requirement matters as much as the stored values.

## Generic Collections


Most modern C# collection types are **generic**. The type parameter in angle brackets tells C# what kind of element the collection stores.

- `List<string>` stores strings.
- `List<int>` stores integers.
- `Dictionary<string, int>` maps string keys to integer values.

Prefer the generic types in `System.Collections.Generic`. Older non-generic collections store values as `object`, which loses helpful compile-time type checking.


````csharp
using System;
using System.Collections.Generic;

List<string> names = new List<string> { "Ana", "Bo" };
Dictionary<string, int> scores = new Dictionary<string, int>
{
    ["Ana"] = 92,
    ["Bo"] = 85
};

Console.WriteLine(names[0]);
Console.WriteLine(scores["Bo"]);
````

```{index} LINQ
```

Once a type can be iterated, it can also be queried. LINQ (Language-Integrated Query) adds filtering, ordering, and grouping operations on top of any `IEnumerable<T>` — including `List<T>` and `Dictionary<TKey,TValue>`. This chapter does not cover LINQ; it is introduced in a later chapter.

## Summary


Data collections organize groups of values. Use arrays when the size is fixed, lists when an ordered sequence can grow, and dictionaries when each value should be found by a key. Other collection families are useful too, but this chapter keeps the main path on lists and dictionaries.
