# Choosing Data Collections


```{contents}
:local:
:depth: 2
```


Arrays are useful when a program has a fixed-size sequence of values. Collections are library types that manage groups of related values when the program needs more flexibility: adding items, removing items, looking up items, or enforcing a particular access pattern.

C# collection types are classes. That means collection variables are reference variables: two variables can refer to the same collection object, and a change through either variable changes that one shared object. This is different from simple value types such as `int`, `double`, and `bool`, where each variable stores its own value.

This chapter focuses on the two collection types introductory C# programmers use most often: `List<T>` for ordered sequences and `Dictionary<TKey,TValue>` for key/value lookup.

## Collection Families


A collection choice usually starts with the way the program needs to retrieve data. Use this table as a map for the collection types you will see in this chapter and later chapters.

| Need | Common type | Where it belongs |
|---|---|---|
| Fixed-size sequential values | array | Chapter 7 |
| Variable-size sequential values | `List<T>` | This chapter |
| Key/value lookup | `Dictionary<TKey,TValue>` | This chapter |
| Unique membership tests | `HashSet<T>` | Applied extension |
| Last-in-first-out or first-in-first-out access | `Stack<T>` or `Queue<T>` | Chapter 15 |

Lists and dictionaries cover many everyday programming tasks. Stacks, queues, sets, trees, and graphs become more important when the access rule or performance requirement matters as much as the stored values.

## Generic Collections


Most modern C# collection types are **generic**. The type parameter in angle brackets tells C# what kind of element the collection stores.

- `List<string>` stores strings.
- `List<int>` stores integers.
- `Dictionary<string, int>` maps string keys to integer values.

Prefer the generic types in `System.Collections.Generic`. Older non-generic collections store values as `object`, which loses helpful compile-time type checking.


````csharp
using System;
using System.Collections.Generic;

List<string> names = new List<string> { "Ana", "Bo" };
Dictionary<string, int> scores = new Dictionary<string, int>
{
    ["Ana"] = 92,
    ["Bo"] = 85
};

Console.WriteLine(names[0]);
Console.WriteLine(scores["Bo"]);
````

## Summary


Data collections organize groups of values. Use arrays when the size is fixed, lists when an ordered sequence can grow, and dictionaries when each value should be found by a key. Other collection families are useful too, but this chapter keeps the main path on lists and dictionaries.
