## Multiple Choice

**1.** What does contiguous storage mean?

a) Elements are stored in linked node objects  
b) Elements are stored in neighboring indexed slots  
c) Elements are stored only in dictionaries  
d) Elements are stored without memory

```{dropdown} Answer
**b)** Arrays store elements in adjacent indexed slots.
```

**2.** What is capacity in a dynamic array?

a) The number of slots currently available in the internal storage  
b) The number of linked references in a node  
c) The value stored at index `0`  
d) The number of methods in a class

```{dropdown} Answer
**a)** Capacity is the number of slots allocated, which may be larger than `Count`.
```

**3.** What happens when a dynamic array is full and a new item is appended?

a) The program must always crash  
b) The array automatically becomes a linked list  
c) Larger storage is allocated and existing values are copied  
d) Existing values are deleted

```{dropdown} Answer
**c)** Resizing allocates larger storage and copies existing values.
```

**4.** What does amortized O(1) append mean for a dynamic array?

a) Every append copies every element  
b) Most appends are cheap, while occasional resize costs are spread over many appends  
c) Appending is impossible  
d) Appending always requires linked traversal

```{dropdown} Answer
**b)** Occasional expensive resizes are averaged across many cheap appends.
```

**5.** What does a linked-list node usually contain?

a) A value and a reference to another node  
b) Only a compiler warning  
c) A database table and a SQL query  
d) A method name and no data

```{dropdown} Answer
**a)** A singly linked node stores a value and a `Next` reference.
```

**6.** What is traversal?

a) Following references from node to node  
b) Doubling an array's capacity  
c) Hiding a private field  
d) Compiling a C# project

```{dropdown} Answer
**a)** Traversal visits nodes by following `Next` references.
```

**7.** Which operation is O(1) for an array but O(n) for a singly linked list?

a) Direct indexed access  
b) Reading the first node  
c) Checking whether a class compiles  
d) Creating a variable

```{dropdown} Answer
**a)** Arrays can jump to an index; singly linked lists must traverse.
```

**8.** Why can middle insertion be expensive in an array?

a) Later elements must shift to make room  
b) Arrays cannot store numbers  
c) Every element must become a node  
d) The compiler removes the array

```{dropdown} Answer
**a)** Inserting in the middle requires shifting later elements.
```

**9.** When is linked-list insertion after a node O(1)?

a) When you already have a reference to the previous node  
b) When the node is stored in a database  
c) Only when the list is sorted  
d) Never

```{dropdown} Answer
**a)** If the node is already known, insertion changes only a few references.
```

**10.** What should guide the choice between an array-backed sequence and a linked list?

a) The workload's most common operations  
b) The alphabetic order of class names  
c) Whether the code has comments  
d) The length of the file name

```{dropdown} Answer
**a)** Representation choice depends on the operations a program performs most often.
```
