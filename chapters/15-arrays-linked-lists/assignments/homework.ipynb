````csharp
// Solution
// 1. Count storage writes during growth.
// Complete the loop: count copied entries only when capacity doubles,
// and count one new-entry write per append. Predict totals for nine appends.
using System;
int capacity = 2;
int count = 0;
int copied = 0;
int writes = 0;
for (int item = 1; item <= 9; item++)
{
    if (count == capacity) { copied += count; writes += count; capacity *= 2; }
    count++;
    writes++;
}
Console.WriteLine($"count={count}, capacity={capacity}, copied={copied}, writes={writes}");
// Then test two and eight appends; explain why amortized cost is not
// a promise that every individual append is cheap.
````
