# Linear Lists

This chapter covers array-backed lists, dynamic arrays, linked nodes, linked lists, and the operation-cost tradeoffs between contiguous and linked sequence representations. The sidebar title **Linear Lists** names the shared idea: data arranged in a single sequence.

- **Contiguous storage**: Array-backed sequences keep elements next to each other for fast indexed access.
- **Dynamic array**: A dynamic array grows by allocating larger storage and copying existing values.
- **Linked node**: A linked structure stores values in nodes connected by references.
- **Insertion and deletion**: Different representations make middle updates cheaper or more expensive.
- **Performance tradeoff**: Choosing a sequence representation depends on access, growth, and update patterns.

Arrays and linked lists make the connection between memory representation and program behavior visible.


> *[Data Structures Easy to Advanced Course - Full Tutorial from a Google Engineer](https://youtube.com/watch?v=RBSGKlAvoiM)* by freeCodeCamp.org.
>
> This video remains hosted on YouTube and is the property of its original creator or rightsholder; it is embedded here via YouTube's official player for educational reference only.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Explain how array-based and linked representations store sequence data
2. Trace resizing, shifting, and the total copy cost of repeated appends
3. Implement generic array-backed and singly linked sequences, preserving count, head, and tail invariants
4. Compare indexed access, search, insertion, and deletion costs
5. Test empty, singleton, duplicate, and invalid-index cases and justify a representation for a specific workload
6. Select linear or binary search and prepare a sorted report with explicit mutation and tie policies.


<h2>Chapter Flow</h2>

- {doc}`15.1. Dynamic Arrays <1501-dynamic-arrays>`
- {doc}`15.2. Linked Nodes <1502-linked-nodes>`
- {doc}`15.3. Representation Comparison Lab <1503-comparison-lab>`
- {doc}`15.4. Searching and Sorting Sequences <1504_searching_sorting>`


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Linear list | A sequence of items arranged in order. |
| 2 | Array representation | A sequence representation that stores items in contiguous indexed positions. |
| 3 | Linked representation | A sequence representation where nodes refer to neighboring nodes. |
| 4 | Dynamic array | An array-backed list that resizes when it needs more capacity. |
| 5 | Capacity | The number of items an underlying array can currently hold. |
| 6 | Resize | Allocating a larger array and copying items into it. |
| 7 | Node | An object that stores one value and one or more links. |
| 8 | Link | A reference from one node to another. |
| 9 | Head | The first node in a linked list. |
| 10 | Tail | The last node in a linked list. |
| 11 | Traversal | Visiting list items from one position to the next. |
| 12 | Insertion | Adding a new item to a sequence. |
| 13 | Deletion | Removing an item from a sequence. |
| 14 | Indexed access | Retrieving an item by numeric position. |
| 15 | Search cost | The work needed to find a value in a sequence. |
| 16 | Amortized cost | Total work over a sequence of operations divided by its length. |
| 17 | Invariant | A representation rule preserved by every completed operation. |
| 18 | Memory locality | Accessing storage locations close together, often helping sequential access. |
| 19 | Linear search | Scan records until a matching item is found or input ends. |
| 20 | Binary search | Repeatedly narrow a sorted random-access interval. |
| 21 | Stable sorting | Preserve arrival order among items with equal comparison keys. |
