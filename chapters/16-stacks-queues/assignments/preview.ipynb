## Multiple Choice

**1.** Which access rule describes a stack?

a) First in, first out  
b) Last in, first out  
c) Random access by index  
d) Sorted access by key

**2.** Which access rule describes a queue?

a) Last in, first out  
b) First in, first out  
c) Smallest value first  
d) Largest value first

**3.** What does `Push` do on a stack?

a) Removes the top item  
b) Adds an item to the top  
c) Removes the front item  
d) Sorts the stored items

**4.** What does `Dequeue` do on a queue?

a) Adds an item to the back  
b) Removes and returns the front item  
c) Removes and returns the newest item  
d) Checks whether two items match

**5.** What does `Peek` usually mean?

a) Look at the next item without removing it  
b) Remove every item  
c) Add a duplicate item  
d) Reverse the collection

**6.** What is a deque?

a) A queue with two fronts only  
b) A double-ended queue  
c) A dictionary with ordered keys  
d) A stack that cannot be empty

**7.** Which structure best models undo history?

a) Stack  
b) Queue  
c) Sorted set  
d) Hash table

**8.** Which structure best models a fair service line?

a) Stack  
b) Queue  
c) Binary tree  
d) Record

**9.** What does a ring buffer help implement efficiently?

a) A queue backed by an array  
b) A class constructor  
c) A nullable reference check  
d) A recursive method

**10.** What is underflow in this chapter's context?

a) Adding too many items to a bounded structure  
b) Removing or peeking when the structure is empty  
c) Sorting values in descending order  
d) Storing duplicate values
