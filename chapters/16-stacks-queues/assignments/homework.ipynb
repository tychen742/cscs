```csharp
// 8. Peek safely
// Print the next customer only if the queue is not empty.

using System;
using System.Collections.Generic;

var customers = new Queue<string>();
customers.Enqueue("Amina");

// Your code starts here.



// Your code ends here.
```

```csharp
// Solution
using System;
using System.Collections.Generic;

var customers = new Queue<string>();
customers.Enqueue("Amina");

if (customers.Count > 0)
{
    Console.WriteLine(customers.Peek());
}
else
{
    Console.WriteLine("No customers waiting.");
}
```
