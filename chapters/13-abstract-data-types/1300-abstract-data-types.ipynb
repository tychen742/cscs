<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Abstract data type | A data type described by its values and operations rather than its implementation. |
| 2 | Contract | The promised behavior an ADT operation provides. |
| 3 | Implementation | The concrete code and representation used to realize an ADT. |
| 4 | Representation | The internal data layout used to store an ADT. |
| 5 | Operation | An action an ADT supports, such as add, remove, or lookup. |
| 6 | Invariant | A rule that must remain true for a representation to be valid. |
| 7 | Precondition | A requirement that must hold before an operation is called. |
| 8 | Postcondition | A result that should hold after an operation completes. |
| 9 | Interface | A named contract that types can implement. |
| 10 | Generic | Reusable code written with type parameters. |
| 11 | Type parameter | A placeholder for a type used by generic code. |
| 12 | Cost | The time or memory needed by an operation. |
| 13 | Client code | Code that uses an ADT through its public operations. |
| 14 | Information hiding | Keeping representation details separate from code that uses the ADT. |
| 15 | Iterator | A method or object that produces a sequence of values one at a time. |
| 16 | `yield return` | A statement that produces one element of an iterator and pauses until the next is requested. |
| 17 | Lazy evaluation | Computing a value only when it is actually requested, not in advance. |
