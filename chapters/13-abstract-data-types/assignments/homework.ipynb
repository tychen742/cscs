````csharp
// 3. Trace this linked structure by hand before running it.
// What output do you expect?

class Node<T>
{
    public T Value { get; }
    public Node<T>? Next { get; }
    public Node(T value, Node<T>? next)
    {
        Value = value;
        Next = next;
    }
}

Node<int>? top = null;
top = new Node<int>(1, top);
top = new Node<int>(2, top);
top = new Node<int>(3, top);

for (Node<int>? current = top; current is not null; current = current.Next)
{
    Console.WriteLine(current.Value);
}
````

Expected output:

```text
3
2
1
```

Each new node becomes the new `top`, so traversal visits the most recently added value first.

````csharp
static T First<T>(List<T> items)
{
    if (items.Count == 0) throw new InvalidOperationException("List is empty.");
    return items[0];
}

Console.WriteLine(First(new List<int> { 8, 13, 21 }));
````
