## Multiple Choice

**1.** What does an abstract data type specify first?

a) The exact memory address of each item  
b) The behavior and operations the type promises  
c) The IDE used to write the code  
d) The file where the type is stored

```{dropdown} Answer
**b)** An ADT starts with values and operations, not storage details.
```

**2.** In an ADT discussion, what is a representation?

a) The hidden storage strategy used by an implementation  
b) The public name of a method  
c) The result printed by `Console.WriteLine`  
d) The project folder containing source files

```{dropdown} Answer
**a)** A representation is how the data is stored internally.
```

**3.** Why is an interface useful for expressing an ADT in C#?

a) It stores all object data in one array  
b) It names required operations without choosing one implementation  
c) It prevents all runtime exceptions  
d) It automatically sorts every collection

```{dropdown} Answer
**b)** An interface names the contract while implementations provide storage and algorithms.
```

**4.** Which statement best describes an invariant?

a) A rule that must remain true for an object to be valid  
b) A variable that can never be read  
c) A method that always returns `void`  
d) A collection that can hold only strings

```{dropdown} Answer
**a)** An invariant is a rule the implementation must preserve.
```

**5.** Which operation belongs naturally to a stack ADT?

a) `Push`  
b) `Join`  
c) `Compile`  
d) `Download`

```{dropdown} Answer
**a)** `Push` adds an item to the top of a stack.
```

**6.** What does representation independence let client code do?

a) Depend on behavior instead of hidden storage details  
b) Skip all tests  
c) Store strings inside `int` variables  
d) Run without compiling

```{dropdown} Answer
**a)** Client code can use the contract even if the implementation changes.
```

**7.** In a linked-node structure, what does a reference connect?

a) One object to another object  
b) A program to a textbook page  
c) A namespace to a file extension  
d) A compiler warning to a comment

```{dropdown} Answer
**a)** Linked structures use references to connect nodes.
```

**8.** What does `T` usually represent in a generic type such as `IStack<T>`?

a) The element type supplied by the caller  
b) The number of items currently stored  
c) The time complexity of every operation  
d) The table name in a database

```{dropdown} Answer
**a)** `T` is a type parameter for the element type.
```

**9.** What does a type constraint do?

a) Limits which types can be used with a generic definition  
b) Deletes items from a collection  
c) Changes a reference type into a value type  
d) Hides all public methods

```{dropdown} Answer
**a)** A constraint requires the type argument to have some capability or category.
```

**10.** Why do data-structure chapters compare operation costs?

a) Different representations make different operations fast or slow  
b) Every representation has exactly the same performance  
c) Cost comparisons replace correctness  
d) C# requires every method to run in O(1)

```{dropdown} Answer
**a)** Choosing a representation means choosing tradeoffs.
```
