```csharp
using System;
using System.IO;

File.WriteAllText("sales.txt", "120\n85\nN/A\n\n42\n4O\n230\n");

static int SumSales(string fileName, out int skipped)
{
    skipped = 0;
    int total = 0;

    if (!File.Exists(fileName))
    {
        return -1;
    }

    var reader = new StreamReader(fileName);
    while (!reader.EndOfStream)
    {
        string line = reader.ReadLine()!.Trim();
        if (line.Length == 0)
        {
            continue;
        }
        if (int.TryParse(line, out int amount))
        {
            total += amount;
        }
        else
        {
            skipped++;
        }
    }
    reader.Close();

    return total;
}

foreach (string fileName in new[] { "sales.txt", "sales_old.txt" })
{
    int total = SumSales(fileName, out int skipped);
    if (total < 0)
    {
        Console.WriteLine($"{fileName}: file not found");
    }
    else
    {
        Console.WriteLine($"{fileName}: total ${total}, skipped {skipped} bad line(s)");
    }
}
```

```csharp
using System;
using System.IO;

File.WriteAllText("orders.txt",
    "Latte,3,5\n" +
    "Muffin, 2 ,3\n" +
    "Espresso,two,4\n" +
    "\n" +
    "Bagel,4\n" +
    "Mocha,1,6\n");

int revenue = 0;
var reader = new StreamReader("orders.txt");
while (!reader.EndOfStream)
{
    string line = reader.ReadLine()!;

    if (string.IsNullOrWhiteSpace(line))
    {
        continue;
    }

    string[] fields = line.Split(',');
    if (fields.Length != 3
        || !int.TryParse(fields[1].Trim(), out int quantity)
        || !int.TryParse(fields[2].Trim(), out int price))
    {
        Console.WriteLine($"Skipped: {line}");
        continue;
    }

    string product = fields[0].Trim();
    int amount = quantity * price;
    revenue += amount;
    Console.WriteLine($"{product}: {quantity} x ${price} = ${amount}");
}
reader.Close();

Console.WriteLine($"Revenue: ${revenue}");
```

```csharp
using System;
using System.IO;

File.WriteAllText("orders.txt",
    "Latte,3,5\n" +
    "Muffin, 2 ,3\n" +
    "Espresso,two,4\n" +
    "Mocha,1,6\n");

var reader = new StreamReader("orders.txt");
var writer = new StreamWriter("report.txt");
int revenue = 0;

while (!reader.EndOfStream)
{
    string[] fields = reader.ReadLine()!.Split(',');
    if (fields.Length == 3
        && int.TryParse(fields[1].Trim(), out int quantity)
        && int.TryParse(fields[2].Trim(), out int price))
    {
        int amount = quantity * price;
        revenue += amount;
        writer.WriteLine($"{fields[0].Trim().ToUpper()},{quantity},{amount}");
    }
}
writer.WriteLine($"TOTAL,,{revenue}");
reader.Close();
writer.Close();

var reportReader = new StreamReader("report.txt");
Console.Write(reportReader.ReadToEnd());
reportReader.Close();
```
