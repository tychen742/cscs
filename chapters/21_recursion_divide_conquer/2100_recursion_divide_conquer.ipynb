# Recursion

This chapter covers recursive methods, base cases, recursive cases, progress measures, call-stack tracing, divide-and-conquer design, binary search as a halving example, merge, and merge sort. The sidebar title **Recursion** names the core mechanism; divide-and-conquer is the main algorithm-design pattern built from it.

- **Base case**: The condition that stops recursion with a direct answer.
- **Recursive case**: The part of the method that calls itself on a smaller input.
- **Progress measure**: A quantity that moves toward the base case on every recursive call.
- **Call stack**: The runtime structure that tracks active method calls and their local variables.
- **Divide-and-conquer**: A design pattern that divides a problem, solves subproblems, and combines results.

Recursion is ordinary method calls plus the call stack. Divide-and-conquer uses that mechanism to split a problem, solve smaller parts, and combine the answers.


> *[Algorithms and Data Structures Tutorial - Full Course for Beginners](https://youtube.com/watch?v=8hly31xKli0)* by freeCodeCamp.org.
>
> This video remains hosted on YouTube and is the property of its original creator or rightsholder; it is embedded here via YouTube's official player for educational reference only.


<h2>Learning Objectives</h2>

1. State recursive input domains, base cases, helper contracts, and well-founded progress measures.
2. Trace separate call frames and returned results, including arithmetic and stack limits.
3. Prove range splits preserve all items and distinguish one-child search from two-child division.
4. Implement stable merge and merge sort with explicit tie and copy policies.
5. Verify order, occurrence preservation, input identity, and stable record order across boundary cases.
6. Relate actual comparisons, writes, calls, and active depth to time and auxiliary-space bounds.


<h2>Chapter Flow</h2>

- {doc}`21.1. Recursive Problem Solving <2101_recursion>`
- {doc}`21.2. Divide-and-Conquer Design <2102_divide_conquer>`
- {doc}`21.3. Merge Sort Lab <2103_merge_sort_lab>`


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Recursion | A technique where a method solves a problem by calling itself on a smaller input. |
| 2 | Base case | A directly solved case that stops recursive calls. |
| 3 | Recursive case | The branch that makes one or more recursive calls. |
| 4 | Progress measure | A value, such as `n` or a range length, that gets closer to the base case with each call. |
| 5 | Call stack | The runtime stack of active method calls. |
| 6 | Stack frame | One active method call and its local state on the call stack. |
| 7 | Divide-and-conquer | An algorithm design pattern that splits a problem into subproblems, solves them, and combines the answers. |
| 8 | Subproblem | A smaller version of the original problem. |
| 9 | Combine step | The step that merges subproblem results into a final answer. |
| 10 | Merge | The process of combining two sorted sequences into one sorted sequence. |
| 11 | Merge sort | A divide-and-conquer sorting algorithm that recursively sorts halves and merges them. |

| 12 | Half-open range | A range [start,end) including start but excluding end. |
| 13 | Helper contract | The promise made by a recursive subproblem method. |
| 14 | Inductive reasoning | A direct base result plus a correct combination of smaller correct results. |
| 15 | Stable sort | A sort preserving arrival order among equal comparison keys. |
| 16 | Occurrence preservation | Keeping every input item with its original multiplicity. |
| 17 | Shared merge buffer | A reusable temporary array used by successive merges. |
| 18 | Shallow array copy | A new array whose element references or values are copied without cloning referenced objects. |
| 19 | Peak active frames | Maximum calls simultaneously retained on the stack, including the base frame here. |


<h2>Slides</h2>

Slides for this chapter will be added after the notebook content stabilizes.
