## Multiple Choice


**1.** What ensures a recursive path reaches a base case?

a) A decreasing well-founded measure  
b) A method name  
c) A print statement  
d) A larger stack  

**2.** What remains in a suspended factorial frame?

a) Its own input and work after the recursive call  
b) All future inputs in one variable  
c) Only the final answer  
d) Nothing  

**3.** What is 0! under the factorial contract?

a) 0  
b) 1  
c) −1  
d) Undefined for this method  

**4.** What is the length of [start,end)?

a) end−start  
b) end+start  
c) end−start+1  
d) start−end  

**5.** Why exclude the midpoint from binary search’s right range?

a) It was already checked and progress must continue  
b) To skip equal keys entirely  
c) To allocate more storage  
d) To make the input sorted  

**6.** What should an empty maximum input do under this chapter’s contract?

a) Be rejected  
b) Return zero for every type  
c) Recurse forever  
d) Return the array  

**7.** What does stable sorting preserve?

a) Arrival order among equal keys  
b) All keys at their original indexes  
c) Only unique records  
d) The call stack  

**8.** Which condition establishes item preservation in a sort test?

a) Equality with an independently sorted sequence including duplicates  
b) Sortedness alone  
c) An empty result  
d) A printed success message only  

**9.** What is range-buffer merge sort’s auxiliary space?

a) Θ(1)  
b) Θ(log n) only  
c) Θ(n)  
d) Θ(n²)  

**10.** Which is an actual operation count rather than a growth estimate?

a) A counter incremented on each key comparison  
b) n log₂ n without inspecting code  
c) One elapsed time labeled Big-O  
d) The number of namespaces
