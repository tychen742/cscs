## Schedule Compatible Meetings

Use earliest finish time to choose a maximum-size set of compatible meetings.


````csharp
using System;
using System.Collections.Generic;
using System.Linq;

var meetings = new[]
{
    new Meeting("A", 1, 4),
    new Meeting("B", 3, 5),
    new Meeting("C", 5, 7),
    new Meeting("D", 0, 6),
    new Meeting("E", 8, 9)
};

var selected = new List<Meeting>();
int lastEnd = int.MinValue;
foreach (Meeting meeting in meetings.OrderBy(m => m.End))
{
    if (meeting.Start >= lastEnd)
    {
        selected.Add(meeting);
        lastEnd = meeting.End;
    }
}

Console.WriteLine(string.Join(", ", selected.Select(m => m.Name)));

record Meeting(string Name, int Start, int End);
````

## Build a Low-Cost Network

Apply Kruskal's algorithm to a small office network.


````csharp
using System;
using System.Collections.Generic;
using System.Linq;

var vertices = new[] { "HQ", "Lab", "Shop", "Depot" };
var edges = new[]
{
    new Edge("HQ", "Lab", 6),
    new Edge("HQ", "Shop", 2),
    new Edge("Lab", "Shop", 3),
    new Edge("Lab", "Depot", 4),
    new Edge("Shop", "Depot", 5)
};

var parent = vertices.ToDictionary(v => v, v => v);
var chosen = new List<Edge>();
foreach (Edge edge in edges.OrderBy(e => e.Cost))
{
    if (Find(edge.From) != Find(edge.To))
    {
        chosen.Add(edge);
        parent[Find(edge.From)] = Find(edge.To);
    }
}

Console.WriteLine($"cost = {chosen.Sum(e => e.Cost)}");
Console.WriteLine(string.Join(", ", chosen.Select(e => $"{e.From}-{e.To}")));

string Find(string vertex)
{
    while (parent[vertex] != vertex) vertex = parent[vertex];
    return vertex;
}

record Edge(string From, string To, int Cost);
````

## Find Delivery Distances

Use Dijkstra's algorithm to compute distances from a warehouse.


````csharp
using System;
using System.Collections.Generic;
using System.Linq;

var graph = new Dictionary<string, List<(string To, int Cost)>>
{
    ["Warehouse"] = new() { ("A", 3), ("B", 8) },
    ["A"] = new() { ("B", 2), ("C", 5) },
    ["B"] = new() { ("C", 1) },
    ["C"] = new()
};

var distance = graph.Keys.ToDictionary(v => v, _ => int.MaxValue);
var queue = new PriorityQueue<string, int>();
distance["Warehouse"] = 0;
queue.Enqueue("Warehouse", 0);

while (queue.Count > 0)
{
    string current = queue.Dequeue();
    foreach ((string to, int cost) in graph[current])
    {
        int candidate = distance[current] + cost;
        if (candidate < distance[to])
        {
            distance[to] = candidate;
            queue.Enqueue(to, candidate);
        }
    }
}

foreach ((string vertex, int cost) in distance.OrderBy(pair => pair.Key))
{
    Console.WriteLine($"{vertex}: {cost}");
}
````
