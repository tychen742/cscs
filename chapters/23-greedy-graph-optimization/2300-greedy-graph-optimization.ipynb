# Greedy Algorithms

This chapter covers greedy choice rules, feasibility, invariants, counterexamples, exchange arguments, minimum spanning trees, Kruskal's algorithm, shortest paths, and Dijkstra's algorithm. The sidebar title **Greedy Algorithms** names the design strategy; graph optimization provides the main application setting.

- **Greedy choice**: A locally best option selected without revisiting earlier choices.
- **Feasible solution**: A partial or complete solution that satisfies the problem constraints.
- **Invariant**: A property that remains true after each greedy step.
- **Minimum spanning tree**: A lowest-cost set of edges that connects every vertex in an undirected weighted graph without cycles.
- **Shortest path**: A lowest-cost route from a source vertex to another vertex under specific edge-weight assumptions.

Greedy algorithms are powerful when the local rule matches the global goal. This chapter studies how to recognize when that is true and when it is a trap.


> *[Graph Theory Algorithms](https://youtube.com/watch?v=09_LlHjoEiY)* by freeCodeCamp.org.
>
> This video remains hosted on YouTube and is the property of its original creator or rightsholder; it is embedded here via YouTube's official player for educational reference only.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:

1. Specify an optimization objective and distinguish feasibility from optimality.
2. Prove maximum-count interval scheduling with an exchange argument and test changed objectives with counterexamples.
3. Trace Kruskal’s union-find invariant and report a minimum spanning forest for disconnected inputs.
4. Explain Dijkstra’s settling argument, reject negative arcs, and skip stale queue entries.
5. Reconstruct reachable paths, represent unreachable vertices explicitly, and verify costs without integer overflow.
6. Choose between network cost and source-to-destination distance using independent small-instance checks.


<h2>Chapter Flow</h2>

- {doc}`23.1. Greedy Choice and Correctness <2301-greedy-choice>`
- {doc}`23.2. Spanning Trees and Shortest Paths <2302-spanning-trees-shortest-paths>`
- {doc}`23.3. Greedy Optimization Lab <2303-optimization-lab>`


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Greedy algorithm | An algorithm that builds a solution by repeatedly choosing the locally best available option. |
| 2 | Greedy choice | The local decision made at one step of a greedy algorithm. |
| 3 | Feasible solution | A solution, or partial solution, that does not violate the problem constraints. |
| 4 | Objective function | The quantity an optimization algorithm tries to minimize or maximize. |
| 5 | Invariant | A fact that remains true after each step of an algorithm. |
| 6 | Exchange argument | A proof technique that shows an optimal solution can be adjusted to include the greedy choice. |
| 7 | Minimum spanning tree | A tree that connects all vertices in an undirected weighted graph with minimum total edge weight. |
| 8 | Kruskal algorithm | A greedy minimum-spanning-tree algorithm that adds safe edges in increasing weight order. |
| 9 | Shortest path | A path with minimum total edge weight between vertices. |
| 10 | Dijkstra algorithm | A greedy shortest-path algorithm for graphs with nonnegative edge weights. |

| 11 | Minimum spanning forest | A minimum-cost tree in each connected component, including isolated vertices. |
| 12 | Union-find | A component representation supporting root lookup and component merging. |
| 13 | Relaxation | Improving a tentative distance through an edge from a reached vertex. |
| 14 | Stale queue entry | An older priority that no longer equals the vertex’s best known distance. |


<h2>Slides</h2>

Slides for this chapter will be added after the notebook content stabilizes.
