# Part IV. Data Structures

Part IV asks how data should be organized so that a program can use it efficiently. You will build the standard structures yourself and learn the trade-offs between them.

<h2>Conceptual Arc</h2>

1. Start with abstract data types and the contracts that separate what a structure does from how it works.
2. Build linear structures: dynamic arrays, linked lists, stacks, and queues.
3. Move to non-linear structures: trees, heaps, hash tables, and graphs.
