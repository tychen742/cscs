# Part IV. Data Structures


Part IV turns object-oriented programming into a study of contracts, representations, costs, and workload choices. It is the foundation for the second course in the sequence.


<h2>Conceptual Arc</h2>

Begin with ADTs and operation costs, then build linear structures. Study basic searching and sorting before comparing trees, heaps, hash collections, and graphs. Traversal completes the foundation; optimization continues in Part V.

<h2>Course Routes</h2>

Parts I–III (Chapters 1–13) form a complete CS1 route. Part IV (Chapters 14–21) supplies the CS2 data-structures foundation. Instructors can select Part V (Chapters 22–27) for additional algorithmic depth according to course goals and available time. The book supports COB, IST, IT, IS, and LIS programs as well as other programming-foundations courses; a standalone CS1 course needs no commitment to a second semester.
