# Part I. Fundamentals

Part I builds the core of programming in C#: how a program is written, run, and debugged, and how it makes decisions and repeats work. It introduces professional responsibility in Chapter 1 and revisits it through technical practice. Every later part assumes these skills.

<h2>Conceptual Arc</h2>

1. Start with computing, responsibility, and a first program, then store and transform data with variables and operators.
2. Organize code into methods, then control the flow with conditionals and loops.
3. Make programs reliable with exceptions, debugging, and testing, and revisit the consequences of failures.
