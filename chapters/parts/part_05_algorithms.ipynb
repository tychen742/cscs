# Part V. Algorithms

Parts I-III (Chapters 1-13) form a complete CS1 route. Part IV (Chapters 14-21) supplies the CS2 data-structures foundation. Instructors can select Part V (Chapters 22-27) for additional algorithmic depth according to course goals and available time.

Part V turns to the methods that operate on data structures: how to measure an algorithm's cost, and the main design strategies for solving problems efficiently.

<h2>Conceptual Arc</h2>

1. Analyze algorithms with Big-O notation, then solve problems by recursion and divide and conquer.
2. Search and sort data, and compare the cost of different approaches.
3. Apply greedy, dynamic programming, and backtracking strategies, and meet the limits of what computers can solve efficiently.
