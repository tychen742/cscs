# Part II. Data and I/O

Part II moves from single values to collections of data and to data that lives outside the program: in files, text, and databases. Business software spends most of its time on these tasks.

<h2>Conceptual Arc</h2>

1. Hold many values in arrays and then in flexible collections such as lists and dictionaries.
2. Read, write, and search files and text, including pattern matching with regular expressions.
3. Store and query structured data in a relational database from C#.
