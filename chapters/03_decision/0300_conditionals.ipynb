# Decisions

A business rule becomes useful only when a program chooses the right action for each input. This chapter develops Boolean conditions, alternative branches, and multiway decisions through familiar eligibility and classification problems. You will trace the selected path and check the boundaries where the answer changes.

- **Boolean expressions**: Conditions evaluate to `true` or `false` and drive program decisions.
- **`if` statements**: An `if` statement runs code only when its condition is true.
- **Multi-branch logic**: `else if`, `else`, and `switch` handle several possible cases.
- **Logical operators**: Operators such as `&&`, `||`, and `!` combine or reverse conditions.

<h2>Learning Objectives</h2>


By the end of this chapter, you will be able to:

1. Write `if`, `else if`, and `else` chains correctly.
2. Construct Boolean expressions with comparison and logical operators.
3. Handle multiple exclusive cases with a `switch` statement.
4. Trace which branch of a conditional statement will execute.
5. Avoid common conditional mistakes such as dangling `else` and boundary errors.
