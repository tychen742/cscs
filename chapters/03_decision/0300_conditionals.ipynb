# Conditionals

Programs become useful when they can make decisions. This chapter introduces the C# constructs that choose different code paths: `if`, `else if`, `else`, `switch`, and the Boolean expressions that control them.

- **Boolean expressions**: Conditions evaluate to `true` or `false` and drive program decisions.
- **`if` statements**: An `if` statement runs code only when its condition is true.
- **Multi-branch logic**: `else if`, `else`, and `switch` handle several possible cases.
- **Logical operators**: Operators such as `&&`, `||`, and `!` combine or reverse conditions.

Conditionals let programs validate input, express business rules, and respond differently as data changes.
