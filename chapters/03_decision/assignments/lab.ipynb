```csharp
using System;

string employee = "Avery";
char department = 'P';
double hours = 46;
double rating = 4.6;
string departmentName = "";
double rate = 0;

// Your code starts here.


// Your code ends here.
```

```csharp
using System;

string employee = "Avery";
char department = 'P';
double hours = 46;
double rating = 4.6;
string departmentName = "";
double rate = 0;
switch (department)
{
    case 'P': departmentName = "Printing"; rate = 20; break;
    case 'D': departmentName = "Delivery"; rate = 18.50; break;
    case 'F': departmentName = "Front Desk"; rate = 16; break;
    default: Console.WriteLine("Unknown department code."); break;
}
if (rate > 0)
{
    double weeklyPay;
    if (hours <= 40)
    {
        weeklyPay = hours * rate;
    }
    else
    {
        weeklyPay = 40 * rate + (hours - 40) * rate * 1.5;
    }
    double percent;
    if (rating >= 4.5) percent = 5;
    else if (rating >= 3.5) percent = 3;
    else if (rating >= 2.5) percent = 1;
    else percent = 0;
    double bonus = weeklyPay * percent / 100;
    Console.WriteLine($"Employee:   {employee}");
    Console.WriteLine($"Department: {departmentName} (${rate:F2}/hour)");
    Console.WriteLine($"Hours:      {hours}");
    Console.WriteLine($"Gross pay:  ${weeklyPay:F2}");
    Console.WriteLine($"Bonus:      ${bonus:F2}");
    Console.WriteLine($"Total pay:  ${weeklyPay + bonus:F2}");
}
```
