**Question 9: Select a service**

Use switch: P means Pickup with zero charge; D means Delivery with charge 12. Unsupported codes print Unknown service and no charge line. Do not use charge == 0 to recognize invalid input: Pickup is valid and free. Check P, D, and X.


````csharp
// Exercise: Select a service
// exercise-id: cscs-decision-homework-coding-4
using System;

string service = "P";
// Your code starts here.
// Implement the contract above and print the requested report.
// Your code ends here.
````

````csharp
using System;

string service = "P";
bool known = true;
string label = "";
decimal charge = 0m;
switch (service)
{
    case "P": label = "Pickup"; charge = 0m; break;
    case "D": label = "Delivery"; charge = 12m; break;
    default: known = false; break;
}
if (known)
{
    Console.WriteLine(label);
    Console.WriteLine($"Charge: {charge:F2}");
}
else Console.WriteLine("Unknown service");
````

**Question 10: Repair a discount boundary**

The intended rule is 10% off at 20 units or more; otherwise no discount. Correct the supplied condition before calculating discount and amount due with F2. Predict 19, 20, and 21 before running. Keep the quantity and subtotal as independent inputs.


````csharp
// Exercise: Repair a discount boundary
// exercise-id: cscs-decision-homework-coding-5
using System;

int quantity = 20;
decimal subtotal = 200m;
// Faulty rule to repair: if (quantity > 20) apply the discount.
// Your code starts here.
// Implement the contract above and print the requested report.
// Your code ends here.
````

````csharp
using System;

int quantity = 20;
decimal subtotal = 200m;
// Faulty rule to repair: if (quantity > 20) apply the discount.
decimal discount = 0m;
if (quantity >= 20) discount = subtotal * 0.10m;
Console.WriteLine($"Discount: {discount:F2}");
Console.WriteLine($"Amount due: {subtotal - discount:F2}");
````
