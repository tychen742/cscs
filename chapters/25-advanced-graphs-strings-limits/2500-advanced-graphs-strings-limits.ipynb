# Advanced Algorithms

This chapter covers graph traversal, breadth-first search, depth-first search, path assumptions, string matching, prefix-aware matching ideas, computational limits, pruning, approximation, and final algorithm selection. The sidebar title **Advanced Algorithms** signals a synthesis chapter rather than a new catalog of unrelated topics.

- **Graph traversal**: Breadth-first and depth-first search explore graph structure in different orders.
- **Reachability**: A question about whether one state, vertex, or item can be reached from another.
- **String matching**: Searching for a pattern inside a larger text.
- **Computational limit**: A point where exact brute-force computation grows too quickly to be practical.
- **Approximation**: A method with a proved quality bound relative to the optimum under stated assumptions.

The goal is not to memorize every advanced algorithm, but to recognize how problem structure shapes what efficient computation can do.


> *[Graph Theory Algorithms](https://youtube.com/watch?v=09_LlHjoEiY)* by freeCodeCamp.org.
>
> This video remains hosted on YouTube and is the property of its original creator or rightsholder; it is embedded here via YouTube's official player for educational reference only.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:

1. Explain BFS discovery and layer invariants and reconstruct valid minimum-hop paths.
2. Trace DFS entry/exit with explicit frames and enforce undirected-component preconditions.
3. Implement direct and prepared-prefix matching under explicit ordinal, overlap, and empty-pattern contracts.
4. Justify prefix fallback and linear comparison bounds, including preparation and output costs.
5. Validate exact and heuristic coverage witnesses and distinguish feasibility, optimality, and proved approximation guarantees.
6. Compare complete workloads and exact growth models without confusing measurements with general lower-bound proofs.


<h2>Chapter Flow</h2>

- {doc}`25.1. Graph Traversal <2501-graph-traversal>`
- {doc}`25.2. Shortest Paths and String Matching <2502-shortest-paths-strings>`
- {doc}`25.3. Algorithms and Computational Limits <2503-limits-project>`


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Breadth-first search | A graph traversal that explores vertices in layers using a queue. |
| 2 | Depth-first search | A graph traversal that explores along one path before backtracking. |
| 3 | Reachability | Whether a target vertex or state can be reached from a starting point. |
| 4 | Traversal order | The order in which an algorithm visits vertices or states. |
| 5 | Shortest path | A path with minimum number of edges or minimum total cost. |
| 6 | String matching | Finding whether and where a pattern occurs in a text. |
| 7 | Pattern | The string or structure being searched for. |
| 8 | Text | The larger string being searched. |
| 9 | Prefix table | A table that records reusable information about matching prefixes in a pattern. |
| 10 | Intractable problem | A problem considered computationally difficult in a stated model; one slow implementation alone does not prove this classification. |
| 11 | Approximation | An algorithm with a proved objective bound relative to the optimum under stated assumptions. |
| 12 | Limit | A constraint, such as time or memory, that shapes algorithm choice. |

| 13 | Discovery | First reaching and marking a vertex before adding it to a worklist. |
| 14 | DFS frame | A vertex and suspended next-neighbor position on the active exploration stack. |
| 15 | Border | A proper pattern prefix that is also a suffix of the matched segment. |
| 16 | Ordinal match | Equality of UTF-16 code units without culture rules, case folding, or normalization. |
| 17 | Heuristic | A practical method used without an optimality or approximation guarantee established here. |


<h2>Slides</h2>

Slides for this chapter will be added after the notebook content stabilizes.
