
Verify a BFS result by checking that the source has distance zero, each reachable non-source has a predecessor one layer earlier, and every examined edge satisfies the layer bound. Check unreachable vertices separately. A shortest path in edge count need not minimize weighted cost; Chapter 23 supplies Dijkstra’s algorithm for nonnegative weights.

For DFS, check that each reached vertex enters and exits exactly once, every child interval is nested in its parent interval, and cycles do not cause repeated discovery. An explicit stack of suspended neighbor positions reproduces recursive order; a stack that pushes all neighbors at once may produce another order.


## Components and Direction

```{index} connected components; directed graph
```


The component method in Chapter 19 requires symmetric adjacency. A directed reachability search does not compute strongly connected components. Choose the mathematical contract before selecting an implementation. Use the Chapter 25 lab to connect traversal evidence with network coverage and the limits of heuristic routing.


```{rubric} Footnotes
```
