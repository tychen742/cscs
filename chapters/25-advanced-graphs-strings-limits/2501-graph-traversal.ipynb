# Graph Traversal



```{contents}
:local:
:depth: 2
```


Graph traversal visits vertices by following edges. Breadth-first search (BFS) explores in layers using a queue. Depth-first search (DFS) follows one path as far as possible before backing up.

Both algorithms need a visited set so cycles do not cause repeated processing.


```{index} breadth-first search
```

## Breadth-First Search

BFS uses a queue. When a vertex is discovered, it waits behind vertices discovered earlier, so the traversal finishes one layer before moving to the next.


````csharp
using System;
using System.Collections.Generic;

var graph = new Dictionary<string, string[]>
{
    ["A"] = new[] { "B", "C" },
    ["B"] = new[] { "A", "D" },
    ["C"] = new[] { "A", "D" },
    ["D"] = new[] { "B", "C" }
};

var queue = new Queue<string>();
var visited = new HashSet<string>();
queue.Enqueue("A");
visited.Add("A");

while (queue.Count > 0)
{
    string current = queue.Dequeue();
    Console.WriteLine(current);

    foreach (string neighbor in graph[current])
    {
        if (visited.Add(neighbor))
        {
            queue.Enqueue(neighbor);
        }
    }
}
````

Because BFS explores by layers, it can find shortest paths in unweighted graphs.


## Depth-First Search

DFS can be written recursively or with an explicit stack. It is useful for reachability, connected components, and exploring decision spaces.


````csharp
using System;
using System.Collections.Generic;

var graph = new Dictionary<string, string[]>
{
    ["A"] = new[] { "B", "C" },
    ["B"] = new[] { "A", "D" },
    ["C"] = new[] { "A", "D" },
    ["D"] = new[] { "B", "C" }
};

var visited = new HashSet<string>();
Visit("A");

void Visit(string vertex)
{
    if (!visited.Add(vertex))
    {
        return;
    }

    Console.WriteLine(vertex);
    foreach (string neighbor in graph[vertex])
    {
        Visit(neighbor);
    }
}
````

The recursive call stack plays the role of the path currently being explored.


```{index} unweighted shortest path
```

## BFS Paths

To reconstruct a BFS path, store the predecessor that first discovered each vertex.


````csharp
using System;
using System.Collections.Generic;

var graph = new Dictionary<string, string[]>
{
    ["A"] = new[] { "B", "C" },
    ["B"] = new[] { "A", "D", "E" },
    ["C"] = new[] { "A", "F" },
    ["D"] = new[] { "B" },
    ["E"] = new[] { "B", "F" },
    ["F"] = new[] { "C", "E" }
};

var previous = new Dictionary<string, string?> { ["A"] = null };
var queue = new Queue<string>();
queue.Enqueue("A");

while (queue.Count > 0)
{
    string current = queue.Dequeue();
    foreach (string neighbor in graph[current])
    {
        if (!previous.ContainsKey(neighbor))
        {
            previous[neighbor] = current;
            queue.Enqueue(neighbor);
        }
    }
}

var path = new Stack<string>();
for (string? at = "F"; at is not null; at = previous[at])
{
    path.Push(at);
}

Console.WriteLine(string.Join(" -> ", path));
````

This path is shortest by number of edges because BFS discovers vertices by distance from the source.

## Connected Components

Repeated traversal can identify connected components in an undirected graph.


````csharp
using System;
using System.Collections.Generic;
using System.Linq;

var graph = new Dictionary<string, string[]>
{
    ["A"] = new[] { "B" },
    ["B"] = new[] { "A" },
    ["C"] = new[] { "D" },
    ["D"] = new[] { "C" },
    ["E"] = Array.Empty<string>()
};

var visited = new HashSet<string>();
int components = 0;

foreach (string vertex in graph.Keys.OrderBy(v => v))
{
    if (visited.Contains(vertex)) continue;

    components++;
    var queue = new Queue<string>();
    queue.Enqueue(vertex);
    visited.Add(vertex);

    while (queue.Count > 0)
    {
        string current = queue.Dequeue();
        foreach (string neighbor in graph[current])
        {
            if (visited.Add(neighbor)) queue.Enqueue(neighbor);
        }
    }
}

Console.WriteLine($"components = {components}");
````

Traversal turns graph structure into answerable questions: reachable or not, how many components, what layer, and what path.
