## Chapter Topics

The chapters progress from computing concepts and C# fundamentals through methods, control flow, error handling, collections, and files, then to object-oriented design and databases, and finally to data structures and algorithms. Preview, Lab, and Homework activities accompany each chapter except Chapter 13, and some chapters add further homework assignments.
