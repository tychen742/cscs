<h2>Glossary</h2>

| Term | Meaning |
| --- | --- |
| Key equality | The policy deciding whether identifiers match. |
| Hash code | A value used to choose a storage bucket. |
| Collision | Distinct keys reaching the same bucket. |
| Load factor | Entry count divided by bucket count. |
| Rehash | Recompute storage positions after growth. |
| Set difference | Members of the first set absent from the second. |
