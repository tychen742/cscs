## Count Repeated Recursive Calls

Start with naive Fibonacci so repeated work is visible.


````csharp
using System;

int calls = 0;
int n = 8;
Console.WriteLine(Fib(n));
Console.WriteLine($"calls = {calls}");

int Fib(int n)
{
    calls++;
    if (n <= 1) return n;
    return Fib(n - 1) + Fib(n - 2);
}
````

## Add Memoization

Use a dictionary cache and compare call counts.


````csharp
using System;
using System.Collections.Generic;

int calls = 0;
var cache = new Dictionary<int, int>();
int n = 30;
Console.WriteLine(Fib(n));
Console.WriteLine($"calls = {calls}");

int Fib(int n)
{
    calls++;
    if (n <= 1) return n;
    if (cache.TryGetValue(n, out int saved)) return saved;

    int result = Fib(n - 1) + Fib(n - 2);
    cache[n] = result;
    return result;
}
````

## Build a Table

Convert the recurrence to a bottom-up table.


````csharp
using System;

int n = 12;
int[] fib = new int[n + 1];
fib[0] = 0;
fib[1] = 1;

for (int i = 2; i <= n; i++)
{
    fib[i] = fib[i - 1] + fib[i - 2];
}

Console.WriteLine(fib[n]);
Console.WriteLine(string.Join(", ", fib));
````

## Search for a Subset

Use backtracking to find whether a target sum can be made.


````csharp
using System;

int[] values = { 3, 7, 10, 12 };
int target = 15;
int calls = 0;

Console.WriteLine(HasSubset(0, target));
Console.WriteLine($"calls = {calls}");

bool HasSubset(int index, int remaining)
{
    calls++;
    if (remaining == 0) return true;
    if (index == values.Length || remaining < 0) return false;

    return HasSubset(index + 1, remaining - values[index])
        || HasSubset(index + 1, remaining);
}
````
