# Dynamic Programming and Backtracking Lab

Compare naive recursion, memoization, tabulation, and pruned search on small but meaningful problem instances.

## Tasks

1. Count a recursive solution's repeated calls.
2. Add memoization and compare the call count.
3. Rewrite the solution with a bottom-up table.
4. Use backtracking to find a subset that meets a target.
5. Test impossible targets and empty inputs.
6. Explain the time and space tradeoffs of caching and pruning.

```csharp
int[] values = { 3, 7, 10, 12 };
int target = 15;
int calls = 0;

bool HasSubset(int index, int remaining)
{
    calls++;
    if (remaining == 0)
        return true;
    if (index == values.Length || remaining < 0)
        return false;

    return HasSubset(index + 1, remaining - values[index])
        || HasSubset(index + 1, remaining);
}

Console.WriteLine($"Found: {HasSubset(0, target)}");
Console.WriteLine($"Calls: {calls}");
```
