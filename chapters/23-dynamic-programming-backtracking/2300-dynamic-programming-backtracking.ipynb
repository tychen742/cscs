# Dynamic Programming

Some problems repeat the same subproblems, while others require systematic search through many possible choices. This chapter introduces dynamic programming and backtracking as two ways to control that complexity.

- **Overlapping subproblems**: Dynamic programming helps when the same smaller problems appear repeatedly.
- **Memoization**: Memoization stores recursive results so repeated calls can reuse them.
- **Tabulation**: Tabulation builds solutions bottom-up in a table.
- **Backtracking**: Backtracking explores a search space by making a choice, recursing, and undoing the choice.
- **Pruning**: Pruning skips branches that cannot lead to a valid or better solution.

These methods trade memory, structure, and search control for problems where direct brute force grows too quickly.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Identify overlapping subproblems and optimal substructure
2. Choose between memoization and tabulation for a small problem
3. Model a backtracking search space with choices and undo steps
4. Use pruning conditions to reduce unnecessary search
5. Explain the time and memory trade-offs of these strategies
