# Dynamic Programming

This chapter covers overlapping subproblems, optimal substructure, memoization, tabulation, small dynamic-programming tables, backtracking, search spaces, undo steps, and pruning. The sidebar title **Dynamic Programming** names the main optimization technique; backtracking is included as the paired strategy for systematic search.

- **Overlapping subproblems**: The same smaller problem appears repeatedly.
- **Optimal substructure**: A larger solution can be built from smaller optimal solutions.
- **Memoization**: A top-down technique that stores recursive results for reuse.
- **Tabulation**: A bottom-up technique that fills a table from small cases to large cases.
- **Backtracking**: A search pattern that makes a choice, explores it, then undoes the choice.

Dynamic programming controls repeated work. Backtracking controls large choice spaces by exploring, undoing, and pruning.


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Dynamic programming | A technique that solves repeated subproblems once and reuses their answers. |
| 2 | Overlapping subproblems | A property where the same smaller problems occur multiple times inside a larger problem. |
| 3 | Optimal substructure | A property where an optimal solution can be built from optimal solutions to smaller problems. |
| 4 | Memoization | Top-down dynamic programming that caches recursive results. |
| 5 | Tabulation | Bottom-up dynamic programming that fills a table in dependency order. |
| 6 | State | The information that identifies one subproblem. |
| 7 | Transition | A rule that computes one state from smaller or earlier states. |
| 8 | Backtracking | A search technique that makes a choice, explores recursively, then undoes the choice. |
| 9 | Choice | One option selected at a search step. |
| 10 | Pruning | Skipping a search branch because it cannot lead to a valid or better solution. |
| 11 | Search tree | The implicit tree of choices explored by a recursive search. |
