# Backtracking and Search Spaces

Backtracking explores a space of choices one decision at a time. When a partial solution cannot lead to a valid result, the algorithm undoes the latest choice and tries another branch.


```{index} backtracking
```

## Choose, Explore, Undo

A backtracking method usually follows this rhythm: make a choice, recurse, then undo the choice before trying the next option.


````csharp
using System;
using System.Collections.Generic;

var chosen = new List<int>();
GenerateSubsets(new[] { 1, 2, 3 }, 0, chosen);

void GenerateSubsets(int[] values, int index, List<int> chosen)
{
    if (index == values.Length)
    {
        Console.WriteLine($"[{string.Join(", ", chosen)}]");
        return;
    }

    chosen.Add(values[index]);
    GenerateSubsets(values, index + 1, chosen);
    chosen.RemoveAt(chosen.Count - 1);

    GenerateSubsets(values, index + 1, chosen);
}
````

The list `chosen` is shared state. Undoing the choice restores it before the next branch begins.

## Subset Sum

Subset sum asks whether some subset of values adds to a target. A backtracking search can include or exclude each value.


````csharp
using System;

int[] values = { 3, 7, 10, 12 };
int target = 15;
Console.WriteLine(HasSubset(values, 0, target));

bool HasSubset(int[] values, int index, int remaining)
{
    if (remaining == 0)
    {
        return true;
    }

    if (index == values.Length || remaining < 0)
    {
        return false;
    }

    return HasSubset(values, index + 1, remaining - values[index])
        || HasSubset(values, index + 1, remaining);
}
````

The branch where `remaining < 0` is pruned because all values are positive in this example.

## Pruning Branches

Pruning skips branches that cannot lead to a useful result. Good pruning depends on honest constraints.


````csharp
using System;

int[] values = { 2, 4, 6, 9 };
int target = 8;
int calls = 0;

Console.WriteLine(HasSubset(0, target));
Console.WriteLine($"calls = {calls}");

bool HasSubset(int index, int remaining)
{
    calls++;
    if (remaining == 0) return true;
    if (index == values.Length) return false;
    if (remaining < 0) return false;

    return HasSubset(index + 1, remaining - values[index])
        || HasSubset(index + 1, remaining);
}
````

This pruning rule is safe only because values are nonnegative. If negative values were allowed, overshooting the target would not necessarily mean failure.

## Permutations

Permutation generation is another backtracking example. Each step chooses one unused item for the next position.


````csharp
using System;
using System.Collections.Generic;

string[] items = { "A", "B", "C" };
var used = new bool[items.Length];
var order = new List<string>();
Permute();

void Permute()
{
    if (order.Count == items.Length)
    {
        Console.WriteLine(string.Join("", order));
        return;
    }

    for (int i = 0; i < items.Length; i++)
    {
        if (used[i]) continue;

        used[i] = true;
        order.Add(items[i]);
        Permute();
        order.RemoveAt(order.Count - 1);
        used[i] = false;
    }
}
````

The `used` array and `order` list are both restored after each recursive exploration.

## Backtracking Cost

Backtracking can still be exponential. Its value is control: it lets you stop early, prune impossible branches, and return actual solutions rather than simply counting possibilities.


````csharp
using System;

int choices = 2;
int depth = 10;
Console.WriteLine($"full binary search tree leaves: {Math.Pow(choices, depth):N0}");
Console.WriteLine("Pruning matters because search spaces grow quickly.");
````
