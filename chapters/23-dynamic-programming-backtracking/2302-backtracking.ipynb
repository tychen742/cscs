# Backtracking and Search Spaces

Backtracking explores a space of choices one decision at a time. When a partial solution cannot lead to a valid result, the algorithm undoes the latest choice and tries another branch.

A good backtracking design makes three things explicit:

1. the current partial solution;
2. the next choices available; and
3. the constraint that allows a branch to continue.

Pruning does not change the answer; it avoids exploring branches that cannot produce one.

```csharp
static bool HasSubset(int[] values, int index, int target)
{
    if (target == 0)
        return true;
    if (index == values.Length || target < 0)
        return false;

    bool include = HasSubset(values, index + 1, target - values[index]);
    bool exclude = HasSubset(values, index + 1, target);
    return include || exclude;
}

int[] values = { 3, 7, 10, 12 };
Console.WriteLine(HasSubset(values, 0, 15));
Console.WriteLine(HasSubset(values, 0, 8));
```
