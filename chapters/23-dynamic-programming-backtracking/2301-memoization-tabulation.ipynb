# Memoization and Tabulation

Dynamic programming applies when a problem has repeated subproblems and useful larger answers can be built from smaller answers. Fibonacci is the standard first example because the repeated work is easy to see.


```{index} overlapping subproblems
```

## Repeated Recursive Work

A naive recursive Fibonacci method recomputes the same values many times.


````csharp
using System;

int calls = 0;
int n = 6;
long value = Fib(n);

Console.WriteLine($"Fib({n}) = {value}");
Console.WriteLine($"calls = {calls}");

long Fib(int n)
{
    calls++;
    if (n <= 1)
    {
        return n;
    }

    return Fib(n - 1) + Fib(n - 2);
}
````

The recurrence branches into two recursive calls. Several smaller values, such as `Fib(3)` and `Fib(2)`, are computed again and again.


## Memoization

Memoization keeps the recursive structure but stores each answer after computing it once.


````csharp
using System;
using System.Collections.Generic;

int calls = 0;
var cache = new Dictionary<int, long>();
int n = 40;
long value = FibMemo(n, cache);

Console.WriteLine($"Fib({n}) = {value}");
Console.WriteLine($"calls = {calls}");
Console.WriteLine($"cached results = {cache.Count}");

long FibMemo(int n, Dictionary<int, long> cache)
{
    calls++;
    if (n <= 1)
    {
        return n;
    }

    if (cache.TryGetValue(n, out long saved))
    {
        return saved;
    }

    long result = FibMemo(n - 1, cache) + FibMemo(n - 2, cache);
    cache[n] = result;
    return result;
}
````

Memoization trades memory for time. The cache stores solved subproblems so repeated calls become dictionary lookups.

## Tabulation

Tabulation fills a table from base cases upward. Instead of asking recursively for smaller values, the method computes smaller values first.


````csharp
using System;

int n = 10;
long[] fib = new long[n + 1];
fib[0] = 0;
fib[1] = 1;

for (int i = 2; i <= n; i++)
{
    fib[i] = fib[i - 1] + fib[i - 2];
}

Console.WriteLine($"Fib({n}) = {fib[n]}");
Console.WriteLine(string.Join(", ", fib));
````

Tabulation requires choosing an order that satisfies dependencies. For Fibonacci, each cell depends only on the previous two cells.

## A Coin-Change Table

For a minimum-coin problem, the table entry `best[amount]` stores the fewest coins needed to make that amount.


````csharp
using System;

int[] coins = { 1, 3, 4 };
int target = 6;
int[] best = new int[target + 1];
Array.Fill(best, int.MaxValue / 2);
best[0] = 0;

for (int amount = 1; amount <= target; amount++)
{
    foreach (int coin in coins)
    {
        if (coin <= amount)
        {
            best[amount] = Math.Min(best[amount], 1 + best[amount - coin]);
        }
    }
}

Console.WriteLine($"fewest coins for {target}: {best[target]}");
Console.WriteLine(string.Join(", ", best));
````

The table works because the best answer for `amount` can be built from the best answer for `amount - coin`.

## Memoization or Tabulation?

Memoization is often easier when recursion naturally expresses the problem. Tabulation is often easier when the dependency order is simple and all smaller states are needed.


````csharp
using System;

string problemShape = "all amounts up to a target";

if (problemShape.Contains("all"))
{
    Console.WriteLine("Tabulation is a natural first choice.");
}
else
{
    Console.WriteLine("Memoization may compute only the states you ask for.");
}
````
