# Memoization and Tabulation

Dynamic programming applies when a problem has **overlapping subproblems** and useful solutions can be built from smaller solutions.

- **Memoization** is top-down: recurse as needed and cache each result.
- **Tabulation** is bottom-up: fill a table from the smallest cases upward.

Both approaches avoid solving the same subproblem repeatedly. The table or cache is part of the algorithm's extra-space cost.

```csharp
using System.Collections.Generic;

static long FibMemo(int n, Dictionary<int, long> cache)
{
    if (n <= 1)
        return n;
    if (cache.TryGetValue(n, out long saved))
        return saved;

    long result = FibMemo(n - 1, cache) + FibMemo(n - 2, cache);
    cache[n] = result;
    return result;
}

long memoized = FibMemo(20, new Dictionary<int, long>());
long[] table = new long[21];
table[1] = 1;
for (int index = 2; index < table.Length; index++)
    table[index] = table[index - 1] + table[index - 2];

Console.WriteLine(memoized);
Console.WriteLine(table[20]);
```
