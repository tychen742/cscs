# Tree Implementation Lab

Test a tree implementation and analyze the effect of different insertion orders.

## Tasks

1. Insert the same values in random order and sorted order.
2. Draw or print the resulting shapes.
3. Implement preorder, inorder, and postorder traversal.
4. Verify that inorder traversal produces sorted output.
5. Test search for present and absent values.
6. Explain why tree shape affects operation cost.

```csharp
using System.Linq;

int[] values = { 8, 3, 10, 1, 6, 14, 4, 7, 13 };
var sorted = values.OrderBy(value => value).ToArray();

Console.WriteLine("Input:   " + string.Join(", ", values));
Console.WriteLine("Sorted:  " + string.Join(", ", sorted));
Console.WriteLine("Question: what traversal should produce the sorted sequence?");
```
