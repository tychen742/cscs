# Tree Structure and Traversal

A tree is a recursive structure made of nodes and relationships. A node may have a parent, zero or more children, and a position within a hierarchy.

Important terms include **root**, **leaf**, **depth**, **height**, and **subtree**. A traversal visits every node in a deliberate order.

- **Preorder:** visit the node before its children.
- **Inorder:** visit the left subtree, node, then right subtree.
- **Postorder:** visit children before the node.

Inorder traversal becomes especially useful for binary search trees because it visits values in sorted order.

```csharp
public sealed class TreeNode
{
    public int Value { get; }
    public TreeNode? Left { get; set; }
    public TreeNode? Right { get; set; }

    public TreeNode(int value)
    {
        Value = value;
    }
}

var root = new TreeNode(8)
{
    Left = new TreeNode(3),
    Right = new TreeNode(10)
};
root.Left!.Left = new TreeNode(1);
root.Left.Right = new TreeNode(6);

void InOrder(TreeNode? node)
{
    if (node is null)
        return;

    InOrder(node.Left);
    Console.WriteLine(node.Value);
    InOrder(node.Right);
}

InOrder(root);
```
