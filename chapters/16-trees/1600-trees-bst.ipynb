# Trees

Trees model hierarchical relationships and recursive structure. This chapter introduces tree vocabulary, traversal, and binary search trees as a bridge from linear data structures to richer organization.

- **Tree**: A tree organizes nodes into parent-child relationships with a root.
- **Recursion**: Tree processing naturally uses recursive structure because each subtree is itself a tree.
- **Traversal**: Preorder, inorder, and postorder traversals visit nodes in different useful orders.
- **Binary search tree**: A BST stores values so smaller values go left and larger values go right.
- **Balance**: Tree shape affects the cost of search, insertion, and removal.

Trees make structure part of the data, not just the code that processes it.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Describe tree nodes, roots, parents, children, leaves, and subtrees
2. Trace recursive tree traversal algorithms
3. Implement or reason about preorder, inorder, and postorder traversal
4. Explain how binary search tree ordering supports search and insertion
5. Describe why tree balance affects performance
