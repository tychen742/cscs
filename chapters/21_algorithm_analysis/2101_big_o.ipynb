## Derive a Count Before Naming a Growth Class



A report compares each invoice with every later invoice. There are n invoices, but not n squared distinct ordered loop-body executions in this particular program. The first invoice has n-1 later partners, the next has n-2, and the final invoice has none. Adding these counts gives n(n-1)/2. This exact count is quadratic even though it is not exactly n squared.

For n = 4, list pairs by index: (0,1), (0,2), (0,3), (1,2), (1,3), (2,3). Six comparisons are needed. This small instance checks whether a proposed formula counts self-pairs or counts each pair twice. It does not itself prove the formula for every n; the decreasing row lengths explain the general sum.

| n | Later-pair count | Rectangular all-pairs count n*n |
| --- | --- | --- |
| 0 | 0 | 0 |
| 1 | 0 | 1 |
| 4 | 6 | 16 |
| 8 | 28 | 64 |

The chosen basic operation matters. Counting loop-body visits, string comparisons, element copies or allocated objects can produce different exact counts. State the model before comparing algorithms. Fixed-width arithmetic is often modeled as constant work here; a proof about very large arbitrary-precision integers must account for growing operand size.

For related explanation, see the further reading below.[^development-reading]

## A Bound Is a Statement with a Domain



Big-O describes an eventual upper bound up to a constant factor. A tight Θ claim also needs a matching lower bound under the stated model. Saying a function is O(n squared) can be true without identifying its tight growth; a linear function also has a quadratic upper bound for sufficiently large n. Use the distinction when a table asks for a tight class rather than any correct upper bound.

For n(n-1)/2 and n at least two, a constant fraction of n squared bounds the count below and one half of n squared bounds it above. Those inequalities connect the exact formula to the tight quadratic class. They are more informative than deleting terms by habit without saying what the approximation means.

Input-dependent cases need their own quantifiers. A first-match scan can stop after one comparison for a leading match and can examine n elements for a missing target. Worst case chooses a maximizing input at each size; expected case requires a probability model. Amortized analysis bounds a sequence of operations and does not require that the sequence be random.

Guided derivation: a loop doubles a counter from one while it remains below n. Write its tested values for n = 8 and n = 9, then state which operation you are counting. Explain why a one-step count difference at a boundary does not contradict logarithmic growth. Finally, distinguish total allocated bytes over a run from peak simultaneously live auxiliary storage.


A finance team can scan every expense once or compare every pair. Before timing either program, define the input and count the work. Analysis describes a cost model; measurement checks how that model behaves on a particular system.

## Choose an Input and a Cost Model




```{index} Choose an Input and a Cost Model
```


Let `n` be the number of expense records. Count one addition when a record contributes to a total. Array access, integer addition, and comparison are treated as constant-cost operations here. String comparisons, database requests, and growing collection operations need their own assumptions.

````csharp
using System;
using System.Linq;

foreach (int n in new[] { 0, 1, 4, 8 })
{
    int[] expenses = Enumerable.Repeat(125, n).ToArray();
    long total = 0, additions = 0;
    foreach (int cents in expenses) { total += cents; additions++; }
    Console.WriteLine($"n={n}, total={total}, additions={additions}");
}
````

The selected operation runs exactly `n` times, including zero times for an empty input. Allocation and initialization in this demonstration are setup; include them when analyzing the entire program. The sum uses `long` to hold many integer amounts.

**Common mistake.** Counting the wrong operation, such as array allocations instead of additions, and then comparing that count with another algorithm's comparison count. A cost model names one operation and one input size; mixing models makes the numbers incomparable.

## Sequential, Rectangular, and Triangular Loops




```{index} Sequential, Rectangular, and Triangular Loops
```


Add costs of sequential phases. Multiply independent loop ranges. For dependent ranges, sum the inner counts instead of multiplying their largest bounds blindly.

````csharp
using System;
using System.Linq;

foreach (int n in new[] { 0, 1, 4, 8 })
{
    long sequential = 0, square = 0, triangle = 0;
    for (int i = 0; i < n; i++) sequential++;
    for (int i = 0; i < n; i++) sequential++;
    for (int i = 0; i < n; i++)
        for (int j = 0; j < n; j++) square++;
    for (int i = 0; i < n; i++)
        for (int j = i + 1; j < n; j++) triangle++;
    Console.WriteLine($"n={n}: sequential={sequential}, square={square}, distinct pairs={triangle}");
}
````

The counts are `2n`, `n²`, and `n(n−1)/2`. The triangular loop compares each unordered pair once, excluding self-pairs. Both pair scans are Θ(n²), despite different constants. A loop inside another loop can still be linear if its inner range is bounded by a fixed constant.

**Common mistake.** Multiplying the two loop bounds for the triangular loop and reporting `n²`. The inner range depends on `i`, so the count is the sum `0 + 1 + ... + (n - 1)`, half of `n²`; both are Θ(n²), but the exact count is what a test can check.

## Logarithmic Loops and Multiple Inputs




```{index} Logarithmic Loops and Multiple Inputs
```


Integer halving performs exactly floor(log₂ n) reductions for positive `n`. Zero performs none. Distinguish reductions from the number of states visited. If there are `n` invoices and `m` vendors, the comparison count is Θ(nm) when both sizes are positive; do not replace `m` with `n` without a reason.

````csharp
using System;
using System.Linq;

foreach (int n in new[] { 0, 1, 2, 3, 8, 9, 16 })
{
    int reductions = 0;
    for (int size = n; size > 1; size /= 2) reductions++;
    Console.WriteLine($"n={n}, reductions={reductions}");
}
int invoices = 3, vendors = 5, comparisons = 0;
for (int i = 0; i < invoices; i++)
    for (int j = 0; j < vendors; j++) comparisons++;
Console.WriteLine($"invoice/vendor comparisons={comparisons}");
````

For asymptotic statements, restrict to sufficiently large positive `n`, where logarithms are defined. A doubling loop has logarithmic iterations too, but an unguarded integer update can overflow; repeated division avoids that problem in this example.

**Common mistake.** Calling a loop over `n` invoices and `m` vendors O(n²) because there are two nested loops. The cost is Θ(nm); when `m` is a small fixed number the loop is linear in `n`, and when `m` grows independently the two symbols cannot be merged.

## Bounds and Tight Growth




```{index} Bounds and Tight Growth
```


For nonnegative costs, `T(n) ∈ O(f(n))` means there are constants `c > 0` and `n₀` such that `T(n) ≤ c f(n)` for all `n ≥ n₀`. Ω supplies a lower bound; Θ supplies both bounds. These symbols describe a chosen cost function, not automatically a worst-case cost.

````csharp
using System;
using System.Linq;

for (long n = 1; n <= 5; n++)
{
    long cost = 3 * n * n + 5 * n + 20;
    Console.WriteLine($"n={n}, 3n^2={3*n*n} <= {cost} <= 28n^2={28*n*n}");
}
````

For `n ≥ 1`, `3n² ≤ 3n² + 5n + 20 ≤ 28n²`, so the cost is Θ(n²). It is also O(n³), an upper bound that loses useful precision. Saying O(n²) alone does not establish that the cost is quadratic: a linear cost also satisfies that upper bound. Logarithm bases differ by constant factors.

**Common mistake.** Reading O(n²) as "the algorithm takes quadratic time". The symbol gives an upper bound only; a linear algorithm is also O(n²). Use Θ when both bounds hold, as the cell's constants 3 and 28 establish for this cost function.

### Simplifying Rules for Growth Classes

Four rules turn an exact count into its growth class, and each follows from the definition with constants c and n₀.

| Rule | Statement | Example |
| --- | --- | --- |
| Constants vanish | If T(n) is O(f(n)) then k × T(n) is O(f(n)) for any constant k | 3n² + 5n + 20 and 28n² are both O(n²) |
| Lower-order terms vanish | If T(n) is O(f(n)) and f grows at least as fast as g, then T(n) + g(n) is O(f(n)) | n² + n log n is O(n²) |
| Sequence adds, so the larger wins | Two phases costing O(f) and O(g) together cost O(max(f, g)) | A sort followed by a scan is O(n log n) |
| Nesting multiplies | A loop of O(f) iterations whose body costs O(g) costs O(f × g) | n iterations of a log n binary search cost O(n log n) |

The rules are shortcuts, not substitutes for the count. The third rule is why a program with several independent passes over the data is judged by its most expensive pass, and the fourth is why "two nested loops" means n² only when both loops really run n times, which the triangular and fixed-width examples on this page showed is not automatic.

The same rules apply to Ω and Θ. Upper bounds and worst cases are different ideas, and confusing them is the most common error in this topic: the worst case is a choice of *input* for a given size, while a bound is a statement about how the cost *grows* across sizes. "Worst-case O(n²)" chooses the maximizing input at each size and then bounds that cost's growth from above; "best-case Θ(n)" chooses the minimizing input and gives its exact growth. Any of the three cases can be paired with any of the three bound symbols.

### A Faster Computer or a Faster Algorithm?

Growth classes answer a practical purchasing question. A nightly report takes an hour on the current machine and the data is about to grow. Will a machine ten times faster keep the report inside the hour? The answer depends entirely on the growth class, because a faster machine does not change how the work grows; it multiplies the budget. If the old machine handled n records in an hour and the new one performs ten times the operations, the new machine handles the largest n' for which T(n') is ten times T(n). The example solves that equation for several growth classes.

````csharp
using System;

long budget = 1000;    // operations the old machine performs in one hour, in thousands
const double Limit = 1e15; // beyond this size the table reports "over 10^15"
Console.WriteLine("growth        old n     new n (10x faster)   gain");
Report("log2 n",  x => Math.Log2(x));
Report("n",       x => x);
Report("n log2 n", x => x * Math.Log2(x));
Report("n^2",     x => x * x);
Report("n^3",     x => x * x * x);
Report("2^n",     x => Math.Pow(2, x));

void Report(string label, Func<double, double> cost)
{
    double oldN = Largest(cost, budget), newN = Largest(cost, budget * 10);
    string Show(double v) => v >= Limit ? "over 10^15" : $"{v:N0}";
    string gain = newN >= Limit ? "" : $"{newN / oldN:F1}x";
    Console.WriteLine($"{label,-9} {Show(oldN),12} {Show(newN),20}   {gain}");
}

static double Largest(Func<double, double> cost, double allowed)
{
    double low = 1, high = Limit;                       // largest n in [1, Limit] whose cost fits
    if (cost(high) <= allowed) return Limit;
    while (high - low > 1)
    {
        double mid = Math.Floor((low + high) / 2);
        if (cost(mid) <= allowed) low = mid; else high = mid;
    }
    return low;
}
````

Read the gain column. A linear algorithm turns a ten-times-faster machine into ten times the data, and an n log n algorithm nearly does. The quadratic algorithm gains only about a factor of three, because √10 is about 3.2, and the cubic algorithm barely doubles its reach. The exponential algorithm gains four more records, since 10 is a little more than 2³. The logarithmic row goes the other way: ten times the budget raises the exponent, and its reach outgrows the table. The decision follows: when growth is quadratic or worse, money spent on hardware buys little, and the same money spent on an algorithm with a lower growth class buys a great deal. When the algorithm is already linear, hardware helps in proportion.

### Growth-Rate Ordering

Arrange these ten cost functions from slowest growing to fastest growing for large n. Group together any that belong to the same growth class, then open the answer.

`n!`, `2ⁿ`, `n³`, `n² log n`, `n log n`, `100n`, `√n`, `log² n`, `log n`, `1,000,000`

```{dropdown} Answer
`1,000,000` (constant), `log n`, `log² n`, `√n`, `100n` (linear; the constant does not matter), `n log n`, `n² log n`, `n³`, `2ⁿ`, `n!`.

Three comparisons are worth remembering. Any power of a logarithm grows more slowly than any positive power of n, so `log² n` is below `√n`. Any polynomial grows more slowly than any exponential, so `n³` is below `2ⁿ`. And `n!` outgrows `2ⁿ` because its factors keep increasing while the exponential's factor stays at 2. For small n the order can be different: `100n` exceeds `n² log n` until n is past 20 or so, which is why the ordering is a statement about large n.
```

### Code Tracing: Count the Loop Bodies

For each fragment, write the exact number of times the inner statement runs as a formula in n, then its tight growth class. Check against the solution, which counts for n = 8 and n = 16.

```csharp
// A
for (int i = 0; i < n; i++)
    for (int j = 0; j < n; j += 2) count++;

// B
for (int i = 1; i < n; i *= 2)
    for (int j = 0; j < i; j++) count++;

// C
for (int i = 0; i < n; i++)
    for (int j = i; j < n; j++)
        if (j % 2 == 0) count++;

// D
for (int i = n; i > 0; i /= 2)
    for (int j = 0; j < n; j++) count++;
```

````csharp
// Solution
using System;

foreach (int n in new[] { 8, 16 })
{
    long a = 0, b = 0, c = 0, d = 0;
    for (int i = 0; i < n; i++) for (int j = 0; j < n; j += 2) a++;
    for (int i = 1; i < n; i *= 2) for (int j = 0; j < i; j++) b++;
    for (int i = 0; i < n; i++) for (int j = i; j < n; j++) if (j % 2 == 0) c++;
    for (int i = n; i > 0; i /= 2) for (int j = 0; j < n; j++) d++;
    Console.WriteLine($"n={n,2}: A={a,4}  B={b,4}  C={c,4}  D={d,4}");
}
````

A runs n × ⌈n / 2⌉ times, 32 and 128 here: the step of two halves the inner count but the class is still Θ(n²). B runs 1 + 2 + 4 + ... up to the largest power of two below n, which is n - 1 for a power of two, so 7 and 15: the outer loop is logarithmic but the inner loop's work doubles each time and the total is Θ(n), a case where multiplying the loop bounds gives the wrong answer. C counts the even j in each suffix, about a quarter of n², giving 16 and 64, Θ(n²). D runs n times for each of ⌊log₂ n⌋ + 1 outer iterations, 32 and 80, Θ(n log n). Doubling n quadrupled A and C, doubled B, and multiplied D by two and a half, which is the quickest empirical check of a growth class.

## Best, Worst, Expected, and Amortized Costs




```{index} Best, Worst, Expected, and Amortized Costs
```


A search can stop early. Best and worst cases vary over inputs of the same size; an expected cost needs a probability model. An amortized cost averages over a sequence of operations and does not require random inputs.

````csharp
using System;
using System.Linq;

int[] ids = { 40, 10, 30, 20 };
foreach (int target in new[] { 40, 20, 99 })
{
    int checks = 0, found = -1;
    for (int i = 0; i < ids.Length; i++)
    {
        checks++;
        if (ids[i] == target) { found = i; break; }
    }
    Console.WriteLine($"target={target}, found={found}, checks={checks}");
}
````

On nonempty arrays, this search has best-case Θ(1) and worst-case Θ(n) comparisons. If a successful target is equally likely at every index, its expected count is `(n+1)/2`, hence Θ(n). A geometric-growth array append has amortized Θ(1) cost even though one resizing append costs Θ(n). Expected hashing requires assumptions about key distribution and table policy; it is a different claim.

**Common mistake.** Reporting the best case as the algorithm's cost because one test found its target at index 0. The checks column shows the count depends on where the target is; a cost claim must say which case it describes.

## Space, Growth, and Measurement




```{index} Space, Growth, and Measurement
```


State whether you count total space or auxiliary space. Auxiliary space excludes the input but includes temporary arrays, returned storage unless explicitly excluded, and active recursion frames. Peak live memory is different from all allocations over a run.

````csharp
using System;
using System.Linq;

foreach (int n in new[] { 8, 16, 32 })
{
    double log = Math.Log2(n);
    Console.WriteLine($"n={n}: log={log:F0}, n={n}, nlogn={n*log:F0}, n^2={(long)n*n}, 2^n={Math.Pow(2,n):F0}");
}
int[] amounts = { 100, 250, 75 };
long sum = 0;
foreach (int value in amounts) sum += value;
int[] copy = amounts.ToArray();
Console.WriteLine($"sum={sum}, copied items={copy.Length}");
````

The sum uses Θ(1) auxiliary space; copying uses Θ(n). A method may have low stack depth and still allocate a large result. Timings depend on JIT compilation, garbage collection, I/O, data, and hardware. Warm up, repeat, report input sizes and conditions, and compare relevant operation counts. A single stopwatch result cannot establish an asymptotic bound.

**Common mistake.** Treating the stack depth as the whole memory cost. The copy in the last line allocates `n` elements while using one stack frame; auxiliary space counts that array, and a method with shallow recursion can still need linear memory.

### Check Your Understanding

Answer each question before opening its dropdown.

1. An algorithm is described as O(n log n) in the worst case and Ω(n) in the best case. Can it be Θ(n log n)?

```{dropdown} Answer
Only for a specific case. Θ requires matching upper and lower bounds on the same cost function; the worst-case cost can be Θ(n log n) while the best-case cost is Θ(n). The algorithm as a whole has no single Θ unless the two cases have the same growth.
```

2. A quadratic report takes 20 minutes on 50,000 records. About how long on 150,000 records, and what would a ten-times-faster machine do to that?

```{dropdown} Answer
About nine times longer, three hours, because tripling n multiplies a quadratic cost by nine. A ten-times-faster machine brings that to 18 minutes, which barely beats today's 20; the next growth in data would exceed the hour again. The durable fix is an algorithm in a lower class.
```

3. Why is a lower bound on a *problem* harder to establish than a lower bound on an *algorithm*?

```{dropdown} Answer
An algorithm's lower bound needs one input family that forces the cost. A problem's lower bound must hold for every algorithm, including ones not yet invented, so it needs an argument about the problem itself, such as that any correct search of unsorted data must examine every record it cannot rule out.
```

4. Two nested loops each run n times, but the body contains a call to a method that scans an array of n elements. What is the growth class?

```{dropdown} Answer
Θ(n³) under the nesting rule: n × n iterations each doing Θ(n) work. The lesson is to count the called method's cost as part of the body; a method call is not a constant-time operation just because it is one line.
```
