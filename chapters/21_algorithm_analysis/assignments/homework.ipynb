````csharp
using System;
using System.Linq;

// Solution
int[] sizes={0,1,5};
foreach(int n in sizes)
{
 long count=0;
 for(int i=1;i<=n;i++)for(int j=0;j<i;j++)count++;
 long expected=(long)n*(n+1)/2;
 if(count!=expected)throw new Exception("Count mismatch.");
 Console.WriteLine($"n={n}, count={count}, formula={expected}");
}
// Theta(n^2) for growing n, despite a dependent inner bound.
````

## 10. Calls versus Reductions

```{index} 10. Calls versus Reductions
```


Count calls in recursive integer halving, including the base call. Reject negative inputs. Test 0,1,3,16 and print calls and reductions. Name the recurrence and stack bound.

````csharp
using System;
using System.Linq;

int[] sizes={0,1,3,16};
// Your code starts here.
/* TODO: complete the task described above. */
// Your code ends here.
````

````csharp
using System;
using System.Linq;

// Solution
int[] sizes={0,1,3,16};
foreach(int n in sizes)
{
 int calls=Calls(n);
 Console.WriteLine($"n={n}, calls={calls}, reductions={calls-1}");
}
int Calls(int n)
{
 if(n<0)throw new ArgumentOutOfRangeException(nameof(n));
 return n<=1?1:1+Calls(n/2);
}
// C(n)=1 for n<=1; otherwise C(floor(n/2))+1. Theta(log n) time/stack for large n.
````
