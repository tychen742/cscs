# Algorithm Analysis

A program can produce the right sample answer and still lack a correctness argument or a useful cost model. This chapter connects specifications and loop invariants to operation counts, asymptotic bounds, and recurrences. You will state the input and counting assumptions behind each claim and distinguish measured performance from general reasoning.

- **Correctness**: A claim that an algorithm satisfies its specification for every valid input.
- **Complexity**: A description of how time or memory use grows as input size grows.
- **Big-O**: An upper-bound growth description that ignores constant factors and lower-order terms.
- **Invariant**: A fact that remains true before and after each loop iteration.
- **Recurrence**: An equation that describes the cost of a recursive algorithm in terms of smaller inputs.

<h2>Learning Objectives</h2>


By the end of this chapter, you will be able to:

1. Define input sizes and a cost model; derive exact counts for sequential, rectangular, triangular, and halving loops.
2. Distinguish upper, lower, and tight bounds and best, worst, expected, and amortized costs.
3. State precise contracts and use initialization, maintenance, exit reasoning, and decreasing variants to establish total correctness.
4. Choose boundary tests and counterexamples that challenge an implementation.
5. Derive recurrences with base cases and rounding; separate modeled work, total calls, stack depth, and other auxiliary storage.
