## Coding

Write each method in its own cell, with at least three test calls that print results, including one boundary or invalid case.

01. `static string Initials(string fullName)` returns the initials of a two-word name with periods, for example `Initials("Ana Diaz")` returns `"A. D."`. Return `"(unknown)"` if the name has no space.
02. `static bool IsStrongPassword(string password)` returns `true` when the password has at least 12 characters, at least one letter, and at least one digit. Use a loop with `char.IsLetter` and `char.IsDigit`.
03. `static string RedactDigits(string note)` returns a copy of a customer-service note with every digit replaced by `#`, so that account and card numbers do not appear in shared notes.
04. `static string SafeDiscountMessage(string percentInput)` uses `int.TryParse` and returns `"Discount set to N%."` for whole numbers from 0 to 50, or a message explaining the allowed range otherwise. The program must never throw an exception.
05. `static string LocalPrice(decimal amount, string cultureName)` returns `amount` formatted as currency for the given culture name, such as `"en-US"` or `"de-DE"`. Print the same amount for three cultures.
