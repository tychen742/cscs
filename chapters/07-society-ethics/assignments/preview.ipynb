## True or False

**1.** A defect that overcharges each customer only a few cents cannot cause serious harm.

```{dropdown} Answer
**False** — a program repeats the same defect for every transaction, so small errors add up across thousands of customers.
```

**2.** Masking a card number as `**** **** **** 1234` is a way to show less personal data.

```{dropdown} Answer
**True** — the display keeps only what a receipt needs.
```

**3.** Logging a user's password helps support staff and is acceptable if the log is private.

```{dropdown} Answer
**False** — logs are read by many people and kept for a long time; secrets must never be logged.
```

**4.** If code you find online has no license, you may reuse it freely.

```{dropdown} Answer
**False** — copyright applies automatically; without a license you have no permission to reuse it.
```

**5.** Using color alone to show an error can exclude some users.

```{dropdown} Answer
**True** — screen-reader users, readers with color blindness, and monochrome terminals miss the meaning.
```

## Multiple Choice

**1.** Which question belongs to the ethical checklist in Section 7.1?

a) Which algorithm is fastest?  
b) Who is affected by this decision?  
c) Which IDE should we use?  
d) How many lines of code will it take?

```{dropdown} Answer
**b)** — identifying stakeholders is the first step.
```

**2.** What is *data minimization*?

a) Compressing data to save disk space  
b) Collecting and keeping only the personal data a task needs  
c) Deleting all data after each run  
d) Encrypting every field

```{dropdown} Answer
**b)** — it is a GDPR principle and a good habit everywhere.
```

**3.** Which input should a quantity check reject?

a) "12"  
b) "1"  
c) "12x"  
d) "500"

```{dropdown} Answer
**c)** — "12x" is not a whole number.
```

**4.** Why is building a query as `"... WHERE Name = '" + name + "'"` dangerous?

a) It is slow  
b) Input can change the meaning of the command (injection)  
c) Strings cannot contain quotes  
d) It uses too much memory

```{dropdown} Answer
**b)** — crafted input such as `x' OR '1'='1` becomes part of the query.
```

**5.** Under the GNU GPL, if you distribute a modified version of GPL code, you must:

a) Pay a fee  
b) Release your version under the GPL  
c) Remove the original author's name  
d) Keep your changes secret

```{dropdown} Answer
**b)** — that requirement is what makes the GPL a copyleft license.
```
