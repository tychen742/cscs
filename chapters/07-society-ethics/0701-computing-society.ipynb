Software now sits between people and many things that matter to them: pay, prices, credit, medical treatment, and access to services. The people affected usually never see the code and cannot check it. That gap is why computing professionals carry special responsibility for what their programs do.

## When Software Fails People

Three well-documented cases show different ways software harms people.

- **Therac-25 (1985–1987).** A radiation therapy machine delivered massive overdoses in six known accidents, killing or seriously injuring patients. Investigators found software race conditions, combined with the removal of hardware safety interlocks that earlier models had relied on and weak engineering practices.[^therac]
- **Knight Capital (2012).** On August 1, 2012, a trading firm deployed new code to seven of its eight order servers. On the eighth, old, unused code was reactivated. In about 45 minutes it sent millions of unintended orders, and the firm lost more than \$460 million. The U.S. Securities and Exchange Commission later settled charges with Knight, which paid a \$12 million penalty.[^knight]
- **The UK Post Office Horizon scandal (1999–2015).** Hundreds of subpostmasters, who run local post offices, were prosecuted for shortfalls that the Horizon accounting system reported. The Post Office itself brought about 700 of these prosecutions. The system had bugs, and the people accused could not see or challenge how it worked. Parliament passed the *Post Office (Horizon System) Offences Act 2024* to quash the convictions.[^horizon]

The lessons differ. Therac-25 shows that removing a safety check because "the software handles it" can be deadly. Knight Capital shows that deployment is part of software quality. Horizon shows the harm done when an organization trusts a system's output over the people it affects.

## Small Defects at Scale

A defect that costs one customer a few dollars can cost customers in total a great deal, because a program repeats the same mistake every time it runs. The discount rule below is meant to give 10% off orders of \$100.00 or more, but the condition uses `>` instead of `>=`:

````csharp
using System;

// Intended business rule: orders of $100.00 or more get 10% off.
static decimal DiscountRate(decimal orderTotal)
{
    if (orderTotal > 100.00m)   // defect: should be >=
    {
        return 0.10m;
    }
    return 0.00m;
}

decimal orderTotal = 100.00m;
decimal overcharge = orderTotal * 0.10m - orderTotal * DiscountRate(orderTotal);
Console.WriteLine($"Overcharge on one $100.00 order: ${overcharge:F2}");

// The same small defect, repeated across a year of orders at exactly $100.00
int ordersPerDay = 25;
int ordersAtThreshold = ordersPerDay * 365;
Console.WriteLine($"Orders affected in a year: {ordersAtThreshold}");
Console.WriteLine($"Customers overcharged in total: ${overcharge * ordersAtThreshold:F2}");
````

No single customer is likely to notice ten dollars, so the defect could run for months. This is exactly the kind of boundary error that the tests in Chapter 6 are designed to catch. Testing the values just below, at, and just above a boundary finds it:

````csharp
using System;

static decimal DiscountRate(decimal orderTotal)
{
    if (orderTotal >= 100.00m)   // fixed: the boundary value now qualifies
    {
        return 0.10m;
    }
    return 0.00m;
}

// Boundary tests: just below, exactly at, and just above the threshold
Console.WriteLine(DiscountRate(99.99m) == 0.00m ? "PASS 99.99" : "FAIL 99.99");
Console.WriteLine(DiscountRate(100.00m) == 0.10m ? "PASS 100.00" : "FAIL 100.00");
Console.WriteLine(DiscountRate(100.01m) == 0.10m ? "PASS 100.01" : "FAIL 100.01");
````

Fixing the code is only part of the response. The company must also decide what it owes the customers who were already overcharged. That is an ethical question, not a technical one.

## Thinking Through an Ethical Question

Ethical decisions in software are rarely about villains. They are usually about trade-offs under deadline pressure. A short checklist keeps the important questions from being skipped:

1. **Who is affected?** List the stakeholders: users, customers, coworkers, the organization, and people outside it.
2. **What could happen, and how bad would it be?** Consider likely and severe outcomes for each stakeholder.
3. **What do we owe them?** Consider laws, contracts, promises made to customers, and professional codes.
4. **Could we explain this decision publicly?** If a decision only works while nobody knows about it, reconsider it.

Applied to the discount defect:

| Question | Answer |
|---|---|
| Who is affected? | 9,125 customers who were overcharged; the company; the developer who must report the defect |
| What could happen? | Customers lose \$91,250 in total; if it comes out later, the company loses trust and may face legal action |
| What do we owe them? | The price the company advertised; honest disclosure; a fix that is tested |
| Could we explain it publicly? | "We found an error, fixed it, and refunded everyone" can be explained. "We fixed it quietly and kept the money" cannot. |

## Professional Codes of Ethics

Professions write down what they expect of their members. The *ACM Code of Ethics and Professional Conduct*, adopted in 2018 by the largest computing society, is the standard reference for computing.[^acm] A few of its principles apply directly to the work in this book:

| Principle | What it means for a programmer |
|---|---|
| 1.2 Avoid harm. | Consider who could be hurt by a feature or a failure before you ship it. |
| 1.3 Be honest and trustworthy. | Report defects and limitations, including your own. |
| 1.6 Respect privacy. | Collect and use personal data only for legitimate, stated purposes. |
| 2.1 Strive to achieve high quality in both the processes and products of professional work. | Test, review, and document, as in Chapter 6. |
| 2.5 Give comprehensive and thorough evaluations of computer systems and their impacts, including analysis of possible risks. | Say what could go wrong, not only what works. |
| 2.9 Design and implement systems that are robustly and usably secure. | Build in security from the start, as in the next section. |

## Summary

Software affects people who cannot see or check it. Small defects multiply at scale, so testing boundaries is an ethical habit as well as a technical one. When a decision affects others, work through who is affected, what could happen, what is owed, and whether the decision could be explained publicly. Professional codes such as the ACM Code describe what others are entitled to expect from you.
