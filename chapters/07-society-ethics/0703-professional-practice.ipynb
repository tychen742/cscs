# Professional Practice

```{contents}
:local:
:depth: 2
```

Professional programmers are responsible for more than whether code runs. They must respect other people's work, be honest about where code came from, and build software that everyone can use. This section covers those responsibilities as they apply to coursework now and to jobs later.

## Who Owns Code

Code is a creative work, so copyright protects it automatically: you may not copy someone else's code just because you can see it. A **license** is the owner's permission to use the code under stated terms. Two common open-source examples:

- The **MIT License** is *permissive*: anyone may use, copy, modify, and sell the code, as long as they keep the copyright and license notice.[^mit]
- The **GNU General Public License (GPL)** is *copyleft*: if you distribute a modified version, you must release it under the GPL as well.[^gpl]

Code written for an employer usually belongs to the employer. Code with no license at all is not free to reuse. A header comment is the conventional place to record ownership, license, and sources:

````csharp
// OrderTotals.cs
// Copyright (c) 2026 Ridgeline Supply Co.
// Licensed under the MIT License. See LICENSE in the project root.
//
// Portions adapted from the "Discount rules" example in Chapter 7.
// AI assistance: an AI assistant suggested the boundary tests; I reviewed,
// ran, and kept them.

using System;

static decimal ApplyDiscount(decimal total)
{
    return total >= 100.00m ? total * 0.90m : total;
}

Console.WriteLine($"{ApplyDiscount(120.00m):F2}");
````

## Using AI Assistants Responsibly

AI assistants can generate code quickly, but they do not take responsibility for it. You do. Four habits keep their use honest and safe:

1. **Follow the policy.** In this course, use AI only when the assignment authorizes it. At work, follow your employer's rules.
2. **Understand and test what you keep.** If you cannot explain a line or show a test for it, it is not yet your work.
3. **Say where code came from.** Record AI assistance and borrowed code in comments or in your submission, as in the header above.
4. **Do not paste private data into AI tools.** Customer records, passwords, and confidential code may leave your organization's control once they are sent to an outside service.

## Designing for Everyone

Software that only works for some people quietly excludes the rest. **Accessibility** means designing software that people with disabilities can use, for example with a screen reader or without a mouse. The *Web Content Accessibility Guidelines* (WCAG 2.2, a W3C Recommendation first published in October 2023) organize accessibility around four principles: content must be *perceivable*, *operable*, *understandable*, and *robust*.[^wcag]

Even a console program can follow these principles. Do not let color carry meaning by itself; say it in words:

````csharp
using System;

// Status messages that do not depend on color alone.
// A screen reader, a monochrome terminal, or a reader with color blindness
// still gets the meaning from the words.
static string StatusLine(bool ok, string message)
{
    string label = ok ? "[OK]" : "[ERROR]";
    return $"{label} {message}";
}

Console.WriteLine(StatusLine(true, "Order 1042 saved."));
Console.WriteLine(StatusLine(false, "Quantity must be from 1 to 500."));
````

Customers also read numbers and dates differently. The same price and due date look different in the United States, Germany, and Japan. .NET's `CultureInfo` formats values the way each audience expects:

````csharp
using System;
using System.Globalization;

decimal price = 1234.50m;

// The same value, formatted for customers in three countries
Console.WriteLine(price.ToString("C", new CultureInfo("en-US")));
Console.WriteLine(price.ToString("C", new CultureInfo("de-DE")));
Console.WriteLine(price.ToString("C", new CultureInfo("ja-JP")));

DateTime due = new DateTime(2026, 11, 3);
Console.WriteLine(due.ToString("d", new CultureInfo("en-US")));   // month first
Console.WriteLine(due.ToString("d", new CultureInfo("de-DE")));   // day first
````

A date written `11/3/2026` means November 3 to an American reader and 11 March to a German one. Formatting for the reader's culture prevents a costly misunderstanding.

## Communication and Sustainability

Two more responsibilities round out professional practice:

- **Communicate honestly.** Write error messages that tell users what happened and what to do next, and report project status accurately, including bad news such as a missed deadline or a defect you introduced.
- **Use resources wisely.** Computation consumes energy and hardware. An inefficient algorithm running on thousands of servers wastes both. Part V of this book shows how to measure and reduce an algorithm's cost.

## Summary

Respect licenses and give credit for code you did not write, including code from AI assistants. Keep confidential data out of outside tools. Design so that everyone can use your software: words instead of color alone, and formats that match each reader's culture. Communicate honestly about what your software does and does not do.
