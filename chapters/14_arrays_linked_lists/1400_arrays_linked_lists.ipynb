# Arrays and Linked Lists

A resizable product list and a chain of linked nodes can represent the same sequence while doing very different work. This chapter develops both representations and compares access, growth, insertion, and removal under explicit invariants. You will test the sequence contract and justify a representation for a workload; Chapter 16 develops search and sorting algorithms separately.

- **Contiguous storage**: Array-backed sequences keep elements next to each other for fast indexed access.
- **Dynamic array**: A dynamic array grows by allocating larger storage and copying existing values.
- **Linked node**: A linked structure stores values in nodes connected by references.
- **Insertion and deletion**: Different representations make middle updates cheaper or more expensive.
- **Performance tradeoff**: Choosing a sequence representation depends on access, growth, and update patterns.

<h2>Learning Objectives</h2>


By the end of this chapter, you will be able to:

1. Explain how array-backed and linked representations store the same sequence contract.
2. Trace resizing, shifting, and the total copy cost of repeated array-backed appends.
3. Implement generic array-backed and singly linked sequences while preserving count, head, and tail invariants.
4. Separate the cost of locating a position from the cost of insertion or deletion at that position.
5. Test empty, singleton, duplicate, and invalid-index cases and justify a representation using workload and memory-locality evidence.
