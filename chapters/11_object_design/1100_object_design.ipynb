# Object Design and Composition

An order and its line items form a collaborating model, not just a collection of individually correct classes. This chapter connects reference sharing and identity to structs, records, composition, and object responsibilities. You will make copying and mutation policies explicit, protect internal collections, and test collaborations; deeper nullability is supporting reading in this same sequence.

- **References and identity**: Trace when variables share one object and when domain equality differs from identity.
- **Structs and records**: Choose copying and equality behavior for a data model.
- **Composition**: Assign coherent responsibilities to collaborating objects.
- **Mutation and ownership**: Protect internal state and recognize shallow copies.
- **Contracts and missing data**: Check observable behavior and make optional data explicit.

<h2>Learning Objectives</h2>


By the end of this chapter, you will be able to:

1. Trace aliases and distinguish object identity from business equality.
2. Explain the effect of sharing mutable and immutable objects, and distinguish struct copying from record-class copying and equality.
3. Design collaborating product, order-line, and order classes.
4. Protect a private collection and explain shallow snapshot behavior.
5. Test boundaries and refactor while preserving observable behavior.

Supporting-reading outcome: distinguish required and optional object data, interpret nullable warnings, and test missing-data contracts.
