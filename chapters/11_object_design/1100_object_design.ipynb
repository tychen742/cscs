# Object Design and Composition

A useful application needs more than individually correct classes. This chapter connects objects into a small business model and makes their ownership, sharing, and failure policies explicit.

- **Reference**: A value identifying an object; several variables may refer to one object.
- **Identity**: Whether references identify the same runtime object.
- **Composition**: Building an object from smaller objects with defined responsibilities.
- **Immutability**: Keeping observable state fixed after construction.
- **Contract**: The promised results and failure behavior of public operations.

<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:

1. Trace aliases and distinguish object identity from business equality.
2. Explain the effect of sharing mutable and immutable objects, and distinguish struct copying from record-class copying and equality.
3. Design collaborating product, order-line, and order classes.
4. Protect a private collection and explain shallow snapshot behavior.
5. Test boundaries and refactor while preserving observable behavior.
6. Distinguish required and optional object data, interpret nullable warnings, and test missing-data contracts.
