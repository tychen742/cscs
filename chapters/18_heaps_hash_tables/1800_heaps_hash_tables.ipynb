# Hashing & Heaps

- **Priority queues**: Select work by urgency and define an explicit policy for ties.
- **Binary heaps**: Preserve complete shape and parent-child order with bubble-up and sink-down.
- **Hash dictionaries**: Associate unique keys with values using consistent equality and hashing.
- **Collisions and growth**: Trace chaining, probing, deletion markers, and rehashing.
- **Workload analysis**: Qualify costs, retain top-k values, and coordinate dispatch indexes.

The service desk connects priority selection to owner lookup and cancellation. The chapter includes complete teaching implementations alongside C# library applications.


> *[Data Structures - Full Course for Beginners](https://youtube.com/watch?v=J2fol8eWo64)* by Tech With Tim.
>
> This video remains hosted on YouTube and is the property of its original creator or rightsholder; it is embedded here via YouTube's official player for educational reference only.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:

1. Trace heap index relationships and both repair paths.
2. Implement and test a binary min-heap with empty, singleton, and duplicate cases.
3. Explain and apply equality, hashing, and duplicate-key contracts.
4. Implement chained lookup, replacement, removal, and rehashing.
5. Explain why open-addressed deletion needs a tombstone or another repair policy.
6. Distinguish expected, amortized, and worst-case operation costs.
7. Build top-k and dispatch workflows with explicit tie and cancellation rules.

<h2>Chapter Flow</h2>

- {doc}`18.1. Priority Queues and Heaps <1801_priority_queues>`
- {doc}`18.2. Hash Functions and Collisions <1802_hashing>`
- {doc}`18.3. Performance Comparison Lab <1803_performance_lab>`


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Priority queue | An ADT that removes items according to priority. |
| 2 | Priority | A value used to decide which item should come out next. |
| 3 | Heap | A nearly complete tree that maintains a parent-child priority rule. |
| 4 | Min-heap | A heap where each parent priority is less than or equal to its children. |
| 5 | Max-heap | A heap where each parent priority is greater than or equal to its children. |
| 6 | Complete tree | A tree filled level by level, left to right. |
| 7 | Bubble up | Move a newly inserted heap item upward until the heap property is restored. |
| 8 | Sink down | Move the root replacement downward until the heap property is restored. |
| 9 | Hash function | A function that converts a key into a hash code or table position. |
| 10 | Bucket | A slot or group of slots in a hash table. |
| 11 | Collision | A situation where different keys map to the same bucket. |
| 12 | Separate chaining | Collision handling that stores colliding entries in a secondary collection per bucket. |
| 13 | Open addressing | Collision handling that probes for another slot in the same table. |
| 14 | Load factor | A measure of how full a hash table is. |
| 15 | Dictionary | A key-value ADT supporting lookup by key. |
| 16 | Set | An ADT storing unique values. |
| 17 | HashSet | A C# collection for fast average-case membership tests. |
| 18 | Dictionary<TKey, TValue> | A C# collection for fast average-case key-value lookup. |

| 19 | Comparer | A policy for ordering values or testing key equality. |
| 20 | Stable tie order | A policy that preserves arrival order among equal priorities. |
| 21 | Rehashing | Recomputing bucket placement when table capacity changes. |
| 22 | Linear probing | Trying consecutive table slots with wraparound. |
| 23 | Tombstone | A deleted-slot marker through which lookup continues. |
| 24 | Top-k | Retaining the k largest or smallest observations. |
| 25 | Lazy deletion | Leaving obsolete queue entries until they are encountered during removal. |

<h2>Lecture Slides</h2>

Lecture slide decks for this chapter are pending.
