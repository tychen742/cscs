# Advanced Algorithms

This chapter develops direct string matching, reusable prefix information, computational growth, and the difference between exact answers and heuristic results. Connect each method to its result contract: which matches count, what work can be reused, and what evidence supports a claimed answer.

- **String matching**: Searching for a pattern inside a larger text.
- **Prefix reuse**: Retaining a proven matching suffix instead of restarting every comparison.
- **Computational limit**: A point where exact brute-force computation grows too quickly to be practical.
- **Approximation**: A method with a proved quality bound relative to the optimum under stated assumptions.

The goal is not to memorize every advanced algorithm, but to recognize how problem structure shapes what efficient computation can do.

<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:

1. Implement direct and prepared-prefix matching under explicit ordinal, overlap, and empty-pattern contracts.
2. Justify prefix fallback and linear comparison bounds, including preparation and output costs.
3. Validate exact and heuristic coverage witnesses and distinguish feasibility, optimality, and proved approximation guarantees.
4. Compare complete workloads and exact growth models without confusing measurements with general lower-bound proofs.
