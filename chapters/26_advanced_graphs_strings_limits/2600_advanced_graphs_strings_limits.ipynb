# String Matching and Computational Limits

A text search can reuse a proved prefix match, while an optimization search may still face rapidly growing work. This chapter connects direct matching and KMP to computational growth, exact witnesses, heuristics, and approximation guarantees. The three core sections emphasize result contracts and evidence; this chapter remains optional for the College of Business pathway.

- **String matching**: Searching for a pattern inside a larger text.
- **Prefix reuse**: Retaining a proven matching suffix instead of restarting every comparison.
- **Computational limit**: A point where exact brute-force computation grows too quickly to be practical.
- **Approximation**: A method with a proved quality bound relative to the optimum under stated assumptions.
