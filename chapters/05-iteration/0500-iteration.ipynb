<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Iteration | Repeating a block of code while a pattern or condition requires it. |
| 2 | Loop | A programming structure that repeats statements. |
| 3 | For loop | A loop commonly used when the number of repetitions is known or counted. |
| 4 | While loop | A loop that continues while a Boolean condition remains true. |
| 5 | Do-while loop | A loop that runs its body once before checking the condition. |
| 6 | Foreach loop | A loop that visits each item in a collection. |
| 7 | Loop variable | A variable used to track loop progress. |
| 8 | Condition | The Boolean test that controls whether a loop continues. |
| 9 | Accumulator | A variable that gathers a result across loop iterations. |
| 10 | Sentinel value | A special input value that tells a loop to stop. |
| 11 | Break | A statement that exits a loop immediately. |
| 12 | Continue | A statement that skips to the next loop iteration. |
| 13 | Infinite loop | A loop whose condition never becomes false. |
| 14 | Off-by-one error | A loop boundary mistake that runs one too many or one too few times. |
