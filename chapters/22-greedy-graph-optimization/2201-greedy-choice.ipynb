# Greedy Choice and Correctness

A greedy algorithm commits to a locally attractive choice. The central question is whether the local rule lines up with the global goal.

A useful greedy explanation names three things: the choice rule, the feasibility condition, and the reason earlier choices never need to be undone.


```{index} greedy algorithm
```

## A First Greedy Rule

In interval scheduling, the goal is to choose as many non-overlapping meetings as possible. A correct greedy rule is: always choose the compatible meeting that ends earliest.


````csharp
using System;
using System.Linq;

var meetings = new[]
{
    new Meeting("A", 1, 4),
    new Meeting("B", 3, 5),
    new Meeting("C", 0, 6),
    new Meeting("D", 5, 7),
    new Meeting("E", 8, 9)
};

int lastEnd = int.MinValue;
foreach (Meeting meeting in meetings.OrderBy(meeting => meeting.End))
{
    if (meeting.Start >= lastEnd)
    {
        Console.WriteLine($"Choose {meeting.Name}: {meeting.Start}-{meeting.End}");
        lastEnd = meeting.End;
    }
}

record Meeting(string Name, int Start, int End);
````

The chosen meetings remain feasible because each new meeting starts after the previous chosen meeting ends.

## When Greedy Fails

Not every local rule is safe. Choosing the highest-value job first can fail when deadlines limit how many jobs fit.


````csharp
using System;
using System.Linq;

var jobs = new[]
{
    new Job("A", Deadline: 1, Value: 80),
    new Job("B", Deadline: 2, Value: 70),
    new Job("C", Deadline: 2, Value: 70)
};

Console.WriteLine("Greedy by value:");
foreach (Job job in jobs.OrderByDescending(job => job.Value))
{
    Console.WriteLine($"Consider {job.Name} value={job.Value} deadline={job.Deadline}");
}

Console.WriteLine("A schedule with B and C earns 140, which beats choosing only A for slot 1.");

record Job(string Name, int Deadline, int Value);
````

The example is deliberately small. The point is to test a greedy rule against the real constraint, not just against a tempting score.

## Greedy Invariants

A greedy invariant states what remains true after every accepted choice. For interval scheduling, one invariant is: the selected meetings do not overlap.


````csharp
using System;
using System.Collections.Generic;
using System.Linq;

var meetings = new[]
{
    new Meeting("A", 1, 3),
    new Meeting("B", 2, 4),
    new Meeting("C", 3, 5),
    new Meeting("D", 5, 6)
};

var selected = new List<Meeting>();
int lastEnd = int.MinValue;

foreach (Meeting meeting in meetings.OrderBy(meeting => meeting.End))
{
    if (meeting.Start >= lastEnd)
    {
        selected.Add(meeting);
        lastEnd = meeting.End;
        Console.WriteLine($"selected: {string.Join(", ", selected.Select(m => m.Name))}");
    }
}

record Meeting(string Name, int Start, int End);
````

The invariant alone does not prove optimality. It proves feasibility. A correctness proof also needs a reason the chosen feasible solution is as good as any alternative.

## Exchange Arguments

An exchange argument shows that if some optimal solution did not make the greedy choice, we can exchange part of that optimal solution for the greedy choice without making it worse.

For interval scheduling, the earliest-ending compatible meeting leaves at least as much room for later meetings as any other first compatible meeting.


````csharp
using System;

int greedyEnd = 4;
int otherEnd = 6;

Console.WriteLine(greedyEnd <= otherEnd
    ? "The greedy meeting leaves no less room for future meetings."
    : "The local rule would need a different justification.");
````

This is a proof sketch, not a full formal proof. The habit matters: do not trust a greedy algorithm until you can explain why the local choice is globally safe.

## Greedy Design Checklist

Before committing to a greedy algorithm, ask:

1. What is the local choice rule?
2. What constraints must remain feasible?
3. What invariant does each choice preserve?
4. Can I build a counterexample?
5. If I think it is correct, what exchange or stay-ahead argument explains why?
