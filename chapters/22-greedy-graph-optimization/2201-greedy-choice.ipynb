# Greedy Choice and Proof

A greedy algorithm commits to a locally attractive choice. Learn to state the choice, identify the invariant it preserves, and explain when the strategy is correct.

```csharp
using System.Linq;

record Meeting(string Name, int Start, int End);

var meetings = new[]
{
    new Meeting("A", 1, 4),
    new Meeting("B", 3, 5),
    new Meeting("C", 0, 6),
    new Meeting("D", 5, 7),
    new Meeting("E", 8, 9)
};

int lastEnd = int.MinValue;
foreach (Meeting meeting in meetings.OrderBy(item => item.End))
{
    if (meeting.Start >= lastEnd)
    {
        Console.WriteLine(meeting.Name);
        lastEnd = meeting.End;
    }
}
```
