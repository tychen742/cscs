# Greedy Algorithms

Greedy algorithms make locally best choices and hope those choices lead to a globally useful solution. This chapter studies when that strategy works and how it applies to graph optimization problems.

- **Greedy choice**: A greedy algorithm commits to a locally best option at each step.
- **Invariant**: A greedy proof often tracks a property preserved by each choice.
- **Exchange argument**: An exchange argument shows that a different optimal solution can be adjusted to include the greedy choice.
- **Minimum spanning tree**: A spanning tree connects all vertices with minimum total edge weight.
- **Shortest path**: Shortest-path algorithms find low-cost routes under specific assumptions.

Greedy design is powerful, but only when the problem structure justifies the choices the algorithm refuses to revisit.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Describe a greedy choice and the local criterion it uses
2. Identify an invariant preserved by a greedy algorithm
3. Explain why a greedy strategy may or may not be correct
4. Apply minimum-spanning-tree ideas to weighted networks
5. Apply shortest-path reasoning under nonnegative edge weights
