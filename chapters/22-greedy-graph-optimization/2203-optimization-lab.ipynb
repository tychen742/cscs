# Greedy Optimization Lab

Compare greedy choices with alternative strategies on scheduling, network, and routing problems.

```csharp
using System.Linq;

record Job(string Name, int Deadline, int Value);

var jobs = new[]
{
    new Job("Report", 2, 80),
    new Job("Audit", 1, 100),
    new Job("Backup", 2, 50),
    new Job("Plan", 3, 40)
};

foreach (Job job in jobs.OrderByDescending(job => job.Value))
    Console.WriteLine($"Consider {job.Name}: deadline {job.Deadline}, value {job.Value}");
```
