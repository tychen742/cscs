## Create and Run Your First Console Application



Open your terminal and navigate to the workspace folder you created in Appendix B. If you are still inside `csharprepl`, enter `exit` first so that you can type shell commands. Run these commands one at a time:

```bash
dotnet new console --use-program-main -n HelloWorld
cd HelloWorld
code .
```

The first command creates a console project in a new folder named `HelloWorld`. The `--use-program-main` option makes the template show the `Program` class and `Main` method explicitly.[^1] The second command moves the shell into that folder. The third opens the folder in VS Code. If `code .` is unavailable, use **File > Open Folder** in VS Code and select `HelloWorld`.

A **project** groups source files and the settings needed to build an application. The `HelloWorld` folder is your project folder. In VS Code's **Explorer**, expand that folder and identify these two files:

| File | What you do with it |
| --- | --- |
| `Program.cs` | Open and edit the C# source code. The `.cs` extension identifies a C# source file. |
| `HelloWorld.csproj` | Inspect the project settings. The `.csproj` extension identifies a C# project file. |

Opening the project folder lets VS Code work with the source files and their build settings together. The folder is where you keep the files; the `.csproj` tells .NET how to build the application. The project name, `HelloWorld`, is not itself an executable C# instruction.

In VS Code, open `Program.cs`. Its contents should look like the following complete program:


`````csharp
namespace HelloWorld;

class Program
{
    static void Main(string[] args)
    {
        Console.WriteLine("Hello, World!");
    }
}
`````

Run the application from the terminal while you are in the `HelloWorld` folder:

```bash
dotnet run
```

The output is `Hello, World!`. The application then ends and the shell prompt returns. In VS Code, you can use **Terminal > New Terminal**; check that its current folder is `HelloWorld`.

## Read the Program from the Outside In



C# programs have both **file organization** and **code organization**. On disk, a project contains source files and a `.csproj` file. Inside a source file, namespaces organize type names, classes contain members such as methods, and methods contain executable statements.

For this first application, the pieces fit together like this:

```text
HelloWorld/                         Project folder opened in VS Code
├── HelloWorld.csproj               Build settings
└── Program.cs                      C# source file
    └── namespace HelloWorld        Groups type names
        └── class Program           Contains the Main method
            └── Main(...)           Application entry point
                └── Console.WriteLine(...)   Executable statement
```

The lower part of this diagram describes code inside `Program.cs`, not additional folders. A namespace is a grouping of names; it does not create a directory. In this application, .NET starts the `Main` method, which executes the greeting statement.

| Part | Role in this application |
| --- | --- |
| `namespace HelloWorld;` | Places the `Program` class in the `HelloWorld` namespace. |
| `class Program` | Declares a class named `Program` that contains `Main`. |
| `static void Main(string[] args)` | Declares the method where execution begins. |
| `Console.WriteLine("Hello, World!");` | Writes the greeting and moves to the next output line. |

C# is case-sensitive. `Main` and `main` are different names, as are `Console` and `console`.

### Namespace: Give Related Types a Shared Name



A **namespace** groups related types, such as classes, under a shared name. `namespace HelloWorld;` applies to the rest of this file; this is a **file-scoped namespace**. The class's **fully qualified name** is `HelloWorld.Program`. Another namespace can contain its own `Program` class.

A namespace organizes names; it does not start execution or create a folder. Keep the generated name for now.

### Class: A Container for Methods



`class Program` declares a class named `Program`. For now, use it as the container for the methods in your application. The opening and closing braces after the declaration mark the class body. The `Main` method is inside those braces.

You will add methods to this class in Chapter 3 and study classes in more detail in Chapter 10.

### Main: Where Execution Begins



A **method** is a named block of code that performs a task. `Main` is this application's **entry point**, the method .NET calls when the application starts.

For now, recognize `static void Main(string[] args)` and put the executable statements inside its braces. `void` says this method returns no value; `static` lets it run without creating a `Program` object. `args` receives command-line arguments, which this application does not use. Methods, arrays, and objects receive their own treatment later.

### Console.WriteLine: Display Output



`Console.WriteLine("Hello, World!");` is a **statement**, an instruction the program executes. `Console` is a .NET library class for console input and output. `WriteLine` displays the supplied value and then ends the output line.

The text inside double quotation marks is a **string literal**, text written directly in the source. The quotation marks are delimiters and are not printed. Parentheses enclose the value supplied to the method; the semicolon ends the statement.

`Console` belongs to the `System` namespace. The generated project enables **implicit usings**, making common namespaces available automatically.[^1] In browser cells, `using System;` makes the examples independent of that project setting. A `using` directive makes names available; a `namespace` declaration groups your own types.

## Statements, Expressions, and Literal Values



An **expression** produces a value. A **literal** writes a value directly: `"Northwind Store"` is text, and `3` is a number. The expression `2 + 1` produces `3`; the statement `Console.WriteLine(2 + 1);` tells the program to display that value.

Text and numbers have different roles. `"2 + 1"` is text containing five characters, including spaces; C# does not calculate the addition inside ordinary quotes. Predict the next two output lines before running.

`````csharp
using System;
Console.WriteLine(2 + 1);
Console.WriteLine("2 + 1");
`````

**String interpolation** inserts expression values into text. Begin the string with `$`, then put each expression inside `{` and `}`. In `$"Items ready: {2 + 1}"`, the braces mark an expression inside text; they do not define a code block. This lets the store label its output clearly without introducing variables yet.

`````csharp
using System;
Console.WriteLine($"Items ready: {2 + 1}");
`````

## Names, Keywords, and Case



An **identifier** names something in a program, such as `Program` or `Main`. C# is **case-sensitive**: `Console`, `console`, and `CONSOLE` are different spellings. Library names must match their declarations exactly, so use `Console.WriteLine`, including the capital `C`, `W`, and `L`.

A **keyword** has a language-defined role. **Reserved keywords**, such as `class`, `static`, and `void`, are reserved for the language itself and cannot be ordinary identifiers. You have already seen them in the first console program; do not replace them with names of your own. Ordinary identifiers begin with a letter or underscore and can include digits after the first character. `OrderDesk` is valid; `2Orders` cannot be an ordinary identifier, and `order-desk` is read as separate tokens.

Use **PascalCase**, capitalizing each word, for your classes and methods, such as `OrderDesk` and `PrintReceipt`. Later, use **camelCase**, starting with a lowercase letter, for local variables and parameters, such as `itemCount`. These naming conventions help readers; they are different from rules enforced by the compiler. You will learn variables in Chapter 2 and method design in Chapter 3.



The collapsed list is a reference, not a memorization task. Learn each keyword as you encounter it. Chapter 2 introduces an optional note on {ref}`escaped identifiers and contextual keywords <keywords-section>`.

## Read a Method Call



In `Console.WriteLine("Northwind Store")`, the dot selects the method by name, the parentheses mark the call, and the text is an **argument**, a value supplied to that call. Commas separate multiple arguments.

Some methods produce a **return value**, a value supplied back to the calling code. `Math.Max(2, 3)` returns the larger of two quantities. The next example displays the larger of two pickup-batch sizes. C# evaluates the inner call first, then passes its returned value to `WriteLine`.

`````csharp
using System;
Console.WriteLine(Math.Max(2, 3));
`````

`WriteLine` produces visible output but returns no value. That differs from `Math.Max`, which returns a value but does not itself print it. For now, practice reading calls; Chapter 3 teaches you to define your own methods.

## Blocks, Formatting, and Comments



A **block** groups code between curly braces, `{` and `}`. In the complete program, one pair encloses the class body and another encloses the `Main` body. Indent the statements inside each block so readers can see the nesting. Braces define that structure for C#; indentation makes it visible.

**Whitespace** includes spaces, tabs, and line breaks. Outside quoted text, extra whitespace often changes readability without changing behavior. You still need separation between words such as `class` and `Program`. Inside a string literal, spaces are part of the displayed text.

Most simple statements end with a **semicolon**. Two statements can share a source line, but use one per line for readability. Wrapping one method call across source lines does not create extra output lines. The closing braces of this class and method need no semicolons.

A **comment** explains intent to a reader and is ignored during execution. `//` begins a comment through the end of the line; `/*` and `*/` enclose a block comment. Explain a useful reason or requirement rather than merely repeating the code.


`````csharp
using System;
// Keep the opening time visible to customers.
Console.WriteLine(
    "The order desk opens at 9:00.");
/* Use a separate line for the collection instruction.
   Customers should not have to infer where to go. */
Console.WriteLine("Collect at the order desk.");
`````

Predict how many lines this prints. Run it to check that source line wrapping and comments add no output lines. Change only indentation and run again; then add a space inside the quoted collection instruction and observe the difference.

## Read and Repair Compiler Messages



A **syntax error** violates the language's writing rules. A missing semicolon prevents compilation; the revised program cannot run until you repair it. A compiler **diagnostic** reports a problem with a code, a message, and usually a source location.

In your saved project, remove the semicolon from one `WriteLine` statement and run `dotnet build`. Look for `CS1002` and `; expected`. Start with the first reported error, inspect that location and the preceding line, restore the semicolon, save, and build again. One mistake can cause several messages, so repair and rebuild before changing unrelated code.

A misspelled name such as `console` is also a compile-time error, but it is a name-resolution problem rather than a punctuation error. It commonly produces `CS0103`, saying the name does not exist in the current context. Compare capitalization with the working example.

## Recognize Top-Level Statements



You may encounter a console application's `Program.cs` containing only:

```csharp
Console.WriteLine("Hello, World!");
```

This uses **top-level statements**, executable statements written outside an explicit class or method. The compiler supplies the entry point. The default `dotnet new console` template uses this shorter form; `--use-program-main` requests the explicit form used in this lesson.[^1]

Both forms can be console applications, and both project forms run with `dotnet run`. For the application you created here, continue writing statements inside `Main`. When later examples show only a few statements, you can place those statements inside `Main` to run them in your project. Replace the example statements rather than appending a second copy of the class or another entry point.
