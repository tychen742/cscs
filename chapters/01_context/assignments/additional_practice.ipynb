## Review Questions

```{index} Review Questions
```

1. What is the input-process-output (IPO) model in computer science and information systems? Apply it to one information system and identify stored data and feedback.
2. What are the basic constructs of programming? Explain sequence, selection, and iteration, and distinguish them from the supporting ideas of variables, types, operators, subroutines, nesting, and arrays. You will study decisions and loops in later chapters; describe their purpose now.
3. What is an algorithm, and why is it important in computer science and information systems? Give instructions precise enough for another person to follow.
4. Why does a hypothetical **32-bit, byte-addressed address space** contain at most `2^32` byte locations (4 GiB)? Explain why this address-space calculation alone does not prove that every computer described as “32-bit” can install no more than 4 GiB of physical RAM. Processor extensions, operating-system limits, and usable address space are separate issues.
5. What is a bit, and what is a byte? How many different patterns can one byte represent? Explain how a pattern can have different meanings under different encodings.
6. Describe compilation and interpretation and give two implementation examples of each. Explain where ordinary C# compilation, intermediate language, and JIT execution fit. Avoid classifying every implementation of a language permanently as compiled or interpreted.
7. Is computer science all about writing computer programs? Describe important areas beyond coding and explain how mathematical reasoning, engineering design, and empirical investigation contribute.

## Check Your Explanations

```{index} Check Your Explanations
```

Use the chapter's worked examples to check your answers. For question four, derive `2^32` before converting bytes to GiB. For question six, examples of compilation include C# source-to-IL and ahead-of-time native C compilation; examples of interpretation include a command shell executing commands and CPython executing bytecode. These examples describe particular execution techniques, not disjoint categories of languages.

For question seven, discuss at least two areas such as algorithms, information representation, computer systems, human-computer interaction, security, or the social effects of computing. Explain an actual question each area investigates.
