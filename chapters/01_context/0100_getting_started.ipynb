# Computing, Programs, and Responsibility

Computing begins with a problem, a precise procedure, and responsibility for the people affected by the result. This chapter introduces those ideas through an appointment reminder, then connects them to a first saved C# program.

- **Computing**: Hardware, software, and data work together to process information.
- **Problem solving**: Inputs, algorithms, and observable outputs define a task you can check.
- **Responsibility**: Privacy, accessibility, attribution, and honest communication shape program design.
- **Execution**: Source code is built and run through a compiler, runtime, editor, and terminal.
- **First program**: Sequential statements produce output you can predict, change, and verify.

<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:

1. Describe hardware, software, data, and the input-process-output model.
2. Express a small business task as a precise procedure with checkable output.
3. Identify stakeholders and apply questions about harm, necessary data, and responsible action.
4. Produce clear output using fictional data and explain privacy, accessibility, and attribution choices.
5. Explain the roles of source code, the compiler, runtime, editor, and terminal.
6. Create, edit, save, and run a C# console project and identify its entry point.
7. Distinguish an AI-generated suggestion from verified behavior and describe how to check it.
8. Read basic C# syntax, predict labeled output, and repair simple compiler errors.
