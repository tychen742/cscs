# Computing and Professional Responsibility


```{contents}
:local:
:depth: 2
```


Computing systems help organizations sell goods, pay employees, manage collections, and provide services. Their results influence what people believe and what they can do. A stock report can cause a manager to reorder supplies. An overdue notice can cause a customer to pay a fee. An account restriction can prevent someone from completing a task. Understanding a program therefore includes understanding the work and decisions around it.

**Professional responsibility** means taking care in the work you perform, explaining its limits honestly, and considering the people affected by it. This section introduces that responsibility before you begin writing larger programs. You do not need C# syntax to understand the examples. Each example connects a computing decision to a consequence; later chapters give you the programming tools to examine and improve such decisions.

The Association for Computing Machinery (ACM) publishes a professional code addressing harm, honesty, fairness, privacy, and the quality of computing work.[^acm] A code supplies principles for judgment. Applying them still requires understanding the particular people, information, and task involved.


## Identify Stakeholders and Possible Harm

```{index} stakeholder; social impact; software responsibility
```


A **stakeholder** is a person or group affected by a system or a decision about it. The person who operates the software is one stakeholder, but others may never touch it. In a bookstore, a cashier uses the sales system, a customer relies on its receipt, a manager uses its reports, and warehouse staff act on its stock requests. A change that saves the cashier time might create extra work for another group.

Consider a bookstore that introduces automatic overdue notices for rented textbooks. Its goal is reasonable: reduce the time staff spend checking rentals and help customers return books. The system compares a stored due date with a recorded return date and selects customers to notify. Before examining any code, we can identify several possible consequences.

| Stakeholder | What they rely on | What could go wrong |
| --- | --- | --- |
| Customer | Accurate return status and a way to correct mistakes | An incorrect notice causes worry or an unfair charge. |
| Desk staff | Records that reflect accepted returns | Staff must investigate repeated complaints. |
| Manager | Reports that distinguish outstanding rentals from record errors | A misleading report leads to a mistaken business decision. |
| Another person using a shared account or device | Appropriate limits on information displayed | A notice exposes someone else's borrowing details. |

**Harm** is a negative consequence, such as financial loss, disclosure of information, exclusion from a service, or damage to trust. A useful analysis identifies the consequence and who experiences it. Saying that a program is “bad” does not explain what needs to change.

Automation can repeat an error across many records before anyone notices. If a return-processing step is delayed, a nightly notice program might contact every affected customer. Faster execution would make that particular problem spread faster. The appropriate response might include checking how return records are updated, reviewing uncertain cases, and giving customers a correction route. It is not necessarily a faster computer.

Four questions help organize an initial review: **Who is affected? What could go wrong? What information is necessary? What should change or be disclosed?** These questions lead from a general concern to an issue that an organization can investigate.


**Check Your Understanding:** A bookstore's overdue report is sent only to its manager. Does that mean customers cannot be affected by an error in it? Explain the connection between the report and a later decision.


```{dropdown} Suggested Answer
Customers can still be affected. The manager might use the report to apply fees or restrict rentals. A person does not have to read a report or operate a program to be a stakeholder in the decisions it supports.
```


## Distinguish Correct Execution from Sound Decisions

```{index} correctness; reliability; fairness; business rules
```


A program can follow its instructions exactly and still support an unsuitable decision. There are several different questions to ask about a result:

- **Implementation:** Did the program perform the specified calculation or procedure?
- **Data quality:** Were the records accurate, complete enough, and current enough for that task?
- **Requirements:** Did the specified procedure express the intended business rule?
- **Consequences:** Is it appropriate to use this result for the proposed decision?

Suppose a fictional store offers a discount on purchases of **at least $100**. A program gives the discount only on purchases **greater than $100**. A $100 purchase exposes a mismatch between the requirement and the implementation. A completed run or an attractive receipt cannot establish that the boundary is correct. Chapter 3 will give you the language tools for expressing such conditions; Chapter 6 develops ways to test them.

Now suppose a payroll report adds every stored work hour correctly, but one employee's approved time sheet has not been imported. The addition can be correct while the reported pay is wrong. Checking the arithmetic alone would miss the absent record. This is why a useful report should make its coverage clear, such as identifying the period and whether all expected time sheets were received.

**Reliability** concerns whether a system performs its intended function consistently under its stated conditions. Repeatedly producing the same answer does not make that answer right. A consistent omission is still an omission. When a result is incomplete or uncertain, a responsible developer communicates that limitation rather than presenting it as a confirmed fact.

**Fairness** concerns how rules and their effects treat people. A rule that applies identically to everyone can still create a barrier. For example, a store might require customers to use a smartphone application to request a correction. Customers without a compatible phone would have no usable route. An alternative desk or telephone process addresses a different issue from fixing the application's arithmetic.

A beginning programmer is not expected to settle every organizational dispute. The useful habit is to identify assumptions, ask who decides the rule, and raise concerns supported by a concrete example. A technical result should not silently become a claim that the underlying policy is appropriate.


## Handle Personal Information with Care

```{index} privacy; personal data; data minimization; security
```


**Personal data** identifies a person or can be connected to one. Names and contact details are familiar examples. Purchase history, borrowing history, and account activity can also describe someone when linked to an identifier. Replacing a name with a customer number does not necessarily make a record anonymous: another record may connect the number to the person.

**Privacy** concerns appropriate collection, use, sharing, and retention of information about people. **Security** concerns protecting information and systems from unauthorized access or changes and keeping services available. The concerns overlap, but they are not interchangeable. Restricting access to a report does not establish that its recipients need every detail in it. Removing a name from a report does not establish that its storage is protected.

**Data minimization** means limiting the information handled to what the purpose needs. Begin with the task, then justify the fields. Consider a manager asking how many notebooks were sold each day last week. Daily totals and product information may answer that question. Individual customer names and telephone numbers do not become necessary merely because the sales system has them.

| Task | Information relevant to the task | A detail that needs a separate justification |
| --- | --- | --- |
| Prepare an order for collection | Order identifier, items, quantities, and collection status | The customer's unrelated purchase history |
| Report daily product sales | Product, date, quantity, and amount | Individual customer contact details |
| Investigate a disputed charge | The relevant transaction and authorized account information | A complete export of every customer's account |

These are starting points for explaining a purpose, not universal field lists. A delivery task, for example, has different information needs from an anonymous daily sales total. The organization must establish who is authorized to receive the information and how it will be handled.

Output includes more than the main screen. A diagnostic log, exported spreadsheet, screenshot, or copied error report can carry information outside its original setting. A developer investigating one failed order might accidentally share an entire customer export. It is better to isolate the relevant technical facts and reproduce the issue with **synthetic data**, deliberately invented data that resembles the needed structure without describing real customers.

For your work in this book, use fictional records and avoid putting real customer details or credentials into demonstrations. Later, the {doc}`privacy and security practice <../09_files_text/0905_privacy_security>` connects these concepts to text processing and program behavior. You will then have concrete input and output to inspect, rather than having to guess how a complete business system works.


## Make Information Understandable and Accessible

```{index} accessibility; inclusive design; communication
```


**Accessibility** means designing a system so people with disabilities can use it. It involves how people perceive information, understand it, navigate the system, and perform actions. A screen that works for its developer may still present a barrier to someone using a keyboard, a screen reader, enlarged text, or another way of interacting.[^accessibility]

Consider an order-status display that uses a green square for “ready” and a red square for “delayed.” A customer who cannot distinguish those colors needs another way to learn the status. Adding the words “Ready for collection” and “Delayed” communicates the distinction in text. Choosing two different colors alone would not address that need. Similarly, an unlabeled button may look obvious beside a picture while giving a screen-reader user no useful description of its action.

Understandable messages are another part of usable communication. Compare the following fictional order-status responses:

| Message | What the reader can determine |
| --- | --- |
| “Error 19” | A problem occurred, but its meaning and next action are unclear. |
| “Invalid!” | The program rejected something, but does not identify what to correct. |
| “Enter a collection date on or after October 16.” | The reader has a specific condition to check. |

A message should also describe what the system actually knows. “Your order was saved” is misleading if saving failed. “Your order has not been saved. Try again or contact the desk” separates the outcome from the next action. The date and contact route in a real application must come from its actual rules and services, not a developer's convenient example.

Context affects interpretation. A date such as 03/04 can be read in different ways. A time without an a.m./p.m. distinction or a needed time zone can cause confusion. Information should be expressed for the intended audience, and a review should include people with different access needs. These concerns belong in design and evaluation, rather than being left until the end as visual polish.

You do not need to build a user interface in this lesson. The immediate connection to computing is that a program's output must communicate a meaningful result to its users. Later chapters will let you produce and format that output yourself.


**Check Your Understanding:** A stock report shows low-stock items only in red. The developer adds the words “Reorder needed” beside those items. What barrier does that change address, and what does it leave unchecked?


```{dropdown} Suggested Answer
The text makes the low-stock status available without requiring readers to distinguish red from other colors. It does not establish that the quantities are correct or that the rest of the report is accessible. For example, a reader may still need meaningful headings or a usable reading order.
```


## Respect Sources and Explain Your Work

```{index} attribution; software license; accountability; professional ethics
```


Computing work builds on other people's ideas, examples, libraries, and documentation. **Attribution** identifies the source of work you use. A software **license** describes permissions and conditions for using software. Giving credit and having permission are related but different questions. Adding an author's name to copied code does not replace checking the terms under which that code was supplied.

In a course, permission also depends on the assignment. An instructor may permit discussing an approach while requiring each student to write a solution independently. An example that is appropriate for private study might not be permitted in a submitted answer. Read the course rules and ask about a specific uncertainty before relying on outside work.

Suppose you find a small example that formats a receipt. A useful record would identify its author or organization, its location, the part you used, and what you changed. If the example has stated reuse conditions, keep the information needed to follow them. “Found online” tells a reviewer too little to locate or evaluate the source.

**Accountability** means being able to explain your decisions and the evidence behind your work. Consider two descriptions of the same program: “It works” and “I checked a purchase below the discount threshold, exactly at it, and above it; returns have not been tested.” The second description lets another person assess both the evidence and its limits. You will learn systematic testing later; even now, you can distinguish a demonstrated observation from a claim about every possible case.

Responsibility also includes responding honestly to mistakes. If a report omitted some transactions, disclose the affected period and help identify which decisions used it. Quietly replacing the file may leave someone relying on an older copy. Detailed incident procedures belong to organizational practice, but the underlying habit is accessible to a beginner: preserve relevant evidence, describe what is known, and avoid claims you cannot support.

## Understand the Role of AI Assistance

```{index} AI assistance; generative AI; large language model; prompt; verification
```


**Artificial intelligence (AI)** is a broad area of computing concerned with systems that perform tasks such as recognizing patterns, making predictions, or generating content. **Generative AI** produces content such as text, images, and source code. A **large language model (LLM)** is a model used to generate text, including programming explanations, from a **prompt**, the instructions and context supplied to it. An AI coding assistant may combine a model with tools for reading files, running programs, or proposing edits.

Using such an assistant while learning is different from building an application that calls an AI service. This course's foundations do not require a paid assistant or a live service. When later activities examine a proposed AI answer, prepared examples let everyone inspect the same material.

An assistant can explain unfamiliar wording or suggest an approach. Its response is still something to evaluate. A fluent explanation may contain a wrong rule, omit an important condition, or claim that a check succeeded without showing evidence. Computing educators therefore distinguish opportunities for learning from risks of over-reliance and difficulties assessing independent understanding.[^ai-education]

For a conceptual example, suppose a fictional store offers a discount on purchases of at least $100. An assistant explains this as “purchases over $100 qualify.” The wording sounds close, but it leaves out the exact $100 case. You can identify the discrepancy from the requirement without knowing any C#. A polite, detailed response would not resolve it. The relevant evidence is how the proposed rule handles the boundary.

In later programming work, separate three questions: **Do I understand the suggestion? Does it meet the requirement? What checks support that conclusion?** A program that compiles has met certain language rules. That alone does not show that its discount, report, or record update is correct. Some assistants can run tests, but you still need to know what was run and what those checks covered.

When your course permits assistance, begin with the task and your own current understanding. “Explain how this example distinguishes hardware from software” identifies a learning need. Asking for a complete submitted answer can bypass the reasoning the assignment is intended to develop. Neither the existence of a tool nor acknowledging it establishes permission to use it on a particular assessment.

Prompts also disclose information to the service receiving them. Use invented examples and follow your organization's rules for approved services. Do not paste credentials, customer records, or private course material simply because a tool asks for more context. Removing a name while retaining an account number and transaction history may leave the record identifiable.

Record assistance as your course requires, including what you used and how you evaluated it. A learner should be able to explain a retained answer and recognize where it might fail. Methods, testing, and text-processing chapters revisit this responsibility with concrete programs, predictions, and checks.


The common thread is that computing work connects a purpose, an implementation, and people who rely on its results. A responsible explanation makes the connections visible: which task the system serves, whose information it handles, what its result means, and which limitations remain. The next section begins the programming workflow. It will give you a way to turn small, stated requirements into executable instructions; these broader questions remain relevant as those programs grow.


```{rubric} Footnotes
```
