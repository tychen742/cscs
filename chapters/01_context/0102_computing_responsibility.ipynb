# Computing and Professional Responsibility


```{contents}
:local:
:depth: 2
```


A program affects people through the decisions it supports, the information it exposes, and the work it repeats. A developer therefore asks both whether a program runs and whether its behavior is appropriate for its purpose. Professional responsibility starts with your first program.[^acm]

Return to the appointment service. Its manager proposes putting the customer's name, telephone number, appointment time, and reason for the visit in every reminder. The information exists in the scheduling record, but that alone does not establish that every reminder needs it.


## Identify Stakeholders and Possible Harm

```{index} stakeholder; social impact; software responsibility
```


A **stakeholder** is a person or group affected by a decision. The reminder service's stakeholders include customers, reception staff, the business, and people who might see a shared screen or notification. Their interests can differ: detailed messages may help staff while exposing information a customer expected to remain private.

Ask four questions before implementing a feature:

1. **Who is affected?** Include people who do not operate the program directly.
2. **What could go wrong?** Consider incorrect results, unwanted disclosure, exclusion, and repeated mistakes.
3. **What information is necessary?** Connect each displayed or stored item to the task.
4. **What should change or be disclosed?** Revise the design and explain remaining limitations honestly.

For this case, a wrong time could cause a missed appointment, and a detailed notification could reveal the reason for a visit to someone nearby. The design needs accuracy and restraint. A technically successful notification can still be inappropriate.

**Checkpoint:** Identify a stakeholder affected by an incorrect overdue notice from a bookstore. Explain one harm besides a wrong number on a screen.


## Use Only the Information the Task Needs

```{index} privacy; personal data; data minimization
```


**Personal data** identifies a person or can be linked to one. **Privacy** concerns appropriate handling of information about people. **Data minimization** means collecting and retaining only what the task needs. Apply the same question to output: information in a record does not automatically belong in a message or log.

A reminder can give the appointment time and a contact route without displaying a diagnosis, account balance, or other unrelated details. The destination used to send a message need not be repeated in its body. Use fictional data when developing and demonstrating the program.

This complete example shows a restrained message. It displays text locally; it sends nothing.


````csharp
using System;

Console.WriteLine("Appointment reminder");
Console.WriteLine("Your appointment is at 9:00.");
Console.WriteLine("Contact the office if you need to reschedule.");
````

Leaving a detail out of this message does not by itself secure the scheduling system. Access controls, validation, and safe storage are later responsibilities. The {doc}`privacy and security practice <../09_files_text/0905_privacy_security>` revisits those responsibilities once you can write methods and process text.


## Make Output Understandable and Accessible

```{index} accessibility; inclusive design; communication
```


**Accessibility** means making a system usable by people with different abilities and access needs. Begin with clear text: explain what happened and what action is available. Do not make color the only way to distinguish success from failure. An error message that says only “red status” fails a reader who cannot perceive that color.

Compare “Error 19” with “The appointment could not be saved. Check the time and try again.” The second gives the user an action. It should also be truthful: do not claim an appointment was saved when the program has not saved it.

Dates, times, and language can be ambiguous. State enough context for the intended audience rather than assuming every reader interprets a short numeric date the same way. You will practice numeric and text formatting in later chapters.

**Checkpoint:** Rewrite “Invalid!” as a useful message for an appointment time outside office hours. Avoid echoing personal details.


## Attribute Work and Check AI Assistance

```{index} attribution; software license; AI assistance; professional ethics
```


**Attribution** identifies where work came from. A software **license** states permissions and conditions for using code. Record sources and check applicable terms before reusing code. Follow your course's rules for collaboration and AI assistance; acknowledging a tool does not replace permission to use it.

**Artificial intelligence (AI)** is a broad area of computing concerned with systems that perform tasks such as recognizing patterns, making predictions, or generating content. **Generative AI** produces content such as text, images, or code. A **large language model (LLM)** is a kind of model used to generate text, including programming explanations and source code, in response to a **prompt**, the instructions and context supplied to it. Using an AI coding assistant is different from building an application that calls an AI service.

A convincing explanation can still contain a mistaken rule, an invented library method, or an incorrect result. Some assistants can run tools and tests, but an answer alone does not establish which checks actually ran. Ask for the evidence and reproduce relevant checks in your own environment. Compiling checks language rules; testing checks selected behaviors; neither automatically establishes that the business requirements are right.[^ai-education]

Use this learning workflow when your course permits AI assistance:

1. State the requirement and predict a small example before asking for help.
2. Attempt the task or identify the exact point you do not understand.
3. Ask a focused question, such as “Explain why these two output lines differ,” rather than requesting an entire assignment.
4. Read the suggestion and explain its inputs, operations, and output yourself.
5. Run it, try a boundary or counterexample, and decide what to keep or change.
6. Record the assistance and the checks you performed as your course requires.

Use synthetic examples in prompts. Remove personal information, credentials, and private course or organizational records; replacing only a person's name may leave identifying details elsewhere. Follow the course policy: permission to discuss a prepared example does not imply permission to generate a submitted solution.

**Practice:** For the reminder program, compare these requests: “Write my assignment” and “Explain what each of these three WriteLine statements displays, without changing the code.” Which better targets a specific learning need? Before reading any answer, predict the three lines yourself. If an assistant proposes a different appointment time, which requirement would you check?

Later chapters provide prepared candidate answers to review, so you can practice these skills without an account or paid subscription. Instructors may permit a live assistant as an additional source of explanations. Keep evidence of independent understanding through code tracing, modifications, and tests. The optional AI application extension belongs after the programming foundation.

Professional codes, including the ACM Code of Ethics, connect technical work with avoiding harm, respecting privacy, and honesty.[^acm] The {doc}`professional practice reference <../appendices/github/professional_practice>` expands licensing, accessibility, and responsible collaboration for project work.

Complete the message below using only `Console.WriteLine` calls. Display the fictional appointment time and one useful action. Do not display the customer's contact details or reason for the visit. Then answer the four questions from this section about your output.

**Expected output:** the appointment is at 10:30, and the reader is told to contact the office to reschedule. Exact wording can differ if the meaning remains clear.


````csharp
// Exercise: A responsible reminder
using System;

// Your code starts here.
// Print the appointment time and a useful next action.

// Your code ends here.
Console.WriteLine("End of reminder.");
````

````csharp
// Solution: A responsible reminder
using System;

Console.WriteLine("Your appointment is at 10:30.");
Console.WriteLine("Contact the office if you need to reschedule.");
Console.WriteLine("End of reminder.");
````

The output names no person and gives no reason for the appointment. It still helps the recipient act. For the wider system, you would also need to confirm that reminders go only to the intended, authorized recipient. Brief output choices and broader system controls solve related but different problems.
