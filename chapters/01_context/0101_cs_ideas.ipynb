A program is a set of instructions a computer executes. Computer science asks how to represent information, describe a procedure precisely, and evaluate whether the result solves the intended problem. Programming makes those ideas concrete.

Consider a small business that sends appointment reminders. A useful program must select the right appointment, produce an understandable message, and avoid exposing unnecessary information. Begin by describing the task before choosing C# statements.

## Computers, Software, and Data




```{index} hardware; software; bit; byte; data representation
```


**Hardware** is the physical equipment, including a processor, memory, storage, and input/output devices. **Software** supplies instructions. An operating system manages hardware and provides services to applications, such as reading a file or displaying text.

Computers encode data as bits, each with two possible values, conventionally written `0` and `1`. Eight bits form a byte. An encoding determines what a sequence of bits means: a number, text, an image, or another kind of data. The same bit pattern can mean different things under different encodings. You will choose C# data types in Chapter 2; a type constrains the values and operations your program uses.

For an appointment service, a date is data, a reminder application is software, and the computer that runs it is hardware. A correctly functioning computer can still execute instructions that send the wrong date. Hardware operation and application correctness are different questions.

**Check Your Understanding:** Identify one input device, one output device, and one stored data item used by the appointment service.


```{dropdown} Suggested Answer
One possible answer is a **keyboard** for entering appointment details, a **screen** for displaying a reminder, and the **scheduled appointment date and time** as a stored data item. Other suitable devices are possible; distinguish physical devices from the data they receive or display.
```

## Describe Input, Processing, and Output




```{index} input-process-output model; IPO
```


The **input-process-output (IPO) model** describes a task through the information it receives, the work it performs, and the result it produces.

| Stage | Appointment reminder example |
| --- | --- |
| Input | Appointment time, contact destination, and permission to send a reminder. |
| Processing | Select the appointment that needs a reminder and compose an appropriate message. |
| Output | A message telling the recipient when to arrive. |

The model separates a problem from its implementation. A receptionist and a computer can follow the same reminder procedure using different tools. Input can also come from a file or another system rather than a keyboard; output can be stored rather than displayed.

Specify an observable result: “A reminder displays the correct appointment time without including the reason for the visit.” That statement supports both a correctness check and a privacy decision. “Make a good reminder app” does not say what to test.

**Check Your Understanding:** Describe input, processing, and output for a bookstore's daily sales summary. Name a result the manager could check.


```{dropdown} Suggested Answer
**Input:** the day's sales records, including item quantities and selling prices. **Processing:** calculate each sale amount and add the amounts for that day. **Output:** a report showing the daily sales total. The manager could check that a day with two sales of $10 and $15 reports $25. A fuller report would need explicit rules for returns, discounts, and taxes.
```


The IPO model is also foundational in systems analysis and software engineering. A payroll system takes employee hours and pay rates as input, calculates wages and deductions, and produces paychecks and reports. Describing the stages helps an analyst identify missing information before writing code.

```text
INPUT                 PROCESSING                  OUTPUT
Hours and rates  -->  Calculate wages        -->   Paychecks
Sales records    -->  Total sales            -->   Daily report
```

Storage connects one run to the next. A sales report can read stored transactions and save a summary that becomes input to another process. A person reviews the output and may correct the input or revise the rules. A useful system model therefore includes storage and feedback, not only a one-way calculation.

```text
             +-----------------------------------------+
             |             COMPUTER SYSTEM             |
User ------> | Input --> Processing --> Output          | --> User
             |               ^            |            |
             |               |            v            |
             |               +---- Storage             |
             +-----------------------------------------+
   ^                                                       |
   +-------------- Review and feedback -------------------+
```

## Thinking Like a Computer Scientist




```{index} Thinking Like a Computer Scientist
```

Computer science is a scientific discipline. Learning to program can therefore mean learning a new way of thinking, not just learning a language's punctuation. Allen Downey describes this as thinking like a computer scientist, bringing together mathematical reasoning, engineering, and natural science.[^0101_cs_ideas-cs-thinking]

Like a mathematician, you use precise notation to describe computations and reason about what follows from stated assumptions. Like an engineer, you design components, assemble systems, and evaluate tradeoffs among competing needs. Like a natural scientist, you form explanations, make predictions, run experiments, and compare observations with those predictions.

The central skill is **problem solving**: formulate the problem, consider possible solutions creatively, and express a chosen solution clearly and accurately. Programming implements an algorithmic solution using a language's notation. An algorithm is the procedure; the program is one implementation of it. A recipe is a useful analogy only when it states the quantities, ordering, decisions, and stopping conditions precisely enough to follow.

For a bookstore report, first decide what “sales” means: gross amounts, paid orders, or amounts after returns? Predict the result for a small set of records. Then implement and test it. If the result disagrees with the prediction, investigate whether the code, the business rule, or the prediction was wrong. This connects mathematical reasoning, system design, and empirical checking.

Computer science also studies what can be computed, how information is represented, how systems communicate, how people interact with software, and how computing affects people. Writing code provides a concrete way to investigate those questions; it does not exhaust the discipline.

## Algorithms and Development Workflow




```{index} algorithm; software development life cycle; SDLC
```


An **algorithm** is a precise procedure for completing a task. A person should be able to follow its steps without inventing missing rules. For the reminder service:

1. Obtain an appointment approved for reminders.
2. Read its scheduled time.
3. Compose a message containing the time and a way to contact the office.
4. Review the message for accuracy and unnecessary personal details.
5. Send it through the approved channel.

“Send reminders to everyone” leaves important decisions unspecified. Which appointments? Which recipients? What information should be shown? Clarifying those rules is part of development.

The **software development life cycle (SDLC)** includes understanding requirements, designing a solution, implementing it, testing it, deploying it, and maintaining it. These activities repeat as you learn more. Writing source code is one activity in that cycle. A small exercise follows the same habit: state the intended result, predict an example, write code, run it, and compare the result with the prediction.[^0101_cs_ideas-workflow]

**Check Your Understanding:** Write a procedure for rescheduling an appointment. Identify one rule you would need the business to clarify.


```{dropdown} Suggested Answer
One possible procedure is to identify the appointment, obtain the requested new time, check availability, confirm the change with the customer, update the stored appointment, and send a confirmation through the approved channel. Stop and ask for another time if the requested slot is unavailable. A rule to clarify is whether rescheduling within 24 hours requires staff approval. Other procedures can be valid when their rules and failure paths are explicit.
```


Application developers and software engineers organize this work into activities. The following seven-stage breakdown makes the build step explicit:

| Activity | Work performed |
| --- | --- |
| Requirements analysis | Define the problem, intended users, goals, constraints, and observable results. |
| Design | Plan the program's features, data, structure, and algorithm. |
| Implementation | Write the source code in an editor. |
| Build and compilation | Check and translate the source, resolving required libraries. |
| Testing and debugging | Compare results with expectations and investigate discrepancies. |
| Deployment | Release the program for its intended users and environment. |
| Maintenance | Correct defects and revise the program as needs change. |

These are activities, not a requirement to finish each one once in a rigid order. Running a program may reveal a coding mistake; a correct calculation may reveal that the original business rule was wrong. Debugging can therefore send you back to the code, the design, or the requirements.

```text
Define problem --> Design --> Write code --> Build --> Run --> Check
       ^                         ^                      |       |
       |                         +---- Repair code -----+-------+
       +--------- Clarify requirements and revise rules --------+
```

A small classroom exercise develops the same habit: define an expected result, predict an example, run the program, and explain any discrepancy.

## Number Systems and Data Representation



A **binary**, or base-2, number uses the digits `0` and `1`. Each position has a value that is a power of two, just as decimal positions represent powers of ten. A byte contains eight bits, so it has `2^8 = 256` possible patterns, from `0000 0000` through `1111 1111`.

The ASCII code for uppercase `A` is decimal 65. Its eight-bit representation is `0100 0001`. Work from right to left to convert the pattern:

| Bit position | 7 | 6 | 5 | 4 | 3 | 2 | 1 | 0 |
| --- | --- | --- | --- | --- | --- | --- | --- | --- |
| Place value | 128 | 64 | 32 | 16 | 8 | 4 | 2 | 1 |
| Bit in `0100 0001` | 0 | 1 | 0 | 0 | 0 | 0 | 0 | 1 |

Only the 64 and 1 positions are set, so the value is `64 + 1 = 65`. The pattern is a number until an encoding gives it an interpretation. Under ASCII it denotes `A`; under another interpretation the same byte can represent the number 65. C# uses Unicode for text; the ASCII characters share the corresponding Unicode values. Chapter 2 connects `char` values with their numeric codes.

The number of bits also limits the number of distinct addresses. A hypothetical byte-addressed machine with 32 address bits has `2^32` distinct addresses, covering 4 GiB of byte locations. A 64-bit address has many more possible patterns. A real computer's usable memory also depends on its processor, operating system, and hardware configuration; a label such as “64-bit” does not promise that every theoretically possible address is available.

**Check Your Understanding:** Convert `0000 1101` to decimal, and explain why `0000 1101` and `0100 0001` can both fit in one byte.

```{dropdown} Suggested Answer
`8 + 4 + 1 = 13`. Both patterns use eight bit positions. The number of available positions determines the byte's capacity; leading zeros still occupy positions.
```

## From Source Code to Execution




```{index} source code; compiler; compilation; runtime; .NET; editor; terminal
```


**Source code** is the program text written in a programming language. An editor lets you change and save that text. A **compiler** translates C# source into a compiled form; the **.NET runtime** executes the compiled program, normally translating intermediate instructions into machine instructions as needed. A **terminal** gives you access to a shell, which accepts commands that start tools and programs.

The .NET **software development kit (SDK)** includes the compiler and development commands. An editor and the SDK have different jobs: saving a file changes its text, while building checks and translates it. Running starts execution. Section 1.3 connects these operations in a saved console project.

Here is a complete small browser-runnable example. `Console.WriteLine` displays a line of text. You will study its structure in Section 1.3.


````csharp
using System;

Console.WriteLine("Appointment reminder");
Console.WriteLine("Please arrive at 9:00.");
````

The source contains two instructions; the output contains their messages. Changing the displayed time changes what the program says. It does not connect this example to a real scheduling system or send a notification.

Installation instructions belong to {doc}`Appendix B <../appendices/command_line>`. You can run this short example in the browser while preparing a local environment, then complete the saved-project workflow in Section 1.3.

### Compilation, Interpretation, and JIT



Early computers executed machine instructions represented as bit patterns. Assembly language gives symbolic names to instructions, while high-level languages let programmers express algorithms with more abstraction. Tools must translate or execute those descriptions.

A **compiler** translates source into another representation before that representation executes. An **interpreter** executes a program representation through another running program. These describe implementation techniques, not a permanent division of all languages into two kinds.

:::{figure} ../../figures/compiler.jpg
:width: 70%

A conceptual compiler pipeline: translate source, then execute the translated result.[^0101_cs_ideas-pipeline-figures]
:::

:::{figure} ../../figures/interpreter.jpg
:width: 60%

A conceptual interpreter pipeline: an interpreter processes a program representation while executing it.[^0101_cs_ideas-pipeline-figures]
:::

C# normally compiles to **intermediate language (IL)** rather than directly to the final processor's instructions. The .NET runtime commonly uses **just-in-time (JIT) compilation** to translate methods into machine instructions when needed. Other .NET deployment modes can compile ahead of time. Similarly, an interpreter can first compile source into bytecode; Python's common CPython implementation does this. “Interpreted” does not mean that no compilation occurs.

The **Common Language Runtime (CLR)** supplies execution services, including automatic memory management through garbage collection, type checking, and exception support. The **Base Class Library (BCL)** supplies reusable types and operations for text, collections, files, and other common work. This is why `Console.WriteLine` is available as a library operation instead of something each author must implement from scratch. The SDK supplies development tools; the runtime executes programs; libraries provide reusable functionality.

**Check Your Understanding:** Does saving `Program.cs` execute it? Does successful compilation prove that its business rules are correct?

```{dropdown} Suggested Answer
Saving writes source text to a file. Building checks and translates the source, and running starts execution. A program can compile and run while calculating the wrong price or displaying the wrong date, so its results must still be checked against requirements.
```

## Execution Order and Correctness




```{index} sequential execution; statement; syntax error; logical error
```


A **statement** instructs the program to do something. In the example, statements execute in order from top to bottom. This is **sequential execution**. A later instruction does not change output already printed by an earlier one.

Other programming constructs select between actions or repeat them. Chapters 3 and 5 teach decisions and iteration. You do not need those constructs to write your first useful output program.

A **syntax error** violates the language's rules and can prevent compilation. A **logical error** occurs when instructions run but produce an unintended result. Printing an outdated appointment time may compile successfully while still misleading a customer. Compilation is one check, not proof that the program is correct. Chapter 6 develops debugging and testing.
