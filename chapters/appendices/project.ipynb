
# Project



After OOP (Chapters 10-12), instructors may select the {doc}`database business application capstone <databases/index>` as a persistence milestone. A separate {doc}`AI application extension <project/ai_extension>` adds one narrowly scoped suggestion feature with an offline demonstration. Neither is required for the core individual or group project.

Using AI during development is a different question. Follow the course policy, understand every submitted part, and record accepted suggestions and verification evidence. Paid AI access is not a requirement for the book's core exercises or review activities.


The course includes two projects that use what you have learned to build a complete program.

| Project | Work | When | Page |
| --- | --- | --- | --- |
| **Individual project** | Alone | Midterm | {doc}`Individual Project <project/individual_project>` |
| **Group project** | Team of 3-4 | Final | {doc}`Group Project <project/group_project>` |

The individual project proves you can build a program on your own. The group project adds teamwork, shared design, and a larger program. Your instructor sets the dates for each project and its checkpoints.


## Platform Compatibility

The core project track uses tested console applications. The individual midterm milestone uses the concepts taught by its due date; the final application brings together collections, files, and collaborating classes. Build and run the program with the .NET SDK, and keep its source and verification evidence in your submitted project. A desktop or graphical interface is an optional extension after the core behavior works. If your course selects that extension, the project instructions must identify the framework, supported operating systems, and required tools before you begin. Not every framework supports every desktop operating system.

For example, .NET MAUI supports Windows and macOS desktop targets, but Microsoft does not officially support Linux desktop apps with MAUI ([supported platforms](https://learn.microsoft.com/en-us/dotnet/maui/supported-platforms?view=net-maui-10.0)). If you use Linux as your primary computer, do not assume a MAUI project will build or run locally. Contact your instructor before committing to an app idea so you can arrange an approved alternative or access to a supported environment. This limitation does not affect the book's regular C# exercises and console projects.
