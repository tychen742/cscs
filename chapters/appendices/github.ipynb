
Use this appendix when you need to track or share saved project files. **Git** records versions locally. **GitHub** hosts Git repositories and supports collaboration. A **repository** holds the files and their history. Saving a file, committing it, and uploading the commit are separate actions.


```{index} version control; why use version control; local and remote repositories
```

## Why Use Version Control


Version control gives you a record of your work that you can inspect, undo, and share. It helps individuals as much as teams: you can see what changed and when, return to a working version, and keep a copy off your own computer. It also replaces carrying files on USB drives or sending e-mail attachments, because you and your instructor can look at the same repository.

The basic idea is to keep a main copy of a project on GitHub (the **remote**) and a working copy on each computer you use (the **local** repository). Several steps stand between editing a file and changing the main copy, and each one is deliberate:

- **Track:** a new file is untracked until you `git add` it.
- **Commit:** changes to tracked files reach your local history only when you `git commit`.
- **Push:** committed changes reach GitHub only when you `git push`.
- **Ignore:** `.gitignore` lists files Git should leave out, such as `bin/` and `obj/`.
- **Pull:** changes made elsewhere reach your computer only when you `git pull`.

Keeping these steps separate protects the main copy from half-finished work. Other version control systems, such as Subversion and Mercurial, follow the same ideas with different commands. This book uses Git because GitHub hosts Git repositories and your course submissions use it.


```{index} Git installation; repository; git init
```

## Prepare a Practice Repository



Install Git using the [Git installation instructions](https://git-scm.com/book/en/v2/Getting-Started-Installing-Git), then open a new terminal and run `git --version`. For GitHub work, use your own account and follow your course's repository and access instructions.

In a new folder separate from existing repositories, create `README.md` in your editor with a project title and one sentence describing its purpose. From that folder, initialize Git and configure the identity recorded on your commits:

```console
git init -b main
git config user.name "Your Name"
git config user.email "your-address@example.com"
git status
```

Replace the example identity with your own choices. Repository-local configuration applies here rather than changing all your projects. You can use a GitHub-provided private email address according to your account settings.

Review the staged diff before committing. A clean status after the commit means there are no uncommitted changes Git currently reports; it does not prove the work was uploaded.

Now edit and save the README. `git diff` shows unstaged changes to tracked files; `git diff --staged` shows the selected changes. New untracked files appear in `git status` and do not appear in the ordinary diff until tracked. Stage the intended files and commit again. A useful message describes the change, such as `Document how to run the report`, rather than `update`.

Commit source, project configuration, and `.gitignore`; build output can be regenerated. Ignore rules do not remove files that are already tracked. Review `git status` before staging to see what the repository actually contains.

A **remote** is a named connection to another repository; `origin` is a conventional name. Verify the URL before pushing. If `origin` already exists, inspect it before deciding whether its URL needs to change. The `-u` option records the branch's upstream connection for later pushes. Open GitHub and confirm that the expected commit and files are present.[^1]

Replace the URL and folder name with the actual ones. Cloning copies the repository and configures `origin`. Do not repeat `git init` or `git remote add origin` inside that clone.

After editing, saving, and checking your program, review and record your changes. Replace `Program.cs` below with the specific files you intend to include:

```console
git diff
git add Program.cs
git diff --staged
git commit -m "Validate order input before calculating totals"
git push
```

This is the recurring cycle for individual work. For team work, use the branch and pull-request workflow below. If `git pull --ff-only` cannot update your branch, inspect the reported situation before continuing; do not discard your work to force an update.


```{index} two computers; working between lab and home
```

### Work on More Than One Computer


GitHub also lets you move between a lab computer and your own. The remote copy is the meeting point, so each session follows the same cycle:

1. **Start:** run `git pull --ff-only` to bring down changes you pushed from another computer.
2. **Work:** create or change files, then run and check your program.
3. **Record:** run `git add` for the files you intend to include, then `git commit -m "message"`.
4. **Finish:** run `git push` before you leave, so the work is on GitHub.

If you forget the last step, your work stays on the first computer. Push at the end of every session. Pulling and committing are separate operations so that you decide when to combine other people's changes with yours. This matters little when you work alone and a great deal in a team; for a team repository, add collaborators in the repository's **Settings** on GitHub as your instructor directs.


```{index} branch; pull request; GitHub workflow
```

## Collaborate Through a Branch and Pull Request



A **branch** names a line of development. A **pull request** proposes combining changes and gives reviewers a place to inspect and discuss them. For an updated clone on `main` with no uncommitted work:

```console
git switch main
git pull --ff-only
git switch -c report-heading
```

Edit the README or project, check the result, and commit the intended changes:

```console
git status
git diff
git add README.md
git diff --staged
git commit -m "Clarify the report heading"
git push -u origin report-heading
```

On GitHub, open a pull request from `report-heading` to the intended base branch. Review the diff, describe the change and its checks, and follow the course's review process before merging. GitHub's [Hello World tutorial](https://docs.github.com/en/get-started/using-github/hello-world) provides a browser-based practice workflow.[^2]

If a pull or merge reports divergent history or conflicts, inspect the situation before proceeding. A conflict marks competing changes that require a deliberate final version. Resolve the content, remove the markers, and check the resulting program. History rewriting is beyond this introductory workflow.


```{index} merge conflicts; conflict markers
```

### Resolve a Merge Conflict


A merge conflict asks you to decide how competing edits should be combined. Inspect the files reported by Git and read both versions around the conflict markers (`<<<<<<<`, `=======`, and `>>>>>>>`). Edit each affected area into the intended final content, then remove the markers.

Save the files, run the relevant program or tests, and review the diff. Stage the resolved files and follow Git's instructions to complete the merge or commit. Do not leave conflict markers in source code, and do not choose a version merely to make the warning disappear.


```{index} GitHub submission; commit verification
```
