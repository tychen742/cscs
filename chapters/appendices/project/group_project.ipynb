# Group Project



The group project is the final project. Work in a team of three or four to extend an approved individual project or build a new small business application. Use the methods, testing, collections, files, classes, and collaboration techniques taught in Chapters 1-12. See {doc}`Project <../project>` for the course sequence and optional platform extensions. Your instructor sets the schedule; the book does not assume a fixed month of project-only classes.


## Objectives

By the end of the project, your team will be able to:

- State one business workflow, its users, inputs, outputs, and failure policies.
- Divide responsibilities among collaborating classes and integrate independently tested parts.
- Use a collection to organize records and files to retain data between runs.
- Verify ordinary, boundary, and invalid cases with reproducible expected results.
- Use Git commits and a shared GitHub repository to coordinate contributions.
- Supply build/run/test instructions and explain the design and each member's contribution.

The starting point is your instructor-approved individual project. If starting afresh, create a normal .NET console application as in {doc}`the first-program lesson <../../01_context/0103_program_structure>`. The project does not depend on a missing game skeleton or a course-specific stub.


## Overview

Choose a bounded workflow such as inventory adjustment, expense reporting, appointment scheduling, or service-ticket routing. Submit a proposal stating the user, required operations, data records, one ordinary case, and two failure cases. Describe which records must survive a restart. Use synthetic data.

Agree on a minimum working version before adding features. Start with a complete console input/process/output path, add collections and validation, then add file persistence and collaborating classes. A team can adapt an earlier project, but it must identify retained code and the new responsibilities and checks.

Keep the core application runnable with the .NET SDK on Windows, macOS, and Linux. A GUI, database, or AI feature is an optional extension after the console workflow and tests work; it does not replace the required core evidence. Use the project overview's platform notes for any approved GUI extension.

Prepare at least two reviewable milestones: a working intermediate version and a final integrated version. Your instructor sets dates and demonstration length. No additional library, database service, paid account, or graphical framework is required for the core project.


## Your Team

Agree on roles and rotate or share them when useful. Everyone contributes substantive code and tests and explains their own work. Each member owns a cohesive piece with a stated input/output and failure contract. Record responsibilities and integration points before coding.

Use a private or course-approved shared GitHub repository as the working history. Submit the repository link and exact commit for each milestone through the course assignment system. Use {doc}`Git and GitHub <../github>` for clone, commit, branch, review, and conflict-resolution practice. Agree on a merge/review process; do not overwrite one another's source with copied numbered folders.


## Roles

1. **Coordinator:** maintains scope, milestones, and unresolved decisions.
2. **Integration lead:** keeps class interfaces and the working branch consistent.
3. **Programmers and testers:** every member implements and verifies a substantial component and reviews another member's change.
4. **Repository coordinator:** checks that source, tests, and README are committed and that milestone commit links are correct. Git history is the version record; Box folders and copied numbered versions are not the submission workflow.
5. **Evidence coordinator:** assembles expected/actual results, known limitations, and the weekly contribution summary.
6. **Documentation coordinator:** checks user instructions and documents public method contracts and project setup.

One person can hold several roles. Ask your instructor early about an unresolved participation or technical problem; do not wait for the final deadline. Record changes in responsibilities in the repository.


## The Process

1. Agree on the problem, core operations, data ownership, and a small acceptance-case table.
2. Choose class responsibilities and write down how the components call one another. State validation and failure policies at each boundary.
3. Implement one working path, commit it, and check its output against an independent expectation.
4. Add one feature at a time. Test the component and the integrated application after each change.
5. Review another member's code before merging. Preserve useful test cases when fixing a defect.
6. Add file persistence and check a save/exit/restart/load cycle with synthetic records. Test a missing file, malformed record, and rejected update according to the stated policies.
7. Prepare the intermediate evidence and use instructor feedback to refine the final version.

Use the project/test workflow in {doc}`Unit Testing <../../06_exceptions_testing/0603_testing>` to create the test project and add references from the solution directory. The tests must exercise production methods, not copied implementations.


## Splitting Up the Coding

Give each component one coherent responsibility: for example, record validation, a collection-backed service, file import/export, or console interaction. Agree on method names, parameter/return types, and failure outcomes before implementing callers. Prefer composition and interfaces when they make collaboration and test doubles useful; an inheritance hierarchy is not required merely to use every language feature.

Each member should supply expected results and tests for their component. Integrate early so type mismatches, shared-state assumptions, and incompatible failure policies are found while the program is still small. Keep deterministic business calculations separate from user input and file operations.


## Weekly Reports

Keep one short team status file such as `reports/2026_10_21.md` in the repository for each assigned reporting date. Your instructor sets the cadence. Record completed work, linked commits/reviews, actual checks and results, and the next concrete tasks with owners. Note deviations from the previous plan and unresolved integration decisions.

Commit the report and submit the requested milestone/commit link through the course system. No external RTF template or Box folder is required. Do not include private peer evaluations in this shared report.


## Intermediate Deliverables

Submit a GitHub commit or tag through the course assignment system. Include:

- A README describing the problem, core operations, class responsibilities, and member roles.
- A runnable console path using collections and at least one tested production method.
- Source and test project files, with the actual build/run/test commands and ordinary/invalid expected results.
- A short issue list identifying the remaining persistence and integration work.
- A contribution summary linking each member's work to commits or reviewed changes.

The intermediate version need not have every final feature. It must build and demonstrate the included features without relying on uncommitted files. State limitations explicitly so feedback can address them.


## Final Deliverables

Submit the repository URL and exact final commit through the course assignment system. Include source and project files, tests, synthetic sample data, and a README. Exclude generated `bin/` and `obj/`, credentials, personal data, and disposable runtime files. A Windows executable or platform-specific binary is not required.

Your README must give the commands for your actual layout. For example, if your solution contains `App/App.csproj` and `App.Tests/App.Tests.csproj`, run from the repository root:

```sh
dotnet build
dotnet test App.Tests/App.Tests.csproj
dotnet run --project App/App.csproj
```

Replace these example paths with the paths you actually submit. Specify the supported .NET SDK and how to prepare/reset sample data. A reviewer with a fresh clone should be able to reproduce the demonstration from these instructions.

Show evidence for:

| Outcome | Required evidence |
| --- | --- |
| Business behavior | One ordinary end-to-end workflow with independently predicted output. |
| Validation | At least two invalid or boundary cases, including a rejected change that preserves state. |
| Collections and collaboration | Explain record identity, class responsibilities, and how components share or own data. |
| File persistence | Save records, exit, restart, and recover them; show the chosen missing/malformed-file policy. |
| Testing | Tests that call production methods and a regression case for a defect your team repaired. |
| Reproducibility | Build, test, run, and data-reset commands that work from a fresh clone. |
| Team process | Contribution summary, review evidence, and the final milestone commit. |

Your instructor sets grading weights and presentation timing. During review, explain and modify a small part of your own implementation and show why a test would detect a plausible defect. Submit each member's individual contribution evaluation privately through the course system; do not place peer evaluations in the shared repository.
