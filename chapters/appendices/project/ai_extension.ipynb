# Optional AI Application Extension


**Extension ID:** `cscs-ai-project-extension`.

After completing the core application and the OOP chapters, you may add one AI-powered feature if your instructor assigns or permits it. Examples include drafting a reply to a fictional service ticket, proposing a product description, or suggesting a category for a synthetic request. Choose one narrow feature with an observable input/output contract.

This is separate from using an AI assistant while writing code. The application itself now consumes a model response. Model training, fine-tuning, autonomous agents, and a survey of machine-learning algorithms are outside this extension.


Finish the non-AI business workflow first. Keep it runnable without an AI account, network connection, or paid subscription. Put the response-producing component behind a small interface, using the collaboration techniques from Chapter 12.

Provide a deterministic demonstration using recorded, synthetic responses. For example, a service-ticket reply feature can read a saved suggestion, validate it, and display it for review. Implement and document a demo mode such as `--demo-ai` in your own project; this book does not supply a live API client or promise a provider-specific command.

A live service is optional and subject to the course policy and available access. If you add one, document its configuration and costs, keep credentials outside submitted source, and retain the recorded-response mode for grading. A student can complete the extension design, parser, interface, and tests with recorded responses alone.


## Specify and Check the Response

```{index} AI output validation; response contract; human review
```


Write the expected response shape before connecting a model. For a reply draft, the response could contain a `suggestion` string that is nonblank and below a stated length limit. For a category feature, define the permitted labels and reject unknown labels.

Treat a response as untrusted external input. Check its structure and your business rules before using it. Require the user to review a proposed message or record change before accepting it. Keep deterministic calculations, inventory updates, and other core business rules in ordinary tested C# code.

Test a valid response, an empty response, malformed data, and a response outside your allowed values or length. If using a live service, also demonstrate the program's behavior when that service is unavailable. Passing these checks verifies the wrapper behavior; it does not establish the factual accuracy or suitability of every possible generated message.

Use only synthetic inputs and recorded examples in the submitted demonstration. Explain what information the feature would need in a real setting and how that differs from the information your demo actually uses.


## Submit and Explain the Evidence

```{index} AI project evidence; reproducibility; attribution
```


Submit through the course project workflow, using a GitHub commit or repository link. Your instructor sets extension credit and timing; omitting this optional feature does not make the core project incomplete.

Include:

- The completed core program and one isolated AI-feature component.
- A synthetic input file and recorded response examples, such as `ai_examples.json`, with their provenance marked as prepared or model-generated.
- Tests for accepted and rejected responses and the deterministic demo mode.
- A README with actual setup, run, demo, and test commands for your implementation.
- A short evaluation showing one useful result, one unsuitable result, the checks performed, and what a human must still review.
- An assistance record identifying the tool/date, suggestions used, revisions made, and independent checks, if AI helped develop the program.

The record should explain your decisions, not simply paste a conversation. Do not submit credentials, personal records, or a requirement that the reviewer buy access to reproduce the demonstration. Explain and modify your own code during any instructor review.
