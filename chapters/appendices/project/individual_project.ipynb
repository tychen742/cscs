# Individual Project



The individual project is the midterm project. You build it alone, in VS Code, and submit it through GitHub. It shows that you can design, write, test, and explain a complete C# program without a team. The group project that follows builds on these skills.


## Objectives

- Choose a small, realistic business or management problem and describe it clearly
- Design a program from input, processing, and output (the IPO model)
- Use variables, selection, loops, methods, and collections to implement it
- Test the program with ordinary and edge-case inputs
- Document how to build and run the program
- Use Git to save your work in steps


## Choosing a Topic

Pick a problem a small organization might actually have. Examples:

- An inventory tracker for a campus store
- A shift scheduler for a restaurant
- An expense report summarizer for a project budget
- A grade or attendance calculator for a training program

Submit a short proposal first (the problem, who uses the program, and the inputs and outputs) so your instructor can catch topics that are too large or too small. A console application is the expected form. A desktop or graphical app is optional; see the platform notes in {doc}`Project <../project>`.


## Requirements

Your program must:

1. Run with `dotnet run` on your computer without errors
2. Read input from the user or from a file and validate it
3. Use at least one method that takes parameters and returns a value
4. Use a loop and a decision (`if` or `switch`)
5. Store related data in a collection such as a `List<T>` or an array
6. Print clear, formatted results
7. Handle at least two kinds of bad input without crashing

Use only the C# features covered so far in the course.


## What to Turn In

- A GitHub repository containing your project folder and a `README.md` that states the problem, how to run the program, and a sample run
- At least five commits with meaningful messages that show how the program grew
- A short written reflection: what you built, one design decision you would change, and one bug you fixed and how
- A short demo of the program in class, if your instructor asks for one

Submit the repository link through the course assignment system. Your instructor sets the due dates for the proposal, a progress check, and the final submission.


## How It Is Graded

| Area | What your instructor looks for |
| --- | --- |
| Correctness | The program runs and produces right results |
| Design | Clear methods, meaningful names, and sensible use of collections |
| Testing | Ordinary and edge-case inputs tried, and bad input handled |
| Documentation | A README someone else can follow |
| Process | Regular, well-described commits |
