## 1. List the Seeded Inventory

```{index} 1. List the Seeded Inventory
```


Implement `List` and `ReadBook`. Read the four named columns, create one `Book` per row, and return books ordered by ID. The seed has `C# for Business` at 2500 cents with stock eight and `Data at Work` at 3200 cents with stock three.

**Checkpoint:** Two books appear, in ID order. Display prices with two decimal places. Explain why the mapper needs the same column order as the query.


## 2. Find a Book

```{index} 2. Find a Book
```


Implement `Find(long id)` using `@id`. Return a book for ID one and `null` for an unknown ID such as 999. Show a helpful message when no book is found.

**Checkpoint:** Searching never changes the row count. Explain why a missing ID is different from an unavailable database file.


## 3. Add a Title

```{index} 3. Add a Title
```


Implement `Add`. Reject blank titles, negative prices, and negative stock. Use parameters and return the generated ID. Add `Manager's SQL Guide` at 1900 cents with stock four, then find it by its returned ID.

**Checkpoint:** The apostrophe is stored unchanged. Use the returned ID instead of assuming it will always be three. Explain why `INSERT OR IGNORE` is used for demonstration seeds but would hide useful failures in an ordinary add operation.


## 4. Update Stock and Verify Boundaries

```{index} 4. Update Stock and Verify Boundaries
```


Implement `SetStock`. Set the new title's stock to six, reject negative stock, and report `false` for a missing ID. Do not update unrelated books.

**Checkpoint:** Run the core checks:

```sh
dotnet run -- --check-core
```

These checks use a fresh in-memory database. Record the output and explain the check showing that updating one book preserves the other book's stock. Reopen the ordinary file database to demonstrate persistence.


## 5. Extension: Record a Sale Atomically

```{index} 5. Extension: Record a Sale Atomically
```


Implement `Sell`. Reject zero or negative quantities. Within one transaction, decrement stock only when `Stock >= @quantity`, then insert the sale. Commit both changes together. Missing books and insufficient stock should return `false` without a sale row.

**Checkpoint:** Sell two copies of a book with stock six. Stock becomes four and exactly one sale is recorded. Attempt to sell five more; stock and sale count must stay unchanged.

```sh
dotnet run -- --check
```

Explain why two individually valid SQL statements still need a transaction. As a further check, force the sale insert to fail inside a disposable test database and confirm that its stock decrement is rolled back.
