# Databases and Bookstore Inventory

Use this appendix as an **optional business application capstone after OOP (Chapters 10-12)**. It connects collections, files, classes, validation, and tests to persistent records in SQLite. An instructor may allocate about one to two weeks when assigning the track, or select it as a project milestone. It is not a prerequisite for completing CS1 or beginning the CS2 data-structures foundation.

The connected example is a bookstore inventory application. Business examples earlier in the book can remain realistic with synthetic data, collections, and files; they do not all need databases.

- **Relational data**: Tables and keys organize persistent business records.
- **SQL**: Queries and change commands read and maintain those records.
- **Parameterized command**: Values travel separately from SQL instructions.
- **Inventory application**: Objects connect database rows to useful operations.

<h2>Learning Objectives</h2>

1. Model books and sales with primary keys, foreign keys, and constraints.
2. Write filtering, aggregation, join, and change commands.
3. Run a C# SQLite project and map query results into objects.
4. Implement and check inventory operations using parameters.
5. Explain why recording a sale and reducing stock should succeed together.

<h2>How to Use This Appendix</h2>

The first two sections cover tables, keys, constraints, basic queries and joins, parameterized changes, and a working C# application. Database administration, ORM frameworks, migrations, and advanced SQL tuning are outside this capstone's scope.

The first two sections explain the concepts and a working application. The third is a connected lab completed in VS Code. The sales transaction is an extension. This appendix does not add a required week to either semester; instructors may use it as a project persistence milestone. Demonstrate that a validated change persists after closing and reopening the file database. Explain how the database enforces a rule that application code also checks. The sales extension illustrates two changes that must commit or roll back together.

<h2>Glossary</h2>

| Term | Meaning |
| --- | --- |
| Schema | The structure and constraints of database tables. |
| Primary key | A value uniquely identifying a row. |
| Foreign key | A reference to a related row. |
| Transaction | A group of changes committed or rolled back together. |
| Parameter | A separately supplied value used by a SQL command. |
