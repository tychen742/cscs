## Data Structures and Algorithms: OpenDSA


[OpenDSA](https://opendsa-server.cs.vt.edu/) is an open eTextbook project from Virginia Tech. Its books combine short readings with interactive slideshows (JSAV visualizations) that animate one operation at a time, plus auto-graded practice exercises. The code is Java, so read it for the algorithm, not the syntax. The content is distributed under an MIT license. Three of its books are references for this text:

- [CS2](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS2/html/): lists, stacks, queues, recursion, elementary sorting, binary trees, and binary search trees. Primary reference for Parts IV and V (Chapters 13 to 20).
- [CS3](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS3/html/): algorithm analysis, sorting, hashing, heaps, graphs, and more advanced structures. Primary reference for Parts IV to VII (Chapters 13 to 26), especially the analysis, hashing, and graph chapters that CS2 does not cover.
- [Introduction to Software Design](https://opendsa-server.cs.vt.edu/OpenDSA/Books/IntroToSoftwareDesign/html/): an objects-first Java CS1. This book's Parts I to III are imperative-first, so use it only as a model for exercise formats, not as parallel reading.

The sections in Parts IV to VII that build an algorithm by hand end their worked example with a **Visualize it** pointer to the matching OpenDSA slideshow. The most useful ones, in reading order:

- [Insertion Sort](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS2/html/InsertionSort.html), [Selection Sort](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS2/html/SelectionSort.html), and [Bubble Sort](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS2/html/BubbleSort.html) for Chapters 7, 16, and 23.
- [Mergesort Concepts](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS3/html/Mergesort.html) and [Quicksort](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS3/html/Quicksort.html) for Chapters 16, 22, and 23.
- [Tracing Recursive Code](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS2/html/Trace.html) for the recursion readings in Chapters 5, 17, and 22.
- [Binary Tree Traversals](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS2/html/BinaryTreeTraversal.html) and [Binary Search Trees](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS2/html/BST.html) for Chapter 17.
- [Heaps and Priority Queues](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS3/html/Heaps.html) for Chapter 18.
- [Open Hashing](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS3/html/OpenHash.html), [Collision Resolution](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS3/html/HashCSimple.html), and [Hashing Deletion](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS3/html/HashDel.html) for Chapter 19. OpenDSA's open hashing is this book's separate chaining, and its closed hashing is this book's open addressing.
- [Graph Traversals](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS3/html/GraphTraversal.html), [Shortest-Paths Problems](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS3/html/GraphShortest.html), and [Kruskal's Algorithm](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS3/html/Kruskal.html) for Chapters 20 and 24.
- [Array-Based List Implementation](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS2/html/ListArray.html), [Linked Lists](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS2/html/ListLinked.html), [Stacks](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS2/html/StackArray.html), and [Queues](https://opendsa-server.cs.vt.edu/OpenDSA/Books/CS2/html/Queue.html) for Chapters 14 and 15.

The chapter-by-chapter mapping to OpenDSA sections is kept in the repository at `authoring/toc_comparison/README.md`.
