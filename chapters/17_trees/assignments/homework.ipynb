````csharp
// Solution
using System;
using System.Collections.Generic;

var root = new Node(40)
{
    Left = new Node(20) { Left = new Node(10), Right = new Node(30) },
    Right = new Node(60)
};
Console.WriteLine(string.Join(", ", InOrder(root)));
Console.WriteLine($"empty count={InOrder(null).Count}");
List<int> InOrder(Node? root)
{
    var result = new List<int>();
    var pending = new Stack<Node>();
    Node? current = root;
    while (current is not null || pending.Count > 0)
    {
        while (current is not null)
        {
            pending.Push(current);
            current = current.Left;
        }
        current = pending.Pop();
        result.Add(current.Id);
        current = current.Right;
    }
    return result;
}
class Node
{
    public int Id { get; }
    public Node? Left { get; set; }
    public Node? Right { get; set; }
    public Node(int id) => Id = id;
}
````

````csharp
// Solution
using System;
var root = new Node(40)
{
    Left = new Node(20) { Left = new Node(10), Right = new Node(30) },
    Right = new Node(60) { Left = new Node(50), Right = new Node(70) }
};
foreach (int target in new[] { 35, 40, 5, 80 })
{
    bool found = TryFloor(root, target, out int id);
    Console.WriteLine($"target={target}, found={found}, floor={id}");
}
bool TryFloor(Node? node, int target, out int id)
{
    bool found = false;
    id = default;
    while (node is not null)
    {
        if (node.Id == target) { id = node.Id; return true; }
        if (node.Id > target) node = node.Left;
        else
        {
            id = node.Id;
            found = true;
            node = node.Right;
        }
    }
    return found;
}
class Node
{
    public int Id { get; }
    public Node? Left { get; set; }
    public Node? Right { get; set; }
    public Node(int id) => Id = id;
}
````
