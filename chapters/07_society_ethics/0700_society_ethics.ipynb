
(society-ethics)=

# Society, Ethics, and the Profession

Programs make decisions about real people: what they pay, whether their order ships, what personal data a company keeps about them, and whether they can use a service at all. The previous chapter showed how to find and prevent defects. This chapter asks the wider questions: who is affected when software is wrong or careless, what professionals owe those people, and which everyday coding habits protect them.

- **Impact**: A small defect repeated across thousands of transactions becomes a large harm.
- **Ethical analysis**: A short, repeatable set of questions helps you reason about a decision instead of guessing.
- **Privacy and security**: Collect, show, and log only what a task needs, and treat every input as untrusted.
- **Professional practice**: Licenses, attribution, responsible use of AI assistants, and accessible design are part of doing the job well.

<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Explain, with real cases, how software defects and design choices can harm people and organizations
2. Apply a four-question checklist to analyze an ethical decision in a software project
3. Write C# methods that mask personal data, validate untrusted input, and log events without leaking secrets
4. Recognize injection as a security risk created by building commands from user input
5. Describe your professional responsibilities for licenses, attribution, AI-assisted code, and accessibility

<h2>Chapter Flow</h2>

- {doc}`7.1. Computing and Society <0701_computing_society>`
- {doc}`7.2. Privacy and Security in Code <0702_privacy_security>`
- {doc}`7.3. Professional Practice <0703_professional_practice>`

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Stakeholder | Anyone affected by a system, including people who never use it directly. |
| 2 | Ethical analysis | A structured way of weighing a decision's effects, duties, and fairness. |
| 3 | Code of ethics | A profession's published standards of conduct, such as the ACM Code. |
| 4 | Personal data | Information that identifies a person or can be linked to one. |
| 5 | Data minimization | Collecting and keeping only the personal data a task needs. |
| 6 | Masking | Showing only part of a value, such as the last four digits of a card. |
| 7 | Untrusted input | Any data that comes from outside the program and must be checked. |
| 8 | Injection | An attack in which input is treated as part of a command. |
| 9 | Least privilege | Giving each user or program only the access it needs. |
| 10 | License | The terms under which others may use, change, and share software. |
| 11 | Attribution | Credit for code or ideas that came from someone or something else. |
| 12 | Accessibility | Designing software that people with disabilities can use. |
