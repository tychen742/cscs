<h2>Glossary</h2>

| Term | Meaning |
| --- | --- |
| Priority queue | An ADT selecting by priority. |
| Min-heap | A complete binary tree with each parent no greater than its children. |
| Sift up | Repair after inserting at the final position. |
| Sift down | Repair after moving the final value to the root. |
| Tie policy | The rule distinguishing equal primary priorities. |
| Top-k | Retain only the k most relevant values. |
