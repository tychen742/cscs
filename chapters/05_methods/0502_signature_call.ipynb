## A Boolean Result Answers a Question



A returned result need not be a number to print. A method can answer a question, and a bool result can become the condition of an if statement. Suppose an order receives free delivery when its subtotal is at least 50. The question is whether an order qualifies, not how much its delivery costs. Keeping those contracts distinct helps explain why two related methods may have different return types.

Trace the boundary: 49.99 does not qualify, 50 qualifies, and 50.01 qualifies. The equality is part of the rule. Writing `>` instead of `>=` would still compile and would pass examples far from the threshold. The boundary cases distinguish the intended rule from that plausible defect.

The general form is `static bool QuestionName(parameterList) { return condition; }`. Here the condition already has type bool. There is no need to convert it into the strings `"true"` and `"false"`; those are text values and cannot serve as an if condition. Printing the answer inside a void method also does not give the caller a Boolean result to use.

Guided variation: qualification additionally requires a nonnegative subtotal. Explain whether that is an input precondition or a condition the method must check. Either design can be explicit, but silently switching between them makes testing and reuse confusing. State the accepted domain before writing your new tests.

For related explanation, see the further reading below.[^development-reading]


````csharp
using System;
Console.WriteLine(Qualifies(49.99m));
Console.WriteLine(Qualifies(50m));
Console.WriteLine(Qualifies(50.01m));
static bool Qualifies(decimal subtotal) { return subtotal >= 50m; }
````

## Returned Values Replace Calls




```{index} Returned Values Replace Calls
```


````csharp
using System;
Console.WriteLine(F(3));
Console.WriteLine(F(3) + F(4));
static int F(int x) { return x * x; }
````

You probably have used mathematical functions in algebra class, but
they all had calculated values associated with them. For instance
if you defined

> F(x)=x{sup}`2`

then it follows that F(3) is 3{sup}`2` = 9, and F(3)+F(4) is
3{sup}`2` + 4{sup}`2` = 9 + 16 = 25.

Method calls in expressions get replaced during evaluation by the value of the method.
That's why when we can directly print out (line# 12-13) the result of the method call in the
following code.

The corresponding definition and examples in C# would be the
following:




The C# syntax to notice is the **return statement**, with the word
`return` followed by an expression, value, or variable(s). Methods that return values
can be used in expressions (e.g., line# 13: F(3) + F(4)), just like in math class. When an
expression with a method call is evaluated, the method call is
effectively replaced temporarily by its returned value. Inside the
C# method, the value to be returned is given by the
expression in the `return` statement.

Since the method returns data, and all data in C# is typed,
there must be a type given for the value returned. Note that the
method header does not start with `static void`.
In place of `void` is `int`. The `void` in earlier method header
meant nothing was returned. The `int` here means that a value *is*
returned and its type is `int`.

After the method `F` finishes executing from inside



```csharp
Console.WriteLine(F(3));
```


it is as if the statement temporarily became



```csharp
Console.WriteLine(9);
```


and similarly when executing



```csharp
Console.WriteLine(F(3) + F(4));
```


the program first evaluates F(3) and effectively replaces the
call by the returned result, 9, as if the statement temporarily
became



```csharp
Console.WriteLine(9 + F(4));
```


and then the program evaluates F(4) and effectively replaces
the call by the returned result, 16, as if the statement
temporarily became



```csharp
Console.WriteLine(9 + 16);
```


resulting finally in 25 being calculated and printed.



The first result is 9; the sum is 25. A mathematical analogy describes how returned values combine, but a C# method may also print or change state. Read its contract rather than assuming all methods behave like pure mathematical functions.

```{figure} ../../figures/method_call_return.svg
:alt: A caller passes digit 4 to num in SquareTheNumber. The returned 16 becomes squaredNum in that caller.

Return to the expression that made this call.
```

| Step in `F(3) + F(4)` | Waiting expression |
| --- | --- |
| Evaluate F(3): x receives 3; return 9 | 9 + F(4) |
| Evaluate F(4): x receives 4; return 16 | 9 + 16 |
| Add the returned values | 25 |

The calls have separate parameter values. The method does not remember the previous x as input for the next call.

## Returned Strings and Formatted Results




```{index} Returned Strings and Formatted Results
```


C# methods can return any type of data, not just numbers, and
there can be any number of statements executed before the return
statement. Read, follow, and run the example program, which has `string` as
`return type`:



````csharp
using System;
Console.WriteLine(LastFirst("Benjamin", "Franklin"));
Console.WriteLine(LastFirst("Andrew", "Harrington"));
static string LastFirst(string firstName, string lastName)
{
    string separator = ", ";
    string result = lastName + separator + firstName;
    return result;
}
````

It is common to want to construct and immediately print a string,
so having `Console.Write` is definitely handy when we want it.
However, sometimes we just want to have the resulting string returned,
so that we can do something else with it. We can use
the C# library method `string.Format`, which does just what we want:
The parameters have the same form as for `Console.Write`,
but the formatted string is *returned*.




Note: The only caveat with `string.Format` is that
there is *no* special method corresponding to `Console.WriteLine`,
with an automatic terminating newline.
You can generate a newline with string.Format: Remember the
escape code `"\n"`. Put it at the end to go on to a new line.



````csharp
using System;
string saved = SumProblemString(2, 3);
Console.WriteLine(saved);
Console.WriteLine(SumProblemString(12345, 53579));
static string SumProblemString(int x, int y)
{
    int sum = x + y;
    return string.Format("The sum of {0} and {1} is {2}.", x, y, sum);
}
````

The returned string can be saved before it is printed. `{0}`, `{1}`, and `{2}` in `string.Format` select the following argument values by position. General syntax is `string.Format(formatText, value0, value1, ...)`; interpolation is another formatting form already used in earlier chapters. Neither form prints by itself.

## Print or Return? Choose the Contract




```{index} Print or Return? Choose the Contract
```


````csharp
using System;
PrintTotal(10m, 2m);
decimal savedTotal = CalculateTotal(10m, 2m);
Console.WriteLine($"Saved plus delivery: {savedTotal + 3m:F2}");
static void PrintTotal(decimal price, decimal charge)
{
    Console.WriteLine($"Total: {price + charge:F2}");
}
static decimal CalculateTotal(decimal price, decimal charge)
{
    return price + charge;
}
````

Both helpers calculate 12.00. PrintTotal writes it immediately and returns no value. CalculateTotal prints nothing; the caller stores 12 and uses it in another expression to print 15.00. A void call cannot be assigned to a decimal. Conversely, calling CalculateTotal and discarding its result is legal, but useless here because that method has no other effect.

**Variation:** change the inputs, store the returned total, and use it twice without repeating user input. **Mistake to diagnose:** `decimal total = PrintTotal(10m, 2m);` asks a void method for a value. The compiler rejects it. Keep calculation reusable and let the caller decide how to display it.

## Decisions, Early Returns, and Complete Return Paths




```{index} Decisions, Early Returns, and Complete Return Paths
```


Methods reuse the decisions and loops you already know. A normal path through a value-returning method must return a compatible value; writing one return somewhere is not enough. A return exits the current call immediately. A void method may use `return;` to finish early, but cannot use `return value;`.

The following valid-input contract accepts a nonnegative subtotal, gives free delivery at 50.00 or above, and charges 4.00 otherwise. The tests cover just below, at, and above the boundary.


````csharp
using System;
Console.WriteLine($"49.99: {DeliveryCharge(49.99m):F2}");
Console.WriteLine($"50.00: {DeliveryCharge(50m):F2}");
Console.WriteLine($"50.01: {DeliveryCharge(50.01m):F2}");
static decimal DeliveryCharge(decimal subtotal)
{
    if (subtotal >= 50m) return 0m;
    return 4m;
}
````

The outputs are 4.00, 0.00, and 0.00. Without the final return, a smaller subtotal could reach the end without a result. A return followed by more work on that same path makes the later work unreachable. An accidental semicolon after `if (condition);` creates an empty decision body, so the next statement can run unconditionally. Trace the paths rather than indenting by appearance. The restored return-fragment variations in Extra Practice apply these rules.

Inputs outside the stated domain need a separate policy. Chapter 6 develops exception-based contracts; here do not invent behavior for a negative subtotal without revising the requirement.

A packing charge is a fixed handling amount of 4.00 **plus** `subtotal * rate`. Assume nonnegative inputs. A prepared candidate returns only `subtotal * rate`; its explanation says, “The method calculates the packing charge from the supplied order amount and rate.”

Before running, calculate the required charge for a subtotal of 30.00 and a rate of 0.05. Then calculate it for a zero subtotal. Explain which part of the requirement the candidate omits. Revise the method while keeping its name, parameters, and return type.

````csharp
// Exercise: Review a packing charge
using System;

static decimal PackingCharge(decimal subtotal, decimal rate)
{
    // Your code starts here.
    return subtotal * rate; // TODO: review against the complete requirement.
    // Your code ends here.
}

Console.WriteLine($"Order: {PackingCharge(30m, 0.05m):F2}");
Console.WriteLine($"Zero subtotal: {PackingCharge(0m, 0.05m):F2}");
````

````csharp
// Solution: Review a packing charge
using System;

static decimal PackingCharge(decimal subtotal, decimal rate)
{
    return 4m + subtotal * rate;
}

Console.WriteLine($"Order: {PackingCharge(30m, 0.05m):F2}");
Console.WriteLine($"Zero subtotal: {PackingCharge(0m, 0.05m):F2}");
````

The required outputs are 5.50 and 4.00. The zero case exposes the missing fixed amount. A signature tells you how to call a method; its contract tells you what it must do.
