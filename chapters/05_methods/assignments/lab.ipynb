## 4. Compose a returned report



Complete ReportLine using NormalizeCode, TicketNumber, and ServiceCost, supplied below. Do not repeat their internal work. Return one formatted report line; keep printing in the caller.


````csharp
// Exercise: Compose a returned report
// exercise-id: cscs-methods-connected-lab-4
using System;

Console.WriteLine(ReportLine(" desk-1042 ", 4));
static string NormalizeCode(string raw) { return raw.Trim().ToUpperInvariant(); }
static string TicketNumber(string code) { return code.Substring(code.IndexOf('-') + 1, 4); }
static int ServiceCost(int jobs) { if (jobs == 0) return 0; return 5 + jobs * 4; }
// Your code starts here.
static string ReportLine(string raw, int jobs) { return ""; }
// Your code ends here.
````

````csharp
using System;

Console.WriteLine(ReportLine(" desk-1042 ", 4));
static string NormalizeCode(string raw) { return raw.Trim().ToUpperInvariant(); }
static string TicketNumber(string code) { return code.Substring(code.IndexOf('-') + 1, 4); }
static int ServiceCost(int jobs) { if (jobs == 0) return 0; return 5 + jobs * 4; }
static string ReportLine(string raw, int jobs)
{
    string code = NormalizeCode(raw);
    string number = TicketNumber(code);
    return $"{code}: ticket {number}, cost {ServiceCost(jobs)}";
}
````
