## True or False

**1.** A `void` method in C# must include a `return` statement.

```{dropdown} Answer
**False** — `void` methods may omit `return`; it is optional.
```

**2.** Method parameters are local variables within the method body.

```{dropdown} Answer
**True** — Parameters behave like local variables scoped to the method.
```

**3.** A string method can return a new string without changing the original string.

```{dropdown} Answer
**True.** Strings are immutable; the caller uses the returned value when a changed result is needed.
```

**4.** A method must always return a value.

```{dropdown} Answer
**False** — Methods declared `void` return nothing.
```

**5.** The method signature includes the method name and its parameter types.

```{dropdown} Answer
**True** — The signature = name + parameter type list (not the return type in C#).
```

**Optional ungraded self-check.**

**6.** You can split your helper methods into a separate `.cs` file without creating a new project.

```{dropdown} Answer
**True** — A project can span many `.cs` files. For this chapter, use separate projects for independently runnable programs. Additional files can define helpers in the same project, but calls must respect their containing type and accessibility. Top-level local functions do not become freely callable helpers across files.
```

## Multiple Choice

**1.** What keyword sends a value back from a method to its caller?

a) send  
b) output  
c) return  
d) yield

```{dropdown} Answer
**c) return** — `return` exits the method and passes the value back.
```

**2.** What is a value supplied at a method call site called?

a) An argument  
b) A parameter declaration  
c) A namespace  
d) A return type

```{dropdown} Answer
**a) An argument** is supplied by the caller; a parameter is the receiving name in the method definition.
```

**3.** What happens when a method is called with fewer arguments than required parameters?

a) Uses null automatically  
b) Compiles but crashes at runtime  
c) Causes a compile-time error  
d) Uses zero for numeric types

```{dropdown} Answer
**c) Causes a compile-time error** — Missing required arguments produce a compile error.
```

**4.** A caller wants a helper to answer a yes/no question for a later decision. Which contract fits?

a) Only print a message  
b) Return a Boolean result  
c) Rename the project  
d) Change a comment

```{dropdown} Answer
**b) Return a Boolean result.** The caller can use that result in a decision; printed text alone is not a returned value.
```

**5.** What does local scope describe?

a) The speed of a method call  
b) Where a local name is usable  
c) The length of its returned text  
d) The number of source files

```{dropdown} Answer
**b) Where a local name is usable.** A parameter/local variable belongs to its method or block; it is not automatically available to every caller.
```



## Optional String Self-Checks — True or False

**1.** `ToUpper()` changes the string it is called on.

```{dropdown} Answer
**False** — strings are immutable; `ToUpper()` returns a new string and the original is unchanged.
```

**2.** For `string s = "Ana";`, the expression `s[0]` is the character `'A'`.

```{dropdown} Answer
**True** — string indexes start at 0.
```

**3.** `Length` is a method, so it is written `s.Length()`.

```{dropdown} Answer
**False** — `Length` is a property and has no parentheses: `s.Length`.
```

**4.** `"ELEC-1042".IndexOf("RED")` returns `-1`.

```{dropdown} Answer
**True** — `IndexOf` returns -1 when the text is not found.
```
These extra recognition and trace questions are ungraded practice after the ten-question Preview.


## Optional String Self-Checks — Multiple Choice

**1.** What does `"  Ana Diaz ".Trim()` return?

a) `"AnaDiaz"`   
b) `"Ana Diaz"`   
c) `"  Ana Diaz"`   
d) `"Ana Diaz "`

```{dropdown} Answer
**b) "Ana Diaz"** — `Trim()` removes whitespace at both ends only, not the space between words.
```

**2.** What does `"ELEC-1042".Substring(5, 4)` return?

a) `"ELEC"`   
b) `"-104"`   
c) `"1042"`   
d) `"042"`

```{dropdown} Answer
**c) "1042"** — it starts at index 5 and takes 4 characters.
```

**3.** Which expression is `true`?

a) `"SKU-7".StartsWith("sku")`   
b) `"SKU-7".Contains("-")`   
c) `"SKU-7".Length == 4`   
d) `"SKU-7".IndexOf('K') == 2`

```{dropdown} Answer
**b)** — the string contains a dash; `StartsWith` is case-sensitive, the length is 5, and `'K'` is at index 1.
```
