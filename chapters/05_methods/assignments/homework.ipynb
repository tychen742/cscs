````csharp
// One solution; compare only after completing your own work.
using System;
static int PackingCost(int parcels)
{
    if (parcels == 0) return 0;
    return parcels < 4 ? 3 * parcels : 10;
}
static bool TryPackingQuantity(string text, out int parcels)
{
    parcels = 0;
    if (!int.TryParse(text, out int value) || value < 0) return false;
    parcels = value; return true;
}
static void ShowCost(int parcels) { Console.WriteLine($"Packing: {PackingCost(parcels)}"); }
Console.WriteLine(PackingCost(0) == 0 && PackingCost(1) == 3 && PackingCost(3) == 9);
Console.WriteLine(PackingCost(4) == 10 && PackingCost(5) == 10);
int quantity;
Console.WriteLine(!TryPackingQuantity("-1", out quantity) && quantity == 0);
Console.WriteLine(!TryPackingQuantity("bad", out quantity));
ShowCost(4);
````

## Question 7: Return name initials



Write Initials(string fullName) returning first and last initials with periods and one separating space, such as A. D. The input contract is two nonempty English names separated by exactly one space. Use IndexOf and indexing; the method must not print. Derive expected strings for Ana Diaz, Li Ng, and Bo Wu before testing. Explain why an empty string is outside this contract; do not add exception handling.


````csharp
// Exercise: Return name initials
// exercise-id: cscs-methods-homework-core-coding-2
using System;
// Your code starts here.
static string Initials(string fullName) { return ""; }
// Your code ends here.
Console.WriteLine(Initials("Ana Diaz"));
Console.WriteLine(Initials("Li Ng"));
Console.WriteLine(Initials("Bo Wu"));
````

````csharp
using System;
static string Initials(string fullName)
{
    int separator = fullName.IndexOf(' ');
    return fullName[0] + ". " + fullName[separator + 1] + ".";
}
Console.WriteLine(Initials("Ana Diaz"));
Console.WriteLine(Initials("Li Ng"));
Console.WriteLine(Initials("Bo Wu"));
````

## Question 8: Extract a returned domain



Write DomainOf(string email) returning the text after @ in lower case. Assume exactly one @, a nonempty local part, and a nonempty English-letter domain. Use IndexOf, Substring, and ToLower. Keep input/output in the caller and preserve the original string. Predict both sample results and the original text; then check a single-letter local part/domain. Malformed addresses are outside this task's contract.


````csharp
// Exercise: Extract a returned domain
// exercise-id: cscs-methods-homework-core-coding-3
using System;
string original = "Ana.Diaz@Example.COM";
// Your code starts here.
static string DomainOf(string email) { return ""; }
// Your code ends here.
Console.WriteLine(DomainOf(original));
Console.WriteLine(DomainOf("b@X"));
Console.WriteLine(original);
````

````csharp
using System;
string original = "Ana.Diaz@Example.COM";
static string DomainOf(string email)
{
    int at = email.IndexOf('@');
    return email.Substring(at + 1).ToLower();
}
Console.WriteLine(DomainOf(original));
Console.WriteLine(DomainOf("b@X"));
Console.WriteLine(original);
````

````csharp
// Exercise: Use a net amount in a larger expression
// exercise-id: cscs-methods-homework-core-coding-5
using System;
decimal subtotal = 100m;
// Your code starts here.
static decimal NetAmount(decimal subtotal, decimal rebate) { return 0m; }
// Write your independent expected-result checks, caller calculation, and output.
// Your code ends here.
````

````csharp
using System;
decimal subtotal = 100m;
static decimal NetAmount(decimal subtotal, decimal rebate)
{
    return Math.Round(subtotal - rebate, 2);
}
// Expectations come from the contract: 100-15=85; 0-0=0;
// 20-2.125=17.875 rounds to the even cent, 17.88.
Console.WriteLine(NetAmount(100m, 15m) == 85m);
Console.WriteLine(NetAmount(0m, 0m) == 0m);
Console.WriteLine(NetAmount(20m, 2.125m) == 17.88m);
decimal amountDue = NetAmount(subtotal, 15m) + 3m;
Console.WriteLine($"Amount due: {amountDue:F2}");
Console.WriteLine($"Subtotal unchanged: {subtotal:F2}");
````
