### Returning string transformations

```{index} Returning string transformations
```

Implement and test the contracts above. Keep printing in the caller so the returned values remain reusable.

````csharp
using System;
static void PrintReps(string text, int count) { for (int i = 0; i < count; i++) Console.Write(text); }
static string StringOfReps(string text, int count) {
    string result = ""; for (int i = 0; i < count; i++) result += text; return result;
}
static string Reverse(string s) {
    string result = ""; for (int i = s.Length - 1; i >= 0; i--) result += s[i]; return result;
}
static string ReverseForeach(string s) {
    string result = ""; foreach (char ch in s) result = ch + result; return result;
}
static string OnlyLetters(string s) {
    string result = "";
    foreach (char ch in s)
        if ((ch >= 'A' && ch <= 'Z') || (ch >= 'a' && ch <= 'z')) result += ch;
    return result;
}
static bool IsPalindrome(string s) {
    string clean = OnlyLetters(s).ToLowerInvariant(); return clean == Reverse(clean);
}
PrintReps("Ok", 3); Console.WriteLine();
Console.WriteLine(StringOfReps("Ok", 3));
Console.WriteLine(Reverse("drab"));
Console.WriteLine(ReverseForeach("drab"));
Console.WriteLine(OnlyLetters("Hello, World!"));
Console.WriteLine(IsPalindrome("A Toyota!"));
Console.WriteLine(IsPalindrome("Hello"));
Console.WriteLine(IsPalindrome("!"));
````

### Dynamic table widths


Implement the table with one common minimal width. Explain why its maximum value can be found from the largest base and power under these positive-input bounds.

````csharp
// Exercise: Dynamic table widths
// Complete the task described above.
using System;
Console.WriteLine("Replace this line with your solution.");
````

````csharp
using System;
static void PowerTable(int nMax, int powerMax) {
    long largest = 1;
    for (int power = 1; power <= powerMax; power++) largest *= nMax;
    int width = Math.Max(largest.ToString().Length, ("n^" + powerMax).Length) + 1;
    for (int power = 1; power <= powerMax; power++) Console.Write(("n^" + power).PadLeft(width));
    Console.WriteLine();
    for (int value = 1; value <= nMax; value++) {
        long product = 1;
        for (int power = 1; power <= powerMax; power++) {
            product *= value; Console.Write(product.ToString().PadLeft(width));
        }
        Console.WriteLine();
    }
}
PowerTable(3, 4);
````
