## Discover a Task Before Naming a Method



A method boundary should express a responsibility that a caller can understand. Imagine an invoice program that reads a quantity, computes a line amount, adds a packing fee, and prints a report. These actions are related, but they do not all need the same information or have the same effects. Separating them lets a calculation be checked without simulating keyboard input or capturing console output.

Write a short sentence for each proposed task. Then identify the information that varies, what the caller needs back, and which effects are intentional. A method named Process does not tell a reader whether it calculates, prints, changes stock, or does all three. A name such as LineTotal makes a more specific promise when accompanied by its price/quantity contract.

| Proposed task | Inputs | Useful result | Intentional external effect |
| --- | --- | --- | --- |
| Calculate line total | Unit price and quantity | Decimal amount | None |
| Format receipt line | Description and amount | String | None |
| Display receipt | Formatted text | No returned value | Console output |

The table is a design tool, not a rule that every program must contain precisely these three methods. A very small program may keep printing in the caller. A reusable formatting policy may deserve a method even when it is initially called once. Repetition is one reason for extraction; giving a complicated idea a clear name is another.

A common mistake is to extract a group of adjacent lines simply because they are adjacent. Instead ask whether they share one coherent purpose and whether the proposed interface hides irrelevant details. If callers need to know every internal step, the boundary may not be helping them.

Guided decision: should CalculateDiscount read a customer name from the keyboard? Under a calculation-only contract, no. The caller obtains input and passes the values required by the rule. This makes the same calculation usable from a console program, a test, or a future interface without rewriting its arithmetic.

## A Repeated Task Needs a Name




```{index} A Repeated Task Needs a Name
```


A shop prints a store name and separator on every receipt. Copying those two statements for every receipt works at first, but a store-name change must then be made in several places. A method names the task once so callers can reuse it. This is **decomposition**: divide a larger program into understandable responsibilities.

Before refactoring, the statements appear twice:

```text
print store name
print separator
print store name
print separator
```

After refactoring, two calls use one definition. Predict the output: defining the helper does not print anything by itself.


````csharp
using System;

PrintHeader();
PrintHeader();

static void PrintHeader()
{
    Console.WriteLine("Campus Books");
    Console.WriteLine("------------");
}
````

The two-line header appears twice, in call order. Change the store name in the definition and run again: both calls use that change. Add a third call without copying the body. This demonstrates reuse rather than merely asserting that reuse is useful.

```{figure} ../../figures/method_decomposition.svg
:alt: Two repeated sets of receipt-header statements become two calls to one PrintHeader definition.

A task name connects each caller to one definition.
```

Read the left column as duplicated instructions and the right as named calls. A method does not automatically make a program better; it should express a coherent responsibility. Separate calculation, input, and presentation when those have different jobs.

### Designing Methods



A caller needs a result without repeating the calculation. Start by stating the method's job, input types, result type, and assumptions. Then write a definition and calls that check that contract. In this section, keep the question “what does the caller receive?” separate from “what appears on the screen?”

For a simple integer square, the contract is: accept an integer in a range whose square fits in int, return its square, and print nothing inside the calculation. We use small inputs here; arbitrary large integers need a stated overflow policy.


You already consume methods such as `Math.Round` and `Console.WriteLine`. Now you supply a definition so a caller can ask your code to do work. The caller needs a contract: what arguments are required, what result is returned, and what effects occur.

This chapter uses top-level statements with explicitly `static` local helper functions for most runnable examples. A **local function** is a named function inside the compiler-generated entry method in this program form. It has no `public` modifier. An explicit static class-member method instead lives inside a class and may use an access modifier. Both forms can demonstrate input, return, and execution flow, but they are not interchangeable declarations.

| Task | Top-level teaching form | Explicit class-based project form |
| --- | --- | --- |
| Begin the program | First top-level statement | First statement in the selected Main |
| Define a helper | `static void PrintHeader() { ... }` beside top-level statements | Same helper inside Program |
| Call it | `PrintHeader();` | Call from Main or another permitted method |
| Explain the wrapper | The compiler supplies the entry method | The source shows Main and Program |

Use one program form at a time in a file. The original class-based examples are retained as console-project reference versions; do not paste them after a complete top-level program. This distinction explains the layout sketch and static/instance terminology that follow. It is not yet a lesson in designing objects.


Methods/functions are reusable, callable, and customizable pieces of code in a computer program. When a function is a part of a class, it is called a method. Class-member functions are called methods. C# also supports local functions, including helpers declared alongside top-level statements, so the program form matters. It is, therefore, by definition, more accurate to call these subroutines methods rather than functions in C# since C# is an OOP language. However, when discussing the construct of subroutine in general, the general term **function** seems appropriate and you will see people use the terms interchangeably from time to time.

In C# or other object oriented programming (OOP) languages, methods are usually associated with a class. In other languages, you may hear the term function. A function is similar to a method, but it is not tied to a class. Due to the similarity in overall usage, it is common to use the terms methods and functions as synonyms.

```text
Method (class member)        Local function                 Standalone function
                             (inside method)                (other languages)
+-----------------------+  +--------------------------+  +-----------------------+
| class Program         |  | class Program            |  | [code...]             |
| {                     |  | {                        |  | int add(int a, int b) |
|   static int Add(...) |  |   static void Main()     |  | {                     |
|   { return a + b; }   |  |   {                      |  |   return a + b;       |
| }                     |  |     int Add(...) { ... } |  | }                     |
+-----------------------+  |   }                      |  | [code...]             |
                           | }                        |  +-----------------------+
                           +--------------------------+
```

You have seen the **Main()** method in C# console app programs generated using the .NET template with `--use-program-main`. In the traditional class-based style, the Main() method is the application’s entry point. When the application is run, execution begins at the start of the Main() method. In modern C#, programs may also use top-level statements without an explicit Main() method.

Up until this unit, you wrote the program steps in top-level statements or an explicit Main method, but now you will be creating new methods that can be called by the main method. Methods are **named** code blocks that can be reused (called) whenever we needed. Methods are an **abstraction** of your code because methods abstract away details (you use the methods by their names only) and that serves to organize your code by function and reduce the repetition of code. In addition, it helps with debugging and maintenance since changes to that block of code only need to happen in one place. Here are some of the main reasons to use multiple methods in your programs:

- Reusing Code: Avoid repetition of code.
- Reducing Complexity: Divide a problem into sub-problems to solve it a piece at a time.
- Maintainability and Debugging: Smaller and specific code blocks are easier to trace and debug. Also, you only need to maintain one piece of code even if it's used all over the place.

When you see duplicate lines of code, that is a signal for you to make a new method. A method is a named set of statements. When we want to execute the statements, we **call** the method using its name. Methods are called in two different ways:

- **Instance methods**: methods that are called using an object, referred to as instance methods or object methods. 
- **static methods**: the methods in this unit are called without an object, so they are static methods. Static methods are also referred to as **class methods**.

Some important characteristics of methods/functions are:

- **SRP**: In general, a method should do a single thing, which we call single responsibility principle (SRP). 
- **Method composition**: You combine a sequence of simple methods to build more complex desired behavior.
- **Method call**: A method body runs when it is **called**. We say we are making a “**function call**” or “method call” when we invoke/use the subroutine.
- **Arguments**: When calling a method, we often pass data into the method. The data passed to the method are called arguments.
- **Parameters**: The variables in the method heading defined to receive those arguments are called parameters. 
- **Return**: We usually design the method to return values when called.

### Method definition




The general syntax for defining a static method is:

```csharp
static returnTypeOrVoid MethodName ( parameter_list )
{
   // statements in the method body...
}
```

Things to know about defining a method in C#:

1. **Parameters**: The *parameter list* can be empty or contain one or more comma separated parameter entries. Each parameter entry has the form 
   > type parameterName

2. **Modifier**: For class-member methods, accessibility determines which callers are permitted. `public static` allows access through the type where that type is accessible; other access levels have other rules. Local helpers beside top-level statements do not take an access modifier.

3. **Return Type**: If **returnTypeOrVoid** in the heading is not `void`, each normally completing path must produce a compatible result; the mere presence of one return statement is insufficient. A return statement has the form

   > `return` *expression* `;`

   where the expression should be of the same type as in **returnTypeOrVoid**.
   Execution of the method terminates immediately when a return statement
   is reached.

4. In the traditional class-based style, execution starts at a method with a heading including

   > `static void Main`

   In modern C#, programs may also use *top-level statements* without an explicit `Main` method.

### Method Signature




One method you encountered in the explicit class-based C# program form is Main, in which *Main* is
the name that you can refer to this set of statements. A Main method is special
because it is designated as the only entry point of a C# application and its selected entry point begins the console program [^0502_signature_call-1].

Note that the **Main method** is contained in a **class** (Program), which is a
standard arrangement for OOP languages.
You define a method by writing the method’s **header** and **body**. The **header** is the complete declaration line. For C# overload identity, the **signature** identifies the name, generic arity when present, and parameter types/passing modes; it does not include the return type or parameter names. [^0502_signature_call-2] When defining methods, you follow a recipe
(standard syntax) for the **method definition**. The
class-member definition pattern is as follows: [^0502_signature_call-3]

``` none
[AccessModifier] static ReturnTypeOrVoid MethodName(ParameterList)
{
   // Method Body: a set of statements enclosed in curly braces { }
}
```

The complete header and body have the following parts:

| Header component | Description |
| --- | --- |
| Access Level | Optional. This determines visibility of methods between classes, It does not determine whether code can instantiate objects. For a local helper, omit access modifiers. For a class-member helper, choose visibility appropriate to its callers. |
| Static modifier | In the form taught here, define the method as a static member method of the class. When not defined as static, it is an instance member. |
| Return Type | If the method is returning a value, this indicates what data type it will be. Not all methods return values. Methods that do not return a value have a return type of `void`. |
| Method Name | The method’s `identifier`; class-member overloads may share a name with different signatures. This is what you use to call your method. |
| Parameter List | A parameter is a placeholder for specific data that the method will act upon. Parameters are optional. In these cases, the `()` are still part of the method call, but remain empty. |
| Method Body | This is where you code your method. Note it is contained between `{}`. |

With method signatures, you can design methods specifically to your needs. But for now, let us analyze the two methods in class TryMethods below to apply our understanding of headers and calls:

````csharp
using System;
MyMethod();
static void MyMethod()
{
    Console.WriteLine("IST 1551");
    Console.WriteLine("T.Y. Chen");
}
````

### Method Calls & Returns




The runnable cell calls MyMethod from top-level statements. The retained class-based version makes the same call inside Main. The call is MyMethod(), not a call to Main.

When making a method call, you use the method **name** followed by
**parentheses**. The method header `static void MyMethod()` indicates the return type is `void` and there are no formal parameters, which means you can call it simply by writing `MyMethod();`.

An important feature of functions/methods is that they can **return** values to their
**callers**. In the example below, `SquareTheNumber(int num)` is defined with a return type of `int`
and is called directly from the top-level code, passing a parameter.

````csharp
using System;
int digit = 4;
int squaredNum = SquareTheNumber(digit);
int squaredAndSummed = squaredNum + SquareTheNumber(digit);
Console.WriteLine(squaredNum);
Console.WriteLine(squaredAndSummed);
Console.WriteLine(SquareTheNumber(5));
static int SquareTheNumber(int num)
{
    return num * num;
}
````

::::{dropdown} Class-based console-project version
The top-level example above is the version to run as a standalone notebook cell. This complete project version retains the explicit `Main` structure. Save it as the only program in a separate console project; do not paste it after top-level statements in the same file.

```csharp
using System;

class Program {
static int SquareTheNumber(int num)
{
    return num * num; // the value to be returned is given by the expression in the return statement.
}

static void Main(){
int digit = 4;
int squaredNum = SquareTheNumber(digit);
int squaredAndSummed = squaredNum + SquareTheNumber(digit);

Console.WriteLine(squaredNum);
Console.WriteLine(squaredAndSummed);
Console.WriteLine(SquareTheNumber(5));
}
}
```
::::


In this example:

- `return num * num;` computes the value that replaces this call in the caller.
- `digit` starts at 4. Passing it binds the value 4 to `num`.
- `squaredNum` is a variable, not a method; it stores the returned 16.
- The addition uses `squaredNum` plus a second call with digit still equal to 4: 16 + 16 = 32.
- The final call uses 5 and returns 25. The three printed values are 16, 32, and 25.

The return type must agree with the returned expression through an allowed conversion. `void` means no returned value; it does not mean the method cannot print or have another effect.

#### Method Calls




1. A method call takes the form

   **MethodName** `(` actual parameter list `)`

   A method call makes the method definition be *executed*.

2. The actual parameter list is a comma separated list of the *same*
   length as the parameter list. Each entry is an expression.
   The entries in an actual parameter list do *not* include type declarations.

   Effectively, the method execution starts by assigning to each
   parameter variable the corresponding value from
   evaluating the actual parameter expression.
   In particular, that means the actual parameter values must be allowed
   in an assignment statement for a variable of the parameter's type!

3. If the method has return type `void`, it can only be used syntactically
   as an entire statement (with a semicolon added). After the method
   call completes, execution continues with the next statement.

4. If there is a non-void return type, then the method call is syntactically
   an expression in the statement where it appears.
   The execution of such a method must reach a return statement. The value
   of the method-call expression is the value of the expression in this
   return statement.

5. A method with a return value can also legally be used as a whole statement.
   In this case the return value is lost. Though legal, this is often an error!

#### Flow of execution




In terms of the construct of **sequential processing**, functions/methods
alter code execution order in several ways: by statements not
being executed as the definition is first read, and then when the
method is called during execution, jumping to the method code,
and back at the end of the method execution. [^0502_signature_call-5]

A class can contain multiple methods. It can be tempting to think the
methods are executed in the order they appear in the class, but this is
not the case. In an explicit class-based console program, ordinary execution begins in the selected Main method; a top-level program begins with its first statement. Each statement in the main is executed sequentially, one at a
time, until you reach a method call. A method call causes the program
execution to jump to the first line of the called method.
Each statement in the called method is then executed in order.
When the called method finishes, execution returns to the point in its caller that was waiting for the result. The caller may itself be another helper. [^0502_signature_call-6]

In other words, the order in which the method definition code blocks does
not matter to C#. It is a human choice. One good practice is to show
`Main` first. This means a human reading in order gets an overview
of what is happening by looking at Main, but does not know the details
until reading the definitions of other methods.
