# Methods

You already call built-in methods, make decisions, and repeat work with loops. Now you will define your own methods to name and reuse those familiar operations. Long programs become unmanageable when all code lives in one place. Methods let you name a block of code, give it inputs, return a result, and reuse the same logic from multiple places.

- **Method definition**: A method packages a named task so the program can call it when needed.
- **Method header and contract**: The full declaration communicates call syntax; the contract states the required behavior. The C# signature is the narrower method identity used for overloading.
- **Parameters**: Parameters let one method work with different input values.
- **Return values**: A value-returning method computes a result that other code can store or combine.

Methods are the first major step from writing a script toward designing a program. They make code easier to read, test, debug, and extend.

<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Define methods with clear names, parameters, and return types
2. Call methods and use their return values in larger expressions
3. Design method signatures that communicate intent clearly
4. Distinguish between `void` methods and value-returning methods
5. Use methods to remove repetition and organize program logic
6. Call string methods to clean, search, and extract parts of text
