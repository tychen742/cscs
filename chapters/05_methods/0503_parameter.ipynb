A method should work with different caller data without reading hidden shared variables. The problem is to keep the task stable while allowing its inputs to vary. Parameters give each call named local inputs; returned values carry results back. The retained Verse comparison makes the repeated pattern visible, followed by business examples and scope traces.


As a young child, you probably heard the "Old MacDonald" song.
Consider the following two verses:

| Verse 1 | Verse 2 |
| --- | --- |
| Old MacDonald had a farm | Old MacDonald had a farm |
| E-I-E-I-O | E-I-E-I-O |
| And on that farm he had a cow | And on that farm he had a duck |
| E-I-E-I-O | E-I-E-I-O |
| With a moo-moo here | With a quack-quack here |
| And a moo-moo there | And a quack-quack there |
| Here a moo, there a moo | Here a quack, there a quack |
| Everywhere a moo-moo | Everywhere a quack-quack |
| Old MacDonald had a farm | Old MacDonald had a farm |
| E-I-E-I-O | E-I-E-I-O |

As you can see, the two verses are in most part identical. We can probably
create a method named `Verse` to abstract the repetitive lines, but the method
will need two **parameters**, which are **placeholders** that allow different
values to be substituted for the animal and noise when the method is called.
The method body will use the parameters as variables to customize the print
statements.

````csharp
using System;
Verse("chicken", "buk");
static void Verse(string animal, string noise)
{
          Console.WriteLine( "Old MacDonald had a farm" );      // method body starting here
          Console.WriteLine( "E-I-E-I-O" );
          Console.WriteLine( "And on that farm he had a " + animal );
          Console.WriteLine( "E-I-E-I-O" );
          Console.WriteLine( "With a " + noise + "-" + noise + " here") ;
          Console.WriteLine( "And a " + noise + "-" + noise + " there" );
          Console.WriteLine( "Here a " + noise + ", there a " + noise );
          Console.WriteLine( "Everywhere a " + noise + "-" + noise );
          Console.WriteLine( "Old MacDonald had a farm" );
          Console.WriteLine( "E-I-E-I-O" );
}
````

Let us review the class-member method definition pattern:

``` none
[AccessModifier] static ReturnTypeOrVoid MethodName(ParameterList)
{
  Method Body
}
```

In this program, when you call the Verse method, you provide values between the
parentheses, called `arguments`, that are copied into the `parameters`
(parameter list in the parentheses after the method name) to
be used in the method body as `variables`. If you call the method more than one time
and supply different arguments, each method call will be run with the arguments
supplied, such as in:

    Verse( "cow", "moo" );
    Verse( "duck", "quack" );

The main method will call the verse method twice, once for the cow and once for the duck.

## Map Arguments to Parameters




```{index} Map Arguments to Parameters
```


Arguments are expressions at the call; parameters are named variables in the definition. Match them by position and compatible type, not by spelling.

| Call | animal in this call | noise in this call |
| --- | --- | --- |
| `Verse("cow", "moo")` | cow | moo |
| `Verse("duck", "quack")` | duck | quack |
| `Verse(chosenAnimal, chosenNoise)` with chosenAnimal = cow and chosenNoise = moo | cow | moo |

General syntax for a two-input helper is `static resultType Name(type1 parameter1, type2 parameter2) { ... }`; the call is `Name(expression1, expression2)`. The call does not repeat type declarations. In the simple required-parameter examples here, argument number, order, and allowed conversions must match. Swapping two strings can compile and still produce the wrong meaning, so type checking does not replace contract checking.

Each ordinary parameter receives a copy of its argument value. With integers this gives independent numerical values. Later chapters explain how a copied reference can still identify a shared object; do not generalize this into deep-copying every argument.


````csharp
using System;
int quantity = 4;
int result = Increase(quantity);
Console.WriteLine($"Caller: {quantity}; returned: {result}");
static int Increase(int count)
{
    count = count + 1;
    return count;
}
````

```{figure} ../../figures/method_parameter_copy.svg
:alt: quantity in the caller stays 4 while a separate copied parameter count changes from 4 to 5 and returns 5.

Different names and separate values in the caller and helper.
```

The first output is `Caller: 4; returned: 5`. Assigning to count does not assign to quantity. To keep the returned increase in quantity, the caller must explicitly write `quantity = Increase(quantity);`. Try that variation, then a zero input. The argument's name could also be count; equal spelling would not merge the two variables.

## Return Values, out, and ref



```{index} gap revision: return values, out, and ref
```


A return value is usually the clearest way to produce one result. Sometimes a method must report both success and a result, or deliberately update a caller variable. C# makes those effects explicit with `out` and `ref`. This connects the `TryParse` call in Chapter 4 to the method definitions you can now write.

General definition and call patterns (schematic):

```csharp
static bool TryRead(string text, out int value) { /* assign value on every normal return */ }
bool success = TryRead(text, out number);
static void Adjust(ref int value) { /* read or replace caller value */ }
Adjust(ref number);
```

| Parameter | Caller obligation | Method obligation | Meaning |
| --- | --- | --- | --- |
| `int value` | Supply a value. | Works with a local copy. | Reassignment does not replace the caller variable. |
| `out int value` | Supply a writable variable, not necessarily initialized. | Assign it before every normal return. | Produces a result in the caller variable. |
| `ref int value` | Supply an initialized writable variable. | May read and replace it. | Deliberately shares the caller variable's storage. |

Both the definition and call show the keyword. Passing an object reference by value later permits mutation of the referenced object but does not permit replacing the caller's reference variable. That is different from a `ref` parameter. Do not add `ref` merely because a value is an object.

The worked example reports failure separately from the parsed value. The method defines a stricter business contract than `int.TryParse`: quantities must be positive. Its failure result is always zero. `AddOne` then demonstrates a deliberate change to an already initialized caller variable.

See the language/API reference for this contract.[^gap-parameters]


````csharp
using System;
static bool TryQuantity(string text, out int quantity)
{
    quantity = 0;
    if (!int.TryParse(text, out int parsed) || parsed <= 0) return false;
    quantity = parsed;
    return true;
}
static void AddOne(ref int value) { value++; }
int amount;
Console.WriteLine(TryQuantity("6", out amount));
AddOne(ref amount);
Console.WriteLine(amount);
Console.WriteLine(TryQuantity("bad", out amount));
Console.WriteLine(amount);
````

## Variable Scopes




A variable may be available for use in some parts of code, but not others.
The `scope` of a variable is the region of the program that the variable is visible,
which means it is accessible by `name` and can be used in the part of code. C# scope rules of variables can be divided into three categories as follows:

- Class Level Scope
- Method Level Scope
- Block Level Scope.

A variable declared inside a method is a `local variable`. A body local belongs to its containing block; a parameter is available in the method body. A nested-block local cannot be used outside that block. You cannot use a local variable before its declaration or read it before it has a value.
In the ordinary noncapturing examples here, each call has its own local values. Do not infer a physical memory address or allocation rule from a scope diagram.
If you called the method again, the old value is not available.

Only the *value* of the actual parameter is passed, not any
variable name, so there is *no need* to have a match between a variable name
(arguments) and the parameter name.

| Region | What can use its local name? | What to check |
| --- | --- | --- |
| Parameter | The called method body | It belongs to this call. |
| Method-body local | Its containing scope after declaration | Another method cannot read it just because it uses the same spelling. |
| Nested block local | That block after declaration | Moving its use outside the block is a compile-time error. |

Scope concerns where a name can be used in source code; lifetime concerns the duration of a value/storage relationship. The later chapters have cases requiring deeper lifetime analysis. Here use separate boxes for caller and helper locals rather than claiming the compiler must allocate/release a specific memory location at each brace.


````csharp
using System;
int x = 1;
int y = Q(5);
Console.WriteLine($"Caller x: {x}; returned y: {y}");
static int Q(int a)
{
    int x = 3;
    x = x + a;
    return x;
}
````

The output is `Caller x: 1; returned y: 8`. The helper's x does not overwrite the caller's x. A static local helper cannot capture the caller's local x; it receives a through its parameter. If a name is needed after a block, declare it in an appropriate enclosing scope and decide how it is initialized, rather than assuming any earlier variable remains available everywhere.

### Scope




1. A variable declared inside a method definition is called a *local variable*.
   This declaration may be in either the parameter
   list or in the body of the method.

2. A local variable comes into existence after the method is called, and ceases
   to exist after that method call terminates. In these noncapturing examples, a local name is visible only in its containing scope, and each call has separate local values. Scope and lifetime are different concepts; nested scopes and later captured-variable examples require more precise rules. Its *value* may be transferred outside of the method scope
   by standard means, principally:

   - If it is the expression in a return statement, its value is
     sent back to the caller.
   - It can be passed as an
     actual parameter to a further method called within its scope.

## Static Variables




You may define `static variables` (variables defined with the word `static`
inside the class, but *outside* of any method definition). These variables are
visible inside all of your methods in the class. Instead of local scope, static variables have *class scope*.
It is good programming practice generally to avoid defining static variables and
instead to put your variables inside methods and explicitly pass
them as parameters where needed. A fixed value shared by the class can be a constant. A `const` field is implicitly static; a mutable static field is not a constant. For example,
A *constant* is a name that you give a fixed data value to and you do not expect the value to change
(e.g., Pi).

````csharp
using System;
Console.WriteLine($"Fixed limit: {CallCounter.Limit}");
CallCounter.Record();
CallCounter.Record();
class CallCounter
{
    public const int Limit = 2;
    private static int calls = 0;
    public static void Record()
    {
        int increment = 1;
        calls += increment;
        Console.WriteLine($"Shared calls: {calls}");
    }
}
````

This example deliberately uses class-member syntax so the fields have a class home. Limit is a fixed constant. The static calls field persists between the two Record calls, so the outputs are 1 and 2. Each call instead has its own local increment. We call through the type name, CallCounter.Record(), without constructing an object. Keep this as a contrast: routine calculations in this chapter should normally receive parameters and use locals, so their results do not depend on a hidden history of calls. Chapters 10–12 develop class design and instance state.

### Static Storage Reference




1. There may be a declaration prefaced by the word `static` that appears
   *inside* a class and *outside* of any method definition in the class.
   Static variables are visible within each method of the class, and may
   be used by the methods.
2. A const field gives a name to a fixed compile-time value and is implicitly static. Mutable static fields instead hold shared state; do not confuse the two.

## Method Tracing




A method can call other methods to help it do its job. Consider the following code:

````csharp
using System;
InchesToCentimeters(10);
static void InchesToCentimeters(double inches)
{
    double centimeters = inches * 2.54;
    PrintInCentimeters(inches, centimeters);
}
static void PrintInCentimeters(double inches, double centimeters)
{
    Console.WriteLine(inches + "-->" + centimeters);
}
````

The output is `10-->25.4`. The caller passes 10 to the first helper; its computed 25.4 becomes an argument of the second helper, which prints and returns control. The second helper does not return a numerical result.

## Reuse a Loop Through a Method Contract




```{index} Reuse a Loop Through a Method Contract
```


Chapter 4's accumulation loop can become a reusable method. For integer n from 0 through 1000, SumToN returns 1 + 2 + ... + n, with an empty sum of zero. The range keeps this example's int arithmetic within bounds. The caller supplies n; the helper owns its accumulator and returns only after all iterations.


````csharp
using System;
Console.WriteLine($"n=0: {SumToN(0)}");
Console.WriteLine($"n=1: {SumToN(1)}");
Console.WriteLine($"n=4: {SumToN(4)}");
static int SumToN(int n)
{
    int total = 0;
    for (int value = 1; value <= n; value++) total += value;
    return total;
}
````

| Call | Accumulator values after each iteration | Returned result |
| --- | --- | --- |
| SumToN(0) | No iterations; initial total is 0 | 0 |
| SumToN(1) | 1 | 1 |
| SumToN(4) | 1, 3, 6, 10 | 10 |

Placing return inside the loop would end the first call iteration rather than accumulate the whole sum. Omitting a final return could leave the zero-iteration path without a result. Trace the decision/loop paths before trying the factorial, repeated-string, and Strange Sequence tasks in Extra Practice. The method wrapper changes reuse and information flow, not the arithmetic rule of the loop.

## Keep Interaction in the Caller




```{index} Keep Interaction in the Caller
```


A reusable calculation or sentence builder should not unexpectedly prompt for input or choose how to print its result. The caller manages the interaction; the helper receives values and returns a result. This is the design needed by the lab's InterviewSentence task, introduced here before the lab asks you to implement it.


````csharp
using System;
string sentence = InterviewSentence("Ana", "10:30");
Console.WriteLine(sentence);
Console.WriteLine(InterviewSentence("Bo", "14:00"));
static string InterviewSentence(string name, string time)
{
    return string.Format("{0} has an interview at {1}.", name, time);
}
````

The outputs are `Ana has an interview at 10:30.` and `Bo has an interview at 14:00.`. No input or printing occurs inside the helper. The same helper can serve a console caller, a test, or later another user interface. Birthday/Verse tasks instead deliberately print a repeated verse; their contracts are different, not automatically wrong.

**Blank-line policy:** printing a blank line inside a verse helper adds it after every call. Printing between calls lets the caller decide which verses need separation. The lab asks you to compare those two placements after this distinction has been taught.
