## Multiple Choice

**1.** What is the root of a tree?

a) A node with no children  
b) The top node with no parent  
c) The deepest node  
d) A duplicate value

**2.** What is a leaf?

a) A node with no children  
b) A node with two parents  
c) The top node  
d) The largest value

**3.** What is a subtree?

a) A sorted array  
b) A node and all nodes below it  
c) A node's parent only  
d) A queue of pending nodes

**4.** Which traversal order is node, left, right?

a) Inorder  
b) Postorder  
c) Preorder  
d) Level order

**5.** Which traversal order is left, node, right?

a) Inorder  
b) Preorder  
c) Postorder  
d) Reverse order

**6.** What is the BST invariant?

a) Every node has exactly two children  
b) Left subtree values are smaller and right subtree values are larger  
c) Values are stored in insertion order  
d) The root is always the smallest value

**7.** What does BST search do when the target is smaller than the current node?

a) Search the right subtree  
b) Search the left subtree  
c) Stop immediately  
d) Visit every node

**8.** What does inorder traversal produce for a valid BST with unique values?

a) Values in sorted order  
b) Values in insertion order  
c) Values in reverse insertion order  
d) Only leaf values

**9.** Why can sorted insertion be a problem for a simple BST?

a) It creates duplicate roots  
b) It can create a tall one-sided chain  
c) It prevents traversal  
d) It removes all leaves

**10.** What does tree height help estimate?

a) How many child references are stored in each node  
b) How long a root-to-leaf path can be  
c) Whether a value is a string  
d) Whether the tree uses arrays
