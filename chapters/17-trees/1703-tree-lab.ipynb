## Build and Inspect a BST

```{index} binary search tree; implementation lab
```


This lab pulls the chapter together: insert values, traverse the result, search for values, and compare tree shapes.


````csharp
using System;
using System.Collections.Generic;

var tree = new BinarySearchTree();
foreach (int value in new[] { 8, 3, 10, 1, 6, 14, 4, 7, 13 })
{
    tree.Insert(value);
}

Console.WriteLine("Inorder:  " + string.Join(", ", tree.InOrder()));
Console.WriteLine("Preorder: " + string.Join(", ", tree.PreOrder()));
Console.WriteLine($"Height: {tree.Height()}");
Console.WriteLine($"Contains 13: {tree.Contains(13)}");
Console.WriteLine($"Contains 12: {tree.Contains(12)}");

class BinarySearchTree
{
    private Node? root;

    public void Insert(int value) => root = Insert(root, value);
    public bool Contains(int value) => Contains(root, value);
    public int Height() => Height(root);

    public List<int> InOrder()
    {
        var values = new List<int>();
        InOrder(root, values);
        return values;
    }

    public List<int> PreOrder()
    {
        var values = new List<int>();
        PreOrder(root, values);
        return values;
    }

    private static Node Insert(Node? node, int value)
    {
        if (node is null) return new Node(value);
        if (value < node.Value) node.Left = Insert(node.Left, value);
        else if (value > node.Value) node.Right = Insert(node.Right, value);
        return node;
    }

    private static bool Contains(Node? node, int value)
    {
        if (node is null) return false;
        if (value == node.Value) return true;
        return value < node.Value ? Contains(node.Left, value) : Contains(node.Right, value);
    }

    private static int Height(Node? node)
    {
        if (node is null) return -1;
        return 1 + Math.Max(Height(node.Left), Height(node.Right));
    }

    private static void InOrder(Node? node, List<int> values)
    {
        if (node is null) return;
        InOrder(node.Left, values);
        values.Add(node.Value);
        InOrder(node.Right, values);
    }

    private static void PreOrder(Node? node, List<int> values)
    {
        if (node is null) return;
        values.Add(node.Value);
        PreOrder(node.Left, values);
        PreOrder(node.Right, values);
    }

    private sealed class Node
    {
        public int Value { get; }
        public Node? Left { get; set; }
        public Node? Right { get; set; }
        public Node(int value) => Value = value;
    }
}
````

`InOrder` should return values in sorted order. `PreOrder` helps reveal the shape created by the insertion order.


## Compare Insertion Orders

```{index} binary search tree; insertion order
```


The same set of values can produce different tree heights. Insertion order is one reason unbalanced BSTs do not guarantee fast operations.


````csharp
using System;

int[] mixed = { 8, 3, 10, 1, 6, 14, 4, 7, 13 };
int[] sorted = { 1, 3, 4, 6, 7, 8, 10, 13, 14 };

Console.WriteLine($"Mixed height: {BuildHeight(mixed)}");
Console.WriteLine($"Sorted height: {BuildHeight(sorted)}");

int BuildHeight(int[] values)
{
    var tree = new BinarySearchTree();
    foreach (int value in values)
    {
        tree.Insert(value);
    }

    return tree.Height();
}

class BinarySearchTree
{
    private Node? root;
    public void Insert(int value) => root = Insert(root, value);
    public int Height() => Height(root);

    private static Node Insert(Node? node, int value)
    {
        if (node is null) return new Node(value);
        if (value < node.Value) node.Left = Insert(node.Left, value);
        else if (value > node.Value) node.Right = Insert(node.Right, value);
        return node;
    }

    private static int Height(Node? node)
    {
        if (node is null) return -1;
        return 1 + Math.Max(Height(node.Left), Height(node.Right));
    }

    private sealed class Node
    {
        public int Value { get; }
        public Node? Left { get; set; }
        public Node? Right { get; set; }
        public Node(int value) => Value = value;
    }
}
````

A sorted insertion order creates a one-sided tree in this simple implementation. A balanced tree implementation would do extra work to prevent that shape.

## Summary

The practical tests for a basic BST are straightforward: inorder traversal should be sorted, search should find present values and reject absent values, and height should reveal when insertion order created a poor shape.
