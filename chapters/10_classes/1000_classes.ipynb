# Classes and Objects

A product has state and rules that should travel together. This chapter introduces classes as definitions of objects with fields, constructors, properties, and methods. You will build and check small business types, distinguish instance state from shared members, and use encapsulation to keep an object valid.

- **Class**: A class defines the fields, properties, constructors, and methods for a kind of object.
- **Instance**: An instance is a concrete object created from a class.
- **Encapsulation**: Encapsulation keeps data inside an object and exposes controlled operations.
- **Constructor**: A constructor initializes a new object into a valid starting state.
- **Property**: A property provides controlled access to object state.

<h2>Learning Objectives</h2>


By the end of this chapter, you will be able to:

1. Define classes with fields, constructors, methods, and properties for a stated business responsibility.
2. Construct valid instances and protect their state with private fields and controlled public operations.
3. Trace changes to independent objects and distinguish instance members from static members.
4. Explain class reference assignment and distinguish a shared object from an independent value.
5. Check constructor and operation contracts using normal and invalid inputs.
