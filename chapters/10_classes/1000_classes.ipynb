# Classes and Objects

Object-oriented programming begins with the class: a blueprint that bundles state and behavior together. This chapter shows how to define classes, create instances, protect data, and build reusable types.

- **Class**: A class defines the fields, properties, constructors, and methods for a kind of object.
- **Instance**: An instance is a concrete object created from a class.
- **Encapsulation**: Encapsulation keeps data inside an object and exposes controlled operations.
- **Constructor**: A constructor initializes a new object into a valid starting state.
- **Property**: A property provides controlled access to object state.

Classes help programs model meaningful concepts instead of passing loose variables through every method.
