<fieldset><legend>1. What does a class define?</legend>
<label><input type="radio" name="cscs-classes-preview-1" value="0"> The members and behavior of a kind of object</label><br>
<label><input type="radio" name="cscs-classes-preview-1" value="1"> Only the values of one local variable</label><br>
<label><input type="radio" name="cscs-classes-preview-1" value="2"> The order of files in a folder</label><br>
<label><input type="radio" name="cscs-classes-preview-1" value="3"> A console window</label><br>
</fieldset>


```{dropdown} Answer
Answer: The members and behavior of a kind of object. A class groups state and behavior into a type.
```


<fieldset><legend>2. What is an instance?</legend>
<label><input type="radio" name="cscs-classes-preview-2" value="0"> A namespace</label><br>
<label><input type="radio" name="cscs-classes-preview-2" value="1"> A particular object created from a class</label><br>
<label><input type="radio" name="cscs-classes-preview-2" value="2"> A compiler setting</label><br>
<label><input type="radio" name="cscs-classes-preview-2" value="3"> A method parameter list</label><br>
</fieldset>


```{dropdown} Answer
Answer: A particular object created from a class. Each constructed instance can have its own state.
```


<fieldset><legend>3. What is the purpose of a constructor?</legend>
<label><input type="radio" name="cscs-classes-preview-3" value="0"> Print every property</label><br>
<label><input type="radio" name="cscs-classes-preview-3" value="1"> Destroy unused objects</label><br>
<label><input type="radio" name="cscs-classes-preview-3" value="2"> Initialize a new object</label><br>
<label><input type="radio" name="cscs-classes-preview-3" value="3"> Make every field static</label><br>
</fieldset>


```{dropdown} Answer
Answer: Initialize a new object. A constructor establishes the starting state.
```


<fieldset><legend>4. Which keyword refers to the current instance?</legend>
<label><input type="radio" name="cscs-classes-preview-4" value="0"> new</label><br>
<label><input type="radio" name="cscs-classes-preview-4" value="1"> class</label><br>
<label><input type="radio" name="cscs-classes-preview-4" value="2"> static</label><br>
<label><input type="radio" name="cscs-classes-preview-4" value="3"> this</label><br>
</fieldset>


```{dropdown} Answer
Answer: this. Use this to refer to the current object, including its fields.
```


<fieldset><legend>5. What does a private field provide?</legend>
<label><input type="radio" name="cscs-classes-preview-5" value="0"> Storage accessed directly only inside its containing type</label><br>
<label><input type="radio" name="cscs-classes-preview-5" value="1"> A globally writable value</label><br>
<label><input type="radio" name="cscs-classes-preview-5" value="2"> A mandatory console output</label><br>
<label><input type="radio" name="cscs-classes-preview-5" value="3"> A second constructor</label><br>
</fieldset>


```{dropdown} Answer
Answer: Storage accessed directly only inside its containing type. Public operations can control access to private state.
```


<fieldset><legend>6. What is a property accessor?</legend>
<label><input type="radio" name="cscs-classes-preview-6" value="0"> An object allocation</label><br>
<label><input type="radio" name="cscs-classes-preview-6" value="1"> A get or set operation for a property</label><br>
<label><input type="radio" name="cscs-classes-preview-6" value="2"> A project folder</label><br>
<label><input type="radio" name="cscs-classes-preview-6" value="3"> A class name</label><br>
</fieldset>


```{dropdown} Answer
Answer: A get or set operation for a property. Accessors read, change, or calculate property values.
```


<fieldset><legend>7. What does a getter-only property allow callers to do?</legend>
<label><input type="radio" name="cscs-classes-preview-7" value="0"> Replace the constructor</label><br>
<label><input type="radio" name="cscs-classes-preview-7" value="1"> Write the backing field directly</label><br>
<label><input type="radio" name="cscs-classes-preview-7" value="2"> Read the property without assigning it</label><br>
<label><input type="radio" name="cscs-classes-preview-7" value="3"> Turn the object into a tuple</label><br>
</fieldset>


```{dropdown} Answer
Answer: Read the property without assigning it. A getter-only property exposes no setter to callers.
```


<fieldset><legend>8. What is encapsulation?</legend>
<label><input type="radio" name="cscs-classes-preview-8" value="0"> Giving every field public access</label><br>
<label><input type="radio" name="cscs-classes-preview-8" value="1"> Putting all code in Main</label><br>
<label><input type="radio" name="cscs-classes-preview-8" value="2"> Copying every object</label><br>
<label><input type="radio" name="cscs-classes-preview-8" value="3"> Protecting state behind controlled public operations</label><br>
</fieldset>


```{dropdown} Answer
Answer: Protecting state behind controlled public operations. Validation in those operations helps preserve valid state.
```


<fieldset><legend>9. What distinguishes a static member?</legend>
<label><input type="radio" name="cscs-classes-preview-9" value="0"> It belongs to the type rather than an individual instance</label><br>
<label><input type="radio" name="cscs-classes-preview-9" value="1"> It must be a private field</label><br>
<label><input type="radio" name="cscs-classes-preview-9" value="2"> It stores a different value for each object</label><br>
<label><input type="radio" name="cscs-classes-preview-9" value="3"> It cannot be called</label><br>
</fieldset>


```{dropdown} Answer
Answer: It belongs to the type rather than an individual instance. Static members are accessed through the type; instance members belong to objects.
```


<fieldset><legend>10. What is the purpose of overriding ToString?</legend>
<label><input type="radio" name="cscs-classes-preview-10" value="0"> To initialize private fields</label><br>
<label><input type="radio" name="cscs-classes-preview-10" value="1"> To provide a useful text representation of an object</label><br>
<label><input type="radio" name="cscs-classes-preview-10" value="2"> To prevent object construction</label><br>
<label><input type="radio" name="cscs-classes-preview-10" value="3"> To share all instance fields</label><br>
</fieldset>


```{dropdown} Answer
Answer: To provide a useful text representation of an object. A report can display that representation with Console.WriteLine(object).
```
