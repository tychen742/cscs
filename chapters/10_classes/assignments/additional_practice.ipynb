````csharp
using System;
using System.Collections.Generic;
public class Animal {
    private readonly string name;
    private readonly List<string> gut = new List<string>();
    public Animal(string name) { this.name = name; }
    public void Greet() { Console.WriteLine($"Hello, my name is {name}."); }
    public void Eat(string food) { gut.Add(food); }
    public void Excrete() {
        if (gut.Count == 0) { Console.WriteLine(""); return; }
        Console.WriteLine(gut[0]); gut.RemoveAt(0);
    }
    public override string ToString() { return "Animal: " + name; }
}
public class TestAnimal {
    public static void Main() {
        Animal froggy = new Animal("Froggy"), robin = new Animal("Robin");
        froggy.Greet(); Console.WriteLine(froggy);
        froggy.Eat("worm"); froggy.Eat("fly");
        Console.WriteLine("Froggy's first removal:"); froggy.Excrete();
        Console.WriteLine("Robin's empty removal:"); robin.Excrete();
        Console.WriteLine("Froggy's second removal:"); froggy.Excrete();
        Console.WriteLine("Froggy's empty removal:"); froggy.Excrete();
    }
}
````
