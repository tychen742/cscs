# Instances, State, and Useful Behavior


```{contents}
:local:
:depth: 2
```


A coffee shop needs each checkout station to maintain its own running total. The class defines the shared rules; constructing an instance creates one station with its own values. Keep the complete CashRegister model beside each runnable example so every cell starts independently. The connected chapter lab applies these ideas to a coffee register.


## Give Each Instance Its Own State

```{index} Give Each Instance Its Own State
```


A constructor establishes a valid station name. Numeric instance fields begin with zero, so this constructor does not need to call Clear. Creating two registers does not make their totals shared: each object owns its own Total and ItemCount. Avoid static fields for values that belong to one register.


````csharp
using System;

CashRegister front = new CashRegister(" Front ");
CashRegister patio = new CashRegister("Patio");
front.Add(4.50m); front.Add(3m); patio.Add(2m);
Console.WriteLine($"{front.Station}: {front.ItemCount} items, {front.Total:F2}");
Console.WriteLine($"{patio.Station}: {patio.ItemCount} item, {patio.Total:F2}");

class CashRegister
{
    public decimal Total { get; private set; }
    public int ItemCount { get; private set; }
    public string Station { get; }
    public CashRegister(string station)
    {
        if (string.IsNullOrWhiteSpace(station)) throw new ArgumentException("Station required.");
        Station = station.Trim();
    }
    public void Add(decimal price)
    {
        if (price < 0m) throw new ArgumentOutOfRangeException(nameof(price));
        Total += price; ItemCount++;
    }
    public decimal Average { get { return ItemCount == 0 ? 0m : Total / ItemCount; } }
    public void Clear() { Total = 0m; ItemCount = 0; }
}
````

## Properties Describe the Public Surface

```{index} Properties Describe the Public Surface
```


Station has a getter and is assigned during construction. Total and ItemCount have public getters and private setters, so callers can read them but must use Add or Clear to change them. This is the idiomatic C# property form of controlled access; separate GetTotal and SetTotal methods would add no useful business behavior here. Validation rejects a negative price before either value changes. A public setter could otherwise let a caller change Total without its corresponding count.


````csharp
using System;

CashRegister register = new CashRegister("Front"); register.Add(5m);
try { register.Add(-2m); } catch (ArgumentOutOfRangeException) { Console.WriteLine("Rejected negative price."); }
Console.WriteLine($"{register.ItemCount} item, {register.Total:F2}");

class CashRegister
{
    public decimal Total { get; private set; }
    public int ItemCount { get; private set; }
    public string Station { get; }
    public CashRegister(string station)
    {
        if (string.IsNullOrWhiteSpace(station)) throw new ArgumentException("Station required.");
        Station = station.Trim();
    }
    public void Add(decimal price)
    {
        if (price < 0m) throw new ArgumentOutOfRangeException(nameof(price));
        Total += price; ItemCount++;
    }
    public decimal Average { get { return ItemCount == 0 ? 0m : Total / ItemCount; } }
    public void Clear() { Total = 0m; ItemCount = 0; }
}
````

## Calculate Behavior from State

```{index} Calculate Behavior from State
```


Average is computed from Total and ItemCount. The empty-register policy returns zero instead of dividing by zero. Computing it on request avoids storing a third value that every update would have to keep synchronized. Clear resets both stored values together. The class maintains business state and calculations; the console program decides how to display them.


````csharp
using System;

CashRegister register = new CashRegister("Front");
Console.WriteLine(register.Average.ToString("F2"));
register.Add(4m); register.Add(6m);
Console.WriteLine(register.Average.ToString("F2"));
register.Clear();
Console.WriteLine($"{register.ItemCount} items, {register.Total:F2}");

class CashRegister
{
    public decimal Total { get; private set; }
    public int ItemCount { get; private set; }
    public string Station { get; }
    public CashRegister(string station)
    {
        if (string.IsNullOrWhiteSpace(station)) throw new ArgumentException("Station required.");
        Station = station.Trim();
    }
    public void Add(decimal price)
    {
        if (price < 0m) throw new ArgumentOutOfRangeException(nameof(price));
        Total += price; ItemCount++;
    }
    public decimal Average { get { return ItemCount == 0 ? 0m : Total / ItemCount; } }
    public void Clear() { Total = 0m; ItemCount = 0; }
}
````
