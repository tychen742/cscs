## Give Each Instance Its Own State

```{index} Give Each Instance Its Own State
```


A constructor establishes a valid station name. Numeric instance fields begin with zero, so this constructor does not need to call Clear. Creating two registers does not make their totals shared: each object owns its own Total and ItemCount. Avoid static fields for values that belong to one register.


````csharp
using System;

CashRegister front = new CashRegister(" Front ");
CashRegister patio = new CashRegister("Patio");
front.Add(4.50m); front.Add(3m); patio.Add(2m);
Console.WriteLine($"{front.Station}: {front.ItemCount} items, {front.Total:F2}");
Console.WriteLine($"{patio.Station}: {patio.ItemCount} item, {patio.Total:F2}");

class CashRegister
{
    public decimal Total { get; private set; }
    public int ItemCount { get; private set; }
    public string Station { get; }
    public CashRegister(string station)
    {
        if (string.IsNullOrWhiteSpace(station)) throw new ArgumentException("Station required.");
        Station = station.Trim();
    }
    public void Add(decimal price)
    {
        if (price < 0m) throw new ArgumentOutOfRangeException(nameof(price));
        Total += price; ItemCount++;
    }
    public decimal Average { get { return ItemCount == 0 ? 0m : Total / ItemCount; } }
    public void Clear() { Total = 0m; ItemCount = 0; }
}
````

## Properties Describe the Public Surface

```{index} Properties Describe the Public Surface
```


Station has a getter and is assigned during construction. Total and ItemCount have public getters and private setters, so callers can read them but must use Add or Clear to change them. This is the idiomatic C# property form of controlled access; separate GetTotal and SetTotal methods would add no useful business behavior here. Validation rejects a negative price before either value changes. A public setter could otherwise let a caller change Total without its corresponding count.


````csharp
using System;

CashRegister register = new CashRegister("Front"); register.Add(5m);
try { register.Add(-2m); } catch (ArgumentOutOfRangeException) { Console.WriteLine("Rejected negative price."); }
Console.WriteLine($"{register.ItemCount} item, {register.Total:F2}");

class CashRegister
{
    public decimal Total { get; private set; }
    public int ItemCount { get; private set; }
    public string Station { get; }
    public CashRegister(string station)
    {
        if (string.IsNullOrWhiteSpace(station)) throw new ArgumentException("Station required.");
        Station = station.Trim();
    }
    public void Add(decimal price)
    {
        if (price < 0m) throw new ArgumentOutOfRangeException(nameof(price));
        Total += price; ItemCount++;
    }
    public decimal Average { get { return ItemCount == 0 ? 0m : Total / ItemCount; } }
    public void Clear() { Total = 0m; ItemCount = 0; }
}
````

## Calculate Behavior from State

```{index} Calculate Behavior from State
```


Average is computed from Total and ItemCount. The empty-register policy returns zero instead of dividing by zero. Computing it on request avoids storing a third value that every update would have to keep synchronized. Clear resets both stored values together. The class maintains business state and calculations; the console program decides how to display them.


````csharp
using System;

CashRegister register = new CashRegister("Front");
Console.WriteLine(register.Average.ToString("F2"));
register.Add(4m); register.Add(6m);
Console.WriteLine(register.Average.ToString("F2"));
register.Clear();
Console.WriteLine($"{register.ItemCount} items, {register.Total:F2}");

class CashRegister
{
    public decimal Total { get; private set; }
    public int ItemCount { get; private set; }
    public string Station { get; }
    public CashRegister(string station)
    {
        if (string.IsNullOrWhiteSpace(station)) throw new ArgumentException("Station required.");
        Station = station.Trim();
    }
    public void Add(decimal price)
    {
        if (price < 0m) throw new ArgumentOutOfRangeException(nameof(price));
        Total += price; ItemCount++;
    }
    public decimal Average { get { return ItemCount == 0 ? 0m : Total / ItemCount; } }
    public void Clear() { Total = 0m; ItemCount = 0; }
}
````

## Statistics Exercise

```{index} Statistics Exercise
```

Modify the project [averager](https://github.com/mstbit/introcs-csharp-examples/tree/master/averager) so the `Averager` class is
converted to `Statistics`. Besides having methods to calculate the count
of data items and average, also calculate the standard deviation with
a method `StandardDeviation`.
It turns out that the only other instance variable needed is
the sum of the squares of the data items, call it `sumOfSqr`.
Before calculating the standard deviation, suppose we
assign the current average to a local variable `avg`.
Then the handiest form of expression for the standard deviation is

```
Math.Sqrt(sumOfSqr/dataCount - avg*avg)
```

Modify the demonstration program to show the standard deviation, too.

This is the **population** standard deviation: variance is the mean of the squared values minus the square of the mean. For values 2 and 4, `sumOfSqr = 20`, `dataCount = 2`, and `avg = 3`, so variance is `20/2 - 9 = 1` and standard deviation is 1. The older expression divided the subtraction by the count and was incorrect. Return `double.NaN` when empty.

For small classroom values, clamp a tiny negative variance caused by rounding with `Math.Max(0, variance)`. For large nearly equal values this formula can lose substantial precision; a running variance algorithm is a later numerical-method improvement, not something a clamp repairs. State whether you intend population or sample statistics; this exercise uses population statistics.

## Unneeded Constructor Code Exercise

```{index} Unneeded Constructor Code Exercise
```

Recall that objects are always initialized. Each instance variable
has a default value assigned before a constructor is even run.
The default value for numeric instance variables is 0, so the
call to `Clear` in the constructor could be left out, leaving the
body of the constructor empty! Try commenting that line out
and test that the behavior of demo program is the same.

Emphasizing the fact that you are thinking about the
initial values of instance variables is not a bad idea. Hence
a common practice is to
explicitly assign even the default values in the constructor, as
we did initially with the call to `Clear` inside the constructor.

If no constructor definition is explicitly provided at all,
the compiler implicitly creates one with no parameters and an empty body.
This means that the entire constructor in `Averager` could be omitted.
Comment the whole constructor out and check.

There is a defensive programming
reason to provide even the do-nothing constructor explicitly:
If you use the implicit constructor and then decide to add a constructor
with parameters, the implicit constructor is no longer defined by the compiler,
so any remaining call to it in your code becomes an error.

## Alternate Internal State Exercise

```{index} Alternate Internal State Exercise
```

The way we represent the internal state for an `Averager` is the
best probably, but if it turns out that the `GetAverage` method
is called a lot more often than a method that changes the state,
we could avoid repeated division by saving the average as an
instance variable. We could keep that *instead of* `sum`
(and still keep `dataCount`). We can manage to
keep the same public interface to the methods. With these
alternate instance variables how would you change
the implementation code and not change the method headings or meanings?
If we keep the assumption that the average of 0
items is
`double.NaN`, you will need to treat adding the first datum as
a special case. The code is simpler if we change the outward assumptions
enough to consider the average of 0 items
to be 0. Try it either way.

In the version with NaN you can avoid testing for NaN,
but if you choose to test for
NaN, you need the boolean `Double.IsNaN` function, because the expression
`double.NaN == double.NaN` is *false*!

This complete demonstration checks the corrected population formula with values two and four. Clear resets all accumulators, and the empty-result policy is checked with IsNaN.

````csharp
using System;
Statistics values = new Statistics();
Console.WriteLine("Empty average is NaN: " + double.IsNaN(values.GetAverage()));
values.Add(2); values.Add(4);
Console.WriteLine($"Count: {values.GetCount()}; Average: {values.GetAverage():F1}; SD: {values.StandardDeviation():F1}");
values.Clear();
Console.WriteLine("Cleared SD is NaN: " + double.IsNaN(values.StandardDeviation()));
public class Statistics {
    private int count;
    private double sum, sumOfSqr;
    public void Add(double value) { count++; sum += value; sumOfSqr += value * value; }
    public int GetCount() { return count; }
    public double GetAverage() { return count == 0 ? double.NaN : sum / count; }
    public double StandardDeviation() {
        if (count == 0) return double.NaN;
        double avg = GetAverage();
        return Math.Sqrt(Math.Max(0, sumOfSqr / count - avg * avg));
    }
    public void Clear() { count = 0; sum = 0; sumOfSqr = 0; }
}
````
