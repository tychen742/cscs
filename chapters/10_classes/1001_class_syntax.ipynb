## Overview




To understand class and object-oriented programming (OOP), some important basic concepts learn first inlcude:

- **Class**: A class is a blueprint, prototype, or template.
- **Class Members**: In C#, a class contains members, which include **fields**, properties, **methods**, and events.
- **Class Instance**: An `object` is an `instance` of a class. Namely, an object is created based on a
  class and therefore is an instance of that class. An object can be created using the `new` keyword.

There are related terminology that you will see about class and OOP:

- **Field**: A field is a variable of any type that is declared directly in a class or struct. Fields are
  members of their containing type. A field stores a piece of data within an object. It acts like a
  variable and may have a different value for each instance of a type.[^1001_class_syntax-field]
- **Property**: Properties expose fields. Fields should (almost always) be kept private to a class and accessed via
  get and set properties.
- The **this** Keyword: In C#, the this keyword refers to the current instance of a class.
- **Reference Types**: A type that is defined as a class is a reference type. Other types include: interface, struct, delegate, or enum.
- **Constructor**: In C#, a constructor is called when an object is created to set default values. A constructor must have the
  **same name** as the enclosing class. Using constructor is not obligatory in C#. If no constructors are specified in a class, the compiler
  automatically creates a parameterless constructor.

## Declaring Classes




To declare a class, you need to use the `class` keyword and give a unique `identifier` (name) to the class. For example,
you want to create a customer class to include some data and methods in one unit:

```csharp
// [access modifier] - [class] - [identifier]

public class Customer
{
   // Fields, properties, methods and events go here...
}
```

Attributes to be considered when creating a class include:

- Access Modifiers: The default access for a class type is internal. Basic levels of access for a class include:[^1001_class_syntax-access]
  - **public**: Access is not restricted. Anyone code can create instances of this class.
  - **protected**: Access is limited to the containing class or types derived from the containing class.
  - **internal**: Access is limited to the current assembly.[^1001_class_syntax-internal]
  - **private**: Access is limited to the containing type.

- Class Identifier: A unique name for the class. An identifier is the name you assign to a type (class, interface, struct, delegate, or enum),
  member, variable, or namespace. An identifier begins with a letter or underscore. As a convention, use PascalCase for class names and method names.

- Class Body: The class body is surrounded by { }. The behavior and data are defined in the class body, including **fields**, properties, **methods**, and events on a class and
  are collectively referred to as **class members**.

- Base Class or Super Class: Optional. Name of the class’s superclass; preceded by a colon (:).

- Interfaces: Interface names implemented by the class; preceded by a colon (:).

The pattern above is the smallest possible class. The runnable version below gives `Customer` one field for the name, one for the credit limit, and one method that uses both. The class is only a description until the `new` expression creates an object from it; the two statements after `new` store values in that object's fields, and the method call reads them back.


````csharp
using System;

Customer first = new Customer();       // create one object from the class
first.Name = "Avery Chen";             // store state in its fields
first.CreditLimit = 2500m;
Console.WriteLine(first.Describe());   // a method reads the object's own fields

public class Customer
{
    public string Name = "";
    public decimal CreditLimit;

    public string Describe()
    {
        return $"{Name} may charge up to {CreditLimit:F2}";
    }
}
````

The output is `Avery Chen may charge up to 2500.00`. Two mistakes are common at this point. Declaring the variable without `new`, as in `Customer first;` followed by `first.Name = ...`, does not compile, because no object exists yet. Naming the class `customer` compiles but breaks the PascalCase convention that lets a reader tell a type from a variable at a glance. The fields here are public so the example stays short; the Properties section shows the protected form that real classes use.

## Using a Class




In the following example, there is a class called `Program` with a Main method. We define another class
called `DoMath` and then, in the Main method of the Program class, 1) instantiate an object of the DoMath class by
using the `new` keyword and 2) use the dot notation (`.`) to access a method defined in DoMath. This is the project form with a `Main` method, shown as a pattern:

```csharp
internal class Program
{
   static void Main(string[] args)
   {
      DoMath doMath = new DoMath();    // create an object using the new keyword
      int aSum = doMath.Sum(2, 2);     // use the [object].[method] syntax to use the methods
      Console.WriteLine(aSum);         // output 4
   }
}

class DoMath                           // class accessability level is default to internal
{
   public int Sum(int num1, int num2)
   {
      var total = num1 + num2;
      return total;
   }
}
```

In addition to store data as local variables and call methods in the same class,
now we have alternatives for storing and accessing data in the methods within
a new class that we write. Just like using the `UI` class or `Random` class, we
now can better design, extend, and organize program functionalities.

When you use classes like `StreamReader`, you create a new `StreamReader` object using the `new`
keyword and then use the `ReadLine` method. In application code, print the value explicitly:

```csharp
StreamReader reader = new StreamReader("ui.cs");
string line = reader.ReadLine();
Console.WriteLine(line);
```

As you can see, we are using the `member access operator (.)` to access a member
of a namespace or a type. Calls to instance methods are always attached to a specific object.
That has always been the part through the `.` of

> *object*`.`*method*`(` … `)`

## Constructors




The constructor is a special method in the class:

1.  Its `name` is the same as the name of its type. For example, if we create a class named `Contact`, then the
    constructor method is also called `Contact`.
2.  Its method `signature` includes only an optional access modifier, the method name, and its parameter list
3.  Its method `signature` has `no return type` (and *no* `static`). Implicitly you are creating a default object from the class.
4.  The constructor can have parameters like a regular method. That means giving values to its fields. For example,
    you may want to store this *state* (the current condition of a system, such as the values of the variables) in
    instance variables `name`, `phone` and `email` if you are creating a Contact type:

```csharp
      public Contact(string fullName, string phoneNumber, string emailAddress)
      {
         name = fullName;
         phone = phoneNumber;
         email = emailAddress;
      }
```

While the local variables in the formal parameters disappear after the constructor terminates,
we want the data to live on as the state of the object. In order to remember state after the
constructor terminates, we must *make sure the information gets into the instance variables*
for the object. This is the basic operation of most constructors: Copy desired formal
parameters in to initialize the state in the fields. That is all our simple
code above does.

To further explain how constructors work, take a look at the following code:

```csharp
internal class Program
{
   static void Main(string[] args)
   {
      CheckID check = new CheckID("taylor swift");
   }
}

class CheckID
{
   public string name;           // this is an instance field (variable)
   public CheckID(string name)
   {
      this.name = name;          // The "this" keyword refers to the current instance of the class
   }

   public void PrintResult()
   {
      // ... if ID for this name exists ...
      Console.WriteLine($"{name} has an existing ID!"); // using the "name" variable in method
   }                             // output "taylor swift has an existing ID!"
   // ....
   // ....
}
```

% Note that `name`, `phone` and `email` are *not* declared as

% local variables. They refer to the *instance* variables, but we are *not* using

% full object notation: an object reference and a

% dot, followed by the field.

% So far always we have always been referring to a built-in type of object

% defined in a different class, like `arrayObject.Length`.

% The constructor is *creating* an object,

% and the use of the bare instance variable names is understood to be giving

% values to the instance variables in this Contact object

% that is being constructed.

% Inside a constructor

% and also inside an instance method

% (discussed below)

% C# allows this shorthand notation without `someObject.`.

`CheckID` stores the name but never shows it. The runnable version below completes the idea: a `Contact` whose constructor copies all three parameters into fields, and a `Print` method that proves each object kept its own state.


````csharp
using System;

Contact first = new Contact("Marie Ortiz", "773-508-7890", "mortiz2@example.edu");
Contact second = new Contact("Otto Heinz", "773-508-9999", "oheinz@example.edu");
first.Print();
second.Print();

class Contact
{
    private string name, phone, email;

    public Contact(string fullName, string phoneNumber, string emailAddress)
    {
        name = fullName;          // copy each parameter into a field
        phone = phoneNumber;
        email = emailAddress;
    }

    public void Print()
    {
        Console.WriteLine($"Name: {name}, Phone: {phone}, Email: {email}");
    }
}
````

Each `new Contact(...)` runs the constructor once with its own arguments, so the first line printed is Marie's record and the second is Otto's. The parameters `fullName`, `phoneNumber`, and `emailAddress` disappear when the constructor returns; the fields keep the values for as long as the object exists.

A common mistake is to declare the fields a second time inside the constructor:

```csharp
public Contact(string fullName, string phoneNumber, string emailAddress)
{
    string name = fullName;    // LOGICAL ERROR: a new local variable, not the field
    string phone = phoneNumber;
    string email = emailAddress;
}
```

This compiles, but the three locals vanish at the closing brace and the fields keep their default value, `null`. `Print` would then show empty values. Fields are declared once, outside every method and constructor; inside them, use the field name without a type.

## Static Classes & Methods




In some conditions you may want to create **static classes** that provide static methods without
instantiation. That means you do not need to use the `new` operator keyword to create a variable
of the class type. Also, without the class instance variable, you access the members of a static
class by using the class name itself. For example, if you have a static class called SomeMath:

```csharp
static class SomeMath
{
   public static int Sum(int a, int b)    // a public static method
   {
      return a + b;
   }
}
```

To use the Sum method in the SomeMath class, you call it directly without creating a new object:

```csharp
internal class Program
{
   static void Main(string[] args)
   {
      int aSum = SomeMath.Sum(2, 2);
      Console.WriteLine(aSum);      // output 4
   }
}
```

As another example, in the .NET Class Library, the `static` [System.Math](https://learn.microsoft.com/en-us/dotnet/api/system.math)
class contains methods that perform mathematical operations, without any requirement to store or
retrieve data that is unique to a particular instance of the [Math](https://learn.microsoft.com/en-us/dotnet/api/system.math)
class. You can use the math methods directly in your code by referring to the static class name `Math`:

```csharp
double dub = -3.14;
Console.WriteLine(Math.Abs(dub));               // 3.14
Console.WriteLine(Math.Floor(dub));             // -4
Console.WriteLine(Math.Round(Math.Abs(dub)));   // 3
```

Run the two forms side by side. `SomeMath.Sum` and the `Math` methods are called through the class name; no object is created anywhere in this cell.


````csharp
using System;

Console.WriteLine(SomeMath.Sum(2, 2));       // class name, then the method
double dub = -3.14;
Console.WriteLine(Math.Abs(dub));
Console.WriteLine(Math.Floor(dub));
Console.WriteLine(Math.Round(Math.Abs(dub)));

static class SomeMath
{
    public static int Sum(int a, int b)       // a public static method
    {
        return a + b;
    }
}
````

The output is `4`, `3.14`, `-4`, and `3`. Compare this with `DoMath` earlier: an instance method needs `new DoMath()` first and then `doMath.Sum`, while a static method is reached as `SomeMath.Sum`. Writing `new SomeMath()` does not compile, because a static class cannot be instantiated. Choose static when the method needs no per-object state, as with arithmetic; choose instances when each object carries its own data, as with contacts and products.

## A Variable Number of Arguments



```{index} gap revision: a variable number of arguments
```


An invoice may contain zero, one, or several charges. Writing a separate `Sum` overload for every possible number of charges does not scale. A parameter array lets one method receive those arguments as an array. It connects familiar array traversal to a more convenient calling form.

General pattern (schematic):

```csharp
static decimal Sum(params decimal[] amounts) { /* process amounts */ }
```

`params` marks the final parameter. In this array form, callers can supply separate compatible values, an existing array, or no values. Inside the method, `amounts` is still an array with a length. The keyword does not mean that one call returns multiple answers. An empty call supplies an empty array; decide explicitly what the method should return for it.

Overloads share a method name but have different parameter lists. For example, `Sum(decimal a, decimal b)` and `Sum(decimal[] amounts)` are distinct overloads, while a changed return type alone cannot distinguish them. A parameter array often avoids a family of repetitive fixed-arity overloads. Keep the simple array form here; broader overload and polymorphism choices are developed later.

See the language/API reference for this contract.[^gap-params]


````csharp
using System;
Console.WriteLine(Charges.Sum());
Console.WriteLine(Charges.Sum(2m, 3m, 4m));
Console.WriteLine(Charges.Sum(new decimal[] { 2m, 3m }));
public static class Charges
{
    public static decimal Sum(params decimal[] amounts)
    {
        decimal total = 0m;
        foreach (decimal amount in amounts) total += amount;
        return total;
    }
}
````

**Guided practice.** Predict `0`, `9`, and `5` before running. Write a parameter-array method that counts charges above a supplied threshold, placing the threshold before the final array parameter. Test an empty call, an exact-threshold value, and an existing array. Compare this design with the retained extra-practice overload exercise.


**Connection to the retained SumManyArgs exercise.** For question nine, define `int Sum(params int[] values)`. The final parameter collects zero or more int arguments into an array for the method body. It is a single parameter declaration with flexible call syntax, not a request to invent many overloads. You may also supply an existing int array. This array form is sufficient for the exercise; modern C# supports additional params collection forms.

## Tuples: When a Class Is Too Much



A class bundles data with the behavior that belongs to it. Sometimes a program only needs to keep two or three values together for a moment, such as a product code and its price, or the lowest and highest sales day of a week. A **tuple** groups a few values into one value without declaring a class.[^1001_class_syntax-tuple-types]

```csharp
(Type1 Name1, Type2 Name2) variableName = (value1, value2);
```

### Creating Tuples



Tuple fields can be unnamed or named. Unnamed fields are read as `Item1`, `Item2`, and so on. Named fields make the code say what each value means, so prefer them.

````csharp
using System;

// Unnamed fields: read them as Item1 and Item2
var item = ("SKU-1042", 19.99m);
Console.WriteLine($"{item.Item1} costs ${item.Item2:F2}");

// Named fields: much clearer
var product = (Sku: "SKU-1042", Price: 19.99m);
Console.WriteLine($"{product.Sku} costs ${product.Price:F2}");
````

### Returning Multiple Values



A method returns one value. Before tuples, returning two results required `out` parameters or a small class written just for that purpose. A tuple return type lets one method report both the lowest and the highest sales day:

````csharp
using System;

static (decimal Low, decimal High) SalesRange(decimal[] dailySales)
{
    decimal low = dailySales[0], high = dailySales[0];
    foreach (decimal sales in dailySales)
    {
        if (sales < low) low = sales;
        if (sales > high) high = sales;
    }
    return (low, high);
}

decimal[] dailySales = { 1250.00m, 980.50m, 1610.25m, 1105.75m, 1432.00m };
var range = SalesRange(dailySales);
Console.WriteLine($"Lowest day: ${range.Low:F2}, highest day: ${range.High:F2}");
````

### Destructuring



**Destructuring** unpacks a tuple into separate variables in one statement. Use the discard `_` for a value you do not need.

````csharp
using System;

static (decimal Low, decimal High) SalesRange(decimal[] dailySales)
{
    decimal low = dailySales[0], high = dailySales[0];
    foreach (decimal sales in dailySales)
    {
        if (sales < low) low = sales;
        if (sales > high) high = sales;
    }
    return (low, high);
}

// Unpack both values into separate variables in one statement
var (low, high) = SalesRange(new[] { 1250.00m, 980.50m, 1610.25m });
Console.WriteLine($"Range: ${low:F2} to ${high:F2}");

// Use _ (a discard) for a value you do not need
var (sku, _) = ("SKU-1042", 19.99m);
Console.WriteLine(sku);
````

### Tuples in Collections



A list of tuples is a quick way to keep paired data, such as sales by representative, when the pairs do not need any behavior of their own:

````csharp
using System;
using System.Collections.Generic;

var salesByRep = new List<(string Rep, decimal Sales)>
{
    ("Avery", 18250.00m),
    ("Jordan", 22410.50m),
    ("Morgan", 19875.25m),
};

var top = salesByRep[0];
foreach (var (rep, sales) in salesByRep)
{
    Console.WriteLine($"{rep}: ${sales:F2}");
    if (sales > top.Sales) top = (rep, sales);
}
Console.WriteLine($"Top rep: {top.Rep} (${top.Sales:F2})");
````

### Tuples or Classes?



| Use a tuple when | Use a class when |
|---|---|
| A method returns two or three related values | The data has behavior (methods) or rules to protect |
| The values are used briefly, inside one method or class | The type is shared across the program |
| Named fields make the meaning clear enough | It needs constructors, properties, inheritance, or interfaces |

If you find yourself passing the same tuple shape through many methods, that is a sign it should become a class. Chapter 20 uses tuples for graph edges, where a pair of locations is all an edge needs.
