## Overview



**Object-oriented programming (OOP)** organizes programs around objects and their interactions. This reading begins with one class at a time: declare the class, create an object, initialize its data, and call its methods. Then compare methods that belong to objects with methods that belong to the class itself.

A customer record needs data, such as a name and credit limit, and operations that use that data. A **class** defines a type that brings related data and behavior together. An **object**, also called an **instance**, is one particular value created from that class. One `Customer` class can describe many customer objects, each with its own name and credit limit.

```{figure} ../../figures/class_instances.svg
:name: fig-class-instances
:width: 584px
:alt: One Account class defines Owner, Balance, and decimal Deposit(decimal amount). Two account objects have separate owners and balances of 2500m and 1000m.

One `Account` class defines the structure and behavior of two separate account objects. Each object has its own `Owner` and `Balance`. `Deposit` accepts a `decimal` amount and returns the updated balance. The **`m` suffix** in `2500m` and `1000m` marks a **decimal literal**, a number written directly in code with type `decimal`, commonly used for monetary calculations. It does not mean money or million. The balances shown are assigned after construction.
```

The declarations inside a class are its **members**. The first examples use two kinds:

- A **field** is a variable declared directly inside a class. An instance field stores part of each object's **state**, meaning the data it currently holds.[^1001_class_syntax-field]
- A **method** defines an operation. An **instance method** runs on a particular object and can use that object's fields.

A **constructor** initializes an object when it is created. The `new` expression creates a class instance and invokes a constructor; its syntax is developed below. A **parameterless constructor** takes no arguments. For the non-static classes taught here, C# supplies a parameterless constructor when you declare no instance constructors yourself.

A class is a **reference type**: a variable of that type stores a reference to an object, rather than containing all the object's data directly. That reference lets the program reach the object's members. Section 10.3 develops how references and independent objects behave.

Section 10.2 introduces **properties**, members that provide controlled access to data or compute a value. Properties do not always correspond to a stored field. Events, which notify other code that something happened, are another kind of member; they are outside this chapter's core examples.

The final part of this reading introduces **tuples**, which group a few related values without declaring a new class. They provide a useful comparison when the data needs no custom behavior.

## Declaring Classes



A class declaration has two parts: a **header** and a **body**. Begin with this general syntax pattern:

```text
accessModifier class ClassName          // Class header
{                                       // Class body begins
    // Class members go here.
}                                       // Class body ends
```

The **class header** is the part before the opening `{`. It names the class and specifies modifiers, such as its access level. The **class body** includes the opening and closing braces and the member declarations inside them. These members hold the class's data and define its behavior. Together, the header and body form the complete **class declaration**.

Before reading the access options, distinguish where a class is declared. A **top-level class** is declared outside any other type, although it may be inside a **namespace**, a named grouping of related types. A **nested class** is declared inside another type. An **assembly** is a compiled unit of .NET code, usually the output of one project. A **derived class** extends another class through **inheritance**; Chapter 12 develops that relationship.

Read the header from left to right:

- **Access modifier:** `accessModifier` is a placeholder for an access level. It controls where the class can be used. These four common modifiers provide a starting map:[^1001_class_syntax-access]

  | Modifier | Accessible from |
  | --- | --- |
  | `public` | Any code that can access the containing type. |
  | `protected` | The containing class and classes derived from it. |
  | `internal` | Code in the same project (compiled into the same assembly). Default for a top-level class when no access modifier is specified. |
  | `private` | The containing type. Default for a class nested inside another class when no access modifier is specified. |

  A top-level class uses `public` or `internal`; omitting the modifier makes it `internal`. Class members and nested classes have additional options, including `protected` and `private`. Inheritance and `protected` access are developed in Chapter 12. `internal` access is limited to the current assembly.[^1001_class_syntax-internal]

- **Class keyword:** Write `class` literally. This keyword tells C# that you are declaring a class.
- **Class name:** `ClassName` is a placeholder for the identifier you choose. An **identifier** is a name assigned to a type, member, variable, or namespace. Follow these rules and conventions when naming a class:

  - Begin with a letter or underscore; subsequent characters can include letters, digits, and underscores.
  - Do not include spaces or punctuation such as hyphens.
  - Names are case-sensitive: `Customer` and `customer` are different identifiers.
  - Do not use a reserved C# keyword as an ordinary identifier. C# permits a keyword escaped with `@`, but choose a descriptive non-keyword name for these examples.
  - Choose a class name that is not already used in the same namespace.
  - By convention, use **PascalCase**, capitalizing the first letter of each word, as in `CustomerAccount`.
  - Choose a descriptive noun or noun phrase that identifies what the class represents, such as `Account` or `CustomerAccount`.


The **class body** is where you declare **class members**, including fields, properties, methods, and events. Write the braces literally, and replace the comment in the pattern with the members your class needs.

For example, substitute `public` for the access modifier and `Customer` for the class name:

```csharp
public class Customer                   // Class header
{                                       // Class body begins
    // Fields, properties, methods, and events go here.
}                                       // Class body ends
```

Here, `public class Customer` is the header. The body contains only a comment for now, so this declaration has no explicitly declared members. The next example adds fields and a method.

**Later header extensions:** A class can name a **base class**, the class it extends, and **interfaces**, contracts describing operations it provides, after a colon (`:`). Chapter 12 develops their syntax and use. The basic declarations here need neither extension.


Now fill the class body with members. The `Customer` example below adds two instance fields to hold state and an instance method to describe that state. Their general declaration patterns inside a class are:

```text
accessModifier FieldType fieldName;
accessModifier FieldType fieldName = initialValue;

accessModifier ReturnType MethodName(ParameterType parameterName, ...)
{
    // Method statements go here.
}
```

`FieldType` and `ReturnType` stand for data types; a method that returns no value uses `void`. Choose names for the fields, method, and parameters. A field initializer supplies its starting value. The parameter list can be empty; the `...` represents additional parameters and is not literal C# code. Omitting `static` makes these fields and methods belong to individual objects.

To create an object and use its members, follow these general patterns:

```text
ClassName variableName = new ClassName(argumentList);
variableName.fieldName = value;
variableName.MethodName(argumentList);
```

The first line has two jobs: `ClassName variableName` declares a reference variable, and `new ClassName(argumentList)` creates an object and invokes its constructor. The assignment stores the resulting reference in the variable. `argumentList` stands for zero or more comma-separated arguments, so a parameterless construction uses `()`. The **member access operator**, the dot (`.`), selects a member of that object. A **parameter** is a variable declared in a method or constructor header; an **argument** is a value or expression supplied by its caller. A method call also needs parentheses; if it returns a value, the call can be used in an expression or passed to `Console.WriteLine`.

The following runnable example uses `Customer` as the class name and `first` as the variable name. Its two assignments store values in that object's fields, and `Describe()` reads those values back.


````csharp
using System;

Customer first = new Customer();       // create one object from the class
first.Name = "Avery Chen";             // store state in its fields
first.CreditLimit = 2500m;
Console.WriteLine(first.Describe());   // a method reads the object's own fields

public class Customer
{
    public string Name = "";
    public decimal CreditLimit;

    public string Describe()
    {
        return $"{Name} may charge up to {CreditLimit:F2}";
    }
}
````

The output is `Avery Chen may charge up to 2500.00`. Two mistakes are common at this point. Declaring the variable without `new`, as in `Customer first;` followed by `first.Name = ...`, does not compile, because the local variable has not been assigned a reference. Naming the class `customer` compiles but breaks the PascalCase convention that lets a reader tell a type from a variable at a glance. The fields here are public so the example stays short; Section 10.2 shows how private fields and public properties control access to stored data.

## Using a Class



Creating an object is also called **instantiation**. Once an object exists, a caller uses its reference to access public members. The complete interaction is: create an object, keep its reference in a variable, and call a method on that object.

General syntax for a call that returns a value:

```text
ClassName objectVariable = new ClassName(constructorArguments);
ResultType resultVariable = objectVariable.MethodName(methodArguments);
```

The first argument list belongs to the constructor; the second belongs to the method. They can contain different values, or be empty. `ResultType` must be compatible with the method's returned value. For a `void` method, omit the result declaration and assignment and write the call as a statement.

The next example applies this pattern to `DoMath`. Its `Sum` method adds two arguments. This version is a complete console-project program: `Main` is the entry method where execution begins. Save alternative complete programs in separate projects rather than combining their entry points.

```csharp
internal class Program
{
   static void Main(string[] args)
   {
      DoMath doMath = new DoMath();    // create an object using the new keyword
      int aSum = doMath.Sum(2, 2);     // use the [object].[method] syntax to use the methods
      Console.WriteLine(aSum);         // output 4
   }
}

class DoMath                           // top-level class defaults to internal
{
   public int Sum(int num1, int num2)
   {
      var total = num1 + num2;
      return total;
   }
}
```

The call `doMath.Sum(2, 2)` runs `Sum` on the object referenced by `doMath`; the returned `4` is stored in `aSum` and then printed. `Sum` is an instance method because its declaration omits `static`. Later in this reading, compare it with a static method that performs the same calculation without an object.

This is the same construction-and-call pattern used with library classes such as `Random` and `StreamReader`. The following file-reading fragment assumes the existing file `ui.cs` and the appropriate `System.IO` import. It illustrates the call pattern, not a standalone file-handling program:

```csharp
StreamReader reader = new StreamReader("ui.cs");
string line = reader.ReadLine();
Console.WriteLine(line);
```

Here, the filename is an argument to the constructor. `ReadLine()` is a separate method call on the resulting reader. Normal file-handling code must also dispose the reader, using the cleanup techniques from Chapter 9.


The browser examples use **top-level statements**, executable statements written outside an explicit `Main` method. The compiler supplies the entry method for them. This describes program statements, whereas a *top-level class* describes where a type is declared. The runnable version below does the same work as `Main` above, then makes a second call with different arguments:


````csharp
using System;

DoMath doMath = new DoMath();      // create an object using the new keyword
int aSum = doMath.Sum(2, 2);       // [object].[method] runs the method on that object
Console.WriteLine(aSum);
Console.WriteLine(doMath.Sum(10, 32));

class DoMath                       // accessibility defaults to internal
{
    public int Sum(int num1, int num2)
    {
        int total = num1 + num2;
        return total;
    }
}
````

The output is `4` and then `42`. `doMath` is a variable of type `DoMath`, exactly as `reader` was a variable of type `StreamReader`; the dot after it selects a member of that one object. Calling `DoMath.Sum(2, 2)` on the class name does not compile here, because `Sum` is an instance method and needs an object to run on. The constructors section next explains initialization. The later static-method section explains calls through the class name.

## Constructors



A constructor establishes an object's initial state. Creating an object and initializing it belong to the same `new` expression: the caller supplies arguments, the constructor receives them as parameters, and its body stores the values the object needs to keep.

General declaration pattern inside a class:

```text
accessModifier ClassName(ParameterType parameterName, ...)
{
    // Initialize the object's fields here.
}
```

Like a method, a constructor has a header and a body. Its header uses exactly the enclosing class name and has no return type, including no `void`. Its parameter list may be empty. This pattern describes an instance constructor, so it does not use `static`.

General construction pattern at the call site:

```text
ClassName objectVariable = new ClassName(argumentList);
```

Arguments supply the corresponding parameter values. The constructor's parameters are local to that call; the object's fields retain the copied values after construction finishes. For example, this fragment belongs inside `Contact`, with `name`, `phone`, and `email` declared as fields:

```csharp
      public Contact(string fullName, string phoneNumber, string emailAddress)
      {
         name = fullName;
         phone = phoneNumber;
         email = emailAddress;
      }
```

Assignments without a type name update the existing fields. Declaring another variable inside the constructor would instead create a local variable, which does not preserve the object's state.

Sometimes a parameter and field have the same name. The keyword **`this`** refers to the object whose constructor or instance method is running. General field-access pattern inside that constructor or method:

```text
this.fieldName = parameterName;
```

The dot selects a field of the current object. `this` makes the destination explicit and resolves a name collision. In the following project example, `this.name` is the field, and `name` by itself is the constructor parameter:

```csharp
internal class Program
{
   static void Main(string[] args)
   {
      CheckID check = new CheckID("taylor swift");
   }
}

class CheckID
{
   public string name;           // this is an instance field (variable)
   public CheckID(string name)
   {
      this.name = name;          // The "this" keyword refers to the current instance of the class
   }

   public void PrintResult()
   {
      // ... if ID for this name exists ...
      Console.WriteLine($"{name} has an existing ID!"); // uses the object's field
   }                             // prints only if PrintResult is called
   // ....
   // ....
}
```

`new CheckID("taylor swift")` stores the argument in that object's field. The program does not call `PrintResult()`, so it prints nothing. Adding `check.PrintResult();` after construction would print the message. The example illustrates storing and displaying a name; it does not actually search an ID database.

When field names and parameter names differ, as in `name = fullName`, the field can be used without `this.` inside the constructor or an instance method. That shorter form still accesses the current object's field.


The next standalone example applies construction to two contacts. Each `new Contact(...)` creates a separate object and initializes its three fields. Calling `Print()` afterward makes the retained state visible:


````csharp
using System;

Contact first = new Contact("Marie Ortiz", "773-508-7890", "mortiz2@example.edu");
Contact second = new Contact("Otto Heinz", "773-508-9999", "oheinz@example.edu");
first.Print();
second.Print();

class Contact
{
    private string name, phone, email;

    public Contact(string fullName, string phoneNumber, string emailAddress)
    {
        name = fullName;          // copy each parameter into a field
        phone = phoneNumber;
        email = emailAddress;
    }

    public void Print()
    {
        Console.WriteLine($"Name: {name}, Phone: {phone}, Email: {email}");
    }
}
````

Each `new Contact(...)` runs the constructor once with its own arguments, so the first line printed is Marie's record and the second is Otto's. The parameters `fullName`, `phoneNumber`, and `emailAddress` disappear when the constructor returns; the fields keep the values for as long as the object exists.

A common mistake is to declare the fields a second time inside the constructor:

```csharp
public Contact(string fullName, string phoneNumber, string emailAddress)
{
    string name = fullName;    // LOGICAL ERROR: a new local variable, not the field
    string phone = phoneNumber;
    string email = emailAddress;
}
```

This compiles, but the three locals vanish at the closing brace and the fields keep their default value, `null`. `Print` would then show empty values. Fields are declared once, outside every method and constructor; inside them, use the field name without a type.

## Static Classes & Methods



An **instance member** belongs to an object; a **static member** belongs to the class itself. A static method has no current object, so it cannot use `this` or directly access an object's instance fields. It can use its parameters, static members, and objects it obtains explicitly.

A **static class** groups static operations and cannot be instantiated. An ordinary class can contain both instance and static methods, but the example here uses a static class to emphasize operations that need no per-object state.

General declaration and call patterns:

```text
accessModifier static class ClassName
{
    accessModifier static ReturnType MethodName(ParameterType parameterName, ...)
    {
        // Method statements go here.
    }
}

ClassName.MethodName(argumentList);
```

`static` is a literal keyword in both declarations. The remaining names and types are placeholders, as in the earlier patterns. A static class cannot be instantiated, and its methods are called through the class name. Use a returned value in an expression when the caller needs the result.

For example, define a static class called `SomeMath`:

```csharp
static class SomeMath
{
   public static int Sum(int a, int b)    // a public static method
   {
      return a + b;
   }
}
```

To use the Sum method in the SomeMath class, you call it directly without creating a new object:

```csharp
internal class Program
{
   static void Main(string[] args)
   {
      int aSum = SomeMath.Sum(2, 2);
      Console.WriteLine(aSum);      // output 4
   }
}
```

As another example, in the .NET Class Library, the `static` [System.Math](https://learn.microsoft.com/en-us/dotnet/api/system.math)
class contains methods that perform mathematical operations, without any requirement to store or
retrieve data that is unique to a particular instance of the [Math](https://learn.microsoft.com/en-us/dotnet/api/system.math)
class. You can use the math methods directly in your code by referring to the static class name `Math`:

```csharp
double dub = -3.14;
Console.WriteLine(Math.Abs(dub));               // 3.14
Console.WriteLine(Math.Floor(dub));             // -4
Console.WriteLine(Math.Round(Math.Abs(dub)));   // 3
```

Run the two forms side by side. `SomeMath.Sum` and the `Math` methods are called through the class name; no object is created anywhere in this cell.


````csharp
using System;

Console.WriteLine(SomeMath.Sum(2, 2));       // class name, then the method
double dub = -3.14;
Console.WriteLine(Math.Abs(dub));
Console.WriteLine(Math.Floor(dub));
Console.WriteLine(Math.Round(Math.Abs(dub)));

static class SomeMath
{
    public static int Sum(int a, int b)       // a public static method
    {
        return a + b;
    }
}
````

The output is `4`, `3.14`, `-4`, and `3`. Compare this with `DoMath` earlier: an instance method needs `new DoMath()` first and then `doMath.Sum`, while a static method is reached as `SomeMath.Sum`. Writing `new SomeMath()` does not compile, because a static class cannot be instantiated. Choose static when the method needs no per-object state, as with arithmetic; choose instances when each object carries its own data, as with contacts and products.


The table collects the differences. The call forms come from the examples on this page; the register comes from {ref}`the class instance examples in 10.3 <class-instance-examples>`.

| | Static method | Instance method |
| --- | --- | --- |
| Declared with | `static` in the method header | No `static` in the method header |
| Called through | The class name: `SomeMath.Sum(2, 2)` | An object reference: `front.Add(4.50m)` |
| Access to object state | No current object; needs an explicit reference to access instance members | Can directly access its current object's fields and properties |
| `this` | Not allowed; there is no current object | Refers to the current object (see 10.2) |
| Choose it when | The result depends only on the arguments, as with `Math.Abs` | The result reads or changes the object's own data, as with a register's `Total` |

Look again at `DoMath.Sum` with this table in hand. It is written as an instance method, so the caller has to create a `DoMath` object first, yet the method uses nothing from that object: the sum depends only on its two arguments. C# allows that, but the last row says it should have been static. An instance method is a useful choice when it works with its object's own data, as `Add` does when it updates the register's `Total` and `ItemCount`.

## A Variable Number of Arguments



```{index} gap revision: a variable number of arguments
```


An invoice may contain zero, one, or several charges. An **overload** is a method that shares a name with another method but has a different parameter list. Writing a separate `Sum` overload for every possible number of charges does not scale. A parameter array lets one method receive those arguments as an array. It connects familiar array traversal to a more convenient calling form.

General pattern (schematic):

```text
static ReturnType MethodName(params ElementType[] parameterName)
{
    // Process parameterName as an array.
}
```

`ReturnType`, `MethodName`, `ElementType`, and `parameterName` are placeholders. Write `params` and `[]` literally; the brackets mean that this parameter is an array. Other parameters, if needed, go before the `params` parameter. For example, the pattern becomes `static decimal Sum(params decimal[] amounts)` for decimal charges.

General call patterns for this static class-member form:

```text
ClassName.MethodName();
ClassName.MethodName(value1, value2, value3);
ClassName.MethodName(existingArray);
```

`params` marks the final parameter. In this array form, callers can supply separate compatible values, an existing array, or no values. Inside the method, `amounts` is still an array with a length. The keyword does not mean that one call returns multiple answers. An empty call supplies an empty array; decide explicitly what the method should return for it.

Overloads share a method name but have different parameter lists. For example, `Sum(decimal a, decimal b)` and `Sum(decimal[] amounts)` are distinct overloads, while a changed return type alone cannot distinguish them. A parameter array often avoids a family of repetitive fixed-arity overloads. Keep the simple array form here; broader overload and polymorphism choices are developed later.

See the language/API reference for this contract.[^gap-params]


````csharp
using System;
Console.WriteLine(Charges.Sum());
Console.WriteLine(Charges.Sum(2m, 3m, 4m));
Console.WriteLine(Charges.Sum(new decimal[] { 2m, 3m }));
public static class Charges
{
    public static decimal Sum(params decimal[] amounts)
    {
        decimal total = 0m;
        foreach (decimal amount in amounts) total += amount;
        return total;
    }
}
````

**Guided practice.** The three calls above produce `0`, `9`, and `5`: an empty sum, a sum of three separate arguments, and a sum of an existing array. Write a parameter-array method that counts charges above a supplied threshold, placing the threshold before the final array parameter. Test an empty call, an exact-threshold value, and an existing array. Compare this design with the retained extra-practice overload exercise.


**Connection to the SumManyArgs exercise.** In the additional-practice SumManyArgs task, define `int Sum(params int[] values)`. The final parameter collects zero or more int arguments into an array for the method body. It is a single parameter declaration with flexible call syntax, not a request to invent many overloads. You may also supply an existing int array. This array form is sufficient for the exercise; modern C# supports additional params collection forms.

## Tuples: When a Class Is Too Much



A class bundles data with the behavior that belongs to it. Sometimes a program only needs to keep two or three values together for a moment, such as a product code and its price, or the lowest and highest daily sales amounts in a week. A **tuple** groups a few values into one value without declaring a class.[^1001_class_syntax-tuple-types]

General syntax pattern for a named tuple variable:

```text
(Type1 Name1, Type2 Name2) variableName = (value1, value2);
```

`Type1` and `Type2` are the component types; `Name1` and `Name2` name the tuple fields. `variableName` names the whole tuple. The values on the right match those types and positions. Parentheses group the components, and commas separate them.




```{index} Creating Tuples
```

### Creating Tuples



Tuple fields can be unnamed or named. Unnamed fields are read as `Item1`, `Item2`, and so on. Named fields make the code say what each value means, so prefer them.

General creation and access patterns:

```text
var variableName = (value1, value2);
var variableName = (Name1: value1, Name2: value2);
variableName.Item1
variableName.Name1
```

The first creation form uses unnamed fields; the second supplies names before colons. `var` asks the compiler to infer the tuple type from its values. The last two lines are expressions that select the first field in each form. The next example uses both forms for a product code and price.


````csharp
using System;

// Unnamed fields: read them as Item1 and Item2
var item = ("SKU-1042", 19.99m);
Console.WriteLine($"{item.Item1} costs ${item.Item2:F2}");

// Named fields: much clearer
var product = (Sku: "SKU-1042", Price: 19.99m);
Console.WriteLine($"{product.Sku} costs ${product.Price:F2}");
````

Both output lines show `SKU-1042 costs $19.99`. `Item1` and `Sku` select corresponding first fields, while `Item2` and `Price` select their decimal second fields. Names improve how a caller reads a value; they do not create a new product class or add validation behavior.

### Returning Multiple Values



A method returns one value. A tuple return type packages related results into that one value. General syntax pattern:

```text
static (Type1 Name1, Type2 Name2) MethodName(ParameterType parameterName)
{
    // Compute the component values.
    return (value1, value2);
}

var result = MethodName(argument);
```

The parenthesized types and names form the return type. The `return` statement supplies values in the same order; the caller accesses them as `result.Name1` and `result.Name2`. Additional parameters follow the usual comma-separated method syntax.

For example, a tuple return type lets `SalesRange` report both the lowest and highest sales amount:


**Input contract:** `dailySales` must be non-null and contain at least one value. The first element supplies both its initial minimum and maximum; an empty array has no first element, so this implementation does not define an empty-case result. A caller must check that contract or use a separately specified failure policy. The following example keeps that assumption rather than returning invented zero minimum and maximum values.


````csharp
using System;

static (decimal Low, decimal High) SalesRange(decimal[] dailySales)
{
    decimal low = dailySales[0], high = dailySales[0];
    foreach (decimal sales in dailySales)
    {
        if (sales < low) low = sales;
        if (sales > high) high = sales;
    }
    return (low, high);
}

decimal[] dailySales = { 1250.00m, 980.50m, 1610.25m, 1105.75m, 1432.00m };
var range = SalesRange(dailySales);
Console.WriteLine($"Lowest day: ${range.Low:F2}, highest day: ${range.High:F2}");
````

The returned value is one tuple: `(980.50m, 1610.25m)`. The declaration names its fields `Low` and `High`, so the caller reads `range.Low` and `range.High`. The loop updates its local candidates; `return (low, high)` packages their final values. The printed line is `Lowest day: $980.50, highest day: $1610.25`. A return type with two fields still returns one value.

### Destructuring



**Deconstruction**, also called **destructuring**, unpacks a tuple into separate variables in one statement. Use the discard `_` for a value you do not need.

General deconstruction patterns:

```text
var (firstVariable, secondVariable) = tupleExpression;
(Type1 firstVariable, Type2 secondVariable) = tupleExpression;
var (firstVariable, _) = tupleExpression;
```

`tupleExpression` supplies one tuple value, possibly returned by a method. Variables receive its components by position. `var` infers their types, while the second pattern declares them explicitly. The literal underscore `_` discards a component. The following example applies the first and third patterns.


````csharp
using System;

static (decimal Low, decimal High) SalesRange(decimal[] dailySales)
{
    decimal low = dailySales[0], high = dailySales[0];
    foreach (decimal sales in dailySales)
    {
        if (sales < low) low = sales;
        if (sales > high) high = sales;
    }
    return (low, high);
}

// Unpack both values into separate variables in one statement
var (low, high) = SalesRange(new[] { 1250.00m, 980.50m, 1610.25m });
Console.WriteLine($"Range: ${low:F2} to ${high:F2}");

// Use _ (a discard) for a value you do not need
var (sku, _) = ("SKU-1042", 19.99m);
Console.WriteLine(sku);
````

`var (low, high) = ...` assigns the two returned components to two local variables. The first line prints `Range: $980.50 to $1610.25`. In `(sku, _)`, `_` discards the price, so the second line prints only `SKU-1042`. Deconstruction does not call the method twice or change the source array. Use the shipment-summary practice below to complete the same mapping for a new pair of results.

### Tuples in Collections



A list can store tuple values as its elements. General syntax patterns for a named two-component tuple list and a deconstructing loop:

```text
var listName = new List<(Type1 Name1, Type2 Name2)>
{
    (value1, value2),
    // Additional tuple elements go here.
};

foreach (var (firstVariable, secondVariable) in listName)
{
    // Use the components of the current element.
}
```

`List<...>` takes the whole tuple type as its element type. Each pair in the initializer is one list element. The loop unpacks each element into two variables in positional order. Replace the placeholder types, names, and values with the data your program needs.

For example, a list of tuples keeps sales paired with their representatives when the pairs do not need behavior of their own:


````csharp
using System;
using System.Collections.Generic;

var salesByRep = new List<(string Rep, decimal Sales)>
{
    ("Avery", 18250.00m),
    ("Jordan", 22410.50m),
    ("Morgan", 19875.25m),
};

var top = salesByRep[0];
foreach (var (rep, sales) in salesByRep)
{
    Console.WriteLine($"{rep}: ${sales:F2}");
    if (sales > top.Sales) top = (rep, sales);
}
Console.WriteLine($"Top rep: {top.Rep} (${top.Sales:F2})");
````

Each `foreach` iteration deconstructs one list element into `rep` and `sales`. The running `top` remains an entire pair, preserving the representative associated with its amount. The final line is `Top rep: Jordan ($22410.50)`. Predict the winner after changing Avery's amount to `25000m`; the winner should change without altering the loop. Ties retain the first encountered pair because the comparison is strict `>`.

### Tuples or Classes?



The choice depends on the role of the data. Use this comparison after working through the class and tuple examples:

| Use a tuple when | Use a class when |
|---|---|
| A method returns two or three related values | The data has behavior (methods) or rules to protect |
| The values are used briefly, inside one method or class | The type is shared across the program |
| Named fields make the meaning clear enough | It needs initialization rules, controlled access to state, or custom behavior |

If you find yourself passing the same tuple shape through many methods, that is a sign it should become a class. Chapter 20 uses tuples for graph edges, where a pair of locations is all an edge needs.

### Guided Variation: Deconstruct a Shipment Summary


A temporary summary needs only the number of parcels and their total mass. A tuple is enough when it is used briefly and has no behavior or validation to enforce. Field names describe meaning; deconstruction still follows position.
