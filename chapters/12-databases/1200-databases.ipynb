# Databases

Persistent information systems need data that outlives one program run. This chapter connects C# objects and collections to relational data, SQL queries, and safer database access patterns.

- **Relational data**: Tables, rows, columns, and keys organize structured data for long-term storage.
- **SQL**: Structured Query Language describes how to filter, sort, join, insert, update, and delete records.
- **Data model**: A data model captures entities, fields, relationships, and constraints.
- **Parameterized commands**: Parameters keep user data separate from SQL text and reduce injection risk.
- **Asynchronous I/O**: Async workflows help programs stay responsive while waiting for files or databases.

Databases extend the semester project from in-memory objects to durable information that can be queried, updated, and shared.

<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Database | An organized collection of data managed for storage, retrieval, and change. |
| 2 | Relational model | A data model that organizes information into related tables. |
| 3 | Table | A named collection of rows and columns. |
| 4 | Row | One record in a table. |
| 5 | Column | One named field shared by rows in a table. |
| 6 | Primary key | A column or set of columns that uniquely identifies each row. |
| 7 | Foreign key | A column that refers to a row in another table. |
| 8 | Relationship | A connection between tables represented through keys. |
| 9 | SQL | Structured Query Language, used to query and modify relational data. |
| 10 | SELECT | An SQL command that retrieves rows and columns. |
| 11 | WHERE | An SQL clause that filters rows. |
| 12 | JOIN | An SQL operation that combines related rows from multiple tables. |
| 13 | Aggregation | Summarizing rows with operations such as count, sum, or average. |
| 14 | INSERT | An SQL command that adds rows. |
| 15 | UPDATE | An SQL command that changes existing rows. |
| 16 | DELETE | An SQL command that removes rows. |
| 17 | Parameterized command | A database command that supplies data separately from SQL text to avoid injection risks. |
| 18 | Connection | A C# object or workflow step that opens access to a database. |
