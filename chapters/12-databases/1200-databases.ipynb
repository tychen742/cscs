# Databases

Persistent information systems need data that outlives one program run. This chapter connects C# objects and collections to relational data, SQL queries, and safer database access patterns.

- **Relational data**: Tables, rows, columns, and keys organize structured data for long-term storage.
- **SQL**: Structured Query Language describes how to filter, sort, join, insert, update, and delete records.
- **Data model**: A data model captures entities, fields, relationships, and constraints.
- **Parameterized commands**: Parameters keep user data separate from SQL text and reduce injection risk.
- **Asynchronous I/O**: Async workflows help programs stay responsive while waiting for files or databases.

Databases extend the semester project from in-memory objects to durable information that can be queried, updated, and shared.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Identify tables, rows, columns, primary keys, and relationships in a relational model
2. Write SQL queries that filter, sort, aggregate, and join records
3. Explain why parameterized commands are safer than string-built SQL
4. Map C# objects and records to persistent data concepts
5. Connect a small C# workflow to stored data and handle query results
