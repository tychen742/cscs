## Multiple Choice

**1.** What does a priority queue remove next?

a) The oldest item  
b) The item chosen by priority  
c) The largest key alphabetically  
d) A random item

**2.** In C#'s default `PriorityQueue<TElement, TPriority>`, which priority comes out first?

a) Lowest priority value  
b) Highest priority value  
c) Most recently inserted  
d) Longest string

**3.** What heap rule keeps a min-heap's smallest item at the root?

a) Every parent is less than or equal to its children  
b) Every child is less than its parent  
c) Every level is sorted left to right  
d) Every node has three children

**4.** What does bubble up repair?

a) A new heap item that may be smaller than its parent  
b) A dictionary collision  
c) A missing hash code  
d) A deleted dictionary key

**5.** What is a hash function used for?

a) Turning a key into a table position or hash code  
b) Sorting a list completely  
c) Removing heap roots  
d) Traversing a tree inorder

**6.** What is a collision?

a) A key that is null  
b) Two different keys mapping to the same bucket  
c) A heap with no root  
d) A queue with too many priorities

**7.** What does `Dictionary<TKey, TValue>` store?

a) Key-value pairs  
b) Only duplicate values  
c) Only sorted values  
d) Parent-child nodes

**8.** What does `HashSet<T>` help test efficiently on average?

a) Membership and uniqueness  
b) Parent pointers  
c) Tree height  
d) Inorder traversal

**9.** What does load factor describe?

a) How full a hash table is  
b) How tall a heap is  
c) How many methods a class has  
d) How many files are open

**10.** Which structure is best for repeatedly processing the next most urgent task?

a) Priority queue  
b) Plain array with arbitrary indexes  
c) String builder  
d) Record type
