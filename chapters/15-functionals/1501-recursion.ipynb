# Recursion

Recursion is a technique where a method calls itself with a simpler input until a **base case** is reached.
Every recursive method needs:

1. A **base case** — stops the recursion and returns a value directly.
2. A **recursive step** — calls itself with a smaller/simpler argument.

Without a base case the program crashes with `StackOverflowException`.


## Factorial

`n! = n × (n-1) × ⋯ × 1`, with `0! = 1`.
Recursive: `factorial(n) = n * factorial(n-1)`, base case `n == 0`.


````
static long Factorial(int n)
{
    if (n == 0) return 1;          // base case
    return n * Factorial(n - 1);  // recursive step
}

Console.WriteLine(Factorial(5));   // 120
Console.WriteLine(Factorial(0));   // 1
Console.WriteLine(Factorial(10));  // 3628800
````

**Call stack for `Factorial(4)`:**

```text
Factorial(4) -> 4 * Factorial(3)
                   -> 3 * Factorial(2)
                              -> 2 * Factorial(1)
                                         -> 1 * Factorial(0) -> 1
```

Unwinds: 1 -> 1 -> 2 -> 6 -> 24.


## Fibonacci

Sequence: `0, 1, 1, 2, 3, 5, 8, 13, ...`

- `fib(0) = 0`, `fib(1) = 1` — base cases
- `fib(n) = fib(n-1) + fib(n-2)` for n > 1


````
static int Fib(int n)
{
    if (n == 0) return 0;
    if (n == 1) return 1;
    return Fib(n - 1) + Fib(n - 2);
}

for (int i = 0; i <= 8; i++)
    Console.Write($"{Fib(i)} ");  // 0 1 1 2 3 5 8 13 21
````

## Recursive GCD

Euclid: `gcd(a, 0) = a`, `gcd(a, b) = gcd(b, a % b)`.


````
static int Gcd(int a, int b)
{
    if (b == 0) return a;
    return Gcd(b, a % b);
}

Console.WriteLine(Gcd(48, 18));   // 6
Console.WriteLine(Gcd(100, 75));  // 25
````

## Recursion vs. Loops

| | Recursion | Loop |
|---|---|---|
| Natural fit | Trees, divide-and-conquer | Counting, accumulation |
| Risk | Stack overflow on deep input | Off-by-one errors |
| Performance | More call overhead | Usually faster |


## Common Mistakes

**Missing base case** → `StackOverflowException`:


````
// BAD: no base case
static int Bad(int n) => n * Bad(n - 1); // never stops!
````

**Moving away from the base case:**


````
// BAD: n+1 moves away from 0
static int BadCount(int n)
{
    if (n == 0) return 0;
    return BadCount(n + 1); // wrong direction!
}
````

## Practice

1. `SumDigits(int n)` — sum of digits. `SumDigits(123)` → `6`.
   Hint: `n % 10` is the last digit; `n / 10` removes it.
2. `ReverseString(string s)` — return reversed string. Base case: length ≤ 1.
3. `Power(double b, int exp)` — compute `b^exp` recursively. `Power(2, 10)` → `1024`.
