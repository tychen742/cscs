# Recursive and Functional Patterns

This chapter introduces three ideas that improve abstraction and expressiveness:

- recursion for self-similar problems
- lambda expressions for concise behavior definitions
- LINQ for querying and transforming collections

## Why these topics are grouped

They all push students beyond step-by-step imperative style toward declarative thinking:

- what the transformation is, not just how to loop it
- what the base and recursive cases are, not just repeated mutation

## Learning goals

By the end of this chapter, you should be able to:

- identify valid recursion structure (base case + recursive step)
- write and read lambda expressions with confidence
- use LINQ for common filtering, projection, and aggregation tasks
- choose between loops and LINQ based on clarity
