## 4. Share Store-Wide Data with Static Members

The shop now rings up a whole receipt at once with `Checkout(decimal[] prices)`, which replaces `RingUp`. Decide where each piece of data belongs:

- **Static field** `nextReceiptNumber`: receipt numbers must be unique across the *whole store*, so all registers share one counter. One copy for the class, not one per register.
- **Static property** `TaxRate`: there is one sales-tax rate for the store. Changing it once (`CashRegister.TaxRate = 0.07m;`) changes it for every register.
- **Static method** `Money`: it formats a number and remembers nothing, so it stays a stateless static helper. Turning it into an instance method would gain nothing.
- **Instance fields** `total` and `receiptCount`: each register's own running results.
- **Local variables** `subtotal`, `tax`, `amountDue`, and `receipt`: they start over for every checkout and mean nothing afterward, so they stay local, just like `lineAmount` did.

```text
Receipt 1001 (Front, Avery): $21.00 + tax $1.68 = $22.68
```
