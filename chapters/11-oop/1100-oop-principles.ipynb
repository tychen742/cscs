# OOP Principles

Classes alone do not make a program object-oriented. This chapter names and illustrates the core object-oriented principles: encapsulation, inheritance, polymorphism, and abstraction, with concrete C# examples.

- **Encapsulation**: Objects protect their internal state and expose a controlled public surface.
- **Inheritance**: A derived class can reuse and specialize behavior from a base class.
- **Polymorphism**: Code can work with a general type while runtime objects provide specific behavior.
- **Abstraction**: Interfaces and abstract classes describe capabilities without committing to one implementation.
- **Composition**: Modern C# often favors combining small objects over building deep inheritance trees.

The goal is conceptual clarity and practical design judgment, especially knowing when interfaces and composition are cleaner than deep hierarchies.


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Object-oriented programming | A style of programming organized around objects that combine state and behavior. |
| 2 | Encapsulation | Protecting object state and exposing safe operations. |
| 3 | Inheritance | Creating a derived class from a base class. |
| 4 | Base class | A class whose members are inherited by another class. |
| 5 | Derived class | A class that extends or specializes another class. |
| 6 | Polymorphism | Using one common interface or base type to work with different concrete types. |
| 7 | Override | A derived-class method that replaces a virtual base-class behavior. |
| 8 | Virtual method | A method designed to be overridden by derived classes. |
| 9 | Interface | A contract that specifies members a type must provide. |
| 10 | Abstraction | Focusing on essential operations while hiding implementation details. |
| 11 | Composition | Building a type by storing and using other objects. |
| 12 | Substitution | Using a derived or implementing type where a base or interface type is expected. |
