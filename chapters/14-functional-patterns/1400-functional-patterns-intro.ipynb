# Functional Patterns

This chapter covers functional programming patterns and query tools available in modern C#: recursion, lambda expressions, and LINQ.

## Learning Objectives

- Write recursive methods and understand base/recursive cases
- Use lambda expressions and delegates
- Query collections with LINQ syntax
