# Functional Patterns

Modern C# includes functional programming patterns that make some problems easier to express. This chapter introduces recursion, lambda expressions, and LINQ as tools for decomposing work and querying data.

- **Recursion**: A recursive method solves a problem by solving smaller versions of the same problem.
- **Lambda expression**: A lambda expression writes a small function inline where behavior is needed.
- **Delegate**: A delegate type represents a callable method or function value.
- **LINQ**: Language Integrated Query provides readable operations for filtering, transforming, and grouping collections.

These patterns let you pass behavior around, express data transformations directly, and recognize recursive structure when loops are not the clearest fit.
