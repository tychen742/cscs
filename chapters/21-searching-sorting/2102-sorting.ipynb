# Sorting

Sorting arranges values according to an ordering rule. Sorted data are easier to scan, compare, display, and search. This section focuses on elementary sorting algorithms because they are simple enough to trace by hand.


```{index} swap operation
```

## Exchanging Array Elements

Most in-place sorting algorithms need a swap operation. A swap uses a temporary variable so that neither value is lost.


````csharp
using System;

int[] values = { 10, 20, 30 };
Swap(values, 0, 2);
Console.WriteLine(string.Join(", ", values));

void Swap(int[] values, int first, int second)
{
    int temp = values[first];
    values[first] = values[second];
    values[second] = temp;
}
````

Swapping is constant-time work, but a sorting algorithm may perform many swaps.

## Selection Sort

Selection sort repeatedly finds the smallest value in the unsorted portion and swaps it into the next position.


````csharp
using System;

int[] values = { 5, 2, 9, 1, 3 };
SelectionSort(values);
Console.WriteLine(string.Join(", ", values));

void SelectionSort(int[] values)
{
    for (int start = 0; start < values.Length - 1; start++)
    {
        int smallest = start;
        for (int i = start + 1; i < values.Length; i++)
        {
            if (values[i] < values[smallest])
            {
                smallest = i;
            }
        }

        Swap(values, start, smallest);
    }
}

void Swap(int[] values, int first, int second)
{
    int temp = values[first];
    values[first] = values[second];
    values[second] = temp;
}
````

Selection sort performs relatively few swaps, but it still compares many pairs of values. Its time growth is `O(n^2)`.

## Insertion Sort

Insertion sort grows a sorted prefix. Each new value is inserted into its correct position among the values already processed.


````csharp
using System;

int[] values = { 5, 2, 9, 1, 3 };
InsertionSort(values);
Console.WriteLine(string.Join(", ", values));

void InsertionSort(int[] values)
{
    for (int i = 1; i < values.Length; i++)
    {
        int current = values[i];
        int j = i - 1;

        while (j >= 0 && values[j] > current)
        {
            values[j + 1] = values[j];
            j--;
        }

        values[j + 1] = current;
    }
}
````

Insertion sort is often efficient for small or nearly sorted arrays, even though its worst-case growth is `O(n^2)`.

## Bubble Sort

Bubble sort repeatedly compares adjacent values and swaps them when they are out of order. Large values move toward the end of the array.


````csharp
using System;

int[] values = { 5, 2, 9, 1, 3 };
BubbleSort(values);
Console.WriteLine(string.Join(", ", values));

void BubbleSort(int[] values)
{
    bool swapped;
    do
    {
        swapped = false;
        for (int i = 0; i < values.Length - 1; i++)
        {
            if (values[i] > values[i + 1])
            {
                Swap(values, i, i + 1);
                swapped = true;
            }
        }
    } while (swapped);
}

void Swap(int[] values, int first, int second)
{
    int temp = values[first];
    values[first] = values[second];
    values[second] = temp;
}
````

Bubble sort is useful for learning because it is easy to trace. It is rarely the best practical sorting choice.

## Counting Comparisons

Counting comparisons helps compare algorithms without depending on machine speed.


````csharp
using System;

int[] values = { 5, 2, 9, 1, 3 };
int comparisons = SelectionSortCount(values);

Console.WriteLine(string.Join(", ", values));
Console.WriteLine($"comparisons = {comparisons}");

int SelectionSortCount(int[] values)
{
    int comparisons = 0;
    for (int start = 0; start < values.Length - 1; start++)
    {
        int smallest = start;
        for (int i = start + 1; i < values.Length; i++)
        {
            comparisons++;
            if (values[i] < values[smallest])
            {
                smallest = i;
            }
        }
        Swap(values, start, smallest);
    }
    return comparisons;
}

void Swap(int[] values, int first, int second)
{
    int temp = values[first];
    values[first] = values[second];
    values[second] = temp;
}
````

For `n` items, selection sort performs about `n * (n - 1) / 2` comparisons.

## Choosing a Sort

Use elementary sorts to learn algorithm mechanics and for small educational examples. In real C# programs, prefer library sorting such as `Array.Sort` or `List<T>.Sort` unless the assignment is specifically about implementing an algorithm.


````csharp
using System;

int[] values = { 5, 2, 9, 1, 3 };
Array.Sort(values);
Console.WriteLine(string.Join(", ", values));
````

Library sorts are tested, optimized, and easier to read. Knowing elementary sorts still helps you understand cost, invariants, and why better algorithms matter.
