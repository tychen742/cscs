# Searching and Sorting

Searching and sorting are classic algorithm families because they appear in nearly every computing domain. This chapter compares ways to locate and order data using correctness and complexity tools from earlier chapters.

- **Linear search**: Linear search checks items one by one and works without sorted data.
- **Binary search**: Binary search repeatedly halves the search range, but requires sorted data.
- **Exchange**: Sorting algorithms often rearrange data through comparisons and swaps.
- **Elementary sorting**: Insertion, selection, and bubble sort reveal common sorting ideas and costs.
- **Algorithm comparison**: Correctness tests and operation counts help compare implementations.

These algorithms give concrete practice applying analysis to code students can trace and test directly.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Select a search algorithm based on data conditions
2. Trace linear and binary search step by step
3. Trace elementary sorting algorithms on small arrays
4. Explain the growth rates of common searching and sorting algorithms
5. Compare implementations using tests and operation counts
