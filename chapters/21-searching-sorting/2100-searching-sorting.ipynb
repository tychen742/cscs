# Search & Sort

This chapter covers linear search, binary search, search preconditions, swap operations, elementary sorting algorithms, comparison counts, correctness tests, and when to use library sorting. The sidebar title **Search & Sort** keeps the two algorithm families visible without crowding the navigation.

- **Linear search**: Checks items one by one until the target is found or the input ends.
- **Binary search**: Uses sorted order to discard half of the remaining search range at each step.
- **Elementary sort**: A simple sorting algorithm, such as selection sort, insertion sort, or bubble sort, that is easy to trace but usually quadratic.
- **Comparison count**: A useful operation count for algorithms that decide order by comparing values.
- **Data condition**: Some algorithms require data to be sorted or structured before they are valid.

Searching locates data. Sorting puts data into an order that makes later work easier, including faster searching.


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Search | The task of finding whether and where a value appears. |
| 2 | Linear search | A search algorithm that checks items sequentially. |
| 3 | Binary search | A search algorithm for sorted data that repeatedly halves the remaining search range. |
| 4 | Sorted order | An arrangement where values follow a chosen ordering rule. |
| 5 | Comparison | A test that decides how two values relate in an order. |
| 6 | Swap | An operation that exchanges values at two positions. |
| 7 | Selection sort | A sorting algorithm that repeatedly selects the smallest remaining value and moves it into place. |
| 8 | Insertion sort | A sorting algorithm that grows a sorted prefix by inserting each new item into its correct position. |
| 9 | Bubble sort | A sorting algorithm that repeatedly swaps adjacent out-of-order values. |
| 10 | Stable sort | A sort that preserves the relative order of equal items. |
| 11 | In-place sort | A sort that uses only small extra storage beyond the input array. |
