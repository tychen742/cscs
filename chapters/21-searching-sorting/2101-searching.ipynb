Searching means locating a value or deciding that the value is not present. The right search strategy depends on what you know about the data before the search begins.

## Linear Search

Linear search checks values one at a time. It works on unsorted data, sorted data, arrays, lists, and many other sequential structures.


````csharp
using System;

int[] values = { 42, 7, 19, 3, 88 };
int target = 19;
int index = LinearSearch(values, target);

Console.WriteLine(index);

int LinearSearch(int[] values, int target)
{
    for (int i = 0; i < values.Length; i++)
    {
        if (values[i] == target)
        {
            return i;
        }
    }

    return -1;
}
````

The return value `-1` is a common sentinel for "not found" because `-1` is never a valid array index.

## Search Contracts

A search method should make its contract clear: what input it accepts, what it returns when the target is present, and what it returns when the target is missing.


````csharp
using System;

string[] names = { "Ada", "Grace", "Katherine", "Radia" };
string target = "Evelyn";
int foundAt = LinearSearch(names, target);

Console.WriteLine(foundAt >= 0
    ? $"Found {target} at index {foundAt}."
    : $"{target} is not in the array.");

int LinearSearch(string[] values, string target)
{
    for (int i = 0; i < values.Length; i++)
    {
        if (values[i] == target)
        {
            return i;
        }
    }

    return -1;
}
````

This method returns the first matching index. If the array contains duplicates, it does not return every match.

## Counting Comparisons

The worst case for linear search happens when the target is missing or at the final position. Then every item must be checked.


````csharp
using System;

int[] values = { 4, 8, 15, 16, 23, 42 };
int target = 99;
int comparisons = 0;
int index = LinearSearch(values, target, ref comparisons);

Console.WriteLine($"index = {index}");
Console.WriteLine($"comparisons = {comparisons}");

int LinearSearch(int[] values, int target, ref int comparisons)
{
    for (int i = 0; i < values.Length; i++)
    {
        comparisons++;
        if (values[i] == target)
        {
            return i;
        }
    }

    return -1;
}
````

Linear search is `O(n)` because the number of checks grows in proportion to the number of items.

## Binary Search

Binary search works only when the data are sorted according to the same ordering used by the comparisons. It checks the middle item, then discards the half that cannot contain the target.


````csharp
using System;

int[] sorted = { 3, 7, 12, 18, 25, 31, 44, 58 };
int target = 25;
int index = BinarySearch(sorted, target);

Console.WriteLine(index);

int BinarySearch(int[] values, int target)
{
    int low = 0;
    int high = values.Length - 1;

    while (low <= high)
    {
        int middle = low + (high - low) / 2;

        if (values[middle] == target)
        {
            return middle;
        }

        if (target < values[middle])
        {
            high = middle - 1;
        }
        else
        {
            low = middle + 1;
        }
    }

    return -1;
}
````

The expression `low + (high - low) / 2` is a safe middle-index pattern that avoids overflow in languages and settings where very large indexes are possible.

## Tracing Binary Search

Trace the low, high, and middle indexes to see how quickly binary search reduces the search range.


````csharp
using System;

int[] sorted = { 3, 7, 12, 18, 25, 31, 44, 58 };
int target = 44;
int low = 0;
int high = sorted.Length - 1;

while (low <= high)
{
    int middle = low + (high - low) / 2;
    Console.WriteLine($"low={low}, high={high}, middle={middle}, value={sorted[middle]}");

    if (sorted[middle] == target)
    {
        Console.WriteLine($"found at {middle}");
        break;
    }

    if (target < sorted[middle])
    {
        high = middle - 1;
    }
    else
    {
        low = middle + 1;
    }
}
````

Binary search is `O(log n)` because each comparison removes about half of the remaining candidates.

## Choosing a Search

Use linear search when the data are unsorted, when the dataset is small, or when you need to inspect every match. Use binary search when the data are already sorted or when paying the cost to sort first is worthwhile for many searches.


````csharp
using System;

int n = 1_000_000;
double linearWorstCase = n;
double binaryWorstCase = Math.Ceiling(Math.Log2(n));

Console.WriteLine($"linear worst-case checks: {linearWorstCase:N0}");
Console.WriteLine($"binary worst-case checks: {binaryWorstCase:N0}");
````

Sorting just to do one search may not be worth it. Sorting once to support many searches can be a strong design choice.
