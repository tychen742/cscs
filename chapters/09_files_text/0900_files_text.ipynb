# Files and Text

An order-import program must find its input, keep file resources under control, and decide whether each text record is usable. This chapter connects reading and writing to paths, resource lifetime, and field validation. Core sections develop file and text workflows; regular expressions and privacy/security cases remain labeled supporting reading.

- **File I/O**: File input and output let a program read stored data and save results.
- **Streams**: Streams provide a sequence-oriented way to read from or write to a source.
- **Text processing**: Trimming, casing, splitting, and replacing help clean and transform text.
- **Structured records**: Delimited text such as CSV stores fields that must be parsed and validated.
- **Regular expressions**: Regex patterns describe text shapes for search, validation, and extraction.

<h2>Learning Objectives</h2>


By the end of this chapter, you will be able to:

1. Read and write text with `File`, `StreamReader`, and `StreamWriter` under an explicit overwrite or append policy.
2. Construct and inspect paths and handle file resources and expected I/O failures intentionally.
3. Clean and split text, validate fields, and convert accepted records into typed values.
4. Check an import against malformed, missing, empty, and boundary cases without silently accepting invalid records.

Supporting-reading outcomes: for the supporting regex reading, select simple patterns and groups for targeted matching and extraction; for the privacy/security reading, explain how to reduce exposed data.
