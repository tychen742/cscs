# File-Driven Responses

This integrated collections-and-files lab follows both Data Collections and Files. It extends the current dictionary helpdesk lab by separating response data from program logic. The legacy names FakeHelp and FakeHelpVerbose describe a deliberately simplistic keyword responder, not a real support service.

## Compare Data in Code and Data in Files

```{index} Compare Data in Code and Data in Files
```

Start with a working responder that builds a list of fallback paragraphs and a dictionary of keyword/response pairs using Add calls. If a word matches a dictionary key, return its response. Otherwise choose a fallback randomly. Compare that version with one that reads the same paragraphs from text files. The visible responses can be the same while the second design permits editing data without rebuilding source.

Download the {download}`starter project <../../../demos/09/help_responses_lab.zip>`, extract it, and open its folder in VS Code. Run `dotnet run` from that folder. Study the complete `fake_help_verbose.cs` first; then set the startup object in the project to `IntroCS.FakeHelp` and implement the three stubs in `file_util.cs`. Keep the supplied namespace and one selected entry point.

The defaults file starts with a welcome paragraph, then a goodbye paragraph, then fallback paragraphs. A blank line separates paragraphs. The responses file contains a keyword line followed by its response paragraph. Inspect both before coding. Use the supplied deliberately synthetic data; do not add real support tickets or personal information.

## ReadParagraph

```{index} ReadParagraph
```

Complete `string ReadParagraph(StreamReader reader)`. Read nonempty lines and append each one plus a newline to the paragraph. Consume the blank separator, but do not consume the first line of the next paragraph. `ReadLine` discards the newline; add it back explicitly. Reading has a side effect: the stream advances, so a second call with the same syntax reads different data.

The first call should return the welcome paragraph, and the second should return the goodbye paragraph. Test this before completing the collections. Also handle a final paragraph that ends at end-of-file without a trailing blank separator. Treat whitespace-only separator lines consistently.

## GetParagraphs

```{index} GetParagraphs
```

Complete `List<string> GetParagraphs(StreamReader reader)` to read all remaining paragraphs. Reuse ReadParagraph rather than rewriting its inner loop. Add each nonempty paragraph to the list and return it. Do not add empty fallback entries when there are consecutive separators. Test zero, one, and multiple paragraphs, and a final paragraph with no blank line after it.

After the welcome and goodbye paragraphs have been read, this method supplies the fallback list. Reject an empty list before selecting randomly: `rand.Next(guessList.Count)` needs a positive count. Reuse one Random object. A fixed seed helps reproduce a test within the same runtime.

## GetDictionary

```{index} GetDictionary
```

Complete `Dictionary<string, string> GetDictionary(StreamReader reader)`. Read a nonblank keyword line, then call ReadParagraph to read its response. Store each key/value pair and continue through the file. Use the same case policy in both loading and lookup. Detect duplicate keys or document a replacement policy; do not silently lose data accidentally.

Test known keywords such as `crash`, input with no recognized keyword, mixed-case input, a last response ending at end-of-file, and duplicate keys. Explain why GetDictionary does not consume paragraphs in exactly the same pattern as GetParagraphs: each dictionary response has a key line before it.

## Integrate and Verify

```{index} Integrate and Verify
```

Run the completed file-driven program. Show the welcome, a known keyword response, a random fallback, and the goodbye. Change one response in the data file and demonstrate that the change appears without editing C# source. Then compare the two designs: the collection operations are the same, but source code and response data now have separate responsibilities.

Submit the completed project, both data files, labeled test transcripts, and a short account of stream position after each helper. Keep the current Data Collections lab as practice in in-memory list/dictionary behavior; this lab adds persistence and loading.

```{rubric} Footnotes
```
