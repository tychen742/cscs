# Graph Vocabulary and Models

A **graph** represents a set of objects and the relationships among them. The objects are **vertices** (or nodes), and the relationships are **edges**.

A graph is a good model when the relationships matter as much as the individual items. A list of cities is just data; a graph of cities and roads can answer questions about direct routes, connections, and later, shortest paths.


```{index} graph
```

## Vertices and Edges

Start by separating the items from the relationships. Vertices name the items. Edges say which pairs of items are connected.


````csharp
using System;

string[] vertices = { "Rolla", "St. Louis", "Chicago", "Kansas City" };
(string From, string To)[] roads = {
    ("Rolla", "St. Louis"),
    ("St. Louis", "Chicago"),
    ("Rolla", "Kansas City")
};

Console.WriteLine("Vertices:");
foreach (string city in vertices)
{
    Console.WriteLine($"- {city}");
}

Console.WriteLine("\nEdges:");
foreach ((string from, string to) in roads)
{
    Console.WriteLine($"{from} <-> {to}");
}
````

The example uses an undirected road model, so `Rolla <-> St. Louis` means the connection works both ways. The tuple stores one edge, but the interpretation comes from the model we choose.


## Directed and Undirected Edges

An **undirected edge** is two-way. A **directed edge** has a direction from one vertex to another. Following someone on a social platform, importing one file from another file, and completing a course prerequisite are usually directed relationships.


````csharp
using System;

(string From, string To)[] prerequisites = {
    ("Programming I", "Programming II"),
    ("Programming II", "Data Structures"),
    ("Discrete Math", "Data Structures")
};

foreach ((string before, string after) in prerequisites)
{
    Console.WriteLine($"Take {before} before {after}.");
}
````

Direction matters here. If `Programming I` is a prerequisite for `Programming II`, that does not mean `Programming II` is a prerequisite for `Programming I`.

## Weighted Edges

A **weighted graph** stores a value with each edge. The weight might be distance, cost, time, capacity, risk, or similarity. The graph structure stores connections; the weight gives each connection a measurable meaning.


````csharp
using System;

(string From, string To, int Miles)[] routes = {
    ("Rolla", "St. Louis", 106),
    ("Rolla", "Kansas City", 193),
    ("St. Louis", "Chicago", 297)
};

foreach ((string from, string to, int miles) in routes)
{
    Console.WriteLine($"{from} to {to}: {miles} miles");
}
````

Later algorithms can use the same graph with different goals. One algorithm might ask whether a route exists at all. Another might ask for the route with the smallest total weight.

## Neighbors, Degree, and Paths

Two vertices are **neighbors** when an edge directly connects them. In an undirected graph, the **degree** of a vertex is the number of incident edges. A **path** is a sequence of vertices where each consecutive pair is connected by an edge.


````csharp
using System;
using System.Collections.Generic;

var neighbors = new Dictionary<string, List<string>>
{
    ["Rolla"] = new() { "St. Louis", "Kansas City" },
    ["St. Louis"] = new() { "Rolla", "Chicago" },
    ["Chicago"] = new() { "St. Louis" },
    ["Kansas City"] = new() { "Rolla" }
};

string city = "Rolla";
Console.WriteLine($"{city} has degree {neighbors[city].Count}.");
Console.WriteLine($"Neighbors: {string.Join(", ", neighbors[city])}");

string[] path = { "Kansas City", "Rolla", "St. Louis", "Chicago" };
Console.WriteLine($"Path: {string.Join(" -> ", path)}");
````

A path is not automatically the best path. It is simply a valid sequence of edges. Later chapters will use graph representations to search for paths and optimize them.

## Model Boundaries

Introductory graph examples usually use **simple graphs**: no duplicate edges between the same vertices and no edge from a vertex to itself. Real systems sometimes need self-loops or multiple edges, but a simple graph is enough for the main representation ideas in this chapter.


````csharp
using System;
using System.Collections.Generic;

var seen = new HashSet<(string, string)>();
(string From, string To) edge = ("A", "A");

if (edge.From == edge.To)
{
    Console.WriteLine("This is a self-loop, so this simple-graph model rejects it.");
}
else if (!seen.Add(edge))
{
    Console.WriteLine("This duplicate edge is already stored.");
}
else
{
    Console.WriteLine("The edge fits the simple-graph model.");
}
````

Choosing model rules early prevents confusing code later. Before implementing a graph, decide whether edges are directed, whether weights matter, and whether duplicates or self-loops are allowed.
