# Graph Vocabulary and Models

A **graph** represents relationships among objects. A graph has **vertices** (also called nodes) and **edges** connecting pairs of vertices.

Important distinctions:

- An undirected edge represents a two-way relationship.
- A directed edge has a source and a destination.
- A weighted edge carries a value such as distance, cost, or time.
- A path is a sequence of connected vertices.
- A cycle is a path that returns to its starting vertex.
- A connected component is a group of vertices linked by paths.

Business examples include transportation routes, organizational relationships,
recommendation networks, and dependencies among tasks.

```csharp
record Edge(string From, string To, int Weight);

var routes = new[]
{
    new Edge("Rolla", "St. Louis", 100),
    new Edge("Rolla", "Kansas City", 300),
    new Edge("St. Louis", "Chicago", 300)
};

foreach (Edge route in routes)
    Console.WriteLine($"{route.From} -> {route.To}: {route.Weight} miles");
```
