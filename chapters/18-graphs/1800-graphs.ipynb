# Graph Data Structures

Graphs model relationships that are not naturally linear or hierarchical. This chapter introduces vertices, edges, graph vocabulary, and the main representations used to store graph data.

- **Vertex**: A vertex represents an entity in a graph.
- **Edge**: An edge represents a relationship or connection between vertices.
- **Adjacency matrix**: A matrix representation records connections in a grid indexed by vertices.
- **Adjacency list**: A list representation stores each vertex with its neighboring vertices.
- **Graph model**: Directed, undirected, weighted, and mixed graphs capture different relationship types.

Graph representations determine what later traversal and optimization algorithms can do efficiently.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Use graph vocabulary to describe vertices, edges, paths, and neighbors
2. Represent a graph with an adjacency matrix
3. Represent a graph with adjacency lists
4. Distinguish directed, undirected, and weighted graphs
5. Choose a graph representation based on storage and operation needs
