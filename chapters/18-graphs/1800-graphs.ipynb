# Graphs

This chapter covers graph vocabulary, directed and undirected relationships, weighted and unweighted edges, adjacency, adjacency matrices, adjacency lists, and small C# graph representations. The sidebar title **Graphs** names the model; the chapter scope is representation rather than traversal algorithms.

- **Vertex**: One item in the graph, such as a city, course, person, page, or device.
- **Edge**: A relationship between two vertices; it may be directed, undirected, weighted, or unweighted.
- **Adjacency**: The set of vertices directly connected to a vertex.
- **Representation**: The concrete structure used to store a graph, commonly an adjacency matrix or adjacency list.
- **Density**: Dense and sparse graphs often call for different storage choices.

Graphs are useful whenever the question is not simply what comes next, but what is connected to what.
