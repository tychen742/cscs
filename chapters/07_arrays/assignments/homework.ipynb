## Find the Last Matching Shipment



**Independent practice.** Shipment quantities arrive as `[6, 2, 6, 9, 6]`. Write `LastIndexOf(int[] values, int target)` without a library search. Return the last matching index or -1, preserving the array. Derive expected results before execution for repeated, missing, first-only, empty, and singleton cases. Explain why returning immediately on the first match violates this contract.


`````csharp
// Exercise: Find the Last Matching Shipment
// TODO: write the method, your tests, and requirement-derived expected results.
`````

`````csharp
// Solution: remember a match and keep scanning.
using System;
static int LastIndexOf(int[] values, int target)
{
    int answer = -1;
    for (int i = 0; i < values.Length; i++) if (values[i] == target) answer = i;
    return answer;
}
Console.WriteLine(LastIndexOf(new[] { 6, 2, 6, 9, 6 }, 6));
Console.WriteLine(LastIndexOf(new[] { 6, 2, 9 }, 6));
Console.WriteLine(LastIndexOf(new[] { 6, 2, 9 }, 7));
Console.WriteLine(LastIndexOf(Array.Empty<int>(), 6));
Console.WriteLine(LastIndexOf(new[] { 6 }, 6));
`````

## Verify a Sorted-ID Lookup



**Independent practice.** Write your own binary-search method for ascending invoice IDs. Use `[11, 24, 24, 40, 63]` and independently derived tests for endpoints, a duplicate, a missing internal value, below/above range, empty, and singleton input. Do not call a library search. Use the half-open candidate interval [low, high), with high initially values.Length. A duplicate match may be at either matching index. Record the low/high/middle trace for missing ID 30. Explain why sorting an arrival-order copy changes the meaning of a reported position.


`````csharp
// Exercise: Verify a Sorted-ID Lookup
// TODO: write the method, your tests, and requirement-derived expected results.
`````

`````csharp
// One implementation; add and submit the requested trace.
using System;
static int BinarySearch(int[] values, int target)
{
    int low = 0;
    int high = values.Length;
    while (low < high)
    {
        int middle = low + (high - low) / 2;
        if (values[middle] == target) return middle;
        if (values[middle] < target) low = middle + 1;
        else high = middle;
    }
    return -1;
}
int[] ids = { 11, 24, 24, 40, 63 };
Console.WriteLine(BinarySearch(ids, 11) == 0);
Console.WriteLine(BinarySearch(ids, 63) == 4);
int duplicate = BinarySearch(ids, 24);
Console.WriteLine(duplicate >= 0 && ids[duplicate] == 24);
Console.WriteLine(BinarySearch(ids, 30) == -1);
Console.WriteLine(BinarySearch(ids, 1) == -1 && BinarySearch(ids, 99) == -1);
Console.WriteLine(BinarySearch(Array.Empty<int>(), 24) == -1);
Console.WriteLine(BinarySearch(new[] { 24 }, 24) == 0);
`````

## Prepare a Descending Stock Report



**Independent practice.** Implement insertion sort in descending order for stock quantities. Copy `[3, 8, 3, 1]` before sorting so the arrival-order array is preserved. Do not use a library sort. Verify the independently predicted report `[8, 3, 3, 1]`, original input, duplicate count, empty/singleton input, and both already/reverse ordered arrays. Explain which comparison changes and why the saved-item assignment must remain.


`````csharp
// Exercise: Prepare a Descending Stock Report
// TODO: write the method, your tests, and requirement-derived expected results.
`````

`````csharp
// Solution: shift values smaller than the saved value.
using System;
static void SortDescending(int[] values)
{
    for (int i = 1; i < values.Length; i++)
    {
        int saved = values[i];
        int j = i - 1;
        while (j >= 0 && values[j] < saved)
        {
            values[j + 1] = values[j];
            j--;
        }
        values[j + 1] = saved;
    }
}
int[] original = { 3, 8, 3, 1 };
int[] report = new int[original.Length];
for (int i = 0; i < original.Length; i++) report[i] = original[i];
SortDescending(report);
Console.WriteLine(string.Join(",", report));
Console.WriteLine(string.Join(",", original));
int[] empty = Array.Empty<int>(); SortDescending(empty); Console.WriteLine(empty.Length);
int[] one = { 3 }; SortDescending(one); Console.WriteLine(one[0]);
int[] ascending = { 1, 2, 3 }; SortDescending(ascending); Console.WriteLine(string.Join(",", ascending));
int[] descending = { 3, 2, 1 }; SortDescending(descending); Console.WriteLine(string.Join(",", descending));
`````

## Question 10: Summarize a rectangular stock table



Each row is a branch and each column is a day. Complete Report(int[,] quantities) to print each row total, each column total, and the grand total, using GetLength for both bounds. Do not assume rows equals columns and do not retain a running row total between rows. Predict totals for the supplied two-by-three table before running. Also check zero rows, one row, and one column; a zero-row table still has three columns with zero totals. Inputs are nonnegative quantities and the test totals fit in int.


`````csharp
// Exercise: Summarize a rectangular stock table
// exercise-id: cscs-arrays-homework-core-coding-5
using System;
Console.WriteLine("Two branches:");
Report(new int[,] { { 2, 0, 3 }, { 1, 4, 0 } });
Console.WriteLine("No branches:");
Report(new int[0, 3]);
Console.WriteLine("One branch:");
Report(new int[,] { { 2, 3 } });
Console.WriteLine("One day:");
Report(new int[,] { { 2 }, { 3 } });
// Your code starts here.
static void Report(int[,] quantities) { }
// Your code ends here.
`````

`````csharp
using System;
Console.WriteLine("Two branches:");
Report(new int[,] { { 2, 0, 3 }, { 1, 4, 0 } });
Console.WriteLine("No branches:");
Report(new int[0, 3]);
Console.WriteLine("One branch:");
Report(new int[,] { { 2, 3 } });
Console.WriteLine("One day:");
Report(new int[,] { { 2 }, { 3 } });
static void Report(int[,] quantities)
{
    int rows = quantities.GetLength(0);
    int cols = quantities.GetLength(1);
    int grandTotal = 0;
    for (int row = 0; row < rows; row++)
    {
        int rowTotal = 0;
        for (int col = 0; col < cols; col++) rowTotal += quantities[row, col];
        grandTotal += rowTotal;
        Console.WriteLine($"Branch {row}: {rowTotal}");
    }
    for (int col = 0; col < cols; col++)
    {
        int dayTotal = 0;
        for (int row = 0; row < rows; row++) dayTotal += quantities[row, col];
        Console.WriteLine($"Day {col}: {dayTotal}");
    }
    Console.WriteLine($"Grand: {grandTotal}");
}
`````
