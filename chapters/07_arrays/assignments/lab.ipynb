```csharp
using System;

string[] branches = { "Downtown", "Airport", "Campus" };
string[] days = { "Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun" };
int[,] sales = {
    { 1200,  950, 1100, 1300, 1650, 2100, 1800 },
    {  800,  870,  900,  950, 1020, 1100, 1250 },
    { 1500, 1450, 1400, 1350,  900,  600,  700 }
};

int rows = sales.GetLength(0);
int cols = sales.GetLength(1);

Console.WriteLine("Branch totals:");
int bestBranch = 0;
int bestTotal = 0;
for (int r = 0; r < rows; r++)
{
    int branchTotal = 0;
    for (int c = 0; c < cols; c++)
    {
        branchTotal += sales[r, c];
    }
    Console.WriteLine($"  {branches[r]}: {branchTotal}");
    if (branchTotal > bestTotal)
    {
        bestTotal = branchTotal;
        bestBranch = r;
    }
}

Console.WriteLine("Day totals:");
for (int c = 0; c < cols; c++)
{
    int dayTotal = 0;
    for (int r = 0; r < rows; r++)
    {
        dayTotal += sales[r, c];
    }
    Console.WriteLine($"  {days[c]}: {dayTotal}");
}

Console.WriteLine($"Top branch: {branches[bestBranch]} ({bestTotal})");
```
