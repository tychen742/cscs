## Ascending Values and Runs

```{index} Ascending Values and Runs
```

`PrintAscendingValues(int[] a)` selects a nondecreasing subsequence, keeping the first value and then only values at least as large as the last printed value. For `{5, 2, 8, 4, 8, 11, 6, 7, 10}`, print `5 8 8 11`. This greedy selection does not sort the input or promise the longest possible subsequence.

`PrintRuns(int[] a)` prints **every** value in input order, starting a new line whenever the next value is smaller than the previous value. For `{2, 5, 8, 3, 9, 9, 8}`, print:

```text
2 5 8
3 9 9
8
```

In the first task compare with the last **selected** value; in the second compare with the previous **input** value. Test empty arrays, one value, equal values, strictly increasing values, and strictly decreasing values. Do not access `a[0]` until you know that an element exists.

````csharp
using System;
static void PairwiseAdd(int[] a, int[] b, int[] sum) {
    if (a.Length != b.Length || a.Length != sum.Length) throw new ArgumentException("Lengths must match.");
    for (int i = 0; i < a.Length; i++) sum[i] = a[i] + b[i];
}
static int[] NewPairwiseAdd(int[] a, int[] b) {
    int[] sum = new int[a.Length]; PairwiseAdd(a, b, sum); return sum;
}
static void PrintAscendingValues(int[] a) {
    if (a.Length == 0) return;
    int last = a[0]; Console.Write(last);
    for (int i = 1; i < a.Length; i++) if (a[i] >= last) { last = a[i]; Console.Write(" " + last); }
    Console.WriteLine();
}
static void PrintRuns(int[] a) {
    for (int i = 0; i < a.Length; i++) {
        if (i > 0) Console.Write(a[i] < a[i - 1] ? "\n" : " ");
        Console.Write(a[i]);
    }
    if (a.Length > 0) Console.WriteLine();
}
static double DotProduct(double[] a, double[] b) {
    if (a.Length != b.Length) throw new ArgumentException("Lengths must match.");
    double sum = 0; for (int i = 0; i < a.Length; i++) sum += a[i] * b[i]; return sum;
}
int[] a = {1, 2, 3}, b = {10, 20, 30}, destination = new int[3];
PairwiseAdd(a, b, destination);
Console.WriteLine(string.Join(" ", destination));
Console.WriteLine(string.Join(" ", NewPairwiseAdd(a, b)));
PrintAscendingValues(new int[] {5, 2, 8, 4, 8, 11, 6, 7, 10});
PrintRuns(new int[] {2, 5, 8, 3, 9, 9, 8});
Console.WriteLine(DotProduct(new double[] {1.5, 2, 3}, new double[] {4, 2, -1}));
````

A testable ReadInts helper can accept a TextReader; the console wrapper passes Console.In. This permits a supplied invalid line and retry without depending on hidden browser input.

````csharp
using System;
using System.IO;
static int[] ReadFrom(string prompt, int n, TextReader input) {
    if (n < 0) throw new ArgumentOutOfRangeException(nameof(n));
    int[] result = new int[n];
    for (int i = 0; i < n;) {
        Console.Write($"{prompt} {i + 1}/{n}: ");
        string? line = input.ReadLine();
        if (line == null) throw new EndOfStreamException("Input ended before the array was filled.");
        if (int.TryParse(line, out int value)) result[i++] = value;
        else Console.WriteLine("Enter an integer.");
    }
    return result;
}
static int[] ReadInts(string prompt, int n) { return ReadFrom(prompt, n, Console.In); }
using StringReader supplied = new StringReader("bad\n7\n-2\n3\n");
Console.WriteLine(string.Join(" ", ReadFrom("Value", 3, supplied)));
Console.WriteLine("Empty array length: " + ReadInts("Value", 0).Length);
````

````csharp
using System;
static void PrintInts(string label, int[] a) { Console.Write(label); foreach (int n in a) Console.Write(" " + n); Console.WriteLine(); }
static int Minimum(int[] a) {
    if (a.Length == 0) throw new ArgumentException("At least one value is required.");
    int minimum = a[0]; for (int i = 1; i < a.Length; i++) if (a[i] < minimum) minimum = a[i]; return minimum;
}
static int CountEven(int[] a) { int count = 0; foreach (int n in a) if (n % 2 == 0) count++; return count; }
static bool IsAscending(int[] a) { for (int i = 1; i < a.Length; i++) if (a[i] < a[i - 1]) return false; return true; }
PrintInts("Nums:", new int[] {3, -1, 5});
Console.WriteLine("Minimum: " + Minimum(new int[] {5, 7, 4, 9}));
Console.WriteLine("Even: " + CountEven(new int[] {-4, 7, 6, 12, 9}));
Console.WriteLine(IsAscending(new int[] {2, 5, 5, 8}));
Console.WriteLine(IsAscending(new int[] {2, 5, 3, 8}));
Console.WriteLine("Empty: " + IsAscending(Array.Empty<int>()));
````
