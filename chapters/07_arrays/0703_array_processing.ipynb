## Process Positions Before Introducing a New Container



An array gives many values one shared organization. To calculate a total, the program still needs a rule for visiting those values and combining them. The accumulator pattern connects arrays to Chapter 4: initialize a running result, visit each relevant element, and update the result exactly once for that element. The array changes where input values come from; it does not change the meaning of a loop or an assignment.

Suppose daily quantities are 4, 0 and 7. Before the first iteration the total is zero because no entries have been processed. After each iteration, total should equal the sum of the processed prefix. This claim explains both the initialization and the update. It also provides a way to find a defect before the final result is printed.

| Processed positions | Next position | Total |
| --- | --- | --- |
| None | 0 | 0 |
| 0 | 1 | 4 |
| 0 and 1 | 2 | 4 |
| 0 through 2 | End | 11 |

A for loop exposes positions and is useful when you must report an index or replace elements. A foreach loop exposes each element value directly and is useful when only those values matter. Do not assume that changing a foreach iteration variable updates an array element. Use explicit indexing when the operation is defined in terms of changing positions.

For related explanation, see the further reading below.[^development-reading]

## Passing an Array Does Not Copy Its Elements



A method parameter of type int[] receives a copy of the reference value. Caller and helper can therefore refer to the same array. Replacing an element through the helper is visible through the caller's reference. Reassigning the helper's parameter to a different array changes that local parameter, not the caller's variable. This connects ordinary parameter passing from Chapter 5 with reference aliasing, without requiring pass-by-reference parameter syntax.

State whether a processing helper may mutate its input. A Total helper normally only reads values. An ApplyAdjustment helper might intentionally replace them. If a reporting task must preserve original values, computing into another array or returning a separate result is a different contract from modifying the original and restoring it afterward.

An empty array is a real collection with Length zero. A total over no quantities naturally returns zero under the sum contract. A request for the first element has no valid result on that same input. Handle the operation's domain rather than giving every array helper the same empty-case behavior. A null reference is different again: there is no array object whose Length can be read.

Guided exercise: replace the quantities with 0, 0 and 7. Predict the accumulator trace. Then predict the result for an empty array. Finally, explain whether the read-only total should change if another variable is assigned the array reference. An alias alone changes no element, but a later mutation through either alias can change what a subsequent total reads.


````csharp
using System;
int[] quantities = { 4, 0, 7 };
int total = 0;
for (int i = 0; i < quantities.Length; i++)
{
    total += quantities[i];
    Console.WriteLine($"After position {i}: {total}");
}
Console.WriteLine($"Empty total: {Total(Array.Empty<int>())}");
static int Total(int[] values)
{
    int result = 0;
    foreach (int value in values) result += value;
    return result;
}
````

## Parameters to Main




The `Main` method may take an array of strings as parameter, as in this example:

```csharp
/// Demonstrate the use of command line parameters.
static void Main(string[] args)
{
   Console.WriteLine("There are {0} command line parameters.", args.Length);
   foreach(string s in args) {
      Console.WriteLine(s);
   }

}
```

By convention, the formal parameter for `Main` is called `args`,
short for arguments.

Compile (`dotnet build`) and run (`dotnet run`) the program from the command line.
Run it with some things at the end of the line in `macOS` like:

``` none
[username]@[computer]:~/workspace/introcscs/Ch08Arrays$ dotnet run hi there 123
```

or in `Windows` like:

``` none
PS C:\Users\[username]\workspace\introcscs\Ch08Arrays> dotnet run hi there 123
```

and it should print for you:

``` none
There are 3 command line parameters.
hi
there
123
```

See what quoted strings do. Use command line parameters (with the quotes)
`"hi there" 123`. This should print for you:

    There are 2 command line parameters.
    hi there
    123

### Command Line Adder




A code block in the Main method of your project that calculates and prints
the sum of three command line parameters taken from user input:

    2 5 22

then the program prints 29.

The code could look like this:

    ////////// args array from command line.
    int sum = 0;
    for (int i = 0; i < args.Length; i++)
    {
          sum = sum + int.Parse(args[i]);
    }
    Console.WriteLine(sum);

Here we are using the `arg` array directly just like the preceding section.

To run the code this time, go down to your project’s bin/Debug/net8.0 directory
from the command line, and run `ls`. You will find the executables (the `.dll` or `.exe`
files) that you can execute. In `Windows`, it should look like:

``` none
PS C:\Users\[username]\workspace\introcscs\Ch08Arrays\bin\Debug\net8.0> Ch08Arrays.exe 2 5 22
```

and you will see the output. You can also run the project by issuing:

``` none
PS C:\Users\[username]\workspace\introcscs\Ch08Arrays> dotnet run 2 5 22
```

as before to see the same output. In earlier version of `macOS`, you run this
with `dotnet run` in your project folder at command line with the arguments:

``` none
[username]@[computer]:~/workspace/introcscs/Ch08Arrays$ dotnet run 2 5 22
```

or in more current versions of `macOS`:

``` none
[username]@[computer]:~/workspace/introcscs/Ch08Arrays/bin/Debug/net8.0$ dotnet Ch08Arrays.dll 2 5 22
```

## References and Aliases




Object variables, like arrays, are references,
and this has important implications for
assignment.

With a primitive type like an `int`, an assignment copies the data:

`{image} ../../figures/intCopy.png :align: center :alt: copying an int :width: 90 pt`

In the diagram, the contents of the memory box labeled `b` is copied to the
memory box labeled `d`. The value of `d` starts off equal to the value of `b`,
but can later be changed independently.

Contrast an assignment with arrays. The value that is copied is the *reference*,
not the array data itself, so both end up pointing at the *same* actual array:

`{image} ../../figures/arrayAlias.png :align: center :alt: copying an array reference :width: 300 pt`

Hereafter, array assignments like:

    b[2] = -10;
    d[1] = 55;

would both change the *same* array. Now `b` and `d` are essentially
names for the same thing (the actual array). The technical term matches English:
The names are *aliases*.

This may seem like a pretty silly discussion. Why bother to give two different
names to the same object? Isn’t one enough? In fact it is very important
in function/method calls. An array reference can be passed as an actual value,
and it is the array *reference* that is copied to the formal parameter, so
the formal parameter name is an **alias** for the actual parameter name.

If an array passed as a parameter to a method has elements changed in the
method, then the change affects the actual parameter array.
The change *remains* in the actual parameter array *after*
the method has terminated.

```{index} array; Scale example
```

For example, consider the following function:

    // Modify a by multiplying all elements by multiplier.
    static void Scale(int[] a, int multiplier)
    {
       for (int i = 0; i < a.Length; i++) {
          a[i] *= multiplier;  // or:  a[i] = a[i] * multiplier
       }
    }

The fragment:

    int[] nums = {2, 4, 1};
    Scale(nums, 5);

would *change* nums, so it ends up containing elements 10, 20, and 5.
