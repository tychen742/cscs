A service desk receives requests, records work, and reports what remains. A computer helps with those tasks by executing instructions and representing information. **Computer science** studies how to represent problems, design algorithms, and reason about what computers can and cannot do. **Information technology** applies computing systems to organizational needs. Writing C# is one way to put those ideas into practice.


## From a Business Need to a Computing Problem


Suppose a manager asks, “How much work is left today?” That request needs clarification before it becomes a program. Does “work” mean requests, hours, or cost? Which requests count as completed? When was the data collected? A **requirement** states the behavior the system must provide. For this first example: report the number of open requests after completed requests are removed from the morning total.

The **input-process-output (IPO) model** helps describe that calculation:

| Stage | Service-desk example | Question to ask |
| --- | --- | --- |
| Input | 18 requests at opening; seven completed | Are both counts accurate and from the same day? |
| Process | Subtract completed requests from the opening count | What about new requests received during the day? |
| Output | 11 requests remain | Does the report identify what the number means? |

This model simplifies the real service desk. The simplification is an **abstraction**: keep the details needed for a particular question and set aside others. Here, requests arriving later are excluded. That assumption must be stated; a correct subtraction can still answer the wrong business question. Programs can also receive input and produce output repeatedly, rather than pass through IPO only once.


````csharp
Console.WriteLine("Open requests at closing:");
Console.WriteLine(18 - 7);
````

`Console.WriteLine(...)` displays the value inside its parentheses and then moves to a new line. Text inside double quotation marks is a **string literal**, a written text value. `18 - 7` is an **expression**, code that produces a value. The quotation marks delimit the text; they are not printed.

**Check your understanding:** If four new requests arrived, what input is missing? The revised calculation would be `18 + 4 - 7`, producing 15. Changing a requirement changes the calculation and its test cases.

## Algorithms and Sequential Execution


An **algorithm** is a precise procedure for solving a problem. “Calculate the remaining work” leaves the actual steps unspecified. The following procedure is precise enough to test:

1. Obtain the opening request count.
2. Obtain the number completed from that opening group.
3. Subtract the completed count from the opening count.
4. Display the result with a label.

A valid input has nonnegative counts, with completed requests no greater than opening requests. Test the ordinary case (18 and seven), the empty case (zero and zero), and the all-completed case (18 and 18). Expected remaining counts are 11, zero, and zero. If completed requests exceed the opening count, investigate the input or revise the model rather than accept a negative workload.

**Sequential execution** means executing statements in their written order. A **statement** tells the program to perform an action. Each call to `Console.WriteLine` below is a statement; the calculation inside the second call is an expression.


````csharp
Console.WriteLine("Service desk: morning summary");
Console.WriteLine(18 - 7);
Console.WriteLine("End of report");
````

C# uses `+` for addition, `-` for subtraction, and `*` for multiplication. Multiplication happens before addition unless parentheses change the grouping. ### Sequence, Selection, and Repetition


Three control structures describe much of a program's behavior:

| Structure | Meaning | Service-desk example |
| --- | --- | --- |
| Sequence | Perform steps in order | Read counts, calculate remaining work, print a report |
| Selection | Choose an action based on a condition | Warn the manager if remaining work exceeds staffing capacity |
| Repetition (iteration) | Repeat an action | Process each request in a list |

This chapter uses sequence. Chapters 4 and 5 teach the C# syntax for selection and repetition. For now, describe a decision in plain language: “If more than 10 requests remain, ask for another technician.” Do not confuse this rule with an instruction to repeat something.

## Hardware, Software, and Data Representation


**Hardware** is the physical equipment: a processor executes instructions, memory holds information currently in use, and storage retains files. **Software** is the programs and their instructions. The **operating system** manages resources and provides services such as access to files, devices, and running programs.

A report passes through layers. Your C# code asks a .NET library to print text; the runtime and operating system arrange for that text to reach an output device. You do not need to program the display electronics to write the report. Each layer offers an abstraction for the layer above it.

A **bit** has two possible values, zero and one. A **byte** contains eight bits, so it has `2 × 2 × 2 × 2 × 2 × 2 × 2 × 2 = 256` possible patterns. A pattern has meaning only under an agreed representation: it might encode part of a number, text, an image, or an instruction.

### Binary and Hexadecimal Numbers


Decimal uses place values that are powers of 10. **Binary** uses powers of two. In the eight-bit pattern `01000001`, the set bits contribute 64 and one, so its unsigned numeric value is 65. **Hexadecimal** uses 16 digits, `0` through `9` and `A` through `F`; its `41` means four sixteens plus one, also 65.

C# prefixes a binary integer literal with `0b` and a hexadecimal integer literal with `0x`. All three statements below display the same decimal value.


````csharp
Console.WriteLine(65);
Console.WriteLine(0b01000001);
Console.WriteLine(0x41);
````

An encoding assigns meaning to a bit pattern. In ASCII, the numeric code 65 represents uppercase `A`. That does not mean every character in every text encoding occupies one byte. Chapter 2 develops numeric and text representations further.

An address identifies a location in memory. With 32 address bits there are `2^32` possible addresses; if each addresses one byte, that address space covers 4 GiB. This calculation concerns an address space, not a universal limit on installed memory. Actual usable memory also depends on the processor, operating system, and how addresses are managed. Likewise, a 64-bit system does not necessarily support physically installing `2^64` bytes.

## From Source Code to Execution


**Source code** is the human-readable program text. A **compiler** translates code into another representation and reports violations of language rules. In a typical .NET workflow, the C# compiler produces **Common Intermediate Language (CIL)** and metadata describing the program. The **Common Language Runtime (CLR)** loads that compiled program. Its **just-in-time (JIT) compiler** translates needed methods into native instructions for the processor.[^1]

```text
C# source → C# compiler → CIL and metadata → .NET runtime → execution
```

The SDK supplies development tools; the runtime supports execution. You will distinguish them in the next section. Other execution arrangements exist, including compiling native code ahead of time.

**Interpretation** executes a program through an interpreter. Implementations can combine translation, interpretation, and compilation. Avoid the rule “a language is either compiled or interpreted.” It is the implementation and workflow that determine how a particular program runs.

## Developing and Checking a Program


The **software development life cycle (SDLC)** organizes work from understanding a need to maintaining the resulting system. A small service-desk report still involves requirements, design, implementation, testing, deployment, and maintenance. These activities repeat: testing may reveal a missing requirement, and maintenance may require a new design.

| Problem | What happens | First response |
| --- | --- | --- |
| Compile-time error | The compiler cannot accept the code, such as a missing closing quote | Read the first diagnostic and inspect the indicated line |
| Runtime error | The program begins running but an operation fails, possibly raising an exception | Identify the operation and the input that caused failure |
| Logic error | The program runs but answers the question incorrectly | Compare actual output with a known expected result |

An **exception** signals a condition that interrupts normal execution. Its detailed handling belongs in Chapter 6. A successful build alone cannot establish that a business calculation is correct.


````csharp
// This runs successfully, but adds completed work instead of subtracting it.
Console.WriteLine("Remaining requests (incorrect calculation):");
Console.WriteLine(18 + 7);
````

The output is 25, but the requirement predicts 11. The defect is a logic error. Repair the expression, rerun the example, and then test the empty and all-completed cases.

**Discussion:** A manager receives the correct count with no date or label. Is the arithmetic wrong? Is the report usable? Testing includes whether the output communicates the required information.
