# Getting Started with C#

A useful program begins with a clear question, a model of the needed information, and a way to check the answer. This chapter connects those ideas to your first C# reports and a local development workflow.

- **Computing concepts**: Model a service-desk problem and distinguish data, algorithms, and execution.
- **Development tools**: Save, build, run, and track a small console application.
- **Program structure**: Read top-level statements and explicit `Main`, and distinguish files, projects, and solutions.


> *[Getting Started with C# & .NET in VS Code (Official Beginner Guide)](https://youtube.com/watch?v=ZVGutgqBMUM)* by Visual Studio Code.
>
> This video remains hosted on YouTube and is the property of its original creator or rightsholder; it is embedded here via YouTube's official player for educational reference only.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:

1. Specify inputs, assumptions, processing steps, and expected output for a small business report.
2. Trace a sequential C# program and distinguish compile-time, runtime, and logic errors.
3. Explain bits, bytes, and the roles of a compiler and runtime.
4. Verify an SDK, edit a console project, and compare build and run results.
5. Identify an entry point and record a reviewed source change in Git.


<h2>Chapter Flow</h2>

- {doc}`1.1. Computing Concepts <0101-cs_ideas>`
- {doc}`1.2. Development Tools <0102-dev_tools>`
- {doc}`1.3. C# Program Structure <0103-program_structure>`
- {doc}`Assignments <assignments/index>`


<h2>Glossary</h2>

| Term | Meaning |
| --- | --- |
| IPO model | A way to describe input, transformation, and output. |
| Algorithm | A precise procedure for solving a problem. |
| Abstraction | A model that keeps details relevant to a question. |
| Bit / byte | A bit has two values; a byte contains eight bits. |
| Compiler | A tool that translates code and checks language rules. |
| Runtime | The environment supporting execution of compiled programs. |
| SDK | Development tools for building and running applications. |
| Terminal / shell | A text interface and the command interpreter it runs. |
| Expression / statement | Code producing a value / code performing an action. |
| Entry point | Where application execution begins. |
| Top-level statements | Starting code for which the compiler generates an entry point. |
| Namespace | A named group of declared types. |
| Project | Source and build settings described by a .csproj file. |
| Solution | A workspace grouping projects, using .slnx or .sln. |
| Git commit | A record of staged changes in local repository history. |


<h2>Lecture Slides</h2>

Slides will be authored after the chapter content is complete. The book owns slide content; Press handles delivery and access controls.
