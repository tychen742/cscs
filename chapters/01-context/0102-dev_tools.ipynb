
Writing a C# program involves more than entering code: you need tools to edit it, build it, run it, and inspect the results. In this section, you will learn how those tools work together. Visual Studio Code (VS Code) is where you edit source files, and its C# extension adds support such as code completion and error checking. The .NET software development kit (SDK) provides the tools to build and run your programs. Through the terminal, you enter commands to manage files and start applications; the C# read-evaluate-print loop (REPL) lets you try expressions and statements interactively. You will use these tools in Section 1.3 to create, edit, and run your first console application.


```{index} .NET; SDK; runtime; dotnet --version
```

## Install and Check the .NET SDK


**.NET** provides the runtime and libraries used by C# applications. The **software development kit (SDK)** includes the compiler and development commands needed to build and run them. The runtime executes compiled applications; installing only the runtime is not enough to develop them.

Install the **.NET 10 SDK** for your operating system and processor from [Microsoft's .NET 10 download page](https://dotnet.microsoft.com/en-us/download/dotnet/10.0). Select the SDK, rather than a runtime-only download. Microsoft provides [installation instructions](https://learn.microsoft.com/en-us/dotnet/core/install/) for Windows, macOS, and Linux.

After installation, open a new terminal and enter:

```console
dotnet --version
```

The version should begin with `10.`. Patch and SDK feature versions can differ; you do not need a particular patch number for these introductory examples. If another installed SDK is selected, inspect `dotnet --list-sdks` and confirm .NET 10 is present. If the command is not found, reopen the terminal and check the installation before proceeding.

**Checkpoint:** Explain the difference between the SDK used to develop a program and the runtime used to execute it.

List its contents with `Get-ChildItem` in PowerShell or `ls` in Bash/Zsh. If you do not already have a `workspace` folder, create it:

```console
mkdir workspace
```

Enter it:

```console
cd workspace
```

Confirm the current location with `Get-Location` in PowerShell or `pwd` in Bash/Zsh. This is where you will create `HelloWorld` in Section 1.3. Do not run `mkdir workspace` again if that folder already exists.

**Checkpoint:** Identify your current folder and its parent. You should be at the operating-system shell prompt, ready to enter commands such as `dotnet --version`.


## Set Up VS Code for C#


After installing and verifying the .NET 10 SDK in the previous section, install [VS Code](https://code.visualstudio.com/download) and open it. Read {doc}`Editing Tools <../appendices/development_tools/editing_tools>` for Explorer, saved files, extensions, and the integrated terminal.

Next, install Microsoft's **C# Dev Kit** (`ms-dotnettools.csdevkit`). It adds project management and testing support. **C# is a VS Code extension** that provides code completion and diagnostics. After installing C# Dev Kit, check that its **C#** and **.NET Install Tool** extensions are installed too; if either still shows **Install**, install it manually. See Microsoft's [C# setup guide](https://code.visualstudio.com/docs/csharp/get-started).

Check the Extensions view to confirm that C# Dev Kit and its two dependencies are installed:

| Extension | Install separately? | What it does |
| --- | --- | --- |
| **C# Dev Kit** by Microsoft | **Yes** | C# project management, testing, and development tools. |
| **C#** by Microsoft | If not installed automatically, install it manually | C# code completion and diagnostics. |
| **.NET Install Tool** by Microsoft | If not installed automatically, install it manually | Supplies .NET components needed by the editor extensions. |

The **.NET 10 SDK** is a separate download, not a VS Code extension. It provides the tools that build and run your C# projects. During setup, VS Code may also ask to install .NET components needed by its extensions; allow the installation and wait for it to finish. The extension-managed runtime supports the editor's tools, while the SDK builds and runs your applications. Seeing a different runtime version in an extension message does not mean your project has changed its target framework. Microsoft's [C# Dev Kit FAQ](https://code.visualstudio.com/docs/csharp/cs-dev-kit-faq) explains these dependencies.

Use **File > Open Folder** to open your `workspace` folder. Use **Terminal > New Terminal** and run `dotnet --version` there to confirm VS Code can access the SDK too. Save a short `notes.txt` file in the folder and locate it in Explorer. You will create the actual C# project in Section 1.3.

When that project exists, open the **HelloWorld project folder**, containing both `Program.cs` and `HelloWorld.csproj`, rather than opening `Program.cs` alone. Allow the C# tools to load the project. If a warning says the active document is outside the workspace, check the open folder and reopen the file from Explorer. After moving files, close old tabs that point to their previous locations.

**Checkpoint:** Confirm C# Dev Kit, C#, and .NET Install Tool appear as installed in Extensions, your saved file appears in Explorer, and the integrated terminal reports a .NET 10 SDK. If C# Dev Kit is installed but C# still shows **Install**, install it before continuing. In Section 1.3, check that typing `Console.` in the loaded project's source offers member completions.

For version control and collaboration, see {doc}`Appendix C: Git and GitHub <../appendices/github>`.


```{index} CSharpRepl; REPL; expressions
```

## Try C# Interactively with CSharpRepl


Read {doc}`Interactive Language Environments (REPLs) <../appendices/development_tools/interactive_language_environments>` to distinguish a language prompt from an operating-system shell. A C# REPL helps you check expressions quickly; your saved console applications will use `Program.cs` and `dotnet run`.

Install [CSharpRepl](https://github.com/waf/CSharpRepl) from the operating-system shell:

```console
dotnet tool install -g csharprepl
```

If it is already installed, you can update it with `dotnet tool update -g csharprepl`. Start it:

```console
csharprepl
```

At its `>` prompt, enter an expression. The following transcript includes the prompt and displayed result; type only `2 + 3`:

```text
> 2 + 3
5
```

Try multiplication too:

```text
> 4 * 10
40
```

The REPL displays expression results automatically. In a console application, you will use `Console.WriteLine(...)` to display them. REPL entries can retain state during a session, so do not assume a snippet works independently unless you have supplied everything it needs.

Enter `exit` to return to the operating-system shell. If `csharprepl` is not found after installation, open a new terminal and follow its [installation troubleshooting](https://github.com/waf/CSharpRepl/wiki/Installation) for the global tool directory.

**Checkpoint:** Evaluate an expression, identify the C# prompt, and return to the shell. You are now ready to create and run your first console application in {ref}`program-structure`.
