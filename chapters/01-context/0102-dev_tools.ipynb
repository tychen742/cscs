## Know Which Tool Does What


| Tool | Responsibility | What it does not establish |
| --- | --- | --- |
| .NET SDK (software development kit) | Provides the compiler, project templates, and `dotnet` development commands | That your calculation satisfies the requirement |
| .NET runtime | Executes compiled .NET applications and supplies runtime services | That you have the SDK needed to build source code |
| VS Code with C# support | Edits files and connects to language and debugging tools | That the SDK is installed correctly |
| Terminal and shell | Accept commands and start tools | That a command is C# source code |
| Git | Records versions of source files | That a commit has been uploaded or backed up elsewhere |

An **integrated development environment (IDE)** brings editing, building, and debugging together. VS Code is an editor that gains those capabilities through extensions. A **debugger** pauses execution and lets you inspect the program while it runs. You will use it more fully in Chapter 6.

### Verify the SDK and Editor


Install the .NET 10 SDK for your operating system and processor from [Microsoft's .NET download page](https://dotnet.microsoft.com/download/dotnet/10.0). Choose the **SDK**, rather than a runtime-only download. Install [VS Code](https://code.visualstudio.com/) and its C# support using Microsoft's [C# setup instructions](https://code.visualstudio.com/docs/languages/csharp).[^1]

Open a new terminal after installation and run these commands, one at a time:

```console
dotnet --version
dotnet --list-sdks
git --version
```

`dotnet --version` should report a 10.x SDK for the file-based workflow in this chapter. `--list-sdks` reports installed SDKs and their locations. Exact patch numbers can differ. If the command is not found, check installation and open a fresh terminal. If a repository selects another SDK through `global.json`, the selected version can differ from the newest one installed.

If Git is missing, use the [Git installation instructions](https://git-scm.com/book/en/v2/Getting-Started-Installing-Git). Record these version results as setup evidence. Do not copy a version number from a classmate.

## Commands, Prompts, and Directories


A **terminal** displays text input and output. A **shell**, such as PowerShell on Windows or Bash/Zsh on macOS and Linux, reads your commands. The shell prompt shows where you can type; it is not part of a command to copy.

A command operates from a **current directory**, the folder your shell is using. A **relative path** is interpreted from that folder; `..` means its parent and `.` means the current folder. Use quotation marks around a path containing spaces. The commands below work in PowerShell and in Bash/Zsh:

| Command | Meaning |
| --- | --- |
| `pwd` | Display your current directory |
| `ls` | List its contents |
| `mkdir ServiceDeskPractice` | Create a new practice folder |
| `cd ServiceDeskPractice` | Enter that folder |
| `cd ..` | Return to its parent |

If `dotnet run` says it cannot find a project, inspect `pwd` and `ls` before reinstalling anything. You may simply be in the wrong folder. Detailed shell reference belongs in the {doc}`Command Line appendix <../appendices/command-line>`.

## Record Your Work with Git


**Version control** records changes so you can inspect earlier versions and understand how work developed. Git stores that history in a **repository**. GitHub is a hosting service for repositories; it is separate from Git.

A local workflow has four distinct states:

```text
Edit a file → inspect the changes → stage selected changes → commit
```

**Staging** selects what belongs in the next commit. A **commit** records that selection in local history. Saving an editor file does not stage it, and committing does not upload it. Uploading requires a configured **remote** repository and a later `git push`.


```{index} git status; git diff; git add; git commit; gitignore
```

### Make a Small Local Commit


For a new, separate practice project that is not already inside another Git repository, create a `.gitignore` file in VS Code containing:

```text
bin/
obj/
```

These folders contain generated build files. Keep the source and project configuration in history rather than their generated copies. Run:

```console
git init
git status
git add Program.cs ServiceDeskPractice.csproj .gitignore
git diff --staged
git commit -m "Add service desk report header"
git status
```

Review the staged diff before committing. If Git asks for your name and email, follow your course's identity instructions; configure them for the repository with `git config user.name "Your Name"` and `git config user.email "your-address@example.com"`, replacing both examples with your own choices.

After committing, change one report line and save it. `git diff` shows unstaged tracked changes; `git diff --staged` shows changes selected for the next commit. New untracked files are listed by `git status` but do not appear in ordinary `git diff` until tracked. This distinction prevents accidentally omitting new files.

**Check your understanding:** Your source is saved and the program runs, but `git status` reports a modified file. Has the new version been committed? No. Has an earlier local commit reached GitHub? You cannot infer that from a successful commit alone.

Team branches, pull requests, and conflict resolution are later workflow topics. This chapter's goal is to save, inspect, and commit a small local change.
