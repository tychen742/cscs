You have now run C# in a browser and seen a console project. The code's **entry point** is where execution begins. Its files and build configuration determine how the tools package and run it. Keep these two questions separate: “Where does execution begin?” and “Which files belong to the application?”

## A Map of Five Useful Workflows


| Workflow | Use it for | What to recognize |
| --- | --- | --- |
| REPL (read-evaluate-print loop) | Optional quick experiments | A tool evaluates entries interactively |
| File-based app | A saved, small single-file experiment | .NET 10 can run `app.cs` without a project file |
| Console project with top-level statements | The beginner project workflow used here | The compiler supplies the entry-point wrapper |
| Console project with explicit `Main` | Reading the entry-point structure directly | A `Main` method contains the starting code |
| Solution workspace | Organizing related projects | A solution groups projects; each application still has an entry point |

These are not five incompatible kinds of C#. File-based apps and projects can both use top-level statements or explicit `Main`. A solution is an organizational layer, and a REPL is an interactive tool. You do not need to install a REPL or create a solution to complete this chapter's lab.

### An Optional REPL Experiment


A REPL can show an expression's value immediately. The following is a transcript, including a prompt and the displayed result, rather than a complete source file:

```console
> 18 - 7
11
```

In an application, write `Console.WriteLine(18 - 7);` to display the value. An expression by itself is not a substitute for an output statement. Book Run cells contain application code, not a REPL transcript.

### Save One File or Create a Project


With the .NET 10 SDK, save the following code as `app.cs` in a fresh experiment folder, separate from your console project:

```csharp
Console.WriteLine("Service desk: daily report");
Console.WriteLine(18 - 7);
```

Run `dotnet run app.cs` from that folder. This is a **file-based app**: the tools build and execute the source without requiring you to create a `.csproj` file.[^1]

For durable course work, use `dotnet new console -o ServiceDeskPractice` and edit the generated `Program.cs`. The `.csproj` file specifies settings such as the target framework and dependencies. A project can contain multiple source files. Do not add several standalone examples to one project unchanged: their top-level statements can conflict.

| Item | Role |
| --- | --- |
| `Program.cs` | Human-readable source for the starter application |
| `ServiceDeskPractice.csproj` | Project configuration |
| `bin/` | Generated compiled output |
| `obj/` | Generated build information |

The filename `Program.cs` is a convention, not a requirement that every C# file have that name.

## Top-Level Statements and Explicit Main


Top-level statements let you put the starting instructions directly in the source file. Read this program from top to bottom, predicting three lines of output before running it.


````csharp
Console.WriteLine("Service desk: daily report");
Console.WriteLine(18 - 7);
Console.WriteLine("End of report");
````

The compiler generates an entry point for those statements. A project may have only one file containing top-level statements; other files can define reusable types.

To expose the traditional entry point, create a separate project using `dotnet new console --use-program-main -o ServiceDeskExplicit`. Its starting code is inside a `Main` method. This complete example produces the same report:

```csharp
using System;

namespace ServiceDesk;

class Program
{
    static void Main(string[] args)
    {
        Console.WriteLine("Service desk: daily report");
        Console.WriteLine(18 - 7);
        Console.WriteLine("End of report");
    }
}
```

Replace that separate project's `Program.cs` with the example and run `dotnet run` in its folder. Do not paste it beneath the other project's top-level statements; compare the two complete versions instead.

### Read the Wrapper Without Memorizing It


| Part | Meaning at this stage |
| --- | --- |
| `using System;` | Makes names from the `System` namespace available without their full prefix |
| `namespace ServiceDesk;` | Places the declared types in a named group |
| `class Program` | Declares a type named `Program` |
| `static void Main(string[] args)` | Declares the starting method; `args` can receive command-line text arguments |
| `{ ... }` | Groups the contents of a class, method, or other block |

A **namespace** groups names to help avoid collisions; it is not a folder command and does not run the report. A **class** defines a type, and a **method** is a named operation. Chapters 3 and 11 develop those ideas. For now, identify the starting statements inside `Main`.

`void` means this method returns no value. `static` permits calling it without constructing a `Program` object. You need to recognize these words in a generated template, not design classes yet. Console templates enable implicit imports for common namespaces, so their starter code can use `Console` without an explicit `using System;`. The example shows the import to make the dependency visible.

## Read and Change Small Programs


A **comment** explains code to a reader: `//` begins a comment that continues to the line's end; `/* ... */` delimits a block comment. Comments do not print themselves. Most simple statements, including the output calls below, end with a semicolon. Braces group statements; they are not quotation marks.

C# names are case-sensitive: `Console.WriteLine` and `console.writeline` are different spellings. Double quotes delimit strings. Spaces inside a string appear in the output; ordinary indentation outside strings helps people read the source.


````csharp
// Print a heading, then the count.
Console.WriteLine("Open requests:");
Console.WriteLine(18 - 7);
/* The closing line is part of the report. */
Console.WriteLine("End of report");
````

**Trace:** The first comment is skipped, the heading prints, the expression evaluates to 11 and prints, the block comment is skipped, and the closing line prints. Moving the closing statement to the beginning changes the report order. Removing the comments does not change its output.

The following is an intentionally invalid example shown as text. It is missing a closing double quote:

```text
Console.WriteLine("Open requests:);
```

Correct the quote before changing the calculation. A compiler diagnostic often identifies where it noticed the problem; the original mistake may begin earlier on that line.

## When a Solution Becomes Useful


A **solution** groups related projects. For example, an application and its tests can be separate projects in one solution. A solution does not replace their `.csproj` files and does not mean that all programs run simultaneously.

This is an optional workspace exercise. From a fresh parent folder, run:

```console
mkdir ServiceDeskWorkspace
cd ServiceDeskWorkspace
dotnet new sln
dotnet new console -o App
dotnet sln add App/App.csproj
dotnet build
dotnet run --project App
```

The .NET 10 solution template creates `ServiceDeskWorkspace.slnx` by default; you may also encounter older `.sln` files. Build the solution from its folder, and select an application project when running. You do not need this extra layer for a single-project assignment.

**Exit check:** Explain why a project file, a namespace, and a solution each serve a different purpose. Then choose a workflow for a one-line experiment and for a saved assignment. A file-based experiment is sufficient for the first; a console project supplies the durable structure for the second.
