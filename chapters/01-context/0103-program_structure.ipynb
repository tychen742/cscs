## The Five C# Structures


C# programs can be written and run in five useful structures. Each one is real C#, but each one is best for a different learning or development situation.

| # | Structure | Best use | Run or create it |
| --- | --- | --- | --- |
| 1 | REPL quick checks | Try one expression or a tiny idea | `csharprepl` |
| 2 | File-based app | Save and rerun one small file | `dotnet run app.cs` |
| 3 | Console project with top-level statements | Start a normal beginner-friendly app | `dotnet new console` |
| 4 | Console project with explicit `Main` | Learn durable C# program structure | `dotnet new console --use-program-main` |
| 5 | Project or solution workspace | Build labs, homework, tests, and multi-file apps | `.csproj` and, when needed, `.sln` |

This book uses more than one structure on purpose. Part I may use the REPL for quick checks such as `a > b`, because requiring a whole project for one expression gets in the way of learning. As programs become larger, the book shifts toward project/application mode so that students practice the structure they will use in VS Code and in submitted work.


```{index} csharprepl; REPL quick checks
```

### REPL Quick Checks


A **REPL** reads one expression or statement, evaluates it, prints the result, and waits for the next entry. `csharprepl` is useful when the goal is to test a small idea quickly:

```csharp
> int a = 7;
> int b = 4;
> a > b
true
```

Use the REPL for quick demonstrations, predictions, and small experiments. Do not use it as the main structure for labs or projects, because project files, source files, and version control matter once the program has more than a few lines.


```{index} file-based C# apps; dotnet run app.cs
```

### File-Based Apps with `dotnet run app.cs`


Starting with .NET 10, C# supports **file-based apps**: a single `.cs` file can be run directly with the .NET CLI, without first creating a `.csproj` project file.[^3]

Create a file named `app.cs`:

```csharp
Console.WriteLine("Hello from one file!");
```

Then run it from the terminal:

```powershell
dotnet run app.cs
```

This workflow is helpful when you want to edit a real source file but still keep setup light. It sits between the REPL and a full project: more durable than a REPL entry, but simpler than a project folder.

```{important}
Use file-based apps for short experiments and small single-file programs. For course projects, labs with multiple files, testing, and versioned submissions, use a normal .NET project.
```


```{index} console project; dotnet new console; top-level statements
```

### Console Projects with Top-Level Statements


A **console project** is the normal starting point for C# applications in this book. The default template uses **top-level statements**, which means the first program can be written directly as executable statements in `Program.cs`:

```csharp
Console.WriteLine("Hello, World!");
```

Create the project from the terminal:

```powershell
dotnet new console
```

A new console project usually contains:

| File or folder | Purpose |
| --- | --- |
| `Program.cs` | The C# source file where the starter program lives |
| `YourProject.csproj` | The project file that records target framework, packages, and build settings |
| `obj/` | Temporary build information created by .NET |
| `bin/` | Compiled output after the project is built or run |

Run the project from the project folder:

```powershell
dotnet run
```

Open the project folder in VS Code:

```powershell
code .
```

```{important}
Top-level statements are allowed in only one source file in a project, and they must appear before namespace and type declarations. If a code example defines classes, interfaces, or other reusable types, it is often clearer to place the running code inside an explicit `Main` method.
```


```{index} explicit Main method; Program class; use-program-main
```

### Console Projects with Explicit `Main`


To see the traditional structure explicitly, create a console project with `--use-program-main`:

```powershell
dotnet new console --use-program-main
```

The generated `Program.cs` uses a `Program` class and a `Main` method:

```csharp
namespace HelloWorld;

class Program
{
    static void Main(string[] args)
    {
        Console.WriteLine("Hello, World!");
    }
}
```

That structure is slightly longer, but it scales better when the program has helper methods, classes, interfaces, or multiple source files.

A conventional `Program.cs` file has several layers:

| Piece | Meaning |
| --- | --- |
| `using System;` | Imports the `System` namespace so common .NET library names can be used directly.[^1] |
| `namespace HelloWorld;` | Groups related code and reduces naming conflicts |
| `class Program` | Defines a class named `Program` |
| `static void Main(string[] args)` | Defines the method where execution begins |
| `{ ... }` | Groups statements into a block |

The important idea is that execution starts at `Main`. The program can then call other methods, create objects, and use classes defined in the same project.

### Project and Solution Workspaces


A **project** is a buildable unit: it has a `.csproj` file, source files, package references, and build settings. A **solution** is a workspace that can contain one or more projects.

```text
Solution
├── Project A
│   ├── Program.cs
│   └── ProjectA.csproj
└── Project B
    ├── Program.cs
    └── ProjectB.csproj
```

Create a solution folder and add projects like this:

```powershell
mkdir MyWorkspace
cd MyWorkspace
dotnet new sln

dotnet new console -o App
dotnet sln add App/App.csproj
```

The resulting structure looks like this:

```text
MyWorkspace/
├── MyWorkspace.sln
└── App/
    ├── App.csproj
    ├── Program.cs
    ├── bin/
    └── obj/
```

Common commands:

| Command | Use |
| --- | --- |
| `dotnet new sln` | Create a solution file |
| `dotnet new console -o App` | Create a console project in a folder named `App` |
| `dotnet sln add App/App.csproj` | Add a project to the solution |
| `dotnet run --project App` | Run a specific project from the solution folder |
| `dotnet build` | Build all projects in the current solution or project |

If you create a project in VS Code using the .NET project tools, a solution may be created automatically. Visual Studio and VS Code use the solution file to show projects together in the explorer and to manage larger applications.[^2]

## Syntax Shared by All Structures


No matter which structure you use, C# programs are built from statements, comments, braces, and semicolons.

- **Single-line comments** start with two forward slashes (`//`). Any text between `//` and the end of the line is ignored by C#.
- **Multi-line comments** start with `/*` and end with `*/`. Any text between `/*` and `*/` is ignored by C#.
- The **curly braces** `{ }` mark the beginning and end of a block of code.
- **Semicolons** work as statement terminators in C#. They are required for most statements because they prevent syntax ambiguities. A semicolon after a method or accessor block, however, is not allowed.

### Compound Statements (Code Blocks)


A code block, or **block**, groups multiple statements into a single unit. In C#, a code block is delimited by curly braces and contains a list of statements.

For example, the following code shows an `if` statement with two blocks: one block for the `if` branch and one block for the `else` branch.


````csharp
double radius = 5.0;  // try a negative value to test the else branch
double area;
if (radius >= 0)
{
    // calculate the area of the circle
    area = Math.PI * radius * radius;
    Console.WriteLine($"The area of the circle is: {area:0.00}");
}
else
{
    Console.WriteLine($"{radius} is not a valid radius.");
}
````
