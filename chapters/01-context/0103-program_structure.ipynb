# C# Program Structure


```{contents}
:local:
```

In {ref}`development-tools`, you learned about the .NET SDK, terminal, VS Code, and C# REPL. Now use those tools to create your first **console application**: a saved program that communicates through text in the console. This first application prints a greeting; later programs will also read input.

The terminal runs a shell that accepts commands such as `dotnet run`. When the shell starts your application, the application writes its output in the same terminal window. After the application finishes, the shell prompt returns. The REPL remains useful for quick experiments, but here you will save source code in a project and run the application.


```{index} console application; dotnet new console; dotnet run
```

## Create and Run Your First Console Application


Open your terminal and navigate to the workspace folder you created in Section 1.2. If you are still inside `csharprepl`, enter `exit` first so that you can type shell commands. Run these commands one at a time:

```bash
dotnet new console --use-program-main -n HelloWorld
cd HelloWorld
code .
```

The first command creates a console project in a new folder named `HelloWorld`. The `--use-program-main` option makes the template show the `Program` class and `Main` method explicitly.[^1] The second command moves the shell into that folder. The third opens the folder in VS Code. If `code .` is unavailable, use **File > Open Folder** in VS Code and select `HelloWorld`.

A **project** groups source files and the settings needed to build an application. The `HelloWorld` folder is your project folder. In VS Code's **Explorer**, expand that folder and identify these two files:

| File | What you do with it |
| --- | --- |
| `Program.cs` | Open and edit the C# source code. The `.cs` extension identifies a C# source file. |
| `HelloWorld.csproj` | Inspect the project settings. The `.csproj` extension identifies a C# project file. |

Opening the project folder lets VS Code work with the source files and their build settings together. The folder is where you keep the files; the `.csproj` tells .NET how to build the application. The project name, `HelloWorld`, is not itself an executable C# instruction.

In VS Code, open `Program.cs`. Its contents should look like the following complete program:


````csharp
namespace HelloWorld;

class Program
{
    static void Main(string[] args)
    {
        Console.WriteLine("Hello, World!");
    }
}
````

Run the application from the terminal while you are in the `HelloWorld` folder:

```bash
dotnet run
```

`dotnet run` builds the current project and starts the application. Its program output is:

```text
Hello, World!
```

The greeting appears in the console, then the application ends and the shell prompt returns. You can also use VS Code's integrated terminal (**Terminal > New Terminal**); check that its current folder is `HelloWorld` before running the command.

```{important}
Type `dotnet` commands in the terminal. Write C# statements in `Program.cs`. Save your edits before running `dotnet run` again.
```

## Read the Program from the Outside In


C# programs have both **file organization** and **code organization**. On disk, a project contains source files and a `.csproj` file. Inside a source file, namespaces organize type names, classes contain members such as methods, and methods contain executable statements.

For this first application, the pieces fit together like this:

```text
HelloWorld/                         Project folder opened in VS Code
├── HelloWorld.csproj               Build settings
└── Program.cs                      C# source file
    └── namespace HelloWorld        Groups type names
        └── class Program           Contains the Main method
            └── Main(...)           Application entry point
                └── Console.WriteLine(...)   Executable statement
```

The lower part of this diagram describes code inside `Program.cs`, not additional folders. A namespace is a grouping of names; it does not create a directory. In this application, .NET starts the `Main` method, which executes the greeting statement.

| Part | Role in this application |
| --- | --- |
| `namespace HelloWorld;` | Places the `Program` class in the `HelloWorld` namespace. |
| `class Program` | Declares a class named `Program` that contains `Main`. |
| `static void Main(string[] args)` | Declares the method where execution begins. |
| `Console.WriteLine("Hello, World!");` | Writes the greeting and moves to the next output line. |

C# is case-sensitive. `Main` and `main` are different names, as are `Console` and `console`.

### Namespace: Give Related Types a Shared Name


A **namespace** groups related types, such as classes, under a shared name. Here, `namespace HelloWorld;` places the types declared in the rest of this file in the `HelloWorld` namespace. This form is called a **file-scoped namespace** because it applies to the rest of the source file without an extra pair of braces.

The class has the short name `Program` and the **fully qualified name** `HelloWorld.Program`. Another namespace could contain its own class named `Program`; the namespace distinguishes the two. A namespace organizes names, rather than starting execution.

The template uses the project name `HelloWorld` as the namespace name. Namespaces conventionally use PascalCase: capitalize the first letter of each word, as in `HelloWorld` or `MethodDemo`. A namespace can span several files, and its name does not have to match a folder name. For this first application, keep the generated name.

### Class: A Container for Methods


`class Program` declares a class named `Program`. For now, use it as the container for the methods in your application. The opening and closing braces after the declaration mark the class body. The `Main` method is inside those braces.

You will add methods to this class in Chapter 3 and study classes in more detail in Chapter 12.


```{index} Main; static; void; command-line arguments
```

### Main: Where Execution Begins


A **method** is a named block of code that performs a task. `Main` is this console application's **entry point**, the method .NET calls when the application starts. You do not add a separate `Program.Main(...)` call to start it.

Read the declaration `static void Main(string[] args)` as follows:

- `static` allows the method to run without first creating a `Program` object.
- `void` means the method does not return a value to its caller.
- `Main` is the entry-point name, with a capital `M`.
- `string[] args` receives command-line arguments as an array of strings. This greeting program does not use them.

These pieces will become more familiar when you study methods, arrays, and objects. At this point, recognize the declaration and put the statements you want to execute inside the `Main` braces.


### Console.WriteLine: Display Output


`Console.WriteLine("Hello, World!");` is a **statement**, an instruction the program executes. `Console` is a .NET library class for console input and output. Its `WriteLine` method displays the supplied text and then ends the output line.

The text between double quotation marks is a **string literal**, text written directly in the source code. The quotation marks delimit the text; they are not printed. The parentheses hold the value passed to `WriteLine`, and the semicolon ends the statement.

`Console` belongs to the `System` namespace. The generated console project enables **implicit usings**, which make several common namespaces, including `System`, available automatically.[^1] That is why the example can say `Console.WriteLine` instead of `System.Console.WriteLine`. In projects without implicit usings, `using System;` can make the short name available. A `using` directive makes names available; a `namespace` declaration groups your own types.

## Edit, Save, and Run Again


Change the greeting and add a second statement inside `Main`. Replace the complete contents of `Program.cs` with this version, then save the file:


````csharp
namespace HelloWorld;

class Program
{
    static void Main(string[] args)
    {
        Console.WriteLine("Welcome to the order desk.");
        Console.WriteLine("The office opens at 9:00.");
    }
}
````

Run `dotnet run` again from the project folder. The output is:

```text
Welcome to the order desk.
The office opens at 9:00.
```

The statements execute from top to bottom. Each `WriteLine` produces one output line. This is the basic working cycle: **edit, save, run, and inspect the result**.

Try changing the office opening time. Before running the program, predict which line will change. Then compare your prediction with the console output.

## Comments, Braces, and Semicolons


A **comment** explains code to a human reader and is ignored during execution. A single-line comment begins with `//` and continues to the end of that line. A block comment begins with `/*` and ends with `*/`.

```csharp
// Display the office opening time.
Console.WriteLine("The office opens at 9:00.");

/* These statements display
   the daily welcome message. */
```

A **block** groups code between curly braces, `{` and `}`. In the complete program, the outer pair encloses the class body and the inner pair encloses the `Main` body. Indentation makes this nesting visible to readers; the braces define it for C#.

Most simple statements, including a method call such as `Console.WriteLine(...)`, end with a **semicolon**. The file-scoped namespace declaration also ends with a semicolon. The closing braces of this class and method do not need semicolons.

Try removing the semicolon from a `WriteLine` statement and running the project. The compiler reports an error instead of running the revised program. Restore the semicolon, save, and run again. This is a **syntax error**, a violation of the language's writing rules.

## Source Files and the Project File


After you run the application, the project folder contains your source and project files plus generated build folders:

```text
HelloWorld/
├── Program.cs
├── HelloWorld.csproj
├── bin/
└── obj/
```

| File or folder | Purpose |
| --- | --- |
| `Program.cs` | Holds the C# source code you edit. |
| `HelloWorld.csproj` | Records build settings, including the target framework and implicit usings. |
| `bin/` | Holds compiled output created by building or running the project. |
| `obj/` | Holds intermediate information created during the build. |

Open `HelloWorld.csproj` in VS Code. It uses **Extensible Markup Language (XML)**, a format that stores settings between named tags. Recognize these settings without changing them:

| Setting | Meaning |
| --- | --- |
| `<OutputType>Exe</OutputType>` | Build an executable application. |
| `<TargetFramework>...</TargetFramework>` | Select the .NET version the project targets. The value depends on the SDK and template used to create it. |
| `<ImplicitUsings>enable</ImplicitUsings>` | Make common namespaces available automatically, including `System`. |

`dotnet run` uses the `.csproj` in the current folder, compiles the project's C# source files, and starts the resulting application. **Compilation** translates source code into compiled code that .NET can execute. The text of `Program.cs` is the source; the compiled application is generated output. Saving a file changes the source, and running the project builds those changes before execution.

A project can contain several `.cs` files. Those files contribute code to the same application rather than each automatically becoming a separate program. A namespace can also span files: `Program.cs` and a later `Order.cs` could both declare `namespace HelloWorld;`. The project controls what is built; the namespace organizes the names in that code.

A **solution** groups related projects for development, for example an application project and a test project. If VS Code displays a solution view, its project entries refer to these buildable units. Your first console application needs only this one project; `dotnet run` works from its folder.

To check your understanding, locate each of these in VS Code: the project folder, the project file, the source file, the namespace declaration, the class declaration, and the entry-point method. Which file would you edit to change the greeting? Which file records the target framework?

## Recognize Top-Level Statements


You may encounter a console application's `Program.cs` containing only:

```csharp
Console.WriteLine("Hello, World!");
```

This uses **top-level statements**, executable statements written outside an explicit class or method. The compiler supplies the entry point. The default `dotnet new console` template uses this shorter form; `--use-program-main` requests the explicit form used in this lesson.[^1]

Both forms can be console applications, and both project forms run with `dotnet run`. For the application you created here, continue writing statements inside `Main`. When later examples show only a few statements, you can place those statements inside `Main` to run them in your project. Replace the example statements rather than appending a second copy of the class or another entry point.
