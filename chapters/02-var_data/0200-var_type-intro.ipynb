# Variables & Types

Programs store and transform data. This chapter introduces the fundamental building blocks of C#: naming values, choosing types, computing with expressions, and reading or displaying console data.

- **Variables**: Names let programs store values and refer to them later.
- **Types**: C# uses types such as `int`, `double`, `bool`, and `string` to define what each value can do.
- **Operators**: Arithmetic, comparison, logical, and string operators let programs compute new values.
- **Console I/O**: `Console.ReadLine` and `Console.WriteLine` connect a program to user input and output.

These ideas are the raw materials for every later C# program: once you can name data, choose its type, transform it, and show a result, you can begin building useful workflows.
