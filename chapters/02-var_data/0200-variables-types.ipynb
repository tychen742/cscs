# Variables, Types, Operators, and I/O

This chapter turns the literal calculations from Chapter 1 into an office-supply invoice with named data, deliberate numeric choices, and readable input/output.

- **Variables and state**: Store inputs, trace assignment, and recalculate derived values.
- **Types and representations**: Preserve counts, money, identifiers, and text appropriately.
- **Operators**: Distinguish arithmetic, assignment, comparisons, and Boolean rules.
- **Input and output**: Parse text under stated assumptions and present an explicitly rounded invoice.


> *[C# variables ✖️](https://youtube.com/watch?v=IxBMVztdlr4)* by Bro Code.
>
> This video remains hosted on YouTube and is the property of its original creator or rightsholder; it is embedded here via YouTube's official player for educational reference only.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:

1. Declare, initialize, and trace local variables with fixed types and appropriate names.
2. Choose numeric and text types and explain range, precision, and conversion tradeoffs.
3. Apply arithmetic, comparisons, logical rules, and assignment in the intended order.
4. Read and parse valid console input while distinguishing assumptions from validation.
5. Calculate rounded invoice components and display them consistently.


<h2>Chapter Flow</h2>

- {doc}`2.1. Variables <0201-variables>`
- {doc}`2.2. Types <0202-data_types>`
- {doc}`2.3. Operators <0203-operators>`
- {doc}`2.4. Input and Output <0206-input_output>`
- {doc}`Assignments <assignments/index>`


<h2>Glossary</h2>

| Term | Meaning |
| --- | --- |
| Variable | A typed name whose stored value can be read or replaced. |
| Initialization | The first assignment of a value. |
| Scope | The region where a declared name can be used. |
| var | Compiler inference of a fixed local type from an initializer. |
| Value / reference type | Assignment copies a value / assignment copies an object reference. |
| Overflow | A numeric result outside the type's representable range. |
| decimal | A finite decimal numeric type used for invoice amounts here. |
| Cast | An explicitly written conversion of an existing value. |
| Parsing | Interpreting text according to a numeric syntax. |
| Short-circuit evaluation | Skipping an unnecessary right Boolean operand. |
| Immutable string | A string whose existing contents cannot be changed. |
| UTF-16 code unit | The 16-bit unit represented by char. |
| Culture | Conventions used when interpreting and displaying values. |
| Rounding policy | When to round and how to resolve midpoint cases. |
| Format specifier | A presentation rule such as F2 for two fractional places. |


<h2>Lecture Slides</h2>

Slides will be authored after content completion. The book owns their content; Press handles delivery and access controls.
