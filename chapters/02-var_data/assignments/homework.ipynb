## Question 6: Recalculate an inventory value



````csharp
// ### Exercise: Recalculate an inventory value
// Start quantity at 40 and price at 2.50m. Save initialValue, remove nine units with -=, then compute revisedValue. Print quantity, initialValue, and revisedValue. Expected: 31, 100.00, 77.50.
// Your code starts here.

// Your code ends here.
````

````csharp
using System.Globalization;
CultureInfo.CurrentCulture = CultureInfo.InvariantCulture;

// ### Solution
int quantity = 40;
decimal price = 2.50m;
decimal initialValue = quantity * price;
quantity -= 9;
decimal revisedValue = quantity * price;
Console.WriteLine(quantity);
Console.WriteLine($"Initial value: {initialValue:F2} USD");
Console.WriteLine($"Revised value: {revisedValue:F2} USD");
````

## Question 10: Apply a stated invoice policy



````csharp
// ### Exercise: Apply a stated invoice policy
// For one unit at 10.10m, calculate subtotal and 5% tax rounded to cents away from zero. Print aligned subtotal, tax, and total rows with F2 and USD identified. Expected: 10.10, 0.51, 10.61. Explain why default midpoint rounding would give a different tax.
// Your code starts here.

// Your code ends here.
````

````csharp
using System.Globalization;
CultureInfo.CurrentCulture = CultureInfo.InvariantCulture;

// ### Solution
int quantity = 1;
decimal unitPrice = 10.10m;
decimal subtotal = quantity * unitPrice;
decimal tax = Math.Round(subtotal * 0.05m, 2, MidpointRounding.AwayFromZero);
decimal total = subtotal + tax;
Console.WriteLine("Amounts in USD");
Console.WriteLine($"{"Subtotal",-10} | {subtotal,10:F2}");
Console.WriteLine($"{"Tax",-10} | {tax,10:F2}");
Console.WriteLine($"{"Total",-10} | {total,10:F2}");
````
