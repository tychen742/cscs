## 5. Print the Final Invoice

Put it all together: apply a 5% loyalty discount to the subtotal with the compound operator `-=`, then add tax and print an aligned invoice using string interpolation with alignment (for example `{label,-10}{amount,10:F2}`).


```csharp
using System;

string customer = "Riverside Dental";
int quantity = 12;
decimal unitPrice = 38.50m;
decimal discountRate = 0.05m;
decimal taxRate = 0.07m;

// Your code starts here.



// Your code ends here.
```

```csharp
using System;

string customer = "Riverside Dental";
int quantity = 12;
decimal unitPrice = 38.50m;
decimal discountRate = 0.05m;
decimal taxRate = 0.07m;

decimal subtotal = quantity * unitPrice;
decimal discount = subtotal * discountRate;
decimal amountDue = subtotal;
amountDue -= discount;
decimal tax = amountDue * taxRate;
amountDue += tax;

Console.WriteLine($"Invoice for {customer}");
Console.WriteLine($"{"Subtotal",-10}{subtotal,10:F2}");
Console.WriteLine($"{"Discount",-10}{-discount,10:F2}");
Console.WriteLine($"{"Tax",-10}{tax,10:F2}");
Console.WriteLine($"{"Total",-10}{amountDue,10:F2}");
```
