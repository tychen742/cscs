## Question 2: Calculate an undiscounted invoice



````csharp
// ### Exercise: Calculate an undiscounted invoice
// For quantity 12 at 38.50m, calculate subtotal, rounded tax at 0.07m, and total. Print all three with F2. Expected values: 462.00, 32.34, 494.34.
// Your code starts here.

// Your code ends here.
````

````csharp
using System.Globalization;
CultureInfo.CurrentCulture = CultureInfo.InvariantCulture;

// ### Solution
string customer = "Riverside Dental";
string item = "Printer paper (case)";
int quantity = 12;
decimal unitPrice = 38.50m;
decimal subtotal = quantity * unitPrice;
decimal tax = Math.Round(subtotal * 0.07m, 2, MidpointRounding.AwayFromZero);
decimal total = subtotal + tax;
Console.WriteLine($"Invoice for {customer}: {item}");
Console.WriteLine($"Subtotal: {subtotal:F2} USD");
Console.WriteLine($"Tax: {tax:F2} USD");
Console.WriteLine($"Total: {total:F2} USD");
````

## Question 4: Check packing and qualification



````csharp
// ### Exercise: Check packing and qualification
// For 12 cases and capacity five, print fullBoxes, leftoverCases, reconstructed quantity, and bool freeShipping for quantity >= 10. Expected: 2, 2, 12, True. These are full boxes and leftovers, not the total boxes required.
// Your code starts here.

// Your code ends here.
````

````csharp
using System.Globalization;
CultureInfo.CurrentCulture = CultureInfo.InvariantCulture;

// ### Solution
int quantity = 12;
const int casesPerBox = 5;
int fullBoxes = quantity / casesPerBox;
int leftoverCases = quantity % casesPerBox;
bool freeShipping = quantity >= 10;
Console.WriteLine($"Full boxes: {fullBoxes}");
Console.WriteLine($"Leftover cases: {leftoverCases}");
Console.WriteLine($"Reconstructed quantity: {fullBoxes * casesPerBox + leftoverCases}");
Console.WriteLine($"Free shipping: {freeShipping}");
````

## Question 5: Deliver the discounted invoice



````csharp
// ### Exercise: Deliver the discounted invoice
// Return to the fixed 12-case order. Calculate subtotal, rounded 5% discount, discounted amount, rounded 7% tax on that amount, and total. Use -= and += for amount updates. Print customer, item, and aligned amount rows with USD indicated. Expected discount 23.10, tax 30.72, total 469.62.
// Your code starts here.

// Your code ends here.
````

````csharp
using System.Globalization;
CultureInfo.CurrentCulture = CultureInfo.InvariantCulture;

// ### Solution
string customer = "Riverside Dental";
string item = "Printer paper (case)";
int quantity = 12;
decimal unitPrice = 38.50m;
decimal subtotal = quantity * unitPrice;
decimal discount = Math.Round(subtotal * 0.05m, 2, MidpointRounding.AwayFromZero);
decimal discounted = subtotal;
discounted -= discount;
decimal tax = Math.Round(discounted * 0.07m, 2, MidpointRounding.AwayFromZero);
decimal amountDue = discounted;
amountDue += tax;
Console.WriteLine($"Invoice for {customer}");
Console.WriteLine($"{quantity} x {item}");
Console.WriteLine($"{"Amount (USD)",-14} | {"Value",10}");
Console.WriteLine($"{"Subtotal",-14} | {subtotal,10:F2}");
Console.WriteLine($"{"Discount",-14} | {-discount,10:F2}");
Console.WriteLine($"{"Tax",-14} | {tax,10:F2}");
Console.WriteLine($"{"Total",-14} | {amountDue,10:F2}");
````
