# Searching and Basic Sorting

Finding a product and preparing a report are different tasks, even when they use the same records. These two sections connect first-match search contracts, sorted-input requirements, and sorting policies to complete query and reporting workloads. You will account for preparation and updates as well as lookup cost, and preserve the original records when a report requires a separate ordered view.

- **Search contract**: Specify the required match and the meaning of a missing result.
- **Sorted input**: Binary search depends on compatible ordering and indexed access.
- **Stable sorting**: Preserve arrival order among records with equivalent keys.
- **Mutation policy**: Distinguish changing a collection from preparing a report snapshot.
- **Complete workload**: Include preparation, queries, and updates in the cost comparison.

<h2>Learning Objectives</h2>


By the end of this chapter, you will be able to:

1. Specify first-match search and missing-result behavior, including the meaning of a library search result.
2. Trace linear search and binary search while enforcing sorted-input and comparer requirements.
3. Implement insertion sorting and verify order, occurrence preservation, and behavior at boundaries.
4. Prepare a report with explicit mutation and tie-order policies, including a separate snapshot when needed.
5. Compare complete search/report workloads, accounting for preparation, repeated queries, updates, and storage.
