<h2>Glossary</h2>

| Term | Meaning |
| --- | --- |
| Search contract | The promised kind of match and missing result. |
| Linear search | Check candidates in sequence. |
| Binary search | Halve a sorted random-access candidate interval. |
| Stable sort | Preserve input order among equivalent keys. |
| Insertion sort | Insert each new value into an ordered prefix. |
| Workload | The mix of queries, updates, and required reports. |
