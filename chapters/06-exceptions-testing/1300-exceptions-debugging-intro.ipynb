# Exceptions & Testing

Programs fail in many ways: invalid input, missing files, incorrect assumptions, and state bugs that only appear in certain cases. This chapter gives you tools for preventing, diagnosing, and testing those failures.

- **Input validation**: Checking values before risky operations prevents many avoidable failures.
- **Exception handling**: `try`, `catch`, and related constructs let a program respond intentionally to runtime errors.
- **Debugging**: Breakpoints, watches, and step-by-step execution help you inspect what a program is actually doing.
- **Unit testing**: Small repeatable tests build confidence that code keeps working as it changes.
- **Nullability**: Nullable annotations and checks make missing values visible in the type system.

As programs grow, correctness depends less on hoping everything works and more on building habits that make errors visible and manageable.
