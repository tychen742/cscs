# Exceptions & Testing

Programs fail in many ways: invalid input, missing files, incorrect assumptions, and state bugs that only appear in certain cases. This chapter gives you tools for preventing, diagnosing, and testing those failures.

- **Input validation**: Checking values before risky operations prevents many avoidable failures.
- **Exception handling**: `try`, `catch`, and related constructs let a program respond intentionally to runtime errors.
- **Debugging**: Breakpoints, watches, and step-by-step execution help you inspect what a program is actually doing.
- **Unit testing**: Small repeatable tests build confidence that code keeps working as it changes.
- **Nullability**: Nullable annotations and checks make missing values visible in the type system.

As programs grow, correctness depends less on hoping everything works and more on building habits that make errors visible and manageable.


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Exception | A runtime event that interrupts normal program flow. |
| 2 | Runtime error | An error that appears while the program is running. |
| 3 | Try block | A block containing code that might throw an exception. |
| 4 | Catch block | A block that handles a specific exception. |
| 5 | Finally block | A block that runs after `try` and `catch`, whether or not an exception occurred. |
| 6 | Throw | To signal an exception from code. |
| 7 | Validation | Checking data before using it in an operation that has requirements. |
| 8 | Debugger | A tool for pausing and inspecting a running program. |
| 9 | Breakpoint | A marked line where the debugger pauses execution. |
| 10 | Step over | A debugging action that runs the current line without entering called methods. |
| 11 | Unit test | A small automated check for one behavior. |
| 12 | Test case | One input and expected-result scenario. |
| 13 | Nullable type | A type annotation that represents whether a value may be missing. |
| 14 | Null check | A condition that verifies a reference is not `null` before use. |
