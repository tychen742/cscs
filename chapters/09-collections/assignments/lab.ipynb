```csharp
using System;
using System.Collections.Generic;

var responses = new Dictionary<string, string>
{
    ["refund"] = "Refunds are processed within 5 business days.",
    ["slow"] = "Try closing other programs and restarting the app.",
    ["crash"] = "Please update to the latest version and try again.",
    ["invoice"] = "You can download invoices from the Billing page.",
    ["password"] = "Use the Forgot Password link on the sign-in page."
};
var messages = new List<string>
{
    "I want a refund, please.",
    "The dashboard is slow.",
    "Hi, I have a question.",
    "Why is the report so slow?",
    "Please send my invoice.",
    "Another crash this morning!",
    "Still slow after a restart.",
    "Nothing works!",
    "Can I get a refund for March?"
};

var demand = new Dictionary<string, int>();
int unmatched = 0;

foreach (string message in messages)
{
    string keyword = FindKeyword(message, responses);
    if (keyword == "")
    {
        unmatched++;
    }
    else if (demand.ContainsKey(keyword))
    {
        demand[keyword]++;
    }
    else
    {
        demand[keyword] = 1;
    }
}

var keywords = new List<string>(demand.Keys);
keywords.Sort();

Console.WriteLine("Keyword demand:");
foreach (string keyword in keywords)
{
    Console.WriteLine($"{keyword}: {demand[keyword]}");
}
Console.WriteLine($"unmatched: {unmatched}");

static string FindKeyword(string message, Dictionary<string, string> responses)
{
    char[] separators = " ,.!?".ToCharArray();
    string[] words = message.ToLower().Split(separators);

    foreach (string word in words)
    {
        if (responses.ContainsKey(word))
        {
            return word;
        }
    }

    return "";
}
```
