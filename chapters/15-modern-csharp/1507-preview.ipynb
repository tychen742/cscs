# Preview Questions

Complete these questions **before** class to prepare for the chapter.
Use the dropdowns to check your answers.

## True or False

**1.** Two `record` instances with identical property values are equal (`==`).

```{dropdown} Answer
**True** — `record` uses value-based equality by default.
```

**2.** A `with` expression mutates the original record.

```{dropdown} Answer
**False** — `with` creates a *copy* with selected properties changed; the original is unchanged.
```

**3.** `x ?? y` returns `y` only when `x` is `null`.

```{dropdown} Answer
**True** — The null-coalescing operator returns the right-hand side only if the left is `null`.
```

**4.** Pattern matching with `switch` expressions was introduced in C# 8.

```{dropdown} Answer
**True** — `switch` expressions and full pattern matching arrived in C# 8.
```

**5.** Enabling nullable reference types makes `string` and `string?` behave identically.

```{dropdown} Answer
**False** — With `#nullable enable`, `string` is non-nullable; `string?` signals that null is allowed and the compiler warns on unsafe dereferences.
```

**6.** A generic type parameter `T` in `List<T>` can be *any* type — value type or reference type — unless a constraint limits it.

```{dropdown} Answer
**True** — Without a `where` constraint, `T` is unconstrained and accepts any type.
```

**7.** An `async` method that returns `Task` begins executing on a background thread as soon as it is called.

```{dropdown} Answer
**False** — `async` methods run synchronously until the first `await` that actually yields. No extra thread is created automatically.
```


## Multiple Choice

**1.** What does a positional `record` provide automatically?

a) Only value equality  
b) Value equality, `ToString`, deconstruction, and `with` expressions  
c) Inheritance and polymorphism  
d) Thread safety

```{dropdown} Answer
**b)** Positional records get all four for free.
```

**2.** What does `order?.ShippingAddress?.City` return if `order` is `null`?

a) Throws `NullReferenceException`  
b) Returns `""`  
c) Returns `null`  
d) Compile error

```{dropdown} Answer
**c)** The null-conditional `?.` short-circuits and returns `null` without throwing.
```

**3.** Which C# keyword tests a value's type and binds it to a new variable?

a) `as`  
b) `is`  
c) `typeof`  
d) `cast`

```{dropdown} Answer
**b) `is`** — `if (obj is string s)` tests and binds in one step.
```

**4.** When should you prefer `record` over `class`?

a) When the type needs mutable state and complex behaviour  
b) When the type is primarily data, identity does not matter, and immutability is preferred  
c) When you need runtime polymorphism  
d) When performance is critical and the object is large

```{dropdown} Answer
**b)** Records shine for immutable, data-centric types.
```

**5.** What does `input ??= "default"` do?

a) Always assigns `"default"` to `input`  
b) Assigns `"default"` to `input` only if `input` is currently `null`  
c) Returns `"default"` without modifying `input`  
d) Compile error

```{dropdown} Answer
**b)** `??=` is null-coalescing assignment — assigns only when the variable is null.
```

**6.** What is the primary benefit of writing `static T Max<T>(T a, T b) where T : IComparable<T>`?

a) It avoids boxing/unboxing  
b) It works for any comparable type without writing separate overloads  
c) It prevents null arguments  
d) It compiles faster

```{dropdown} Answer
**b)** A single generic method replaces multiple type-specific overloads while remaining type-safe.
```

**7.** What return type should an `async` method use when it produces a value of type `string`?

a) `string`  
b) `void`  
c) `Task<string>`  
d) `async<string>`

```{dropdown} Answer
**c)** `Task<string>` lets callers `await` the method and receive the `string` result.
```
