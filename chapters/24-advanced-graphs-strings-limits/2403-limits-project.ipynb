## Growth Limits

Some exact algorithms become impractical because the number of possibilities grows too quickly.


````csharp
using System;

int[] inputSizes = { 10, 20, 30, 40 };

foreach (int size in inputSizes)
{
    double linear = size;
    double quadratic = size * size;
    double exponential = Math.Pow(2, size);
    Console.WriteLine($"n={size,2}  n={linear,6:N0}  n^2={quadratic,8:N0}  2^n={exponential,15:N0}");
}
````

The table is not a benchmark. It is a warning sign: exponential search spaces get large quickly.

## Exact, Pruned, or Approximate

When exact search is too expensive, common responses include pruning, dynamic programming, greedy heuristics, approximation, or changing the problem requirements.


````csharp
using System;

string requirement = "good route quickly";

if (requirement.Contains("exact"))
{
    Console.WriteLine("Expect higher cost or stronger input limits.");
}
else if (requirement.Contains("quickly"))
{
    Console.WriteLine("A heuristic or approximation may be appropriate.");
}
else
{
    Console.WriteLine("Clarify the quality and runtime requirements.");
}
````

Algorithm design is also requirements design. A vague request for "best" may need to become a concrete request for exact, approximate, fast, explainable, or robust.

## A Small Comparison Example

The code below compares reachability by BFS and DFS on the same graph. Both answer reachability, but their exploration order differs.


````csharp
using System;
using System.Collections.Generic;

var graph = new Dictionary<string, string[]>
{
    ["A"] = new[] { "B", "C" },
    ["B"] = new[] { "D" },
    ["C"] = new[] { "E" },
    ["D"] = Array.Empty<string>(),
    ["E"] = Array.Empty<string>()
};

Console.WriteLine($"BFS reaches E? {BfsReachable("A", "E")}");
Console.WriteLine($"DFS reaches E? {DfsReachable("A", "E", new HashSet<string>())}");

bool BfsReachable(string start, string target)
{
    var queue = new Queue<string>();
    var visited = new HashSet<string> { start };
    queue.Enqueue(start);

    while (queue.Count > 0)
    {
        string current = queue.Dequeue();
        if (current == target) return true;
        foreach (string neighbor in graph[current])
        {
            if (visited.Add(neighbor)) queue.Enqueue(neighbor);
        }
    }
    return false;
}

bool DfsReachable(string current, string target, HashSet<string> visited)
{
    if (!visited.Add(current)) return false;
    if (current == target) return true;

    foreach (string neighbor in graph[current])
    {
        if (DfsReachable(neighbor, target, visited)) return true;
    }
    return false;
}
````

A good project conclusion says when the algorithms agree, when one gives extra information, and what assumptions make the recommendation valid.
