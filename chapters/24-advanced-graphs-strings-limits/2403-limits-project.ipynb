# Algorithms and Computational Limits Project

Compare multiple algorithms on one data problem, document correctness and complexity, and explain which parts remain computationally difficult.

## Deliverable

- a problem statement and input model;
- two or more algorithms with tests;
- a correctness argument for each solution;
- time and extra-space analysis;
- measurements on increasing input sizes; and
- a discussion of approximation, pruning, or domain restrictions when exact computation becomes impractical.

```csharp
int[] inputSizes = { 10, 100, 1_000, 10_000 };

foreach (int size in inputSizes)
{
    long linearWork = size;
    long quadraticWork = (long)size * size;
    Console.WriteLine($"n={size}, linear={linearWork}, quadratic={quadraticWork}");
}
```
