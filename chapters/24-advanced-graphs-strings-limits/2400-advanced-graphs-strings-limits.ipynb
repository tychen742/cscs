# Advanced Algorithms

This chapter covers graph traversal, breadth-first search, depth-first search, path assumptions, string matching, prefix-aware matching ideas, computational limits, pruning, approximation, and final algorithm selection. The sidebar title **Advanced Algorithms** signals a synthesis chapter rather than a new catalog of unrelated topics.

- **Graph traversal**: Breadth-first and depth-first search explore graph structure in different orders.
- **Reachability**: A question about whether one state, vertex, or item can be reached from another.
- **String matching**: Searching for a pattern inside a larger text.
- **Computational limit**: A point where exact brute-force computation grows too quickly to be practical.
- **Approximation**: A strategy that seeks a good enough answer when exact optimization is too expensive.

The goal is not to memorize every advanced algorithm, but to recognize how problem structure shapes what efficient computation can do.


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Breadth-first search | A graph traversal that explores vertices in layers using a queue. |
| 2 | Depth-first search | A graph traversal that explores along one path before backtracking. |
| 3 | Reachability | Whether a target vertex or state can be reached from a starting point. |
| 4 | Traversal order | The order in which an algorithm visits vertices or states. |
| 5 | Shortest path | A path with minimum number of edges or minimum total cost. |
| 6 | String matching | Finding whether and where a pattern occurs in a text. |
| 7 | Pattern | The string or structure being searched for. |
| 8 | Text | The larger string being searched. |
| 9 | Prefix table | A table that records reusable information about matching prefixes in a pattern. |
| 10 | Intractable problem | A problem whose exact solutions become impractical as input size grows. |
| 11 | Approximation | A method that trades exactness for practical runtime or simpler implementation. |
| 12 | Limit | A constraint, such as time or memory, that shapes algorithm choice. |
