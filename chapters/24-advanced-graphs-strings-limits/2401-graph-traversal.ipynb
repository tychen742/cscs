# Graph Traversal

Breadth-first search (BFS) explores a graph in layers using a queue. Depth-first search (DFS) follows one path as far as possible before backtracking.

Both algorithms need a visited set to avoid processing the same vertex repeatedly. BFS is useful for unweighted shortest paths; DFS is useful for reachability, cycle reasoning, and exploring components.

```csharp
using System.Collections.Generic;

var graph = new Dictionary<string, string[]>
{
    ["A"] = new[] { "B", "C" },
    ["B"] = new[] { "A", "D" },
    ["C"] = new[] { "A", "D" },
    ["D"] = new[] { "B", "C" }
};

var queue = new Queue<string>();
var visited = new HashSet<string>();
queue.Enqueue("A");

while (queue.Count > 0)
{
    string current = queue.Dequeue();
    if (!visited.Add(current))
        continue;

    Console.WriteLine(current);
    foreach (string neighbor in graph[current])
        if (!visited.Contains(neighbor))
            queue.Enqueue(neighbor);
}
```
