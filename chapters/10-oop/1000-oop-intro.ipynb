# OOP Principles

Classes alone do not make a program object-oriented. This chapter names and illustrates the core object-oriented principles: encapsulation, inheritance, polymorphism, and abstraction, with concrete C# examples.

- **Encapsulation**: Objects protect their internal state and expose a controlled public surface.
- **Inheritance**: A derived class can reuse and specialize behavior from a base class.
- **Polymorphism**: Code can work with a general type while runtime objects provide specific behavior.
- **Abstraction**: Interfaces and abstract classes describe capabilities without committing to one implementation.
- **Composition**: Modern C# often favors combining small objects over building deep inheritance trees.

The goal is conceptual clarity and practical design judgment, especially knowing when interfaces and composition are cleaner than deep hierarchies.
