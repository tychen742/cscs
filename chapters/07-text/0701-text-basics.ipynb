# Text Basics

Before parsing text, clean it first.

## Common cleanup operations

```csharp
string raw = "   Alice,  93  ";

string trimmed = raw.Trim();              // "Alice,  93"
string upper = trimmed.ToUpper();         // "ALICE,  93"
string lower = trimmed.ToLower();         // "alice,  93"
bool empty = string.IsNullOrWhiteSpace(raw); // false
```

Use these rules:

- `Trim()` before splitting fields
- `IsNullOrWhiteSpace()` before parsing
- normalize case when comparisons should be case-insensitive

## Accessing characters safely

```csharp
string s = "cat";

for (int i = 0; i < s.Length; i++)
{
    Console.WriteLine($"{i}: {s[i]}");
}
```

Never index a string unless you know bounds are valid.

## String immutability

Strings are immutable in C#. Methods return new strings.

```csharp
string a = "hello";
string b = a.ToUpper();  // b is "HELLO", a is still "hello"
```

## Mini practice

1. Read a line from the user.
2. Print original, trimmed, upper, and lower versions.
3. If the trimmed result is empty, print a warning and stop.
