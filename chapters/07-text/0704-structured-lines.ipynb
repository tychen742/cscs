# Structured Lines

Structured lines are text records with predictable field order.

Example record:

```text
Alice,98,CS
```

Field model:

1. name (`string`)
2. score (`int`)
3. major (`string`)

## Parsing pattern

```csharp
public static bool TryParseStudent(string line, out string name, out int score, out string major)
{
    name = "";
    major = "";
    score = 0;

    if (string.IsNullOrWhiteSpace(line)) return false;

    string[] parts = line.Split(',');
    if (parts.Length != 3) return false;

    name = parts[0].Trim();
    major = parts[2].Trim();

    return int.TryParse(parts[1].Trim(), out score);
}
```

## Using parsed records

```csharp
if (TryParseStudent(line, out string name, out int score, out string major))
{
    Console.WriteLine($"{name} ({major}) scored {score}");
}
else
{
    Console.WriteLine($"Skipped malformed line: {line}");
}
```

## Design guidance

- parse with `Try...` style methods
- validate field count and types
- skip bad rows with clear reporting
- avoid crashing the whole job for one bad line
