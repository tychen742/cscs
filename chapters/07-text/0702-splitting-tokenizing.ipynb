# Splitting and Tokenizing

Splitting turns one text line into meaningful fields.

## Basic `Split`

```csharp
string line = "Ada,98,CS";
string[] parts = line.Split(',');
// parts[0] = "Ada", parts[1] = "98", parts[2] = "CS"
```

## Trimming and empty-field handling

```csharp
string line = " Ada , 98 , CS ";
string[] parts = line.Split(',');

for (int i = 0; i < parts.Length; i++)
{
    parts[i] = parts[i].Trim();
}
```

When input may contain repeated delimiters, remove empties:

```csharp
string line = "Ada,,CS";
string[] parts = line.Split(',', StringSplitOptions.RemoveEmptyEntries);
```

## Token validation pattern

```csharp
if (parts.Length != 3)
{
    Console.WriteLine("Invalid record: expected 3 fields.");
    return;
}
```

Validate shape before parsing values.

## Mini practice

Given `"Name,Score,Major"` records:

1. split each line
2. trim each field
3. validate field count is exactly 3
4. parse score to `int` with `TryParse`
