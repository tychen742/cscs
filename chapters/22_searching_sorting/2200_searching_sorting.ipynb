# Search & Sort

This chapter covers linear search, binary search, search preconditions, swap operations, elementary sorting algorithms, comparison counts, correctness tests, and when to use library sorting. The sidebar title **Search & Sort** keeps the two algorithm families visible without crowding the navigation.

- **Linear search**: Checks items one by one until the target is found or the input ends.
- **Binary search**: Uses sorted order to discard half of the remaining search range at each step.
- **Elementary sort**: A simple sorting algorithm, such as selection sort, insertion sort, or bubble sort, that is easy to trace but usually quadratic.
- **Comparison count**: A useful operation count for algorithms that decide order by comparing values.
- **Data condition**: Some algorithms require data to be sorted or structured before they are valid.

Searching locates data. Sorting puts data into an order that makes later work easier, including faster searching.


> *[Algorithms and Data Structures Tutorial - Full Course for Beginners](https://youtube.com/watch?v=8hly31xKli0)* by freeCodeCamp.org.
>
> This video remains hosted on YouTube and is the property of its original creator or rightsholder; it is embedded here via YouTube's official player for educational reference only.


<h2>Learning Objectives</h2>

1. Specify first, any, and all-match search contracts with compatible equality and ordering policies.
2. Trace half-open binary search and derive lower/upper bounds for duplicate groups and insertion positions.
3. Explain selection, insertion, and optimized bubble sorts using invariants and occurrence preservation.
4. Derive exact selection comparisons and relate insertion shifts and adjacent swaps to inversions.
5. Verify stable record order, sorted permutations, and explicit mutation/library return policies.
6. Compare complete workloads including preparation, queries, input order, updates, and storage.


<h2>Chapter Flow</h2>

- {doc}`22.1. Searching <2201_searching>`
- {doc}`22.2. Sorting <2202_sorting>`
- {doc}`22.3. Searching and Sorting Comparison Lab <2203_comparison_lab>`


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Search | The task of finding whether and where a value appears. |
| 2 | Linear search | A search algorithm that checks items sequentially. |
| 3 | Binary search | A search algorithm for sorted data that repeatedly halves the remaining search range. |
| 4 | Sorted order | An arrangement where values follow a chosen ordering rule. |
| 5 | Comparison | A test that decides how two values relate in an order. |
| 6 | Swap | An operation that exchanges values at two positions. |
| 7 | Selection sort | A sorting algorithm that repeatedly selects the smallest remaining value and moves it into place. |
| 8 | Insertion sort | A sorting algorithm that grows a sorted prefix by inserting each new item into its correct position. |
| 9 | Bubble sort | A sorting algorithm that repeatedly swaps adjacent out-of-order values. |
| 10 | Stable sort | A sort that preserves the relative order of items with equal comparison keys. |
| 11 | In-place sort | A sort that uses only small extra storage beyond the input array. |

| 12 | Any-match search | Returns a matching index without guaranteeing the first duplicate. |
| 13 | Lower bound | First sorted position with a key at least the target, or the length. |
| 14 | Upper bound | First sorted position with a key greater than the target, or the length. |
| 15 | Probe | One selected equality/comparer call in a search. |
| 16 | Inversion | A pair of positions whose earlier key is greater than the later key. |
| 17 | Shift | An insertion-sort move of a greater predecessor one position right. |
| 18 | Sorted permutation | Ordered output containing every input occurrence with the same multiplicity. |
| 19 | Preparation cost | Work to construct a search representation before answering queries. |
| 20 | Sorted snapshot | An ordered view that may need rebuilding or maintenance after updates. |


<h2>Slides</h2>

Slides for this chapter will be added after the notebook content stabilizes.
