```csharp
using System;

static double CalcWeeklyPay(double hours, double rate)
{
    if (hours <= 40)
    {
        return hours * rate;
    }
    else
    {
        return 40 * rate + (hours - 40) * rate * 1.5;
    }
}

static double BonusPercent(double rating)
{
    if (rating >= 4.5) return 5;
    else if (rating >= 3.5) return 3;
    else if (rating >= 2.5) return 1;
    else return 0;
}

string employee = "Avery";
char department = 'P';
double hours = 46;
double rating = 4.6;

string departmentName = "";
double rate = 0;

switch (department)
{
    case 'P':
        departmentName = "Printing";
        rate = 20.00;
        break;
    case 'D':
        departmentName = "Delivery";
        rate = 18.50;
        break;
    case 'F':
        departmentName = "Front Desk";
        rate = 16.00;
        break;
    default:
        Console.WriteLine("Unknown department code.");
        break;
}

if (rate > 0)
{
    double pay = CalcWeeklyPay(hours, rate);
    double bonus = pay * BonusPercent(rating) / 100;
    Console.WriteLine($"Employee:   {employee}");
    Console.WriteLine($"Department: {departmentName} (${rate:F2}/hour)");
    Console.WriteLine($"Hours:      {hours}");
    Console.WriteLine($"Gross pay:  ${pay:F2}");
    Console.WriteLine($"Bonus:      ${bonus:F2}");
    Console.WriteLine($"Total pay:  ${pay + bonus:F2}");
}
```
