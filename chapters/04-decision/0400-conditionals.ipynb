<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Conditional statement | A statement that chooses which code to run based on a Boolean condition. |
| 2 | Boolean expression | An expression that evaluates to `true` or `false`. |
| 3 | Comparison operator | An operator such as `==`, `<`, or `>=` that compares values. |
| 4 | Logical operator | An operator such as `&&`, `||`, or `!` that combines or changes Boolean values. |
| 5 | If statement | A conditional statement that runs a block when its condition is true. |
| 6 | Else branch | A branch that runs when previous conditions in the chain are false. |
| 7 | Else-if chain | A sequence of conditional tests that selects among multiple cases. |
| 8 | Nested conditional | A conditional statement placed inside another conditional statement. |
| 9 | Switch statement | A selection statement that chooses a branch based on matching cases. |
| 10 | Case label | A value or pattern used by a `switch` statement to select code. |
| 11 | Default case | The fallback branch of a `switch` statement. |
| 12 | Boundary condition | A value at the edge of a range where conditional mistakes often occur. |
