# Recursion

Recursion is a technique where a method calls itself with a simpler input until a **base case** is reached.
Every recursive method needs:

1. A **base case** — stops the recursion and returns a value directly.
2. A **recursive step** — calls itself with a smaller/simpler argument.

Without a base case the program crashes with `StackOverflowException`.

## Factorial

`n! = n × (n-1) × ⋯ × 1`, with `0! = 1`.
Recursive: `factorial(n) = n * factorial(n-1)`, base case `n == 0`.

```python
static long Factorial(int n)
{
    if (n == 0) return 1;          // base case
    return n * Factorial(n - 1);  // recursive step
}

Console.WriteLine(Factorial(5));   // 120
Console.WriteLine(Factorial(0));   // 1
Console.WriteLine(Factorial(10));  // 3628800
```
