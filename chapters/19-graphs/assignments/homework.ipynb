- False

**Question 4: Duplicate direction**

In a simple undirected graph, adding A–B and then B–A should produce two logical edges.

- True
- False

```{dropdown} Answer
False. Those endpoint pairs describe the same logical edge. The representation may store two reciprocal entries, but the logical edge count is one.
```

- True
- False

```{dropdown} Answer
False. Neighbor maps enumerate stored entries in Θ(deg(u)) time; a matrix scans V cells. Hash lookup costs require their own assumptions.
```
