````csharp
// Solution
using System;
using System.Collections.Generic;
foreach (string text in new[] { "([])", "([)]", ")(", "((", "" })
    Console.WriteLine($"{text}: {IsBalanced(text)}");
static bool IsBalanced(string text)
{
    var openings = new Stack<char>();
    foreach (char ch in text)
    {
        if (ch == '(' || ch == '[' || ch == '{') openings.Push(ch);
        else if (ch == ')' || ch == ']' || ch == '}')
        {
            if (!openings.TryPop(out char open)) return false;
            bool matches = (open == '(' && ch == ')')
                        || (open == '[' && ch == ']')
                        || (open == '{' && ch == '}');
            if (!matches) return false;
        }
    }
    return openings.Count == 0;
}
````

````csharp
// Solution
using System;
using System.Collections.Generic;
var urgent = new Queue<string>(new[] { "U1", "U2", "U3" });
var normal = new Queue<string>(new[] { "N1", "N2" });
int urgentSinceNormal = 0;
while (urgent.Count > 0 || normal.Count > 0)
{
    if (normal.Count > 0 && (urgentSinceNormal == 2 || urgent.Count == 0))
    {
        Console.WriteLine(normal.Dequeue()); urgentSinceNormal = 0;
    }
    else
    {
        Console.WriteLine(urgent.Dequeue());
        urgentSinceNormal = Math.Min(2, urgentSinceNormal + 1);
    }
}
````

```{dropdown} Scheduling reasoning
At most two urgent tickets are served between normal services while normal work is waiting. A normal ticket with r normal tickets ahead of it is therefore bypassed by at most `2(r + 1)` urgent completions before it is served, under this sequential policy. New normal arrivals join behind it. This count does not bound elapsed time when service durations vary or a service never finishes.
```
