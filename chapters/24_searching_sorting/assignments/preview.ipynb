## Multiple Choice

```{index} Multiple Choice
```

**1.** What does first-match linear search return on absence?

a) −1  
b) Length for all implementations  
c) Zero  
d) The last index  

**2.** What must binary search’s ordering agree with?

a) The order used to sort the input  
b) The ID’s printed color  
c) The original arrival order only  
d) The variable name  

**3.** Which interface locates all equal sorted keys?

a) Lower and upper bounds  
b) Any-match binary search alone  
c) A swap  
d) Selection’s smallest index  

**4.** What may a lower bound return above every key?

a) Length  
b) Only −1  
c) Always zero  
d) An invalid midpoint  

**5.** Which is selection’s exact key comparison count?

a) n(n−1)/2  
b) n log₂ n  
c) n−1 always  
d) 2n  

**6.** Which items does stable insertion shift?

a) Strictly greater predecessors  
b) Equal predecessors first  
c) Every earlier item  
d) Only smaller predecessors  

**7.** What allows optimized bubble to stop early?

a) A pass with no swaps  
b) One successful swap  
c) A duplicate key  
d) An empty print buffer  

**8.** What does stability preserve?

a) Arrival order within equal-key groups  
b) Every item at its original index  
c) Only numeric values  
d) The input size alone  

**9.** Why clone data before comparing in-place sorts?

a) Give each algorithm the same initial order  
b) Make them asymptotically faster  
c) Avoid checking correctness  
d) Guarantee stable selection  

**10.** What belongs in a sort-then-search comparison?

a) Preparation, queries, and relevant update/storage costs  
b) Queries only  
c) One stopwatch value without inputs  
d) A universal item-count cutoff
