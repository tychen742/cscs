# Lambda Expressions

Lambda expressions are compact ways to define behavior inline.

## Basic syntax

```csharp
x => x * x
```

Left side is parameter(s). Right side is expression or statement block.

## Expression vs statement lambda

Expression lambda:

```csharp
Func<int, int> square = x => x * x;
```

Statement lambda:

```csharp
Func<int, int, int> max = (a, b) =>
{
    if (a > b) return a;
    return b;
};
```

## `Func<>` and `Action<>`

- `Func<...>` returns a value.
- `Action<...>` returns `void`.

```csharp
Func<string, string> normalize = s => s.Trim().ToLower();
Action<string> print = msg => Console.WriteLine(msg);
```

## Common uses

- pass comparison rules to sorting
- define predicates for filtering
- transform values without creating one-off named methods

## Readability guidance

- If logic is short and local, lambda is good.
- If logic is long or reused, use a named method.

## Mini practice

1. Create a `Func<int, bool>` that checks even numbers.
2. Create an `Action<string>` that prints uppercase text.
3. Rewrite a small helper method as a lambda, then decide which is clearer.
