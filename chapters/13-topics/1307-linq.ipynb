# LINQ

LINQ (Language Integrated Query) lets you query collections with a readable pipeline style.

## Why LINQ

LINQ helps you express:

- filtering
- projection (shape changes)
- ordering
- aggregation

with fewer mutable variables and less loop boilerplate.

## Core operators

### `Where` (filter)

```csharp
var passing = scores.Where(s => s >= 60);
```

### `Select` (transform)

```csharp
var labels = scores.Select(s => $"Score={s}");
```

### `OrderBy` / `OrderByDescending`

```csharp
var sorted = scores.OrderByDescending(s => s);
```

### Aggregation

```csharp
bool anyFailing = scores.Any(s => s < 60);
int countPassing = scores.Count(s => s >= 60);
int total = scores.Sum();
```

## End-to-end example

```csharp
List<string> lines = new() { "Ada,98", "Bob,87", "Kai,72" };

var topStudents =
    lines
    .Select(line => line.Split(','))
    .Where(parts => parts.Length == 2 && int.TryParse(parts[1], out _))
    .Select(parts => new { Name = parts[0], Score = int.Parse(parts[1]) })
    .Where(x => x.Score >= 90)
    .OrderByDescending(x => x.Score);
```

## When to use loops instead

Prefer explicit loops when:

- side effects are central
- branching is complex
- debugging each step is easier with explicit mutation
