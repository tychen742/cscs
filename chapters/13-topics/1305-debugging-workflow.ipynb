# Debugging Workflow

Debugging is a repeatable process, not guesswork.

## A practical workflow

Use this sequence every time:

1. Reproduce reliably.
2. Minimize the failing case.
3. Observe state with tools.
4. Form a specific hypothesis.
5. Patch one thing.
6. Re-test and check regressions.

## 1) Reproduce reliably

Capture:

- exact input
- exact command
- expected behavior
- actual behavior

Without reproducibility, debugging quality collapses.

## 2) Breakpoints and stepping

In VS Code:

- set a breakpoint before the suspected bug
- use Step Over for linear flow
- use Step Into for method internals
- use Step Out when you have enough detail

## 3) Watch and call stack

Inspect:

- key variables before and after transformations
- call stack to confirm control flow assumptions
- branch conditions (`if`, loop conditions) at decision points

## 4) Typical bug patterns

- off-by-one index errors
- stale state reused across loop iterations
- incorrect assumptions about null/empty input
- parsing without validation

## 5) Validate the fix

After patching:

- re-run original failing case
- run nearby edge cases
- run unit tests

## Mini practice

Take a method that parses a CSV line. Add one intentional bug, then debug it using:

- breakpoint at parse start
- watch on `parts.Length`
- step into `TryParse`
