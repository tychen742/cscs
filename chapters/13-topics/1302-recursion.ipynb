# Recursion

Recursion is a technique where a method calls itself with a simpler input until a base case is reached.

## Recursive GCD

Euclid's definition maps naturally to recursion:

- `gcd(a, 0) = a` (base case)
- `gcd(a, b) = gcd(b, a % b)` (recursive step)



```
/// Return the greatest common divisor of nonnegative numbers, not both 0.
public static int GreatestCommonDivisor(int a, int b)
{
    if (b == 0)
    {
        return a; // base case
    }

    return GreatestCommonDivisor(b, a % b); // recursive step
}

```

Why this works:

- Each call reduces the second argument from `b` to `a % b`.
- The second argument eventually reaches `0`.
- At that point, the base case returns the result directly.
