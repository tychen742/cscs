# Priority Queues and Heaps

A **priority queue** removes the item with the most important priority, not necessarily the item that arrived first. A heap is a common representation because it keeps the next item near the root without fully sorting every item.

For a min-priority queue, lower priority values are removed first. The heap operations restore the heap invariant after insertion or removal.

- insertion adds an item and moves it toward the root when necessary;
- removal takes the root and repairs the remaining heap;
- peeking reads the root without removing it.

The abstraction is useful for scheduling, event simulation, and selecting the next task.

```csharp
var tasks = new PriorityQueue<string, int>();
tasks.Enqueue("Process invoice", 2);
tasks.Enqueue("Respond to outage", 0);
tasks.Enqueue("Send summary", 3);

while (tasks.Count > 0)
{
    string task = tasks.Dequeue();
    Console.WriteLine(task);
}
```
