# Hash Functions and Collisions

A **hash table** uses a hash function to map a key to a bucket. The goal is to support fast lookup, insertion, and removal by key.

A **collision** occurs when two different keys map to the same bucket. Common strategies include separate chaining and open addressing. Performance depends on the hash function, table capacity, and **load factor**.

Dictionaries and sets expose hash-table behavior through an interface. Users should choose keys with stable equality and hash behavior.

```csharp
var customersById = new Dictionary<int, string>
{
    [1042] = "Amina",
    [1077] = "Diego",
    [1120] = "Mina"
};

int lookupId = 1077;
if (customersById.TryGetValue(lookupId, out string? customer))
    Console.WriteLine($"Customer {lookupId}: {customer}");
else
    Console.WriteLine("Customer not found");

Console.WriteLine($"Hash code for {lookupId}: {lookupId.GetHashCode()}");
```
