# Performance Comparison Lab

Compare priority-queue and hash-table workloads using measurements and tests.

## Tasks

1. Use a priority queue to process tasks in priority order.
2. Use a dictionary to retrieve records by a stable key.
3. Construct a small collision example and explain how the table handles it.
4. Record operation counts or elapsed time for a meaningful workload.
5. Explain why a priority queue is not a replacement for a dictionary, and why a dictionary is not a replacement for a priority queue.

Focus on the operations the application needs rather than claiming that one structure is universally faster.

```csharp
using System.Diagnostics;

var stopwatch = Stopwatch.StartNew();
var records = new Dictionary<int, string>();
for (int id = 0; id < 10_000; id++)
    records[id] = $"Record {id}";

bool found = records.TryGetValue(9_999, out string? result);
stopwatch.Stop();

Console.WriteLine($"Found: {found}, value: {result}");
Console.WriteLine($"Elapsed ticks: {stopwatch.ElapsedTicks}");
```
