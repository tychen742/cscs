# Heaps and Hash Tables

Some structures are designed around priority or direct lookup rather than sequence order. This chapter studies heaps for priority queues and hash tables for dictionary-style access.

- **Priority queue**: A priority queue removes the item with the most important priority, not necessarily the oldest item.
- **Heap**: A heap stores priorities in a nearly complete tree so the next item stays near the root.
- **Hash table**: A hash table maps keys to array positions for fast average-case lookup.
- **Collision**: A collision happens when two different keys map to the same bucket.
- **Workload fit**: Heaps help when order changes by priority; hash tables help when lookup by key dominates.

The chapter's main theme is representation for speed: a small rule about shape or hashing can make an operation much cheaper for the workload that needs it.


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Priority queue | An ADT that removes items according to priority. |
| 2 | Priority | A value used to decide which item should come out next. |
| 3 | Heap | A nearly complete tree that maintains a parent-child priority rule. |
| 4 | Min-heap | A heap where each parent priority is less than or equal to its children. |
| 5 | Max-heap | A heap where each parent priority is greater than or equal to its children. |
| 6 | Complete tree | A tree filled level by level, left to right. |
| 7 | Bubble up | Move a newly inserted heap item upward until the heap property is restored. |
| 8 | Sink down | Move the root replacement downward until the heap property is restored. |
| 9 | Hash function | A function that converts a key into a hash code or table position. |
| 10 | Bucket | A slot or group of slots in a hash table. |
| 11 | Collision | A situation where different keys map to the same bucket. |
| 12 | Separate chaining | Collision handling that stores colliding entries in a secondary collection per bucket. |
| 13 | Open addressing | Collision handling that probes for another slot in the same table. |
| 14 | Load factor | A measure of how full a hash table is. |
| 15 | Dictionary | A key-value ADT supporting lookup by key. |
| 16 | Set | An ADT storing unique values. |
| 17 | HashSet | A C# collection for fast average-case membership tests. |
| 18 | Dictionary | A C# collection for fast average-case key-value lookup. |
