# Heaps and Hash Tables

Some data structures are designed around priority or direct lookup rather than sequence order. This chapter studies heaps for priority queues and hash tables for dictionary-style access.

- **Priority queue**: A priority queue removes the highest-priority item rather than the oldest or newest item.
- **Heap**: A heap stores priority-ordered data in an array-friendly tree shape.
- **Heap repair**: Insertion and removal restore heap order through local swaps.
- **Hash function**: A hash function maps a key to a table position.
- **Collision handling**: Hash tables need a strategy for keys that map to the same position.

Heaps and hash tables show how specialized structure can make important operations dramatically faster for the right workload.
