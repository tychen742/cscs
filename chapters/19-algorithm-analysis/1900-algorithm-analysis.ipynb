# Algorithm Analysis

An algorithm is a precise procedure for solving a problem. Algorithm analysis asks whether the procedure produces the required result and how its time and memory needs grow as the input grows.

- **Correctness**: Correctness asks whether an algorithm always produces the required result under its stated assumptions.
- **Specification**: Preconditions and postconditions state what an algorithm expects and promises.
- **Invariant**: An invariant is a fact that remains true through repeated algorithm steps.
- **Asymptotic analysis**: Growth-rate notation compares algorithms independently of a specific machine.
- **Recurrence**: A recurrence describes the cost of a recursive algorithm in terms of smaller inputs.

Later chapters use these tools to compare searching, sorting, graph, and optimization algorithms without relying on guesswork.
