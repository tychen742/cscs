# Analysis

This chapter covers algorithm correctness, specifications, preconditions, postconditions, loop invariants, operation counting, Big-O/Big-Theta notation, space growth, and recurrence relations. The sidebar title **Analysis** names the skill: reasoning about whether an algorithm works and how its cost grows.

- **Correctness**: A claim that an algorithm satisfies its specification for every valid input.
- **Complexity**: A description of how time or memory use grows as input size grows.
- **Big-O**: An upper-bound growth description that ignores constant factors and lower-order terms.
- **Invariant**: A fact that remains true before and after each loop iteration.
- **Recurrence**: An equation that describes the cost of a recursive algorithm in terms of smaller inputs.

Algorithm analysis asks two questions: does the procedure always produce the required result, and how do its time and memory needs grow as the input grows?


<h2>Glossary</h2>

```{glossary}
Input size
  A measure of how large the input is, usually written as `n`.

Growth rate
  A description of how work or memory changes as input size increases.

Big-O
  An asymptotic upper bound on growth.

Big-Theta
  An asymptotically tight growth description.

Precondition
  A condition that must be true before an algorithm runs.

Postcondition
  A condition that should be true after an algorithm finishes.

Loop invariant
  A fact that is true before and after every loop iteration.

Recurrence relation
  An equation that defines a recursive algorithm's cost using the cost of smaller inputs.
```


<h2>Slides</h2>

Slides for this chapter will be added after the notebook content stabilizes.
