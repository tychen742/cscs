# Big-O and Complexity

Complexity analysis estimates how runtime and memory grow as input size increases. Measured runtime depends on hardware and environment; asymptotic notation gives a machine-independent description of growth.

- `O(1)`: constant work
- `O(log n)`: repeatedly reduce the problem by a factor
- `O(n)`: one pass through the input
- `O(n log n)`: common in efficient comparison sorting
- `O(n^2)`: compare many pairs or use nested passes

Analyze the dominant growth term and state whether you are discussing time or extra space.

## Why complexity matters

Measured runtime depends on machine and environment. Complexity gives machine-independent
growth behavior.

## Big-O quick reference

- `O(1)`: constant work
- `O(n)`: linear work
- `O(n log n)`: typical efficient sorting
- `O(n^2)`: nested loops over same data

## Common patterns

Single pass:

```csharp
for (int i = 0; i < n; i++) { ... } // O(n)
```

Nested pass:

```csharp
for (int i = 0; i < n; i++)
    for (int j = 0; j < n; j++) { ... } // O(n^2)
```

Halving loop:

```csharp
while (n > 1) n /= 2; // O(log n)
```

## Space complexity

Track extra memory, not input storage:

- fixed temporaries: `O(1)`
- additional array of size `n`: `O(n)`

## Practical use

Use complexity to choose algorithms first, then benchmark with realistic data.
