# Big-O and Complexity

Complexity analysis estimates how runtime or memory grows as input size increases. A measured time such as `0.3 milliseconds` depends on the computer, the runtime, and the current workload. A growth rate such as `O(n)` describes the shape of the work in a machine-independent way.


```{index} Big-O notation
```

## Counting Operations

To analyze an algorithm, choose a basic operation and count how many times it can run as input size grows. The exact operation depends on the algorithm: comparisons, assignments, array accesses, hash lookups, or loop iterations may be useful counts.


````csharp
using System;

int n = 8;
int operations = 0;

for (int i = 0; i < n; i++)
{
    operations++;
}

Console.WriteLine($"n = {n}");
Console.WriteLine($"operations = {operations}");
````

The loop body runs once per item, so the count grows linearly with `n`. Doubling `n` roughly doubles the work.

## Common Growth Rates

The most common first-pass growth rates are constant, logarithmic, linear, linearithmic, quadratic, and exponential.


````csharp
using System;

int[] sizes = { 10, 100, 1000 };

foreach (int n in sizes)
{
    double log = Math.Log2(n);
    double nLogN = n * log;
    double square = n * n;

    Console.WriteLine($"n={n,4}  log2(n)={log,6:F1}  n log n={nLogN,8:F1}  n^2={square,9:F0}");
}
````

Growth-rate differences become more important as `n` grows. For small inputs, simple code with a worse growth rate may still be fast enough. For large inputs, growth usually dominates.

## Loop Patterns

A single loop over `n` items is usually linear. Nested loops over the same input are often quadratic. A loop that repeatedly halves the input is logarithmic.


````csharp
using System;

int n = 6;
int singleLoop = 0;
int nestedLoop = 0;
int halvingLoop = 0;

for (int i = 0; i < n; i++)
{
    singleLoop++;
}

for (int i = 0; i < n; i++)
{
    for (int j = 0; j < n; j++)
    {
        nestedLoop++;
    }
}

for (int size = n; size > 1; size /= 2)
{
    halvingLoop++;
}

Console.WriteLine($"single loop: {singleLoop}");
Console.WriteLine($"nested loop: {nestedLoop}");
Console.WriteLine($"halving loop: {halvingLoop}");
````

The halving loop runs only a few times because each iteration discards about half of the remaining size. This pattern is the core reason binary search has logarithmic time.

## Big-O and Big-Theta

`O(f(n))` says the algorithm grows no faster than a constant multiple of `f(n)` after inputs are large enough. `Theta(f(n))` says the growth is tightly described by `f(n)` from above and below.

For an introductory course, use Big-O when you want a safe upper bound and Big-Theta when you want the actual growth class.


````csharp
using System;

int n = 10;
int exact = 3 * n * n + 5 * n + 20;
int dominant = n * n;

Console.WriteLine($"exact count estimate: {exact}");
Console.WriteLine($"dominant term n^2: {dominant}");
Console.WriteLine("3n^2 + 5n + 20 is O(n^2) and Theta(n^2).");
````

Constants and lower-order terms still affect real programs, but asymptotic notation focuses on how work scales. That makes it useful for choosing among algorithm families.

## Space Complexity

Space complexity tracks extra memory used by an algorithm, not the memory already occupied by the input. A few variables are `O(1)` extra space. A new array proportional to the input is `O(n)` extra space.


````csharp
using System;

int[] values = { 4, 1, 7, 3 };
int[] doubled = new int[values.Length];

for (int i = 0; i < values.Length; i++)
{
    doubled[i] = values[i] * 2;
}

Console.WriteLine(string.Join(", ", doubled));
Console.WriteLine("The new array uses O(n) extra space.");
````

Sometimes an algorithm trades space for time. A dictionary can use more memory than a list, but it may reduce lookup time dramatically.
