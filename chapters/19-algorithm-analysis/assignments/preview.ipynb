## Multiple Choice

**1.** What does complexity analysis describe?

a) How work or memory grows as input size grows  
b) Which IDE theme is easiest to read  
c) Whether a file exists on disk  
d) The exact runtime on every computer

**2.** What does a precondition describe?

a) A fact that must be true before an algorithm runs  
b) A fact that is checked only after the program exits  
c) The color of console output  
d) A random test case

**3.** Which loop pattern is usually `O(n^2)`?

a) One loop from `0` to `n`  
b) A loop that halves `n` each time  
c) Two nested loops over the same `n` items  
d) One dictionary lookup

**4.** What is a loop invariant?

a) A fact that remains true before and after each loop iteration  
b) A variable that cannot be declared  
c) A method that never returns  
d) A syntax rule for arrays

**5.** What does a recurrence relation describe?

a) The cost of a recursive algorithm in terms of smaller inputs  
b) The number of namespaces in a file  
c) The memory address of an object  
d) The return type of `Main`
