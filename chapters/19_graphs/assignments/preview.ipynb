
When should BFS mark a vertex as discovered?

a) When it enters the queue  
b) After every predecessor finishes  
c) Only when removed from the queue  
d) Only if it has no neighbors  

```{dropdown} Answer
**a)** Discovery-time marking prevents duplicate queue entries around cycles.
```
