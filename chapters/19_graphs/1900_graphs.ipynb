# Graphs

- **Modeling relationships**: Define vertices, direction, weights, and the simple-graph boundary.
- **Local and global structure**: Distinguish adjacency from connectivity, and walks from paths.
- **Storage choices**: Compare matrix cells with actual neighbor collections and operation costs.
- **Complete updates**: Preserve reciprocal weights, counts, and isolates during edge and vertex changes.
- **Representation verification**: Compare a route graph with a nullable matrix and validate supplied walks.

An office facilities model connects the representations to a business task. Traversal, shortest paths, and spanning trees remain in the later algorithm chapters.


> *[Graph Theory Algorithms](https://youtube.com/watch?v=09_LlHjoEiY)* by freeCodeCamp.org.
>
> This video remains hosted on YouTube and is the property of its original creator or rightsholder; it is embedded here via YouTube's official player for educational reference only.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:

1. Distinguish directed/undirected, weighted/unweighted, and simple/multiple-edge models.
2. Compute degrees and explain adjacency, paths, cycles, connectivity, and isolates.
3. Compare representation costs using V, E, and degree rather than a fixed density cutoff.
4. Implement edge replacement/removal and vertex removal while preserving graph invariants.
5. Preserve zero weights and isolated vertices when converting to a matrix.
6. Test graph updates and validate the cost of a supplied walk without claiming route discovery.
7. Trace BFS and DFS, reconstruct reachable paths, and identify undirected components.


<h2>Chapter Flow</h2>

- {doc}`19.1. Graph Vocabulary and Models <1901_graph_vocabulary>`
- {doc}`19.2. Adjacency Representations <1902_adjacency_representations>`
- {doc}`19.3. Graph Structures Lab <1903_graph_structures_lab>`
- {doc}`19.4. Graph Traversal <1904_graph_traversal>`


<h2>Glossary</h2>

| # | Term | Explanation |
| ---: | --- | --- |
| 1 | Graph | Vertices together with their modeled relationships. |
| 2 | Vertex | One item represented in the graph. |
| 3 | Edge | A relationship joining two endpoints. |
| 4 | Adjacent | Directly connected by an edge in the required direction. |
| 5 | Neighbor | A directly adjacent vertex. |
| 6 | Degree | Number of incident edges; distinct neighbors in the simple undirected model. |
| 7 | Indegree | Number of incoming arcs at a vertex. |
| 8 | Outdegree | Number of outgoing arcs at a vertex. |
| 9 | Directed graph | A graph with source-to-destination arcs. |
| 10 | Undirected graph | A graph whose edges represent two-way relationships. |
| 11 | Weighted graph | A graph with a value and stated units on each edge. |
| 12 | Unweighted graph | A graph whose stored edges indicate connection only. |
| 13 | Walk | A valid consecutive edge sequence that may repeat vertices. |
| 14 | Path | A walk with no repeated vertices under this chapter's convention. |
| 15 | Cycle | A closed walk with no repeated edges or vertices except its start/end. |
| 16 | Isolated vertex | A present vertex with no incident edges. |
| 17 | Connected component | A maximal connected group in an undirected graph. |
| 18 | Simple graph | A graph without loops or multiple edges for the same endpoint pair. |
| 19 | Self-loop | An edge from a vertex to itself. |
| 20 | Multigraph | A model permitting distinct edges with the same endpoints. |
| 21 | Adjacency matrix | A table indexed by vertex pairs. |
| 22 | Adjacency list | A vertex-to-neighbor collection representation. |
| 23 | Density | Actual edges divided by possible edges under the model. |
| 24 | Symmetry | Equal reciprocal matrix cells or undirected adjacency entries. |
| 25 | Defensive copy | A returned snapshot that callers can change without mutating stored state. || 26 | Breadth-first search | Discover vertices in nondecreasing edge-count distance. |
| 27 | Depth-first search | Finish exploration of a child before the next sibling. |
| 28 | Predecessor | The vertex that first discovers another vertex on a search path. |


<h2>Lecture Slides</h2>

Lecture slide decks for this chapter are pending.
