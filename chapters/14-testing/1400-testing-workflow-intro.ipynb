# Testing

<!-- chapter-intro -->

Writing code is only half the job. The other half is proving that it works and collaborating safely as changes accumulate.

This chapter pairs unit testing discipline with team workflow using version control.

<h2>Why these topics are paired</h2>

Testing and version control reinforce each other:

- tests protect behavior during refactors
- small commits make failures easier to isolate
- pull requests become stronger with tests attached
- a disciplined workflow lets teams move fast without breaking things

<h2>Learning goals</h2>

By the end of this chapter, you should be able to:

- write small, focused unit tests for method behavior
- run tests regularly while coding to catch regressions
- organize commits around logical units of change
- use a branch-and-review workflow without losing work
