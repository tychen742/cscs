# Generics, Nullability & Async

Modern C# programs need to stay type-safe, avoid null-reference bugs, and remain responsive while waiting on slow operations. This chapter covers generics, nullable reference types, and async/await.

- **Generics**: Generic methods and classes use type parameters such as `<T>` to reduce duplication safely.
- **Type constraint**: A constraint limits which types can be used with a generic definition.
- **Nullability**: Nullable annotations and operators make missing reference values explicit.
- **Async/await**: Asynchronous methods let programs wait for slow operations without blocking the current workflow.
- **Responsive I/O**: File, network, and database operations often benefit from async structure.

These features appear constantly in real C# libraries, so reading them well is as important as writing them from scratch.
