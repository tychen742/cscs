# Linked Nodes

A linked sequence stores each element in a **node**. A node contains a value and a reference to another node. Unlike an array, the nodes do not need to occupy adjacent memory locations.

Insertion and deletion can be efficient when the position is already known. Finding a position usually requires traversal from the first node.

The example below implements a small singly linked list so that the references are visible.

```csharp
public sealed class Node<T>
{
    public T Value { get; }
    public Node<T>? Next { get; set; }

    public Node(T value)
    {
        Value = value;
    }
}

Node<string> first = new Node<string>("first");
Node<string> third = new Node<string>("third");
first.Next = new Node<string>("second");
first.Next.Next = third;

for (Node<string>? current = first; current is not null; current = current.Next)
{
    Console.WriteLine(current.Value);
}
```
