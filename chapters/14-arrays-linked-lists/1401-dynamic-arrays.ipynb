# Dynamic Arrays

Study contiguous storage, capacity, resizing, indexing, and amortized insertion.

```csharp
public sealed class DynamicIntArray
{
    private int[] items = new int[2];
    public int Count { get; private set; }
    public int Capacity => items.Length;

    public void Add(int value)
    {
        if (Count == items.Length)
        {
            int[] larger = new int[items.Length * 2];
            Array.Copy(items, larger, items.Length);
            items = larger;
        }

        items[Count] = value;
        Count++;
    }

    public int this[int index]
    {
        get
        {
            if (index < 0 || index >= Count)
                throw new IndexOutOfRangeException();
            return items[index];
        }
    }
}

var numbers = new DynamicIntArray();
for (int value = 10; value <= 50; value += 10)
{
    numbers.Add(value);
    Console.WriteLine($"count={numbers.Count}, capacity={numbers.Capacity}");
}

Console.WriteLine(numbers[3]);
```
