# Representation Comparison Lab

Compare array-based and linked representations against realistic workloads.

## Tasks

1. Implement indexed lookup for an array-backed sequence.
2. Implement traversal and insertion for a linked sequence.
3. Record the number of element visits for each operation.
4. Explain which representation is better for frequent indexing, frequent insertion, and unknown collection size.
5. Write tests for empty, one-element, and multi-element sequences.

Your conclusion should describe the workload that led to your choice. A data structure is not universally better; its usefulness depends on the operations a program performs most often.

```csharp
int[] values = { 12, 7, 19, 4, 15 };

int visits = 0;
int target = 15;
for (int index = 0; index < values.Length; index++)
{
    visits++;
    if (values[index] == target)
    {
        Console.WriteLine($"Found {target} at index {index} after {visits} visits.");
        break;
    }
}

Console.WriteLine("Ask: how would the visit count change for a linked sequence?");
```
