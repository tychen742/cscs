# Linear Lists

This chapter covers array-backed lists, dynamic arrays, linked nodes, linked lists, and the operation-cost tradeoffs between contiguous and linked sequence representations. The sidebar title **Linear Lists** names the shared idea: data arranged in a single sequence.

- **Contiguous storage**: Array-backed sequences keep elements next to each other for fast indexed access.
- **Dynamic array**: A dynamic array grows by allocating larger storage and copying existing values.
- **Linked node**: A linked structure stores values in nodes connected by references.
- **Insertion and deletion**: Different representations make middle updates cheaper or more expensive.
- **Performance tradeoff**: Choosing a sequence representation depends on access, growth, and update patterns.

Arrays and linked lists make the connection between memory representation and program behavior visible.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Explain how array-based and linked representations store sequence data
2. Trace dynamic-array resizing and copying
3. Trace linked-node insertion, deletion, and traversal
4. Compare indexed access, search, insertion, and deletion costs
5. Choose a sequence representation for a specific workload
