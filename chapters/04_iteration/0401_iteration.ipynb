Iteration, or looping, is one of the three basic constructs in computer
programming (see {ref}`programming-constructs`). The basic purpose of iteration is code repetition, which
humans usually find difficult but comes very easy for computers. After
starting, the repetitive execution of a block of code or a set of
instructions/statements will not stop until certain criterion is met.
The termination of iteration can be either **count**-controlled (for loops)
or **condition**-controlled (while loops). In general, when you know
exactly how many times you are going to loop through a
block of code, you use the `for` loop; otherwise, you would choose a `while` loop.

The need for code repetition can be seen from the following code:



````csharp
Console.Write("1 ");
Console.Write("2 ");
Console.Write("3 ");
Console.Write("4 ");
Console.Write("5 ");
Console.Write("6 ");
Console.Write("7 ");
Console.Write("8 ");
Console.Write("9 ");
Console.Write("10 ");
Console.WriteLine();
````


Obviously, to code the print statement repetitively for 10 times is not a efficient use
of time for a developer, not to mention the repetition could go for more than 10 times.

## C# Iteration Statements




The C# language reference states that iteration
statements include `for`, `foreach`, `do`, and `while` and that the iteration
statements "repeatedly execute a statement or a block of statements": [^0401_iteration-iteration-1]

- The `for` statement [^0401_iteration-c-style-for-loop] executes its body of an iteration statement while a specified
  Boolean expression evaluates to true.
- The `foreach` statement enumerates the elements of a collection and executes
  its body for each element of the collection.
- The `do` statement conditionally executes its body one or more times.
- The `while` statement conditionally executes its body zero or more times.

As a solution to the preceding code repetition problem, a `for` loop can be used as below.
Comparing the highlighted for statement with the, you can see the need for iteration.
The syntax of the `for` loop statement will be explained in a later section.



````csharp
for (int i = 1; i <= 10; i = i + 1)
{
    Console.Write(i + " ");
}
Console.WriteLine();
````

Read the loop as a table. Each row is one pass: the condition is tested, the body prints, then the step runs and the next test uses the new value.

| Pass | `i` at the test | `i <= 10`? | Printed | `i` after the step |
| ---: | ---: | --- | --- | ---: |
| 1 | 1 | true | `1 ` | 2 |
| 2 | 2 | true | `2 ` | 3 |
| ... | ... | ... | ... | ... |
| 10 | 10 | true | `10 ` | 11 |
| 11 | 11 | false | nothing; the loop ends | |

The eleventh test fails and the body does not run, which is why the loop prints exactly ten numbers and `i` is 11 afterward. Every counting loop in this chapter can be traced with the same four columns.

## Iterator/Step



The iterator, also known as **step**, section can contain zero or more expressions, separated by commas.
Commonly used iterators include:

- prefix or postfix increment expression, such as ++i or i++
- prefix or postfix decrement expression, such as --i or i--
- assignment

The increment operator `++` adds one and `--` subtracts one. In `i++`, the expression produces the old value of `i`, but the stored variable is updated during that expression's evaluation. In `++i`, the expression produces the updated value. Neither form delays the variable's update until another iteration.

In a `for` loop, the update part is executed *after* the body. That location explains why the next body execution sees the new counter; it is separate from whether increment uses prefix or postfix syntax.

### Separate the Expression Result from Stored State


````csharp
using System;
int counter = 2;
int old = counter++;
Console.WriteLine($"postfix result={old}, counter={counter}");
int updated = ++counter;
Console.WriteLine($"prefix result={updated}, counter={counter}");
````

The output is `postfix result=2, counter=3`, then `prefix result=4, counter=4`. After the first assignment, two different values exist: `old` receives the expression result 2, while `counter` already holds 3. A common mistake is to conclude that both variables still hold 2 until the next loop iteration. There is no loop in this example, and the update is already observable.


````csharp
int num = 3;
Console.WriteLine(num);
num++;       // postfix expression, the value remains
Console.WriteLine(num);  // i is evaluated again. Now it shows 4.
````

Observe the code below to compare the behavior of postfix and prefix operations and you will see
the difference between the postfix expression and prefix expression:

````csharp
int i = 0;
Console.WriteLine($"Before: {i}");
Console.WriteLine($"Postfix value: {i++}");
Console.WriteLine($"After: {i}");
int j = 0;
Console.WriteLine($"Before: {j}");
Console.WriteLine($"Prefix value: {++j}");
Console.WriteLine($"After: {j}");
````


When setting a "**step**, you may also use assignment statement (see {ref}`assignment-operators`) to step, which give you
more flexibility in stepping. For example, in a small runnable C# example:


````csharp
int n = int.Parse(Console.ReadLine());
  for (var i = 0; i <= n; i += 10)
  {
      Console.WriteLine(i);
  }
````

## `break` and `continue`




`break` exits the nearest loop or switch. `continue` skips the rest of the current loop body. In a `for` loop, the update runs before the next condition check; in a `while` loop, the next condition is checked directly. Ensure an update is not skipped accidentally.

`return` exits a method and is taught with user-defined methods in Chapter 5. It is not needed for this chapter's programs.


Both statements change where control goes without changing the loop header. In the example, `continue` skips the closed ticket and `break` stops at the first ticket over the limit:


````csharp
using System;

int[] tickets = { 12, 0, 7, 45, 3 };
int reviewed = 0;
foreach (int minutes in tickets)
{
    if (minutes == 0) continue;        // closed ticket: skip the rest of this pass
    if (minutes > 30) break;           // over the limit: leave the loop entirely
    reviewed++;
    Console.WriteLine($"Reviewed a {minutes}-minute ticket");
}
Console.WriteLine($"Reviewed {reviewed} tickets");
````

The output is two `Reviewed` lines, for 12 and 7 minutes, and then `Reviewed 2 tickets`. The 0 was skipped by `continue`, and the 45 ended the loop before 3 was ever examined. A `continue` placed after the counter instead of before it would still count the skipped ticket; place the skip before the work it is meant to skip.

## Loop Planning Rubric



Before reading the checklist, see it applied. A store adds each day's sales to a running total and stops at the first day the total passes 1,000:


````csharp
using System;

int[] dailySales = { 320, 410, 150, 275, 500 };   // data
int total = 0;                                     // initialization
int day = 0;
while (day < dailySales.Length && total <= 1000)   // continuation condition
{
    total += dailySales[day];                      // repeated action
    day++;                                         // update
}
Console.WriteLine($"Passed 1,000 after day {day} with {total}");   // done once, after the loop
````

The output is `Passed 1,000 after day 4 with 1155`. Each comment names the rubric question it answers:

| Rubric question | Answer in this loop |
| --- | --- |
| Data and loop variables | `dailySales`, `total`, `day` |
| Initialization | `total = 0`, `day = 0` |
| Continuation condition | `day < dailySales.Length && total <= 1000` |
| Done once | Print the result after the loop |
| Repeated action | Add one day's sales |
| Update | `day++` |
| Will the condition eventually fail? | `day` grows by one each pass, so it reaches the array length even if the total never passes 1,000 |

The second half of the condition is the one students forget: without the `day < dailySales.Length` guard, a week that never reaches 1,000 reads past the end of the array.



Looking ahead to more complicated and interesting problems,
here is a list of questions to ask yourself when
designing a program with a loop:

1. Data & loop variables: What data is involved? Make sure you give good variable names.
2. Initialization: What needs to be initialized and how? This certainly includes any
   variable tested in the condition.
3. Condition: What is the condition that will allow the loop to *continue*? It
   may be easier to think of the condition that will *stop* the loop.
   That is fine - but remember to *negate* it (with `!`) to turn it
   into a proper *continuation* condition.
4. Distinguish: What is the code that should only be executed once?
   What action do I want to repeat?
5. Repetition: How do I write the repeating action so I can modify it for the next time
   through the loop to work with new data?
6. Update: What code is needed to do modifications to make the same action code work
   the next time through the loop?
7. Locale variables: Have I thought of variables needed in the middle and declared them;
   do other things need initialization?
8. Condition again: Will the continuation condition eventually fail?
   *Be sure to think about this!*
9. Structure: Separate the actions to be done once before the repetition (code before the
   loop) from repetitive actions (in the loop) and from actions not repeated, but
   done after the loop (code after the loop). Missing this distinction
   is a *common error*!

This is quite a sequence of steps. You should at least note a more basic split into two parts:

1. Writing the code to make sure each desired iteration is reached
   (loop dynamics: initialized data for the test condition, the test condition itself,
   preparing for the next iteration)
2. Doing the work for a specific item/iteration (mostly of the body of the loop)

You generally only need to concentrate on one of those at a time.
If you set up how to reach all of the repetitions, giving names for the variables
to be acted on, then you can separately just concentrate on dealing with the the action/code
to deal with the *current* situation.

### Apply the Loop Plan to a Closing-Time Countdown


A service desk sends reminders when 6, 4, and 2 minutes remain before closing. First plan the behavior without choosing C# punctuation:

| Planning question | Decision and reason |
| --- | --- |
| What changes? | `minutes`, the time remaining. |
| Initial state? | 6, the first reminder time. |
| Continue when? | `minutes >= 2`; zero must not produce a reminder. |
| Repeated action? | Print the current time remaining. |
| Update? | Subtract 2 after the reminder. |
| Why stop? | 6 → 4 → 2 → 0, and 0 fails the condition. |

The loop is count-controlled because the schedule is known. In `for (initialization; condition; update)`, the initialization runs once, the condition is checked before every body execution, and the update runs after each completed body. Section 4.2 develops this syntax in detail.

Guided repair: the starter prints 6, 4, **and no 2**. Change one comparison to implement the plan. Then predict the output if the initial value is 1.


````csharp
using System;
for (int minutes = 6; minutes > 2; minutes -= 2) // TODO: include the two-minute reminder.
    Console.WriteLine($"Closing in {minutes} minutes");
````

````csharp
using System;
// Solution: include the boundary, but do not include zero.
for (int minutes = 6; minutes >= 2; minutes -= 2)
    Console.WriteLine($"Closing in {minutes} minutes");
````

The solution prints `Closing in 6 minutes`, `Closing in 4 minutes`, and `Closing in 2 minutes`. Starting at 1 prints nothing because the condition fails before the body. The error was the continuation boundary, not the update. For a variation using `continue`, ensure it does not skip the only update in a `while` body; in a `for` loop the update part still runs after `continue`. Use the planning table to identify which state must change before testing again.
