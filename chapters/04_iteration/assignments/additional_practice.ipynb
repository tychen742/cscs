````csharp
using System;
Random r = new Random(17);
int total = 10, groupSize = 4, heads = 0;
for (int i = 0; i < total; i++)
{
    bool isHead = r.Next(0, 2) == 0;
    if (isHead) heads++;
    Console.Write(isHead ? "Heads" : "Tails");
    if ((i + 1) % groupSize == 0 || i + 1 == total) Console.WriteLine();
    else Console.Write(" ");
}
Console.WriteLine($"Heads: {heads}; Tails: {total - heads}; Total: {total}");
````

## Number Guessing Game

```{index} Number Guessing Game
```

Two players model the problem: one chooses a positive integer less than `big`, and the other repeatedly guesses. After an incorrect guess, the first player must truthfully say Higher or Lower. The computer takes the first role.

For a secret of 29, guesses 50, 25, 30, and 29 produce Lower, Higher, Lower, and a win. Each answer narrows the possible interval. With a secret of 68, guesses 60, 72, 66, and 68 win on guess **four**.

3. Read guesses until the correct value is found. Use `TryParse` to reject non-integers; handle exhausted input by ending the program.
4. Count valid guesses and print the count when the player wins. State whether out-of-range guesses count; use the same policy throughout.
5. Extend the program with an outer replay loop. Generate a fresh secret for each game and reset the counter; retain the same random generator.

During testing, assign a fixed secret first so you can exercise Higher, Lower, and Correct deliberately. Restore random selection only after those branches work.

````csharp
using System;
int secret = 68, count = 0, guess = 60;
while (true)
{
    count++;
    Console.WriteLine($"Guess {count}: {guess}");
    if (guess == secret) { Console.WriteLine($"Good job! You win on guess {count}!"); break; }
    Console.WriteLine(guess < secret ? "Higher!" : "Lower!");
    if (count == 1) guess = 72;
    else if (count == 2) guess = 66;
    else guess = 68;
}
````

## Sum from One to n


For `n = 5`, the successive sums are `1 + 2 = 3`, `3 + 3 = 6`, `6 + 4 = 10`, and `10 + 5 = 15`. The next integer and the partial sum both change, so name them `i` and `sum`.

Each iteration must remember `sum + i`. You could first save this as `newSum`, then assign `sum = newSum`; because the intermediate value is used once, `sum = sum + i` does both jobs. Increment `i` so the next iteration adds the next number.

Start `sum = 0` and `i = 1` so zero terms also work. Continue while `i <= n`. Before each iteration, `sum` contains the integers from one through `i - 1`. After the addition, it contains the integers through `i`.

## Loan Amortization Table

```{index} Loan Amortization Table
```

A fixed periodic payment covers interest first and reduces principal with the remainder. Interest for a period depends on the principal left after the previous payment. With principal 100, rate 0.10 per period, and payment 20: interest is 10, the amount owed is 110, and the new principal is 90. The next period's interest is 9, leaving principal 79 after payment.

Print payment number, beginning principal, interest, and actual payment. When principal plus interest is smaller than the regular payment, make that smaller final payment and stop. The rate is per **payment period**, not automatically an annual rate.

Use `decimal` for amounts and rate. Decimal fractions such as `0.10m` are representable exactly, but interest can still have more than two decimal places. For this exercise, round interest to cents with `Math.Round(principal * rate, 2, MidpointRounding.AwayFromZero)`. This explicit midpoint policy matches the worked schedules below; the two-argument overload rounds exact midpoints to even and can produce different balances. Keep the selected policy consistent.

If a positive balance's payment does not exceed the rounded interest, principal will not decrease. Report that the payment cannot pay off the loan; do not enter an endless loop. Require positive principal and payment and nonnegative rate.

For 100 at 10% with payment 20, there are eight payments; the last is 5.65. For 1000 at 5% with payment 196, there are seven; the last is 7.28. Try a payment of 46 in the second case and explain why it fails.

### Loan schedule

```{index} Loan schedule
```

Produce both schedules and test a payment that fails to reduce principal. Explain each variable update before writing the loop.

````csharp
// Exercise: Loan schedule
// Complete the task described above.
using System;
Console.WriteLine("Replace this line with your solution.");
````

````csharp
using System;
decimal principal = 100m, rate = 0.10m, payment = 20m;
int number = 1;
Console.WriteLine("Number Principal Interest Payment");
while (principal > 0m)
{
    decimal interest = Math.Round(principal * rate, 2, MidpointRounding.AwayFromZero);
    if (payment <= interest) { Console.WriteLine("Payment does not reduce principal."); break; }
    decimal actual = Math.Min(payment, principal + interest);
    Console.WriteLine($"{number,6} {principal,9:F2} {interest,8:F2} {actual,7:F2}");
    principal = principal + interest - actual;
    number++;
}
````
