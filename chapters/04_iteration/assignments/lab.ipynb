```csharp
using System;
// Sample input: ten | 4.5 | 10
int quantity = 0;
bool valid;
string? text;
do
{
    Console.Write("Quantity: ");
    text = Console.ReadLine();
    if (text == null) break;
    valid = int.TryParse(text, out quantity);
    if (!valid) Console.WriteLine("Enter a whole number.");
} while (!valid);
if (text == null) Console.WriteLine("No more input.");
else Console.WriteLine($"Quantity: {quantity}");
```

## 5. Build an Order Entry Console

Combine the validation patterns in one outer do-while loop. Read an item, quantity (1-50), and unit price in whole dollars (1-500). Accumulate line count and total, and repeat only when the answer is exactly lowercase y. Validate y/n responses with an inner loop. Null input stops without adding an incomplete line. Each complete line prints its item, quantity, price, and line total; finish with line count and order total.


```csharp
using System;

Console.WriteLine("Order Entry");

// Your code starts here.


// Your code ends here.
```

```csharp
using System;
// Sample input: Desk lamp | 0 | 3 | 25 | y | Paper | x | 10 | 4 | n
int lineNumber = 0;
int orderTotal = 0;
string? answer = "n";
do
{
    Console.Write($"Line {lineNumber + 1} item: ");
    string? item = Console.ReadLine();
    if (item == null) break;
    int quantity = 0;
    int unitPrice = 0;
    bool valid;
    string? text;
    do
    {
        Console.Write("Quantity (1-50): ");
        text = Console.ReadLine();
        if (text == null) break;
        valid = int.TryParse(text, out quantity) && quantity >= 1 && quantity <= 50;
        if (!valid) Console.WriteLine("Enter a whole number from 1 to 50.");
    } while (!valid);
    if (text == null) break;
    do
    {
        Console.Write("Unit price in dollars (1-500): ");
        text = Console.ReadLine();
        if (text == null) break;
        valid = int.TryParse(text, out unitPrice) && unitPrice >= 1 && unitPrice <= 500;
        if (!valid) Console.WriteLine("Enter a whole number from 1 to 500.");
    } while (!valid);
    if (text == null) break;
    int lineTotal = quantity * unitPrice;
    orderTotal += lineTotal;
    lineNumber++;
    Console.WriteLine($"  {item}: {quantity} x ${unitPrice} = ${lineTotal}");
    do
    {
        Console.Write("Add another line? (y/n) ");
        answer = Console.ReadLine();
        if (answer == null) break;
        if (answer != "y" && answer != "n") Console.WriteLine("Please answer y or n.");
    } while (answer != "y" && answer != "n");
} while (answer == "y");
Console.WriteLine($"Order complete: {lineNumber} lines, total ${orderTotal}");
```
