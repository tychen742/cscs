**Question 7: Count order marks**

Traverse the string and count uppercase Y marks, lowercase y marks, and all other characters. Print all three counts. Do not use arrays or LINQ. Check empty text, one Y, and text containing only other marks.


````csharp
// Exercise: Count order marks
// exercise-id: cscs-iteration-homework-coding-2
using System;

string marks = "Yyn?Y";
// Your code starts here.
// Implement the contract above and print the requested report.
// Your code ends here.
````

````csharp
using System;

string marks = "Yyn?Y";
int upper = 0, lower = 0, other = 0;
foreach (char mark in marks)
{
    if (mark == 'Y') upper++;
    else if (mark == 'y') lower++;
    else other++;
}
Console.WriteLine($"Upper Y: {upper}");
Console.WriteLine($"Lower y: {lower}");
Console.WriteLine($"Other: {other}");
````

**Question 8: Count down a balance**

Use while to subtract a fixed payment from a nonnegative balance until the balance is zero. On the last step pay only the remaining balance. Print each payment, final balance, and payment count. Use a positive payment and amounts at most 1000. Check balance zero, a balance below one payment, and an exact multiple.


````csharp
// Exercise: Count down a balance
// exercise-id: cscs-iteration-homework-coding-3
using System;

int balance = 11;
int payment = 4;
// Your code starts here.
// Implement the contract above and print the requested report.
// Your code ends here.
````

````csharp
using System;

int balance = 11;
int payment = 4;
int count = 0;
while (balance > 0)
{
    int actual = Math.Min(payment, balance);
    balance -= actual;
    count++;
    Console.WriteLine($"Payment: {actual}");
}
Console.WriteLine($"Balance: {balance}");
Console.WriteLine($"Payments: {count}");
````

**Question 9: Read a valid batch size**

Repeat a prompt until int.TryParse succeeds and quantity is from 1 through 5. On exhausted input print No more input and stop. For other invalid input print Retry. On success print Accepted: followed by the value. Test lines bad, 0, 5, then test exhausted input separately. No exception handling is needed.


````csharp
// Exercise: Read a valid batch size
// exercise-id: cscs-iteration-homework-coding-4
using System;

// Sample input lines: bad | 0 | 5
// Your code starts here.
// Implement the contract above and print the requested report.
// Your code ends here.
````

````csharp
using System;

// Sample input lines: bad | 0 | 5
int quantity = 0;
string? text;
bool valid;
do
{
    Console.WriteLine("Quantity (1-5):");
    text = Console.ReadLine();
    if (text == null) break;
    valid = int.TryParse(text, out quantity) && quantity >= 1 && quantity <= 5;
    if (!valid) Console.WriteLine("Retry");
} while (!valid);
if (text == null) Console.WriteLine("No more input");
else Console.WriteLine($"Accepted: {quantity}");
````

**Question 10: Stop at the closing mark**

Scan marks from left to right. Stop at # with break. Skip spaces with continue. Print each remaining character on its own line, then print Processed and its count. Check empty text, # as the first character, and input with no #.


````csharp
// Exercise: Stop at the closing mark
// exercise-id: cscs-iteration-homework-coding-5
using System;

string marks = "A B#C";
// Your code starts here.
// Implement the contract above and print the requested report.
// Your code ends here.
````

````csharp
using System;

string marks = "A B#C";
int processed = 0;
foreach (char mark in marks)
{
    if (mark == '#') break;
    if (mark == ' ') continue;
    Console.WriteLine(mark);
    processed++;
}
Console.WriteLine($"Processed: {processed}");
````
