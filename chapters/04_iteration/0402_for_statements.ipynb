# The `for` Statement




The `for` statement executes a statement or a block of statements
while a specified **Boolean expression** evaluates to true. Such evaluation involves
comparing the local loop counter variable (an **integer counter** that increments/decrements over every loop)
to a preset value. Since the number of loops is controlled by comparing the counter with the **preset value**,
you need to specify the preset value to decide the number of times you want to run the code block when defining
a `for` loop statement.

## For Statement Syntax




`for` loops are usually favored for iteration because the
all the information about the local loop variable are available in one place at the top,
which helps quickly visualize the overall sequence in the loop.

The C# for loop statement syntax is:

```csharp
for ( *initializer* ; *condition* ; *iterator* )
{
// code block (statements)
}
```

Let us practice with `csharprepl`, `Run C#`, or a small VS Code project:


````csharp
for (int i = 1; i <= 5; i++)
{
      Console.WriteLine(i);
}
````

You can see that the `for` loop has four parts, a three-section semicolon-separated **header**
and a **body**:

- **Initializer**: This section is a **variable declaration statement** that
  declares a `local loop variable` (or `iterator variable` or the counter) and initialize it with an initial value.
- **Condition**: The condition section is a **boolean expression** that determines
  if the next iteration in the loop should be executed or should the loop be terminated. If it evaluates to true,
  the next iteration is executed; otherwise, the loop is exited.
- **Iterator**: The iterator (**iteration expression**) section that defines what happens to the counter after each
  execution of the body of the loop. A common behavior is reassigning
  (increment or decrement) the value of the counter.
- The **body** of the loop, which must be a statement or a block of statements.

In the example above, the *initializer* is `i = 1`, *condition* is `i <= n`,
and *iterator* is `i++` (see {ref}`arithmetic operators <arithmetic-operators>`).

Note that the initializer section is executed **only once**, before entering the loop.
Also, the declared variable is local to the `for` loop heading and
the loop body and can not be accessed from outside the for statement.

The two semicolons are always needed in the `for` heading, but any of the
parts they normally separate may be omitted.
If the condition part is omitted, the condition is
interpreted as always true, leading to an infinite loop, that can only
terminate due to a `return` or {ref}`break statement <break-continue>` in the body.

**Header variation**

The declaration of a for loop header is flexible. There may be several variables of the
same type initialized, separated by commas. Also, at the end of the `for` loop heading,
the iteration portion may include more than one expression, also separated by commas.
For example:

````csharp
for (int i = 0, j = 10; i < j; i = i+2, j++) {
   Console.WriteLine("{0} {1}", i, j);
}
````

## `break` and `continue` Statements




`break` exits the nearest enclosing loop. Execution resumes after that loop. `continue` skips the remaining body and proceeds to the next update/condition check. Decisions inside loops can select when either action occurs. Method returns are taught in Chapter 5 and are not needed here.

Before searching text, introduce its two access operations: `text.Length` counts UTF-16 code units, and `text[i]` reads one `char` at zero-based index `i`. Index zero is first; the last valid index is `text.Length - 1`. An empty string has length zero and no valid index. Use ordinary English letters/digits in these examples; some Unicode characters occupy multiple `char` values.

The next examples search a string directly, without requiring arrays or the later Strings and Their Methods lesson.


````csharp
string text = "INV-204";
char target = '-';
bool found = false;
int foundIndex = -1;

for (int i = 0; i < text.Length; i++)   // loop for text.Length times
{
   if (text[i] == target)               // if one character == target
   {
      found = true;                    // set found to true
      foundIndex = i;
      break;                           // break out of the enclosing loop
   }
}

if (found)                             // if found == true (from the previous block)
{
   Console.WriteLine("Target found at index " + foundIndex);
}
else
{
   Console.WriteLine("Target not found");
}
````

When `text[i]` matches `target`, `break` stops the loop and the program reports the found index. Change the text to `INV204` or an empty string to check unsuccessful search. The alternate implementation below includes `!found` in the loop condition instead of using `break`.


````csharp
string text = "INV-204";
char target = '-';
bool found = false;
int foundIndex = -1;

for (int i = 0; i < text.Length && !found; i++) {
   if (text[i] == target) {
      found = true;
      foundIndex = i;
   }
}

if (found) {
   Console.WriteLine("Target found at index " + foundIndex);
} else {
   Console.WriteLine("Target not found");
}
````

As you can see, the code exit because the condition section of the if statement header
has an expression `!found` ({ref}`logical-operators`), meaning found is not true.
The shows that since `break` statements rely on the logic of the conditional statement,
if the condition can be embedded in the header of the loop, you don't have to use break.
However, if you are designing a loop that has multiple exit criteria, using break statements
can make the code much less verbose in the header's condition section, and hence easier to
follow because the if statement conditions and the immediate break action may be clearly
presented.

## Nested `for` Loop




There will be times when **nested** loops are are required for the problem scenario.
A nested loop can look like this:



```text
outer-Loop
{
   // body of outer-loop
   inner-Loop
   {
      // body of inner-loop
   }
... ... ...
}
```


Continuing with our discussion on `break`, let's say we are in a situation like the following:



```text
for (....) {

   // some statements of outer for loop

   for (....) {
      ...
      if (...) {
      ...
      break;
      }
      ...
   }

   // some statements of outer for look
}
```


The break statement is in the inner loop. If it is reached, the inner loop ends,
but the **inner loop** is just a **single statement** inside the outer loop,
and **the outer loop continues**.
If the outer loop continuation condition remains true,
the inner loop will be executed again. As an example:



````csharp
for (int i = 0; i <= 3; i++)
{
   for (int j = 0; j <= 3; j++)
   {

      if (i == 2)
      {
         break;
      }

      Console.WriteLine("{0} -- {1}", i, j);
   }
}
````

in your tests folder.

`continue` Statement

For completeness we mention the much less used `continue` statement:

> `continue;`

A `continue` statement:
\- does not break out of the whole loop statement.
\- break/skips the execution of the rest of the *body* in the current enclosing loop iteration.
\- starts the next enclosing loop iteration.

In the simplest situations, a `continue` statement just avoids an extra `else` clause.
It can considerably shorten code if the test is inside of complicated, deeply nested
`if` statements. As an example:



````csharp
for (int i = 0; i <= 3; i++)
{
   for (int j = 0; j <= 3; j++)
   {
      if (i == 2)
      {
           continue;
      }

      Console.WriteLine("{0} -- {1}", i, j);
   }
}
````

### `foreach` Statement




A string can supply characters to a loop. For these introductory examples, use ordinary English letters, digits, and punctuation. `text.Length` is the number of UTF-16 code units, and `text[i]` reads one `char` at zero-based index `i`. The valid indexes run from zero through `text.Length - 1`; an empty string has length zero and no valid index. Some Unicode characters require more than one `char`.

A counted loop visits the positions:

```csharp
string text = "ABC";
for (int i = 0; i < text.Length; i++)
{
    Console.WriteLine(text[i]);
}
```

A `foreach` visits the values without managing an index:

```csharp
string text = "ABC";
foreach (char letter in text)
{
    Console.WriteLine(letter);
}
```

Both print A, B, and C on separate lines. `foreach` is useful when each value receives the same treatment; an indexed loop is useful when positions or reverse order matter. Strings are immutable sequences, not arrays. Arrays and other collections follow in Chapters 7-8. Cleaning, searching, and extracting text are developed in Chapter 5; none is required here.

### Formatting Tables




Reports commonly include tables, often with successive lines generated by
a consistent formula and therefore a good example for coding. As a simple first table, we can show the square,
cube, and square root of numbers 1 through 10.
The Math class has a method `Sqrt`, so we take the square root with the `Math.Sqrt`
method. The pattern is consistent, so we can loop easily:



````csharp
for ( int n = 1; n <= 10; n++)
{
    Console.WriteLine("{0} {1} {2} {3}", n, n*n, n*n*n, Math.Sqrt(n));
}
````


The numbers will be there, but the output is not pretty:

> `,`

## Repeated Trials and Random Values



```{index} gap revision: repeated trials and random values
```


A simulation repeats a model of one event and summarizes what happened. A coin-flip exercise therefore combines two ideas you already know: a counted loop controls the number of trials, and a conditional classifies each trial. A pseudorandom generator produces a deterministic sequence that looks irregular; it is a model, not a source of guaranteed balance.

General patterns (schematic):

```csharp
Random generator = new Random();
int value = generator.Next(inclusiveLowerBound, exclusiveUpperBound);
```

Create the generator once before the loop and call `Next` for each trial. `Next(1, 7)` can produce 1 through 6, never 7. `Next(0, 2)` produces either zero or one. A named seed, as in `new Random(42)`, helps reproduce an investigation in the same runtime environment. Do not treat a particular seeded sequence as a portable specification across all .NET versions.

| Model | Call | Interpretation |
| --- | --- | --- |
| Coin | `Next(0, 2)` | Zero is tails; one is heads. |
| Die | `Next(1, 7)` | Six equally possible integer faces. |
| Guessing game | `Next(1, 101)` | Secret integer from 1 through 100. |

The following example checks a stronger claim than “heads should be five”: every trial contributes to exactly one counter, so heads plus tails must equal the number of trials. A short run need not be balanced. Recreating a generator on each iteration obscures the intended single stream of trials; use one generator for the experiment.

See the language/API reference for this contract.[^gap-random]


````csharp
using System;
Random generator = new Random(42);
int heads = 0;
int tails = 0;
for (int trial = 0; trial < 10; trial++)
{
    if (generator.Next(0, 2) == 1) heads++;
    else tails++;
}
Console.WriteLine(heads + tails);
Console.WriteLine(heads >= 0 && tails >= 0);
````

**Guided practice.** Before running, explain why the two output lines must be `10` and `True` even though the individual counters can vary. Model 30 die throws and count sixes. Explain why a result other than five sixes is not automatically a bug. The extra-practice coin and guessing tasks apply these same bounds and loop ideas.


**Equivalent encoding in the retained coin exercise.** The extra-practice activity reverses the labels: zero is Heads and one is Tails. Either convention models the same two outcomes when applied consistently. Its original explanation follows.

A simulation uses generated values to model possible outcomes. Create one `Random` object and reuse it throughout the experiment. For `low < high`, `r.Next(low, high)` includes `low` and excludes `high`, so `r.Next(0, 2)` can return only zero or one. Map zero to Heads and one to Tails.

First print one flip, then 10 flips. Count the two outcomes; the counts must add to the number of flips. Randomness does not guarantee equal counts in a short run. A seed makes a run repeatable in the same runtime environment; it does not make the outcomes alternate.
