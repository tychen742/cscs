# Iteration

Processing a range of values or repeatedly asking for valid input requires controlled repetition. This chapter connects loop initialization, continuation conditions, and updates to counted work, string traversal, and running totals. You will explain why a loop stops and test its first, last, and empty cases; the GCD lesson provides additional supporting practice.

- **Loop control**: A loop needs a starting state, a continuation condition, and an update that moves it toward stopping.
- **`for` loops**: A `for` loop works well when the number of repetitions or index range is known.
- **`while` loops**: A `while` loop works well when repetition depends on a condition that may change during execution.
- **Loop safety**: Careful updates and stopping conditions prevent infinite loops and off-by-one errors.

<h2>Learning Objectives</h2>


By the end of this chapter, you will be able to:

1. Write counted loops with `for` and open-ended loops with `while`.
2. Trace loop variables through repeated iterations.
3. Use loops to process strings, numeric ranges, and repeated prompts.
4. Use `break` and `continue` only when they clarify loop behavior.
5. Recognize and avoid infinite loops and off-by-one mistakes.
