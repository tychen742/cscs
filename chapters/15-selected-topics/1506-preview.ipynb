## True or False

**1.** A recursive method must have a base case to prevent infinite recursion.

```{dropdown} Answer
**True** — Without a base case the call stack will overflow.
```

**2.** Lambda expressions in C# always require an explicit `return` statement.

```{dropdown} Answer
**False** — Single-expression lambdas return implicitly: `x => x * 2`.
```

**3.** LINQ queries can only be used with in-memory collections, not databases.

```{dropdown} Answer
**False** — LINQ to Entities / EF Core translates LINQ queries to SQL.
```

**4.** Pattern matching with `switch` expressions was introduced in modern C# (C# 8+).

```{dropdown} Answer
**True** — `switch` expressions and pattern matching arrived in C# 8.
```

**5.** A C# tuple can hold values of different types.

```{dropdown} Answer
**True** — `(int, string) t = (1, "hello")` mixes types in one tuple.
```



## Multiple Choice

**1.** What is the base case in a recursive method?

a) The first method call  
b) The condition that stops further recursion  
c) The return value type  
d) The method name

```{dropdown} Answer
**b) The condition that stops further recursion** — The base case returns a result directly without another recursive call.
```

**2.** Which LINQ method filters elements based on a predicate?

a) Select()  
b) OrderBy()  
c) Where()  
d) First()

```{dropdown} Answer
**c) Where()** — `Where(x => condition)` keeps only elements satisfying the condition.
```

**3.** What does the lambda `x => x * 2` do?

a) Takes no parameters and returns 2  
b) Takes one parameter `x` and returns `x * 2`  
c) Takes two parameters  
d) Returns only 2 always

```{dropdown} Answer
**b) Takes one parameter `x` and returns `x * 2`** — `x =>` declares one parameter; the body `x * 2` is returned implicitly.
```

**4.** Which C# feature matches a value against a series of patterns?

a) Generics  
b) Delegates  
c) Pattern matching  
d) Reflection

```{dropdown} Answer
**c) Pattern matching** — Pattern matching (`switch` expressions, `is` patterns) allows structural matching.
```

**5.** How do you access the second element of a tuple `(int, string) t`?

a) t.First  
b) t[1]  
c) t.Item2  
d) t.Second

```{dropdown} Answer
**c) t.Item2** — Tuple elements are accessed with `Item1`, `Item2`, etc. by default.
```
