# Selected Topics

<!-- chapter-intro -->

This chapter covers advanced C# features not covered in earlier chapters. Each section is self-contained — you can read them in any order based on your needs.

<h2>Why this chapter matters</h2>

Modern C# is expressive and concise. These features appear frequently in production codebases and interviews:

- **Recursion** — elegantly solves problems defined in terms of themselves
- **Lambdas** — pass behavior as a value without defining a named method
- **LINQ** — query and transform collections with a readable, SQL-like syntax
- **Pattern matching** — replace complex if-chains with structured, safe dispatch
- **Tuples** — return multiple values from a method without a wrapper class

<h2>Learning goals</h2>

By the end of this chapter, you should be able to:

- write a recursive method with a correct base case and recursive step
- use lambda expressions with `Func<>`, `Action<>`, and LINQ methods
- filter, project, and sort a collection using LINQ query or method syntax
- apply `switch` expressions and `is` patterns to replace conditional chains
- use tuples to pass multiple values without defining a dedicated class
