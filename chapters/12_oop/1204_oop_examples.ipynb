## Programming Paradigms and Separation of Concerns




```{index} Programming Paradigms and Separation of Concerns
```

Languages such as Java, C++, C#, and Python are general-purpose high-level programming languages and they support multiple
programming **paradigms**; which means you can use C# to develop computer programs using
different styles such as:

- procedural programming: (focused on procedures and functions),
- object-oriented programming (based on objects and classes), and
- functional programming (emphasizing mathematical functions and immutability).

Object-oriented programming (OOP) is one of the popular programming paradigms.
**Separation of concerns** is an important design goal that classes and objects can support. Separation of concerns is a design principle to separate a computer program
into different sections with each section addresses a separate concern. For OOP design,
the concerns are separated into objects. Architectural patterns such as model-view-controller (MVC) can separate stored state and business behavior from presentation and interaction. These design goals also apply outside OOP.

The most popular model of OOP is a class-based model. In OOP, a class is a
**template** for creating objects, and an **object** entity is
an instance of a class. Each object follows the member definitions of its class. Creating an instance is **instantiation**; inheritance instead relates a derived class to a base class. In .NET, some classes
(e.g., Console) are supplied by the standard libraries and users can also define classes (e.g., UI).

While procedural programming is about writing procedures or methods that perform
operations on the data, object-oriented programming (OOP) provides a clear structure
for the code to reduce **complexity** and to make the programs **reusable**.

The four basic principles of object-oriented programming are: [^1204_oop_examples-oop-microsoft]

- **Abstraction**: Modeling the relevant attributes and interactions of entities as classes
  to define an abstract representation of a system.
- **Encapsulation**: Hiding the internal state and functionality of an object and only
  allowing access through a public set of functions.
- **Inheritance**: Ability to create new abstractions based on existing abstractions; which
  enables code **reusability**.
- **Polymorphism**: Ability to implement inherited properties or methods in different
  ways across multiple abstractions.

**Data abstraction** describes operations and their meaning while hiding representation choices. Abstract classes and interfaces can express contracts, but abstraction is a broader design idea and does not require either keyword.

To further explain the four pillars of OOP [^1204_oop_examples-oop-techtarget]:
The **encapsulation** principle states that all important information is contained inside
an object and only select information is exposed. The implementation and state of
each object are privately held inside a defined class. Client code uses the accessible members rather than directly changing private fields. Encapsulation can protect state invariants and reduce accidental changes; it is not a security boundary against arbitrary hostile code.

The **abstraction** principle states that objects only reveal internal mechanisms that
are relevant for the use of other objects, hiding any unnecessary implementation code.
An implementation can change while its public contract remains stable. This can let clients keep using the same operations after internal changes.

The **inheritance** principle states that classes can reuse code and properties from
other classes. Relationships and subclasses between objects can be assigned,
enabling developers to reuse common logic, while still maintaining a unique hierarchy.
Inheritance can reuse common behavior, but it does not automatically improve accuracy or design. The derived type still needs to preserve the base contract.

The **polymorphism** principle states that objects are designed to share behaviors,
and they can take on more than one form. The program determines which meaning or
usage is necessary for each execution of that object from a parent class, reducing
the need to duplicate code. A child class is then created, which extends the
functionality of the parent class. Polymorphism enables different types of objects
to pass through the same interface.

## Inheritance: Vehicle and Car





Retain the original simple inheritance example before comparing it with the current substitution contracts. Vehicle supplies a brand and Honk behavior; Car adds a model name. A Car instance has access to the public inherited members. This demonstration uses public fields to expose the mechanics; the current business models show stronger encapsulation and validation.

````csharp
using System;
Car car = new Car(); car.Honk(); Console.WriteLine(car.Brand + " " + car.ModelName);
class Vehicle {
    public string Brand = "Ford";
    public void Honk() { Console.WriteLine("Tuut, tuut!"); }
}
class Car : Vehicle { public string ModelName = "Mustang"; }
````

## Nonvirtual Hiding and Virtual Overriding




```{index} Nonvirtual Hiding and Virtual Overriding
```

The original animal-sound example first declared methods with the same name without virtual/override, then revised the methods for runtime dispatch. Preserve that comparison: a nonvirtual member is selected from the reference's compile-time type; an override is selected from the object's runtime type. Use the explicit new modifier when hiding is intentional so the compiler does not have to warn about an accidental name collision.

In the demonstration, a base-typed reference to a hiding Pig calls the base sound. A base-typed reference to an overriding Pig calls the pig's sound. Do not conclude that every same-named derived method is polymorphic overriding.

````csharp
using System;
HidingAnimal hidden = new HidingPig(); hidden.Sound();
VirtualAnimal[] animals = {new VirtualAnimal(), new VirtualPig(), new VirtualDog()};
foreach (VirtualAnimal animal in animals) animal.Sound();
class HidingAnimal { public void Sound() { Console.WriteLine("The animal makes a sound"); } }
class HidingPig : HidingAnimal { public new void Sound() { Console.WriteLine("The pig says: wee wee"); } }
class VirtualAnimal { public virtual void Sound() { Console.WriteLine("The animal makes a sound"); } }
class VirtualPig : VirtualAnimal { public override void Sound() { Console.WriteLine("The pig says: wee wee"); } }
class VirtualDog : VirtualAnimal { public override void Sound() { Console.WriteLine("The dog says: bow wow"); } }
````

## Overloading by Type and Number of Arguments




```{index} Overloading by Type and Number of Arguments
```

Start with separately named PlusMethodInt and PlusMethodDouble: both add, but each accepts a different numeric type. Then give the two operations one name, PlusMethod, with distinct parameter types. The compiler chooses an applicable overload using the arguments. A different return type alone is not a distinct overload.

The original MethodOverLoading example also varies the **number** of int arguments and supplies float and string overloads. Adding strings concatenates text; it does not perform numeric addition. The spelling of the declared class and the constructed type must match. This corrects a casing/spelling defect in the historical listing while retaining all four variations.

````csharp
using System;
class OverloadDemo {
    static int PlusMethodInt(int x, int y) { return x + y; }
    static double PlusMethodDouble(double x, double y) { return x + y; }
    static int PlusMethod(int x, int y) { return x + y; }
    static double PlusMethod(double x, double y) { return x + y; }
    public static void Main() {
        Console.WriteLine("Int: " + PlusMethodInt(8, 5));
        Console.WriteLine("Double: " + PlusMethodDouble(4.3, 6.26));
        Console.WriteLine("Int overload: " + PlusMethod(8, 5));
        Console.WriteLine("Double overload: " + PlusMethod(4.3, 6.26));
        MethodOverLoading methods = new MethodOverLoading();
        Console.WriteLine("Add two int parameters: " + methods.Add(3, 2));
        Console.WriteLine("Add three int parameters: " + methods.Add(3, 2, 8));
        Console.WriteLine("Add two float parameters: " + methods.Add(3f, 22f));
        Console.WriteLine("Add two string parameters: " + methods.Add("hello", "world"));
    }
}
class MethodOverLoading {
    public int Add(int a, int b) { return a + b; }
    public int Add(int a, int b, int c) { return a + b + c; }
    public float Add(float a, float b) { return a + b; }
    public string Add(string a, string b) { return a + " " + b; }
}
````

## Calling base and Sealing an Override




```{index} Calling base and Sealing an Override
```

A class inherits directly from one base class and can implement multiple interfaces. A virtual base method permits an override with a matching signature. A derived implementation may call base.Draw() to preserve the base action and then extend it.

Use **sealed override** to close one virtual member to further overriding while still permitting inheritance from the class. This differs from a sealed class, which cannot be inherited at all. In the original shape example, Rectangle seals Draw. Square may derive from Rectangle and inherits that implementation, but it cannot override Draw again.[^1204_oop_examples-sealed]

Predict the output below. Circle calls the base action before its own message; Square uses Rectangle's sealed implementation. Triangle supplies a separate override.

````csharp
using System;
Shape[] shapes = {new Circle(), new Rectangle(), new Triangle(), new Square()};
foreach (Shape shape in shapes) shape.Draw();
class Shape { public virtual void Draw() { Console.WriteLine("Drawing a shape"); } }
class Circle : Shape {
    public override void Draw() { base.Draw(); Console.WriteLine("Drawing a circle"); }
}
class Rectangle : Shape {
    public sealed override void Draw() { Console.WriteLine("Drawing a rectangle"); }
}
class Triangle : Shape { public override void Draw() { Console.WriteLine("Drawing a triangle"); } }
class Square : Rectangle { }
````

## Abstract Classes and Interfaces Compared




```{index} Abstract Classes and Interfaces Compared
```

The original comparison table is retained in corrected form. Both abstract classes and interfaces can express abstractions. An abstract class can supply instance state, constructors for derived classes, and shared implementations. An interface can describe a capability implemented by otherwise unrelated types. Neither can be instantiated directly as a complete concrete object.

| Feature | Abstract class | Interface |
| --- | --- | --- |
| Type relationship | A class derives directly from at most one class. | A class can implement multiple interfaces; an interface can inherit other interfaces. |
| Instance state | Can store instance fields. | Does not supply per-object instance fields. |
| Construction | Can initialize shared state through a constructor invoked by a derived class. | Does not provide instance constructors for implementing objects. |
| Operations | Can combine concrete, abstract, and virtual members. | Can declare contracts; modern C# also permits default implementations and other advanced members. |
| Access | Can expose public/protected members and hide private implementation. | Ordinary signatures used in this course are public contracts; modern interfaces have additional access and member rules. |
| Reuse | Can share state and behavior within a meaningful subtype family. | Can share a capability across unrelated implementations. |
| Evolution | Adding an abstract requirement can break derived implementations; adding suitable concrete behavior may not. | Adding a required member can break implementers; default implementations can sometimes provide a migration path. |

The current chapter deliberately uses ordinary interface signatures. The table does not turn advanced interface features into prerequisites, but it avoids the old claim that every interface contains only method signatures.[^1204_oop_examples-interfaces]

The earlier abstract Animal example also combined an abstract Sound with a concrete Sleep method. Here is that small variation, separate from the current payment example.

````csharp
using System;
Pig pig = new Pig(); pig.Sound(); pig.Sleep();
abstract class Animal {
    public abstract void Sound();
    public void Sleep() { Console.WriteLine("Zzz"); }
}
class Pig : Animal { public override void Sound() { Console.WriteLine("The pig says: wee wee"); } }
````

### Additional Abstract and Interface Variations




```{index} Additional Abstract and Interface Variations
```

The original abstraction lesson also illustrated a Circle area, a Rectangle with both area and perimeter, a failed attempt to construct an abstract class, and an animal interface. Retain those small cases as well as the current payment/capability examples.

Use different type names in this one-cell comparison so the two contracts do not conflict: AreaShape requires area only; MeasuredShape requires area and perimeter. Circle stores its radius and returns pi times radius squared. MeasuredRectangle stores width and height and implements both required operations. An interface supplies a contract for a capability, rather than an instance-field representation or a constructor.

**Intentional compile-time error:** `MeasuredShape shape = new MeasuredShape();` attempts to instantiate an abstract class and produces CS0144. Construct a concrete MeasuredRectangle instead. A concrete derived class must implement every inherited abstract operation.

The interface variation calls sound and running behavior through IAnimal. Public implementations do not use override simply because they satisfy an interface. Compare this with the abstract Animal/Sleep example above and with the current IDiscountPolicy contract.

````csharp
using System;
AreaShape circle = new Circle(2); Console.WriteLine($"Circle area: {circle.GetArea():F2}");
MeasuredShape rectangle = new MeasuredRectangle(3, 4);
Console.WriteLine($"Rectangle area: {rectangle.GetArea()}; perimeter: {rectangle.GetPerimeter()}");
IAnimal animal = new RunningPig(); animal.AnimalSound(); animal.Run();
abstract class AreaShape { public abstract double GetArea(); }
class Circle : AreaShape {
    private readonly double radius;
    public Circle(double radius) { if (!double.IsFinite(radius) || radius < 0) throw new ArgumentOutOfRangeException(nameof(radius)); this.radius = radius; }
    public override double GetArea() { return Math.PI * radius * radius; }
}
abstract class MeasuredShape { public abstract double GetArea(); public abstract double GetPerimeter(); }
class MeasuredRectangle : MeasuredShape {
    private readonly double width, height;
    public MeasuredRectangle(double width, double height) {
        if (!double.IsFinite(width) || !double.IsFinite(height) || width < 0 || height < 0) throw new ArgumentOutOfRangeException(nameof(width));
        this.width = width; this.height = height;
    }
    public override double GetArea() { return width * height; }
    public override double GetPerimeter() { return 2 * (width + height); }
}
interface IAnimal { void AnimalSound(); void Run(); }
class RunningPig : IAnimal {
    public void AnimalSound() { Console.WriteLine("The pig says: wee wee"); }
    public void Run() { Console.WriteLine("The pig runs"); }
}
````
