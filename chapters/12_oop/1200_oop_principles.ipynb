# Inheritance, Interfaces, and Polymorphism

This chapter builds on ordinary composition to support clients that work with different implementations of one behavior.

- **Inheritance**: A derived type builds on a base type while preserving its behavioral promises.
- **Polymorphism**: A common client call uses the implementation associated with an actual object.
- **Abstract class**: A general type that requires concrete derived behavior and can supply shared state or methods.
- **Interface**: A named capability contract implemented by different classes.
- **Substitution**: Using one implementation without breaking the expectations of clients of the general type.
