# Inheritance, Interfaces, and Polymorphism

A reporting client should be able to use different implementations without changing its expectations. Building on composition, this chapter develops inheritance, virtual dispatch, abstract classes, and interfaces as ways to express and share behavior. You will test substitution and choose a relationship from the requirements rather than inheriting merely to reuse code.

- **Inheritance**: A derived type builds on a base type while preserving its behavioral promises.
- **Polymorphism**: A common client call uses the implementation associated with an actual object.
- **Abstract class**: A general type that requires concrete derived behavior and can supply shared state or methods.
- **Interface**: A named capability contract implemented by different classes.
- **Substitution**: Using one implementation without breaking the expectations of clients of the general type.

3. Implement concrete behavior for an abstract class.
4. Supply interchangeable components through interfaces and constructors.
5. Test shared contracts and choose composition or inheritance from requirements.
