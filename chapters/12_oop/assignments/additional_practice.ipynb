````csharp
using System;
Shape[] shapes = {new Triangle(3, 4), new Rectangle(3, 4), new Triangle(0, 4)};
foreach (Shape shape in shapes) Console.WriteLine($"{shape.GetType().Name}: {shape.GetArea()}");
try { new Rectangle(-1, 4); } catch (ArgumentOutOfRangeException) { Console.WriteLine("Rejected negative dimension"); }
public abstract class Shape { public abstract double GetArea(); }
public class Triangle : Shape {
    private readonly double basis, height;
    public Triangle(double basis, double height) {
        if (!double.IsFinite(basis) || !double.IsFinite(height) || basis < 0 || height < 0) throw new ArgumentOutOfRangeException(nameof(basis));
        this.basis = basis; this.height = height;
    }
    public override double GetArea() { return basis * height / 2; }
}
public class Rectangle : Shape {
    private readonly double width, height;
    public Rectangle(double width, double height) {
        if (!double.IsFinite(width) || !double.IsFinite(height) || width < 0 || height < 0) throw new ArgumentOutOfRangeException(nameof(width));
        this.width = width; this.height = height;
    }
    public override double GetArea() { return width * height; }
}
````
