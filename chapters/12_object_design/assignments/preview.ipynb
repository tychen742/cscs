<fieldset><legend>1. What is an alias?</legend>
<label><input type="radio" name="cscs-object-design-preview-1" value="0"> An independent field copy</label><br>
<label><input type="radio" name="cscs-object-design-preview-1" value="1"> Another reference to the same object</label><br>


<fieldset><legend>3. What is an immutable object?</legend>
<label><input type="radio" name="cscs-object-design-preview-3" value="0"> An object whose observable state stays fixed</label><br>
<label><input type="radio" name="cscs-object-design-preview-3" value="1"> An object with no constructor</label><br>
<label><input type="radio" name="cscs-object-design-preview-3" value="2"> A static variable</label><br>
<label><input type="radio" name="cscs-object-design-preview-3" value="3"> An object that cannot be shared</label><br>
</fieldset>


```{dropdown} Answer
Answer: An object whose observable state stays fixed. Sharing fixed state does not permit later mutation.
```


<fieldset><legend>4. Which relationship describes composition?</legend>
<label><input type="radio" name="cscs-object-design-preview-4" value="0"> Is-a</label><br>
<label><input type="radio" name="cscs-object-design-preview-4" value="1"> Overrides</label><br>
<label><input type="radio" name="cscs-object-design-preview-4" value="2"> Has-a</label><br>
<label><input type="radio" name="cscs-object-design-preview-4" value="3"> Compiles-as</label><br>
</fieldset>


```{dropdown} Answer
Answer: Has-a. An order has lines.
```


<fieldset><legend>5. What is a class responsibility?</legend>


<fieldset><legend>7. What does a precondition describe?</legend>
<label><input type="radio" name="cscs-object-design-preview-7" value="0"> A required input or starting state</label><br>
