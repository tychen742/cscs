````csharp
using System;
using System.Linq;

// Solution
int[][] reports={new[]{1200,500,1200,300,800},new[]{-500,-200,-900},new[]{42},Array.Empty<int>()};
foreach(int[] values in reports)
{
 try{Console.WriteLine($"maximum={Max(values)}");}
 catch(ArgumentException){Console.WriteLine("empty report rejected");}
}
int Max(int[] values)
{
 if(values is null)throw new ArgumentNullException(nameof(values));
 if(values.Length==0)throw new ArgumentException("Nonempty report required.");
 int best=values[0];
 // At loop head, best is max of [0,i); variant is Length-i.
 for(int i=1;i<values.Length;i++)if(values[i]>best)best=values[i];
 return best;
}
````

````csharp
using System;
using System.Linq;

// Solution
int[] expenses={1200,500,1200,300,800};
foreach(int target in new[]{1200,800,999})
{
 var result=Find(expenses,target);
 Console.WriteLine($"target={target}, index={result.index}, checks={result.checks}");
}
(int index,int checks) Find(int[] values,int target)
{
 int checks=0;
 // Before index i, no value in [0,i) matches.
 for(int i=0;i<values.Length;i++)
 {checks++;if(values[i]==target)return(i,checks);}
 return(-1,checks);
}
// Best Theta(1), worst Theta(n) for nonempty reports. Early return is first match; normal exit proves absence.
````

````csharp
using System;
using System.Linq;

// Solution
int[] sizes={1,5,8};
foreach(int n in sizes)
{
 long calls=0;int active=0,peak=0;
 long work=Review(n);
 Console.WriteLine($"n={n}, modeled work={work}, calls={calls}, peak={peak}");
 long Review(int size)
 {
  calls++;active++;peak=Math.Max(peak,active);
  long result=size==1?1:Review(size/2)+Review(size-size/2)+size;
  active--;return result;
 }
}
// Model: W(1)=1, W(n)=W(floor(n/2))+W(ceil(n/2))+n.
// Modeled work Theta(n log n); this counter visits Theta(n) nodes, with Theta(log n) frames.
````
