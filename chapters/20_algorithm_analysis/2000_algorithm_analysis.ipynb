# Analysis

This chapter covers algorithm correctness, specifications, preconditions, postconditions, loop invariants, operation counting, Big-O/Big-Theta notation, space growth, and recurrence relations. The sidebar title **Analysis** names the skill: reasoning about whether an algorithm works and how its cost grows.

- **Correctness**: A claim that an algorithm satisfies its specification for every valid input.
- **Complexity**: A description of how time or memory use grows as input size grows.
- **Big-O**: An upper-bound growth description that ignores constant factors and lower-order terms.
- **Invariant**: A fact that remains true before and after each loop iteration.
- **Recurrence**: An equation that describes the cost of a recursive algorithm in terms of smaller inputs.

Algorithm analysis asks two questions: does the procedure always produce the required result, and how do its time and memory needs grow as the input grows?


> *[Data Structures and Algorithms In C#: Big-O Notation](https://youtube.com/watch?v=IcDcNp8gZ7U)* by Teddy Smith.
>
> This video is a quick supplement for the growth-rate ideas in this chapter. The chapter develops the notation, correctness language, and recurrence reasoning in the book's sequence.


<h2>Learning Objectives</h2>

1. Define input sizes and a cost model; derive exact counts for sequential, rectangular, triangular, and halving loops.
2. Distinguish upper, lower, and tight bounds and best, worst, expected, and amortized costs.
3. State precise contracts and use initialization, maintenance, exit reasoning, and decreasing variants to establish total correctness.
4. Choose boundary tests and counterexamples that challenge an implementation.
5. Derive recurrences with base cases and rounding; separate modeled work, total calls, stack depth, and other auxiliary storage.


<h2>Chapter Flow</h2>

- {doc}`20.1. Big-O and Complexity <2001_big_o>`
- {doc}`20.2. Correctness and Invariants <2002_correctness>`
- {doc}`20.3. Recurrences and Recursive Cost <2003_recurrences>`


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Input size | A measure of how large the input is, usually written as `n`. |
| 2 | Growth rate | A description of how work or memory changes as input size increases. |
| 3 | Big-O | An asymptotic upper bound on growth. |
| 4 | Big-Theta | An asymptotically tight growth description. |
| 5 | Constant time | A cost that does not grow with input size. |
| 6 | Linear time | A cost that grows proportionally with input size. |
| 7 | Logarithmic time | A cost that grows by repeatedly reducing the problem size by a factor. |
| 8 | Precondition | A condition that must be true before an algorithm runs. |
| 9 | Postcondition | A condition that should be true after an algorithm finishes. |
| 10 | Loop invariant | A fact that is true before and after every loop iteration. |
| 11 | Recurrence relation | An equation that defines a recursive algorithm cost using smaller inputs. |

| 12 | Big-Omega | An asymptotic lower bound. |
| 13 | Cost model | The operations and assumptions used to count work. |
| 14 | Auxiliary space | Storage used in addition to the input, under a stated output convention. |
| 15 | Variant | A nonnegative integer measure that strictly decreases on continuing loop iterations. |
| 16 | Partial correctness | The postcondition holds if execution terminates on a valid input. |
| 17 | Total correctness | Partial correctness plus guaranteed termination. |
| 18 | Worst-case cost | Maximum cost over valid inputs of a given size. |
| 19 | Expected cost | A cost average under a stated probability model. |
| 20 | Amortized cost | Cost per operation averaged over an operation sequence. |
| 21 | Recursion depth | Maximum simultaneously active calls, under a stated counting convention. |


<h2>Slides</h2>

Slides for this chapter will be added after the notebook content stabilizes.
