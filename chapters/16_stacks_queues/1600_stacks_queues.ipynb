# Stacks & Queues

This chapter covers stacks, queues, deques, restricted-access ADTs, array-based implementations, linked implementations, and applications such as undo, scheduling, buffering, backtracking, and breadth-first processing. The sidebar title focuses on the two central access rules: last-in-first-out and first-in-first-out.

- **Stack**: A stack uses last-in, first-out behavior for undo, nested processing, and backtracking.
- **Queue**: A queue uses first-in, first-out behavior for service lines and scheduled work.
- **Deque**: A deque supports insertion and removal at both ends, making it a flexible boundary structure.
- **Implementation choice**: Array-based and linked implementations can provide the same ADT operations with different costs.
- **Application fit**: Restricted-access structures are useful when the rule for who comes next is part of the problem.

Stacks, queues, and deques show that a data structure is not only storage. It is also a policy for how a program is allowed to use that storage.


> *[C# Data Structures - Arrays, Lists, Dictionaries, HashSets, Stacks, Queues](https://youtube.com/watch?v=G0bD3Af1DRc)* by Mark Inman.
>
> This video remains hosted on YouTube and is the property of its original creator or rightsholder; it is embedded here via YouTube's official player for educational reference only.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:

1. Specify stack, queue, and deque access rules and failure behavior.
2. Trace count, capacity, stack top, and ring-buffer wraparound.
3. Implement generic array stacks, bounded ring queues, and linked queues while preserving their invariants.
4. Distinguish worst-case and amortized operation costs and compare representation tradeoffs.
5. Apply and test delimiter matching, undo/redo histories, queue worklists, and scheduling policies.


<h2>Chapter Flow</h2>

- {doc}`16.1. Stacks <1601_stack>`
- {doc}`16.2. Queues and Deques <1602_queue_deque>`
- {doc}`16.3. Applications Lab <1603_applications_lab>`


<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Stack | A collection where the most recently added item is the first removed. |
| 2 | Queue | A collection where the earliest added item is the first removed. |
| 3 | Deque | A double-ended queue that supports adding and removing at both ends. |
| 4 | LIFO | Last in, first out; the access rule used by stacks. |
| 5 | FIFO | First in, first out; the access rule used by queues. |
| 6 | Push | Add an item to the top of a stack. |
| 7 | Pop | Remove and return the top item from a stack. |
| 8 | Peek | Look at the next item without removing it. |
| 9 | Enqueue | Add an item to the back of a queue. |
| 10 | Dequeue | Remove and return the front item from a queue. |
| 11 | Front | The next queue item to be removed. |
| 12 | Back | The end where new queue items are normally added. |
| 13 | Ring buffer | An array-based queue representation that wraps indexes around a fixed array. |
| 14 | Overflow | A condition where a bounded structure has no space for another item. |
| 15 | Underflow | A condition where code tries to remove an item from an empty structure. |
| 16 | Backtracking | A strategy that uses a stack to remember choices that may need to be undone. |
| 17 | Worklist | A collection of pending tasks or states to process. |
| 18 | Breadth-first order | Processing nearest or earliest-discovered items before later layers, often with a queue. |
| 19 | Invariant | A representation rule that holds after every completed operation. |
| 20 | Amortized cost | Total cost over an operation sequence divided by its length. |
| 21 | Snapshot | A saved state that can be restored, such as a prior report version. |
| 22 | Starvation | A waiting task keeps being bypassed by other work under a scheduling policy. |


<h2>Lecture Slides</h2>

Lecture slide decks for this chapter are pending.
