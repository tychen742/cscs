# Preface

This book teaches you to think like a computer scientist while you write real C# programs. It assumes no programming experience. The first course develops the skills to build a small multi-class information system that stores its data in a database. Further study introduces data structures and the algorithmic foundations needed to choose and use them. An optional algorithms sequence develops deeper analysis and problem-solving techniques.


The book is about computer science and information technology, not only C# syntax. Each chapter pairs a programming skill with the idea behind it: why a program is split into methods, why a dictionary finds a key faster than a list, why one algorithm finishes in a second while another would run for years. The examples come from business and management settings such as sales, inventory, payroll, scheduling, and delivery routes, so the problems look like the ones organizations actually solve with software.

The code cells in this book run as written. Change a value, run the cell again, and see what happens; that habit is how programming is learned.

## Why Coding When AI Writes Code

AI assistants now generate working code from a short description. That makes the skills in this book more important, not less. Someone still has to state precisely what a program should do, decides whether the code an assistant produces is correct, work with the agents to make sure the development is in the right direction, and judge which tradeoffs make better sense. 

Those are the skills this book practices. You will trace method calls by hand (Chapter 3), test and debug code systematically (Chapter 6), state preconditions and loop invariants, and compare operation costs using Big-O notation (Chapter 14), with deeper analysis in Chapter 20. A developer who can do these things uses AI as a fast assistant. A developer who cannot is left trusting output they are unable to check.

## How the Book Is Organized

The book has five parts. The first three develop an applied programming foundation; the fourth connects data structures to the algorithms needed to use them; the fifth extends that foundation with deeper algorithm analysis and design.

| Part | Chapters | Role in the sequence |
| --- | --- | --- |
| I. Fundamentals | 1-7 | Introduce computing concepts and C# programs, then develop methods, decisions, iteration, testing, and responsible professional practice. |
| II. Data and I/O | 8-11 | Organize records in arrays and collections, and read, write, and process files and text, then store and query relational data. |
| III. Object-Oriented Programming | 12-13 | Model application data and behavior with classes, apply object-oriented principles, and organize the earlier data workflows into reusable components. |
| IV. Data Structures | 14-19 | Compare abstract data types, linear lists, stacks, queues, trees, heaps, hash tables, and graphs. Learn basic cost analysis, sequence search and library sorting, recursive calls, and graph traversal where those ideas support the structures. |
| V. Algorithms | 20-25 | Develop formal cost bounds, correctness arguments, recurrences, divide and conquer, detailed sorting analysis, greedy methods, dynamic programming, backtracking, and advanced graph and string applications. |

The {doc}`Contents <contents>` lists the chapters and sections, and {doc}`Chapter Topics <chapter-topics>` provides a more detailed overview. The appendices supply resources, command-line and GitHub guidance, the group project, and instructor-facing curriculum alignment.

## Course Pathways



**Programming and Information Systems Foundations (Chapters 1-13)** can serve as a first-semester course. Students progress from small console programs to methods, collections, files, classes, and persistent application data. Integration, testing, documentation, and a project demonstration bring these skills together.

**Data Structures and Algorithms Foundations (Chapters 14-25)** can serve as a second course for students who already have the first course's programming foundation. Part IV develops representation choices and their algorithmic foundations before Part V deepens analysis and design. An instructor using this portion independently should establish equivalent preparation in methods, collections, classes, and interfaces.

**For the College of Business pathway, Part V is optional.** Chapters 1-13 provide the applied foundation, and instructors select Part IV depth to fit their course goals. Part IV includes basic Big-O, linear and binary search, library sorting, recursion for tree operations, and breadth-first and depth-first graph traversal. Students can study these foundations without taking Part V. Instructors may use the optional algorithms chapters for a technical specialization or further study.

A chapter is a weekly organizing unit, with roughly 100-120 minutes of planned lecture and additional practice. Course schedules also need time for review, assessment, and project integration; instructors can adjust the depth and pace rather than assign every topic in the volume.

The {ref}`Curriculum Alignment appendix <cs2023-alignment>` documents the book's contributions to both Information Systems (IS2020) and Computer Science (CS2023) curriculum guidance. The mappings help instructors identify relevant material and coverage boundaries; neither represents complete degree-program coverage.

## How Each Chapter Works

Each chapter is one week of work. A chapter opens with a landing page that lists its sections, followed by content sections, usually three, each sized for one class meeting. The first section introduces the chapter's main ideas, and the later sections each take one focused topic. Every section includes code you can run and modify.

The assignments follow the content:

- **Preview**: a short multiple-choice quiz on the chapter's key ideas, taken before or early in the week.
- **Lab**: connected coding practice on one theme, where each step builds on the previous one.
- **Homework**: true/false questions and coding problems that reinforce the chapter after class.
- **Project** (some chapters): a deliverable built in VS Code and submitted through GitHub, connecting the chapter to a larger application.

To get the most out of a chapter, read a section before class, run every example, then change it: use different input, break it on purpose, and read the error message. The labs and homework assume you have done this.

## Why C\#

C# is a general-purpose, statically typed language at the center of Microsoft's open-source, cross-platform .NET platform. It is a good first language for this book for four reasons:

- **Types are visible.** Every variable has a declared type that the compiler checks before the program runs. Beginners see how data is represented, and many mistakes are caught as compile errors instead of surprising results.
- **It is used where business software is built.** C# and .NET power enterprise web services, internal business applications, cloud back ends, and database-driven systems, the same kind of information systems this book builds toward.
- **It is free and runs everywhere.** The .NET SDK and VS Code run on Windows, macOS, and Linux, so the tools you install in Chapter 1 are the tools professionals use.
- **The ideas transfer.** Classes, interfaces, generics, collections, and exceptions in C# map closely to Java, TypeScript, and other modern languages. The data structures and algorithms in Part IV and Part V apply in any language.

C# is widely used beyond business systems: the Unity game engine, ASP.NET web applications, and desktop and mobile apps are written in it. C# also holds a steady place among the most used languages. The chart below shows the ten most used languages in the Stack Overflow Developer Survey[^stack-overflow-survey].

````
using System.Net.Http;
using System.Text.RegularExpressions;
using Plotly.NET;
using Plotly.NET.Interactive;
using Plotly.NET.LayoutObjects;

var surveyUrl = "https://survey.stackoverflow.co/2025";
var topLangs = new List<(string Language, double Percent)>();

var targetLanguages = new HashSet<string>(StringComparer.OrdinalIgnoreCase)
{
    "JavaScript", "HTML/CSS", "SQL", "Python", "Bash/Shell",
    "TypeScript", "Java", "C#", "C++", "PowerShell"
};

try
{
    using var http = new HttpClient();
    http.DefaultRequestHeaders.UserAgent.ParseAdd("Mozilla/5.0 (compatible; CSCSBook/1.0)");
    var html = await http.GetStringAsync(surveyUrl);

    // Parse language entries such as "JavaScript 66%" that appear in chart payload/content.
    var matches = Regex.Matches(
        html,
        @"(JavaScript|HTML/CSS|SQL|Python|Bash/Shell(?:\s*\(all shells\))?|TypeScript|Java|C#|C\+\+|PowerShell)\s*([0-9]+(?:\.[0-9]+)?)%",
        RegexOptions.IgnoreCase
    );

    foreach (Match m in matches)
    {
        var rawLanguage = m.Groups[1].Value.Trim();
        var language = rawLanguage.StartsWith("Bash/Shell", StringComparison.OrdinalIgnoreCase) ? "Bash/Shell" : rawLanguage;
        if (!targetLanguages.Contains(language)) continue;

        var pct = double.Parse(m.Groups[2].Value);
        if (!topLangs.Any(x => x.Language.Equals(language, StringComparison.OrdinalIgnoreCase)))
            topLangs.Add((language, pct));
    }
}
catch
{
    // Network or parsing failures should not break the notebook build.
}

if (topLangs.Count < 10)
{
    // Fallback top-10 values (Stack Overflow Developer Survey 2025, published summary/chart).
    topLangs = new List<(string, double)>
    {
        ("JavaScript", 66.0),
        ("HTML/CSS", 61.9),
        ("SQL", 58.6),
        ("Python", 57.9),
        ("Bash/Shell", 48.7),
        ("TypeScript", 43.8),
        ("Java", 29.5),
        ("C#", 27.9),
        ("C++", 23.6),
        ("PowerShell", 23.3),
    };
}

var ordered = topLangs
    .OrderByDescending(x => x.Percent)
    .Take(10)
    .OrderBy(x => x.Percent)
    .ToArray();

var langs = ordered.Select(x => x.Language).ToArray();
var pcts = ordered.Select(x => x.Percent).ToArray();

Chart2D.Chart.Bar<double, string, string, string, string>(
    values: pcts,
    Keys: langs,
    Name: "% Used"
 )
.WithLayout(Layout.init<string>(
    Width: 500, Height: 320,
    Title: Title.init("Top 10 Languages - Stack Overflow Survey 2025", X: 0.5, Y: 0.95),
    Margin: Margin.init<int, int, int, int, int, int>(Top: 60, Right: 30, Bottom: 50, Left: 140)
)).WithXAxis(LinearAxis.init<string, string, string, string, string, string, string, string>(
    Title: Title.init("% of Respondents")
))
.WithYAxis(LinearAxis.init<string, string, string, string, string, string, string, string>(
))
.WithConfig(Config.init(Responsive: false))
.Display();
````
