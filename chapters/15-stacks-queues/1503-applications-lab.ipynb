# Applications Lab

Use stacks, queues, and deques to model service lines, undo behavior, and scheduling.

## Tasks

1. Use a stack to reverse a sequence and explain why LIFO produces the result.
2. Use a queue to simulate customers waiting for a service operation.
3. Use a deque to process urgent work from the front and newly discovered work from the back.
4. Test empty, one-item, and multiple-item cases.
5. Explain which operations the application requires and why the chosen structure fits them.

```csharp
var customers = new Queue<string>(new[] { "Amina", "Diego", "Mina" });
int served = 0;

while (customers.Count > 0)
{
    string customer = customers.Dequeue();
    served++;
    Console.WriteLine($"Serving {customer}");
}

Console.WriteLine($"Customers served: {served}");
```
