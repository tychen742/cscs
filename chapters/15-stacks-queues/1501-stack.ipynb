# Stacks

A **stack** is a restricted-access sequence that follows **last in, first out** (LIFO). The most recently added item is the first item removed.

Core operations:

- `Push` adds an item to the top.
- `Pop` removes and returns the top item.
- `Peek` returns the top item without removing it.
- `Count` reports how many items are present.

Stacks model undo histories, nested function calls, and matching delimiters. The structure is useful because it prevents callers from reaching into the middle of the sequence.

```csharp
var history = new Stack<string>();
history.Push("Open report");
history.Push("Add chart");
history.Push("Save report");

Console.WriteLine($"Next undo: {history.Pop()}");
Console.WriteLine($"Remaining action: {history.Peek()}");
Console.WriteLine($"Actions remaining: {history.Count}");
```
