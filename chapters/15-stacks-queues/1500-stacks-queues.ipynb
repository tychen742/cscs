# Stacks & Queues

Some problems need restricted access rather than arbitrary indexing. This chapter studies stacks, queues, and deques: sequence-like abstract data types that model order, waiting, undo, and scheduling.

- **Stack**: A stack uses last-in, first-out behavior for operations such as undo and nested processing.
- **Queue**: A queue uses first-in, first-out behavior for waiting lines and scheduled work.
- **Deque**: A deque allows insertion and removal at both ends.
- **Implementation choice**: Array-based and linked implementations support the same ADT with different trade-offs.
- **Application**: Restricted-access structures appear in parsing, scheduling, traversal, and history features.

Stacks and queues show how limiting operations can make a structure match a problem more precisely.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Specify stack, queue, and deque operations
2. Trace last-in, first-out and first-in, first-out behavior
3. Choose an implementation strategy for a restricted-access sequence
4. Apply stacks and queues to practical programming problems
5. Test stack and queue behavior against their ADT contracts
