```csharp
using System;
using System.Collections.Generic;

string expression = "((price + tax) * quantity)";
var openings = new Stack<char>();
bool balanced = true;

foreach (char ch in expression)
{
    if (ch == '(')
    {
        openings.Push(ch);
    }
    else if (ch == ')')
    {
        if (openings.Count == 0)
        {
            balanced = false;
            break;
        }

        openings.Pop();
    }
}

balanced = balanced && openings.Count == 0;
Console.WriteLine(balanced);
```
