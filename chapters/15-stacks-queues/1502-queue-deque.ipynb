# Queues and Deques

A **queue** is a restricted-access sequence that follows **first in, first out** (FIFO). The first item added is the first item removed.

A **deque** (double-ended queue) permits insertion and removal at both ends.

- Queues model service lines, print jobs, and event processing.
- Deques support worklists that can grow or shrink from either end.
- The interface should make invalid operations on an empty structure explicit.

The implementation choice depends on which end operations the application performs.

```csharp
var serviceLine = new Queue<string>();
serviceLine.Enqueue("Amina");
serviceLine.Enqueue("Diego");
serviceLine.Enqueue("Mina");

Console.WriteLine($"Serving: {serviceLine.Dequeue()}");
Console.WriteLine($"Waiting next: {serviceLine.Peek()}");

var worklist = new LinkedList<string>();
worklist.AddLast("normal task");
worklist.AddFirst("urgent task");
Console.WriteLine(worklist.First!.Value);
```
