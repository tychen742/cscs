# Advanced Searching and Sorting

A search result must say which match it promises, and a sorted report must preserve more than order. This chapter deepens Chapter 16 through duplicate-aware bounds, elementary-sort invariants, exact counts, and complete workload comparisons. Shell sort and quicksort remain supporting readings after the core sequence.

- **Match policy**: Distinguish first, any, and all matches under explicit equality rules.
- **Search bounds**: Locate duplicate groups and insertion positions in sorted data.
- **Sort invariants**: Explain why selection, insertion, and bubble sorts preserve every occurrence.
- **Stability and mutation**: Preserve tied-record order and state which collection changes.
- **Workload comparison**: Count preparation, repeated queries, updates, and auxiliary storage.

<h2>Learning Objectives</h2>


By the end of this chapter, you will be able to:

1. Specify first, any, and all-match search contracts with compatible equality and ordering policies.
2. Trace half-open binary search and derive lower/upper bounds for duplicate groups and insertion positions.
3. Explain selection, insertion, and optimized bubble sorts using invariants and occurrence preservation.
4. Derive exact selection comparisons and relate insertion shifts and adjacent swaps to inversions.
5. Verify stable record order, sorted permutations, and explicit mutation/library return policies.
6. Compare complete workloads including preparation, queries, input order, updates, and storage.
