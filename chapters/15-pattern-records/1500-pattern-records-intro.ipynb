# Pattern Matching & Records

<!-- chapter-intro -->

Modern C# gives you concise, expressive ways to test structure and model data. This chapter covers two
features that pair naturally: pattern matching for structural dispatch, and record types for data-carrying
types with built-in equality and immutability.

<h2>Why these features matter</h2>

- **Pattern matching** — replaces chains of `if`/`else` and `is`/`as` casts with clear, structural dispatch
- **Record types** — eliminate boilerplate for data-carrying types; you have been using tuples since Ch.9 — now we name what a full `record` gives you
- Together, records deconstruct naturally inside `switch` expressions and property patterns — the two features are designed to be used side by side

<h2>Learning goals</h2>

By the end of this chapter, you should be able to:

- use `switch` expressions and `is` patterns to match on type and structure
- explain what `record` adds over `class`: value equality, `with`, deconstruction, and `init`-only properties
- choose between `record`, `class`, and `struct` for a given design
- combine pattern matching and records to dispatch over a record hierarchy
