# Pattern Matching & Records

Modern C# gives you concise ways to test structure and model data. This chapter pairs pattern matching for structural dispatch with record types for data-carrying values.

- **Pattern matching**: Patterns replace many chains of casts and conditionals with clear structural checks.
- **Switch expression**: A switch expression selects a result from multiple pattern-based cases.
- **Record type**: A record reduces boilerplate for data-carrying types with value-style equality.
- **Immutability**: `init` properties and `with` expressions support safer updates by copying values.
- **Deconstruction**: Records and patterns can unpack grouped values for readable dispatch.

Pattern matching and records work especially well together when a program needs to name structured data and handle its different shapes.
