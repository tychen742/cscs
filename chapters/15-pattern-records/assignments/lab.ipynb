## Part 2: Record Types

**2a.** Define `record Temperature(double Celsius)`. Add a computed property `Fahrenheit`. Verify that two `Temperature(100)` instances are `==`.

**2b.** Define `record Person(string Name, int Age)`. Starting from `Person("Sam", 20)`, use `with` to produce ages 21, 22, 23 and print each.

````csharp
record Temperature(double Celsius)
{
    public double Fahrenheit => Celsius * 9.0 / 5.0 + 32;
}

var a = new Temperature(100);
var b = new Temperature(100);
Console.WriteLine(a == b);         // True
Console.WriteLine(a.Fahrenheit);   // 212

record Person(string Name, int Age);
var sam = new Person("Sam", 20);
// TODO: produce sam at ages 21, 22, 23 using `with`
````
