# Lab: Pattern Matching & Records — Solutions

Full solutions for both parts of the Ch.15 lab: pattern matching and records.

## Part 1: Pattern Matching

```python
static string Fizzbuzz(int n) => (n % 3, n % 5) switch
{
    (0, 0) => "FizzBuzz",
    (0, _) => "Fizz",
    (_, 0) => "Buzz",
    _      => n.ToString()
};

for (int i = 1; i <= 20; i++) Console.Write(Fizzbuzz(i) + " " );

static string GradeLabel(int score) => score switch
{
    >= 90 => "A",
    >= 80 => "B",
    >= 70 => "C",
    >= 60 => "D",
    _     => "F"
};
```

## Part 2: Record Types

```python
record Temperature(double Celsius)
{
    public double Fahrenheit => Celsius * 9.0 / 5.0 + 32;
}

var a = new Temperature(100);
var b = new Temperature(100);
Console.WriteLine(a == b);         // True
Console.WriteLine(a.Fahrenheit);   // 212

record Person(string Name, int Age);
var sam = new Person("Sam", 20);
for (int age = 21; age <= 23; age++)
{
    sam = sam with { Age = age };
    Console.WriteLine(sam);
}
```

## Stretch Challenge

```python
abstract record Shape;
record Circle(double Radius) : Shape;
record Rectangle(double Width, double Height) : Shape;
record Triangle(double Base, double Height) : Shape;

double Area(Shape s) => s switch
{
    Circle(var r) => Math.PI * r * r,
    Rectangle(var w, var h) => w * h,
    Triangle(var b, var h) => 0.5 * b * h,
    _ => 0
};

string Describe(Shape? s) => s switch
{
    null => "No shape",
    var shape => $"Area: {Area(shape)}"
};
```
