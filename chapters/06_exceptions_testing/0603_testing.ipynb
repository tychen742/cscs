## Introduction




The purpose of testing in software development is to establish developer’s confidence that the
code continues to have high quality and less errors, hence to ensure that the requirements of
the project are met.

Testing is a critical part of software development. In JetBrain’s The State of
Developer Ecosystem 2023 survey [1], for example, 80% of the
respondents say that “testing plays an integral role in their software development
projects.” Also, 63% of respondents say they use **Unit testing** in their projects while
47% say they use Integration testing.

It is also noteworthy that while 41% of respondents say they have less
than 1 tester/Quality Assurance Engineer per 10 developers, 32% of the respondents say
that they have 1-3 testers/QA per 10 software developers. Testing, especially unit testing,
therefore, is a common practice in real world software development.

Amazon AWS defines unit testing as:

> Unit testing is the process where you test the smallest functional unit of code.
> Software testing helps ensure code quality, and it’s an integral part of software
> development. It’s a software development best practice to write software as small,
> functional units then write a unit test for each code unit. You can first write
> unit tests as code. Then, run that test code automatically every time you make
> changes in the software code. This way, if a test fails, you can quickly isolate
> the area of the code that has the bug or error. Unit testing enforces modular
> thinking paradigms and improves test coverage and quality. Automated unit
> testing helps ensure you or your developers have more time to concentrate on coding. [2]

The notion of unit testing is straightforward in principle. When you write a program
in general, the program comprises what are properly known as units of development.
Each language has its own definition of what units are but most modern programming
languages view the **class** concept as the core unit of testing, although it can be done
at lower level such as **method** and higher level such as package. Once we have a class,
we can test it and all of the parts associated with it, especially its methods.

Unit testing is usually done by using testing frameworks. For C#, popular testing
frameworks include: MSTest, NUnit, and xUnit.

## Simple Testing




Assuming that you have a killer app like:

    namespace IntroCSCS
    {
       public class Calculator
       {
          public int Add(int x, int y)
          {
             return x + y;
          }
       }
    }

Let’s say you want to know that this app is doing what you intend for it to do. The best
way to achieve that validation is probably by providing a set of validation code
that does nothing but running the app to **test** if the results come out right, such as the
following:

````csharp
namespace IntroCSCS
{
    public class Calculator
    {
        public int Add(int x, int y)
        {
            return x + y;
        }
    }

    internal class Program
    {
        static void Main(string[] args)
        {
            var calculator = new Calculator();

            if (calculator.Add(15, 20) != 35)
            {
                throw new InvalidOperationException();
            }
            Console.WriteLine("OK");
        }
    }
}
````

Currently, there is only one method (Add) in the Calculator class. As you add more methods,
    you will need more cases and your Program.cs will become disorganized. This is where the
    established tooling comes in.

## The Unit Testing Process




One of the fundamental principles of adopting unit testing is to follow a
TDD (Test Driven Development) approach where you will:

1. Write a failing test (to prove the point that the test works)
2. Implement the functionality to pass the test
3. Refactor the code while keeping the test green

As an example, you can prepare test cases such as the code below. Note that the forth method
will fail.



````csharp
    public class BasicMaths
    {
       public double Add(double num1, double num2) {
          return num1 + num2;
       }
       public double Subtract(double num1, double num2) {
          return num1 - num2;
       }
       public double divide(double num1, double num2) {
          return num1 / num2;
       }
       public double Multiply(double num1, double num2) {
          // To trace error while testing, writing + operator instead of * operator.
          return num1 + num2;
       }
    }
````

## Design Tests Independently of the Implementation



```{index} gap revision: design tests independently of the implementation
```


A unit test checks a small behavior in isolation, such as one pricing method. An integration test checks collaborating parts, such as parsing a file and submitting its records. Passing a method test does not prove that the file path, parser, and caller work together. Start with a written contract and calculate expected results without calling the method under test.

Suppose an order of zero items costs zero, one through four items costs two dollars per item, and five or more costs nine dollars in total. A candidate that tests `quantity > 5` misses exactly five. Cases just below, at, and just above a decision boundary distinguish that defect from an arithmetic mistake.

| Case | Requirement-derived expectation | What it checks |
| --- | --- | --- |
| Zero | 0 | Empty order. |
| Four | 8 | Lower-price rule before threshold. |
| Five | 9 | Inclusive threshold. |
| Six | 9 | Upper rule after threshold. |
| Negative | Rejection | Invalid-domain contract. |

For debugging, record input, expected result, actual result, a hypothesis, and an observation that can disprove it. Put a breakpoint at the branch and watch the condition for five. After fixing the comparison, rerun the same cases as regression checks. Adding the failing result to the expected values would hide the defect rather than fix it.

See the language/API reference for this contract.[^gap-mstest]


**Guided practice.** Create a test table for a different threshold of seven before seeing code. Explain which three adjacent quantities distinguish `>` from `>=`. Translate one row into Arrange, Act, Assert in MSTest. Preserve the original failing test while repairing the method.

The complete MSTest workflow below replaces the earlier mixed NUnit/MSTest example. `dotnet new mstest` creates the test project and its framework dependencies; `dotnet add ... reference ...` adds a reference to application code, not a test framework. MSTest discovers `[TestClass]` classes and `[TestMethod]` methods. In these introductory tests, each method is public, returns void, and takes no parameters. ### Prepare Tests



Replace the template test source in MathAppTest with the complete listing below. If the template generated an example test file under another name, remove that example so the intended suite contains exactly these four tests. Keep its generated project file and package references.

The TestClass and TestMethod attributes let the test framework discover tests. Each test creates a fresh BasicMath, calls one method, and checks an independently chosen expected result. `Assert.AreEqual(expected, actual)` belongs to MSTest and runs as part of the tests; it is not a debug-only assertion or an interactive error dialog.

```csharp
using Microsoft.VisualStudio.TestTools.UnitTesting;
using SomeMath;
namespace MathAppTest;
[TestClass]
public class UnitTest1
{
    [TestMethod]
    public void Test_AddMethod()
    {
        BasicMath bm = new BasicMath();
        double result = bm.Add(10, 10);
        Assert.AreEqual(20.0, result);
    }
    [TestMethod]
    public void Test_SubtractMethod()
    {
        BasicMath bm = new BasicMath();
        double result = bm.Subtract(10, 10);
        Assert.AreEqual(0.0, result);
    }
    [TestMethod]
    public void Test_DivideMethod()
    {
        BasicMath bm = new BasicMath();
        double result = bm.Divide(10, 5);
        Assert.AreEqual(2.0, result);
    }
    [TestMethod]
    public void Test_MultiplyMethod()
    {
        BasicMath bm = new BasicMath();
        double result = bm.Multiply(10, 10);
        Assert.AreEqual(100.0, result);
    }
}
```

These integer-valued results are exactly representable as doubles. For calculations involving rounding, use an appropriate tolerance and explain it; do not assume every floating-point computation should match exactly.

**Contract:** Positive order quantities below 10 cost 5.00 to ship. Quantities of 10 or more ship free. Reject zero and negative quantities with `ArgumentOutOfRangeException`.

A prepared candidate uses `units > 10` and offers only tests for 3 and 12 units. Those tests pass, but do not cover the full contract. Derive expected results from the contract before looking at the candidate's result.

**Possible AI prompt:** “Review this shipping contract and these two tests. Suggest missing boundary and invalid-input cases with expected outcomes derived from the contract. Explain each choice.” Verify the suggestions independently; generating implementation and expected results from the same mistake can make a faulty program appear correct.


````csharp
// Prepared candidate: intentionally incomplete behavior; both ordinary tests pass.
using System;

static decimal ShippingFee(int units)
{
    return units > 10 ? 0m : 5m;
}

static void CheckFee(int units, decimal expected)
{
    decimal actual = ShippingFee(units);
    Console.WriteLine($"{(actual == expected ? "PASS" : "FAIL")} fee for {units}: {actual:F2}");
}


CheckFee(3, 5m);
CheckFee(12, 0m);
````

````csharp
// Solution: Improve tests, then repair shipping
using System;

static decimal ShippingFee(int units)
{
    if (units <= 0) throw new ArgumentOutOfRangeException(nameof(units));
    return units >= 10 ? 0m : 5m;
}

static void CheckFee(int units, decimal expected)
{
    decimal actual = ShippingFee(units);
    Console.WriteLine($"{(actual == expected ? "PASS" : "FAIL")} fee for {units}: {actual:F2}");
}

static void CheckRejected(int units)
{
    try
    {
        ShippingFee(units);
        Console.WriteLine($"FAIL accepted {units}");
    }
    catch (ArgumentOutOfRangeException)
    {
        Console.WriteLine($"PASS rejected {units}");
    }
}

CheckFee(3, 5m);
CheckFee(12, 0m);
CheckFee(10, 0m);
CheckRejected(0);
CheckRejected(-1);
````
