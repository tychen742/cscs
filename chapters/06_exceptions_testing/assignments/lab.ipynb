## Preparing Project

The idea of TDD is to write tests first, so that you will be focused with the requirements and not
to be distracted by your own coding. That's why we start with simple test cases in the test project to test the
units in the application project. After you make sure the application tests can pass, you then improve the
units (methods, classes) in the application project.

The process of creating a unit test involves several steps:

1. Create a lab folder, call it `Ch06ReliabilityLab`, inside `[USERNAME]\workspace\introcscs\`.

2. Create a solution folder:

   1. Inside Ch06ReliabilityLab, create a folder `UnitTestProject1` and run:

      ```bash
      dotnet new sln
      ```

   2. To open the solution in VS Code: `View` → `Command Palette` → `.NET: Open Solution`.

3. Create an **application project** folder `MathApp`, and a **test project** folder `MathAppTest` inside UnitTestProject1.

4. From the `UnitTestProject1` directory that contains the solution, initialize the projects without changing directories:

   ```bash
   dotnet new console -o MathApp
   dotnet new mstest -o MathAppTest
   ```

5. Still in `UnitTestProject1`, add both projects to the solution and link the test project to application code:

   ```bash
   dotnet sln add MathApp/MathApp.csproj
   dotnet sln add MathAppTest/MathAppTest.csproj
   dotnet add MathAppTest/MathAppTest.csproj reference MathApp/MathApp.csproj
   ```

The `-o` option chooses an output folder; it does not change the terminal's working directory. Keep all commands above and `dotnet test` in `UnitTestProject1`. If your terminal is already inside either project, run `cd ..` first. The solution extension depends on the installed SDK; `dotnet sln` finds the solution in the current directory. For the complete source files and their intentional defect, use the {doc}`MSTest project lesson <../0603_testing>` before the walkthrough linked below.
