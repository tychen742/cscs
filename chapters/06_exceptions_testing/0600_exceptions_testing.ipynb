# Testing, Debugging, and Error Handling

An invalid quantity, a failing file operation, and a wrong total call for different responses. This chapter brings the earlier checking habits together into validation, intentional exception handling, evidence-based debugging, and repeatable unit tests. You will distinguish reporting an expected failure from investigating a defect and verify that a repair preserves working cases.

- **Input validation**: Checking values before risky operations prevents many avoidable failures.
- **Exception handling**: `try`, `catch`, and related constructs let a program respond intentionally to runtime errors.
- **Debugging**: Breakpoints, watches, and step-by-step execution help you inspect what a program is actually doing.
- **Unit testing**: Small repeatable tests build confidence that code keeps working as it changes.
- **Missing input**: Basic null checks distinguish absent input from valid values. Deeper nullable-type and object-contract discussion follows in Chapter 11.

<h2>Learning Objectives</h2>


By the end of this chapter, you will be able to:

1. Choose validation or exception handling for a stated failure and use `TryParse` for expected invalid input.
2. Trace `try`, matching `catch`, and `finally` paths and report or throw failures intentionally.
3. Reproduce a defect, form a hypothesis, and use breakpoints, watches, stepping, and the call stack to test it.
4. Write repeatable method tests with explicit expected results and meaningful normal, boundary, and failure cases.
5. Verify a repair with regression checks and evaluate whether proposed tests challenge the actual contract.
