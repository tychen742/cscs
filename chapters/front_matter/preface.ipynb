## Why Coding When AI Writes Code

AI assistants now generate working code from a short description. That makes the skills in this book more important, not less. Someone still has to state precisely what a program should do, decides whether the code an assistant produces is correct, work with the agents to make sure the development is in the right direction, and judge which tradeoffs make better sense. 

Those are the skills this book practices. You will trace method calls by hand (Chapter 5), test and debug code systematically (Chapter 6), state preconditions and loop invariants, and compare operation costs using Big-O notation (Chapter 13), with deeper analysis in Chapter 21. A developer who can do these things uses AI as a fast assistant. A developer who cannot is left trusting output they are unable to check.

## Course Pathways



**Programming and Information Systems Foundations (Chapters 1-12)** can serve as a first-semester course. Students progress from small console programs to methods, collections, files, classes, and object-oriented design. Integration, testing, documentation, and a project demonstration bring these skills together.

**Data Structures and Algorithms Foundations (Parts IV-V, Chapters 13-20)** can serve as a second course for students who already have the first course's programming foundation. Parts IV-V develop representation choices and their algorithmic foundations. Parts VI-VII (Chapters 21-26) provide optional deeper analysis and design. An instructor using this portion independently should establish equivalent preparation in methods, collections, classes, and interfaces.

**CS1 can use Parts I-III; CS2 can use Parts IV-V with selected Parts VI-VII chapters.** Chapters 1-12 provide the applied foundation, and instructors select Parts IV-V depth to fit their course goals. Parts IV-V include basic Big-O, linear and binary search, library sorting, recursion for tree operations, and breadth-first and depth-first graph traversal. Students can study these foundations without taking Parts VI-VII. Instructors may use the optional algorithms chapters for a technical specialization or further study.

**Recommended CS2 sorting route:** include {doc}`Recursive Methods <../05_methods/0506_recursive_methods>` before {doc}`Merge Sort <../16_searching_sorting/1603_merge_sort>`. These supporting readings supply the call-stack preparation and a complete efficient comparison sort without requiring Parts VI-VII. A lighter applied route may retain the core search, insertion-sort, and library-workload treatment; it should state that narrower outcome. For an algorithms route, select Chapters 21-23 and include {doc}`Quicksort and Partitioning <../23_searching_sorting/2305_quicksort>` alongside Chapter 22's merge-sort analysis. Instructors allocate practice and assessment time for selected extensions explicitly; the standard assignment scope is unchanged.

A chapter is a weekly organizing unit, with roughly 100-120 minutes of planned lecture and additional practice. Course schedules also need time for review, assessment, and project integration; instructors can adjust the depth and pace rather than assign every topic in the volume.

The {ref}`Curriculum Alignment appendix <cs2023_alignment>` documents the book's contributions to both Information Systems (IS2020) and Computer Science (CS2023) curriculum guidance. The mappings help instructors identify relevant material and coverage boundaries; neither represents complete degree-program coverage.

## Why C\#

C# is a general-purpose, statically typed language at the center of Microsoft's open-source, cross-platform .NET platform. It is a good first language for this book for four reasons:

- **Types are visible.** Every variable has a declared type that the compiler checks before the program runs. Beginners see how data is represented, and many mistakes are caught as compile errors instead of surprising results.
- **It is used where business software is built.** C# and .NET power enterprise web services, internal business applications, cloud back ends, and database-driven systems, the same kind of information systems this book builds toward.
- **It is free and runs everywhere.** The .NET SDK and VS Code run on Windows, macOS, and Linux, so the tools you prepare with Appendix B and use for the first program in Chapter 1 are the tools professionals use.
- **The ideas transfer.** Classes, interfaces, generics, collections, and exceptions in C# map closely to Java, TypeScript, and other modern languages. The data structures and algorithms in Parts IV-VII apply in any language.

C# is widely used beyond business systems: the Unity game engine, ASP.NET web applications, and desktop and mobile apps are written in it. C# also holds a steady place among the most used languages.
