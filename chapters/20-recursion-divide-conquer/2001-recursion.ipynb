# Recursive Problem Solving

Recursion solves a problem by asking the same question about a smaller input. Every recursive method needs:

- a **base case** that can be answered directly;
- a **recursive step** that moves toward the base case; and
- a clear meaning for the value returned by the recursive call.

Trace the call stack before writing code. This makes the order of calls and returns visible and helps distinguish recursion from an accidental infinite call.

```csharp
static int SumDown(int value)
{
    if (value <= 0)
        return 0;

    return value + SumDown(value - 1);
}

Console.WriteLine(SumDown(4));
Console.WriteLine(SumDown(0));
```
