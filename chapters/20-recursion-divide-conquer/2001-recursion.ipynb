A recursive method needs two promises. First, it must know when to stop. Second, every recursive call must move closer to that stopping point.

Recursion is not magic; it is ordinary method calls plus the call stack. The hard part is choosing a smaller version of the problem that has the same shape as the original problem.


## Base Case and Recursive Case

A **base case** returns an answer directly. A **recursive case** calls the same method on a smaller input and uses that result to build the current answer.


````csharp
using System;

Console.WriteLine(SumDown(4));
Console.WriteLine(SumDown(0));

int SumDown(int value)
{
    if (value <= 0)
    {
        return 0;
    }

    return value + SumDown(value - 1);
}
````

The progress measure is `value`. Each recursive call subtracts one, so the method eventually reaches the base case.

## Tracing the Call Stack

The call stack stores each active method call until that call returns. A trace can make the order of calls and returns visible.


````csharp
using System;

Console.WriteLine($"Result: {Factorial(4)}");

int Factorial(int n)
{
    Console.WriteLine($"call Factorial({n})");

    if (n <= 1)
    {
        Console.WriteLine($"return 1 from Factorial({n})");
        return 1;
    }

    int result = n * Factorial(n - 1);
    Console.WriteLine($"return {result} from Factorial({n})");
    return result;
}
````

Calls move downward toward the base case. Returns move upward as each suspended call finishes its remaining work.

## Progress Measures

A progress measure proves that recursion is moving in the right direction. The measure does not have to be a single number, but it must get closer to a base case.


````csharp
using System;

int[] values = { 2, 4, 6, 8, 10 };
Console.WriteLine(Contains(values, 6, 0));
Console.WriteLine(Contains(values, 7, 0));

bool Contains(int[] values, int target, int index)
{
    if (index >= values.Length)
    {
        return false;
    }

    if (values[index] == target)
    {
        return true;
    }

    return Contains(values, target, index + 1);
}
````

The progress measure is the number of unchecked positions. Each call increases `index`, so fewer positions remain.

## Recursion and Iteration

Many recursive methods can be written with loops. Use recursion when the problem naturally breaks into smaller problems of the same shape, or when recursion makes the structure easier to see.


````csharp
using System;

int n = 5;
Console.WriteLine($"recursive: {RecursiveSum(n)}");
Console.WriteLine($"iterative: {IterativeSum(n)}");

int RecursiveSum(int n)
{
    if (n <= 0)
    {
        return 0;
    }

    return n + RecursiveSum(n - 1);
}

int IterativeSum(int n)
{
    int total = 0;
    for (int current = n; current > 0; current--)
    {
        total += current;
    }
    return total;
}
````

For simple counting, a loop is often clearer and avoids call-stack growth. Recursion becomes more useful for nested structures, search spaces, and divide-and-conquer algorithms.

## Recursive Risks

A recursive method with a missing base case or no progress can call itself forever until the program runs out of stack space. In C#, recursive calls are not something to use casually for huge linear loops.


````csharp
using System;

Console.WriteLine(SafeCountdown(3));

int SafeCountdown(int n)
{
    if (n <= 0)
    {
        return 0;
    }

    return 1 + SafeCountdown(n - 1);
}
````

Before writing a recursive method, say the base case out loud and name the progress measure. That habit catches many infinite-recursion bugs before they start.
