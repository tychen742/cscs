# Recursion

Recursive programs solve a problem by solving smaller versions of the same problem. This chapter connects recursive program structure to divide-and-conquer algorithm design.

- **Base case**: A base case stops recursion when the problem is small enough to answer directly.
- **Recursive case**: A recursive case makes progress by calling the same method on a smaller input.
- **Call stack**: The call stack keeps track of active recursive calls and their local state.
- **Divide and conquer**: A divide-and-conquer algorithm splits a problem, solves parts, and combines results.
- **Merge sort**: Merge sort demonstrates recursive division plus a combining step.

Recursion is both a programming technique and a way to design algorithms whose structure mirrors the problem itself.


<h2>Learning Objectives</h2>

By the end of this chapter, you will be able to:
1. Identify base cases and recursive progress in a method
2. Trace recursive calls with a call-stack model
3. Design a divide-and-conquer algorithm for a simple problem
4. Explain how merge sort divides, solves, and combines data
5. Connect recursive structure to recurrence-based cost analysis
