This lab implements merge sort in stages. The goal is not just to get sorted output, but to connect code structure to the divide-and-conquer pattern: base case, divide, recursive solve, and combine.

## Start With Merge

Write and test the combine step first. If merge is wrong, merge sort will also be wrong.


````csharp
using System;

int[] left = { 1, 5, 9 };
int[] right = { 2, 4, 8, 10 };
int[] merged = Merge(left, right);

Console.WriteLine(string.Join(", ", merged));

int[] Merge(int[] left, int[] right)
{
    int[] result = new int[left.Length + right.Length];
    int i = 0;
    int j = 0;
    int k = 0;

    while (i < left.Length && j < right.Length)
    {
        if (left[i] <= right[j])
        {
            result[k++] = left[i++];
        }
        else
        {
            result[k++] = right[j++];
        }
    }

    while (i < left.Length)
    {
        result[k++] = left[i++];
    }

    while (j < right.Length)
    {
        result[k++] = right[j++];
    }

    return result;
}
````

## Add the Recursive Structure

Merge sort's base case is an array of length zero or one. Such an array is already sorted.


````csharp
using System;

int[] values = { 7, 2, 6, 3, 1, 5, 4 };
int[] sorted = MergeSort(values);

Console.WriteLine(string.Join(", ", sorted));

int[] MergeSort(int[] values)
{
    if (values.Length <= 1)
    {
        return values;
    }

    int middle = values.Length / 2;
    int[] left = values[..middle];
    int[] right = values[middle..];

    int[] sortedLeft = MergeSort(left);
    int[] sortedRight = MergeSort(right);
    return Merge(sortedLeft, sortedRight);
}

int[] Merge(int[] left, int[] right)
{
    int[] result = new int[left.Length + right.Length];
    int i = 0;
    int j = 0;
    int k = 0;

    while (i < left.Length && j < right.Length)
    {
        result[k++] = left[i] <= right[j] ? left[i++] : right[j++];
    }

    while (i < left.Length)
    {
        result[k++] = left[i++];
    }

    while (j < right.Length)
    {
        result[k++] = right[j++];
    }

    return result;
}
````

## Test the Result

A sorting test should check more than one example. Include empty arrays, one-item arrays, repeated values, and already sorted values.


````csharp
using System;
using System.Linq;

int[][] tests = {
    Array.Empty<int>(),
    new[] { 4 },
    new[] { 3, 1, 2 },
    new[] { 1, 1, 3, 2 },
    new[] { 1, 2, 3, 4 }
};

foreach (int[] test in tests)
{
    int[] sorted = MergeSort(test);
    Console.WriteLine($"[{string.Join(", ", test)}] -> [{string.Join(", ", sorted)}] ok={IsSorted(sorted)}");
}

bool IsSorted(int[] values)
{
    return values.Zip(values.Skip(1), (a, b) => a <= b).All(result => result);
}

int[] MergeSort(int[] values)
{
    if (values.Length <= 1)
    {
        return values;
    }

    int middle = values.Length / 2;
    return Merge(MergeSort(values[..middle]), MergeSort(values[middle..]));
}

int[] Merge(int[] left, int[] right)
{
    int[] result = new int[left.Length + right.Length];
    int i = 0;
    int j = 0;
    int k = 0;

    while (i < left.Length && j < right.Length)
    {
        result[k++] = left[i] <= right[j] ? left[i++] : right[j++];
    }

    while (i < left.Length)
    {
        result[k++] = left[i++];
    }

    while (j < right.Length)
    {
        result[k++] = right[j++];
    }

    return result;
}
````

A test can show whether the output is sorted. If the assignment also requires preserving the original items, add a count-based test for duplicates.

## Trace the Splits

A short trace helps connect the code to the recurrence `T(n) = 2T(n / 2) + O(n)`.


````csharp
using System;

int[] values = { 6, 2, 5, 1 };
TraceMergeSort(values, 0);

void TraceMergeSort(int[] values, int depth)
{
    Console.WriteLine($"{new string(' ', depth * 2)}sort [{string.Join(", ", values)}]");

    if (values.Length <= 1)
    {
        return;
    }

    int middle = values.Length / 2;
    TraceMergeSort(values[..middle], depth + 1);
    TraceMergeSort(values[middle..], depth + 1);
}
````

The trace shows the recursive shape without the merge details. For four items, the split tree has three levels: size 4, size 2, and size 1.

## Compare Counts

For a small input, estimate the number of comparison opportunities in merge sort and a quadratic nested-loop sort. This is a count-based comparison, not a timing benchmark.


````csharp
using System;

int n = 32;
double mergeEstimate = n * Math.Log2(n);
int quadraticEstimate = n * n;

Console.WriteLine($"n = {n}");
Console.WriteLine($"n log2 n estimate: {mergeEstimate:F0}");
Console.WriteLine($"n^2 estimate: {quadraticEstimate}");
````

The exact number of comparisons depends on implementation and data order. The growth pattern is the main lesson: `n log n` grows much more slowly than `n^2`.
