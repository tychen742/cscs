# OOP

<!-- chapter-intro -->

Classes alone do not make a program object-oriented. The four pillars — encapsulation, inheritance, polymorphism, and abstraction — turn a collection of classes into a flexible, extensible design. This chapter covers each pillar with concrete C# examples.


<h2>Why this chapter matters</h2>

OOP is the dominant paradigm in enterprise and systems programming. Mastering it unlocks:

- reusable base classes that capture shared behavior
- derived classes that extend or specialize without rewriting
- interfaces and abstract classes that define contracts
- polymorphism that lets a single method call work across many types


<h2>Learning goals</h2>

By the end of this chapter, you should be able to:

- apply encapsulation using access modifiers and properties
- create a class hierarchy using `class` and `: BaseClass` syntax
- override methods and use `virtual` / `override` correctly
- write code against an interface or abstract class
