<h2>Glossary</h2>

| # | Term | Explanation |
|---:|---|---|
| 1 | Collection | An object that stores multiple values and provides operations for working with them. |
| 2 | Generic type | A type written with a type parameter, such as `List<T>`. |
| 3 | Type parameter | A placeholder for the element type a generic collection stores. |
| 4 | List | A resizable ordered collection accessed by index. |
| 5 | Dictionary | A collection that maps keys to values. |
| 6 | Key | A lookup value used to find an associated dictionary value. |
| 7 | Value | The data associated with a dictionary key. |
| 8 | Add | An operation that inserts a new item into a collection. |
| 9 | Remove | An operation that deletes an item from a collection. |
| 10 | Contains | A membership test that checks whether a collection has an item or key. |
| 11 | Count | The number of items currently stored in a collection. |
| 12 | Foreach loop | A loop that visits each item in a collection. |
| 13 | Lookup | Retrieving a value by index or key. |
| 14 | Access pattern | The way code most often reads, writes, searches, or traverses data. |
| 15 | Collection contract | The operations and behavioral rules a collection promises. |
| 16 | Element | One item stored in a collection. |
| 17 | Sequence | A group whose elements have positions in an order. |
| 18 | Linked list | A sequence organized as linked nodes rather than indexed positions. |
| 19 | Node | An element together with links to neighboring nodes in a linked structure. |
| 20 | Set | A collection of distinct elements used for membership tests. |
| 21 | Stack | A collection whose most recently added item is removed first. |
| 22 | Queue | A collection whose earliest added item is removed first. |
| 23 | Priority queue | A collection that selects the next item according to priority. |
| 24 | Type argument | The actual type supplied for a generic type parameter. |
| 25 | Traversal | Visiting a collection's elements, also called enumeration. |
| 26 | Alias | Another variable referring to the same collection object. |
