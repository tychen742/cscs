# Data Collections

An inventory report may need ordered items, a lookup by product ID, or a record of which IDs have already appeared. Arrays are one collection family, but different access rules support different tasks. This chapter compares those families and their shared operations, then develops lists and dictionaries through creation, updates, traversal, missing-data policies, and reference behavior.

- **Collection families**: Arrays, lists, linked lists, dictionaries, sets, stacks, queues, and priority queues provide different access rules.
- **List**: A `List<T>` stores an ordered sequence that can grow or shrink.
- **Dictionary**: A `Dictionary<TKey, TValue>` maps keys to values for fast lookup.
- **Generics**: Generic type parameters keep collections type-safe while making them reusable.
- **Access pattern**: Choosing a collection depends on whether you need order, lookup by key, insertion, removal, or traversal.

<h2>Learning Objectives</h2>


By the end of this chapter, you will be able to:

1. Match the introductory collection families to their access rules and choose a family for a stated business task.
2. Compare shared collection operations while using the correct API for an array, list, dictionary, set, stack, or queue.
3. Create and update type-safe `List<T>` values and explain indexing, count, capacity, and traversal behavior.
4. Store and retrieve `Dictionary<TKey, TValue>` entries using an explicit key and missing-value policy.
5. Distinguish collection reference assignment from copying elements, and explain the limits of a shallow copy.
6. Check collection operations with empty inputs, duplicates, missing keys, and mutations that affect traversal.



<h2>Glossary</h2>


| # | Term | Explanation |
|---:|---|---|
| 1 | Collection | An object that stores multiple values and provides operations for working with them. |
| 2 | Generic type | A type written with a type parameter, such as `List<T>`. |
| 3 | Type parameter | A placeholder for the element type a generic collection stores. |
| 4 | List | A resizable ordered collection accessed by index. |
| 5 | Dictionary | A collection that maps keys to values. |
| 6 | Key | A lookup value used to find an associated dictionary value. |
| 7 | Value | The data associated with a dictionary key. |
| 8 | Add | An operation that inserts a new item into a collection. |
| 9 | Remove | An operation that deletes an item from a collection. |
| 10 | Contains | An API name used for membership checks by several collections; dictionary key lookup uses `ContainsKey` or `TryGetValue` instead. |
| 11 | Count | The number of items currently stored in a collection. |
| 12 | Foreach loop | A loop that visits each item in a collection. |
| 13 | Lookup | Retrieving a value by index or key. |
| 14 | Access pattern | The way code most often reads, writes, searches, or traverses data. |
| 15 | Collection contract | The operations and behavioral rules a collection promises. |
| 16 | Element | One item stored in a collection. |
| 17 | Sequence | A group whose elements have positions in an order. |
| 18 | Linked list | A sequence organized as linked nodes rather than indexed positions. |
| 19 | Node | An element together with links to neighboring nodes in a linked structure. |
| 20 | Set | A collection of distinct elements used for membership tests. |
| 21 | Stack | A collection whose most recently added item is removed first. |
| 22 | Queue | A collection whose earliest added item is removed first. |
| 23 | Priority queue | A collection that selects the next item according to priority. |
| 24 | Type argument | The actual type supplied for a generic type parameter. |
| 25 | Traversal | Visiting a collection's elements, also called enumeration. |
| 26 | Alias | Another variable referring to the same collection object. |
| 27 | Push | Add an item at the top of a stack. |
| 28 | Pop | Remove and return the top stack item. |
| 29 | Peek | Inspect the next item without removing it. |
| 30 | Enqueue | Add an item at the back of a queue. |
| 31 | Dequeue | Remove and return the front queue item. |
| 32 | Shallow copy | A new collection containing copied values or references rather than independent copies of every referenced object. |


<h2>Lecture Slides</h2>


The chapter lecture deck is in preparation. Use the content sections and their worked examples for the current reading sequence.
