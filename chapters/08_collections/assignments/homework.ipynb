**Coding 1: Reorder Report**

Implement the request-count and threshold report described above. Apply ordinal uppercase normalization to product codes; retain repeated requests. Check empty input and a threshold of 2.


````csharp
// Exercise: Reorder Report
// exercise-id: cscs-collections-transfer-1
using System;
using System.Collections.Generic;
List<string> requests = new List<string> { "PEN", "BOOK", "PEN", "BAG", "BOOK", "PEN" };
int threshold = 2;
// Your code starts here.

// Your code ends here.
````

````csharp
// Solution: Reorder Report
using System;
using System.Collections.Generic;
List<string> requests = new List<string> { "PEN", "BOOK", "PEN", "BAG", "BOOK", "PEN" };
int threshold = 2;
Dictionary<string, int> counts = new Dictionary<string, int>();
foreach (string request in requests)
{
    string key = request.ToUpperInvariant();
    counts.TryGetValue(key, out int previous);
    counts[key] = previous + 1;
}
List<string> keys = new List<string>(counts.Keys);
keys.Sort(StringComparer.Ordinal);
Console.WriteLine("All requests:");
foreach (string key in keys) Console.WriteLine($"{key}: {counts[key]}");
Console.WriteLine("At threshold:");
foreach (string key in keys)
    if (counts[key] >= threshold) Console.WriteLine($"{key}: {counts[key]}");
````

**Coding 2: Cancel One Request**

Remove only the first request equal to PEN; then insert CARD at position 1 and print the sequence and count. Check a missing cancellation code and an initially empty list; define how you would handle an invalid insertion index.

````csharp
// Solution: Cancel One Request
using System;
using System.Collections.Generic;
List<string> requests = new List<string> { "PEN", "BOOK", "PEN", "BAG" };
string canceled = "PEN";
bool removed = requests.Remove(canceled);
if (requests.Count >= 1) requests.Insert(1, "CARD");
else requests.Add("CARD");
Console.WriteLine($"Canceled: {removed}");
Console.WriteLine(string.Join(", ", requests));
Console.WriteLine($"Count: {requests.Count}");
````

````csharp
// Solution: Record Revised Prices
using System;
using System.Collections.Generic;
Dictionary<string, decimal> prices = new Dictionary<string, decimal> { ["PEN"] = 2m, ["BOOK"] = 10m };
Dictionary<string, decimal> updates = new Dictionary<string, decimal> { ["PEN"] = 3m, ["BAG"] = 5m };
foreach (KeyValuePair<string, decimal> update in updates) prices[update.Key] = update.Value;
List<string> keys = new List<string>(prices.Keys);
keys.Sort(StringComparer.Ordinal);
foreach (string key in keys) Console.WriteLine($"{key}: {prices[key]:F2}");
Console.WriteLine($"Products: {prices.Count}");
````
