Arrays are fine if you know ahead of time how long your sequence of items is.
Then you create your array with that length, and you are all set.

If you want a variable sized container, you are likely to want a `List`.
List is a type of collection that is used when the number of elements is unknown.
As with arrays, you might want a collection of any particular type.
Unfortunately, you cannot use the simple notation of arrays to specify
the type of element in a `List`. Array syntax is
*built into* the language. Lists are handled in the *library* of types
provided by C# from .NET.
There are many situations where one general operation should work with different element types.

```{index} generics < > for generics
```



A bookstore basket starts empty and changes as a customer selects or cancels items. We need the indexed access and traversal learned with arrays, but we cannot predict the final number of items. The big idea is a **resizable ordered sequence**: `List<T>` provides indexed positions while managing changes in size.

Follow one distinction throughout this lesson: changing a value at an existing position is an update; inserting or removing an item changes the sequence and can change later positions. Use the operation comparison in {ref}`collection-operation-syntax` as a map while learning these calls.

The library operation contracts are documented in the .NET API reference.[^0802_list-list-api]

## Generics



C# collection types use *generics*: a type in angle brackets after a class name tells the compiler what element type the collection holds. The generic collection types live in the `System.Collections.Generic` namespace, which some project templates import through implicit usings. Include `using System.Collections.Generic;` when those imports are not enabled.

The generic list syntax is:

```csharp
List<T>
```

To create a list, supply a type argument for the elements it will store:

The declaration and construction patterns are:

```text
List<elementType> name = new List<elementType>();
List<elementType> name = new List<elementType> { item1, item2 };
List<elementType> copy = new List<elementType>(existingSequence);
```

These are templates. Replace `elementType` with a type such as `string` and supply actual values. Angle brackets choose the stored type. Parentheses supply constructor arguments; an empty argument list creates an empty list. The initializer's braces supply starting items. Copying into a new list differs from assigning another variable to the same list; the last example copies the elements into a new outer collection.

## Lists Basics



A `List<T>` starts with an element type and then supports the everyday sequence operations you expect: add an item, count the items, read an item by index, and loop through the items. The index rules are the same as arrays: the first item is at index `0`.


````csharp
using System;
using System.Collections.Generic;

List<int> numbers = new List<int>();
numbers.Add(10);
numbers.Add(20);
numbers.Add(30);

Console.WriteLine(numbers[0]);
Console.WriteLine(numbers.Count);

foreach (int number in numbers)
{
    Console.WriteLine(number);
}
````

A list can also be initialized with values. A `for` loop is useful when the index is part of the output or logic; a `foreach` loop is simpler when you only need each value.

`Count` is the number of stored elements. `Capacity` is the number of positions reserved internally before another growth allocation is needed; it can exceed `Count`. `Clear()` removes the elements but normally keeps that reserved capacity. Use `Count` for valid element indexes, not `Capacity`. You do not need to tune capacity for the examples in this chapter.


````csharp
using System;
using System.Collections.Generic;

List<string> names = new List<string> { "Alice", "Bob", "Charlie" };

for (int i = 0; i < names.Count; i++)
{
    Console.WriteLine($"{i}: {names[i]}");
}
````

Starting from `10, 20, 30`, insert `15` at index 1 and then remove the element at index 0 so the output is `15, 20, 30` with a count of 3.


````csharp
using System;
using System.Collections.Generic;

List<int> numbers = new List<int> { 10, 20, 30 };

// Your code starts here.
// Insert 15 at index 1, then remove the element at index 0.
// Your code ends here.

Console.WriteLine(string.Join(", ", numbers));
Console.WriteLine(numbers.Count);
````

````csharp
using System;
using System.Collections.Generic;

// Solution
List<int> numbers = new List<int> { 10, 20, 30 };

numbers.Insert(1, 15);   // 10, 15, 20, 30
numbers.RemoveAt(0);     // 15, 20, 30

Console.WriteLine(string.Join(", ", numbers));
Console.WriteLine(numbers.Count);
````

Unlike an array, the list's `Count` changed twice, up to 4 and back to 3, and the list did the shifting itself. `RemoveAt(0)` takes an index; `Remove(10)` would take a value, and the two are easy to confuse when the values are small integers.

## Lists Constructors and Methods



We can explore `List` methods with standalone C# examples. When you want to display the contents of a `List`, use a loop or `string.Join(...)`; printing the `List` object itself displays its type name, not its items.


````csharp
using System;
using System.Collections.Generic;
List<string> words = new List<string>();
Console.WriteLine($"Count: {words.Count}");
Console.WriteLine($"Items: [{string.Join(", ", words)}]");
````

With the no-parameter `constructor` (the parentheses at the end of the variable declaration), the `List` is empty to start:


````csharp
using System;
using System.Collections.Generic;
List<string> words = new List<string>();
Console.WriteLine($"Count: {words.Count}");
Console.WriteLine($"Items: [{string.Join(", ", words)}]");
````

You can add elements, and keep count with the `Count` property
as the size changes:

````csharp
using System;
using System.Collections.Generic;
List<string> myNames = new List<string>();
myNames.Add("Alice");
myNames.Add("Bob");
myNames.Add("Charles");
Console.WriteLine($"Count: {myNames.Count}; names: {string.Join(", ", myNames)}");
````

You can reference and change elements by index, like with `arrays`:

````csharp
using System;
using System.Collections.Generic;
List<string> words = new List<string> { "Apple", "Banana", "Cherry" };
Console.WriteLine(words[0]);
Console.WriteLine(words[2]);
words[2] = "Coconut";
Console.WriteLine(string.Join(", ", words));
````

You can use `foreach` like with arrays or other sequences:

````csharp
using System;
using System.Collections.Generic;
List<string> myNames = new List<string> { "Alice", "Bob", "Charles" };

foreach (string name in myNames)
{
    Console.WriteLine(name);
}
````

````csharp
using System;
using System.Collections.Generic;
List<string> words = new List<string> { "Apple", "Banana", "Cherry" };
Console.WriteLine(words);
````

Next, compare `Remove`, which finds the first matching element and removes it,
and `RemoveAt`, which removes the element at a specified index.
`Remove` returns whether the List has been changed:

````csharp
using System;
using System.Collections.Generic;
List<string> words = new List<string> { "Apple", "Banana", "Cherry" };
words.Remove("Apple");
Console.WriteLine($"Count={words.Count}; Items=[{string.Join(", ", words)}]");
words.Add("Avocado");
words.Add("Durian");
Console.WriteLine($"Count={words.Count}; Items=[{string.Join(", ", words)}]");
words.RemoveAt(3);
Console.WriteLine($"Count={words.Count}; Items=[{string.Join(", ", words)}]");
````

The three displays show `[Banana, Cherry]`, then `[Banana, Cherry, Avocado, Durian]`, then `[Banana, Cherry, Avocado]`. The first removal searches by value; the last removes position 3. The count changes 3 → 2 → 4 → 3. Do not infer contents from a collection's type-name display.

| Action on the original list | Positions afterward |
| --- | --- |
| Start | 0: Apple; 1: Banana; 2: Cherry |
| `Remove("Apple")` | 0: Banana; 1: Cherry |
| Append Avocado and Durian | 0: Banana; 1: Cherry; 2: Avocado; 3: Durian |
| `RemoveAt(3)` | 0: Banana; 1: Cherry; 2: Avocado |

Trace a position after every size-changing operation. An index identifies a current location, not a permanent item identity.


Removing does not leave a "hole" in the `List`: The list closes up,
so the index decreases for the elements after the removed one:

````csharp
using System;
using System.Collections.Generic;
List<string> words = new List<string> { "Banana", "Cherry", "Avocado" };
Console.WriteLine(words.Count);
````

`Add` always appends to the end. To place an element at a specific position instead, use `Insert`, which shifts every later element one position to the right:

````csharp
using System;
using System.Collections.Generic;
List<string> words = new List<string> { "Apple", "Banana", "Cherry" };
words.Insert(1, "Avocado");
Console.WriteLine(string.Join(", ", words));
````

```{figure} ../../figures/list_position_changes.svg
:alt: PEN, BOOK, BAG become PEN, CARD, BOOK, BAG after inserting CARD at index 1; removing index 2 then removes BOOK and shifts BAG left.

Insertion and removal change later positions.
```

Read each row as the complete list after that operation. The picture uses the same rules as the fruit example: insertion shifts a suffix right; removal closes a gap. `Insert` permits an index from 0 through `Count`, including appending at `Count`; retrieval and `RemoveAt` require an index below `Count`.

General operation patterns:

```text
list.Add(item);
list.Insert(index, item);
bool removed = list.Remove(item);
list.RemoveAt(index);
elementType item = list[index];
list[index] = replacement;
int count = list.Count;
```

`Count` is a property, not `Count()`. `Remove` returns success, but `RemoveAt` returns no value. On an integer list, `Remove(2)` selects the first value equal to 2; `RemoveAt(2)` selects the third position.


````csharp
using System;
using System.Collections.Generic;

List<int> byValue = new List<int> { 2, 7, 2, 9 };
List<int> byPosition = new List<int>(byValue);
bool removed = byValue.Remove(2);
byPosition.RemoveAt(2);
Console.WriteLine($"By value ({removed}): {string.Join(", ", byValue)}");
Console.WriteLine($"By position: {string.Join(", ", byPosition)}");
Console.WriteLine($"Missing value removed: {byValue.Remove(100)}");
````

The first list loses its first 2; the second loses the 2 at position 2. Removing a missing value returns `false` and leaves the list unchanged. As a variation, change the repeated value and predict which occurrence is removed. With an empty list, value removal can return `false`, but position removal at 0 is invalid.


You can check for membership in a `List` with `Contains`:

````csharp
using System;
using System.Collections.Generic;
List<string> words = new List<string>();
Console.WriteLine(words.Contains("Apple"));
Console.WriteLine(words.Contains("Banana"));
````

`Contains` only tells you whether a value is present. When you need its position, use `IndexOf`, which returns the index of the first match or `-1` if the value is not found:

````csharp
using System;
using System.Collections.Generic;
List<string> words = new List<string> { "Apple", "Banana", "Cherry" };
Console.WriteLine(words.IndexOf("Banana"));
Console.WriteLine(words.IndexOf("Mango"));
````

You can also remove all elements at once:

````csharp
using System;
using System.Collections.Generic;
List<string> words = new List<string> { "Apple", "Banana", "Cherry" };
words.Clear();
Console.WriteLine($"Count={words.Count}; Items=[{string.Join(", ", words)}]");
````

## Common Pitfall: Modifying a List While Iterating



`foreach` keeps an internal position in the list as it loops. Adding or removing elements through that same list while a `foreach` is still running invalidates that position, and C# detects it: the loop throws an `InvalidOperationException` rather than silently skipping or duplicating elements.

````csharp
using System;
using System.Collections.Generic;
List<int> numbers = new List<int> { 1, 2, 3, 4, 5 };
try
{
    foreach (int n in numbers)
    {
        if (n % 2 == 0)
        {
            numbers.Remove(n);  // modifying the list mid-foreach
        }
    }
}
catch (InvalidOperationException ex)
{
    Console.WriteLine($"Error: {ex.Message}");
}
````

Loop over the indices backwards instead, so removing an element does not disturb the indices you still need to visit. (`List<T>` also has a `RemoveAll` method that removes every matching element in one call — the optional {doc}`query reading <0805_querying_collections>` introduces lambda syntax, and Chapter 13 develops delegate policies.)

````csharp
using System;
using System.Collections.Generic;
List<int> numbers = new List<int> { 1, 2, 3, 4, 5 };
for (int i = numbers.Count - 1; i >= 0; i--)
{
    if (numbers[i] % 2 == 0)
    {
        numbers.RemoveAt(i);
    }
}
Console.WriteLine(string.Join(", ", numbers));
````

```{index} side effect
```

An aside on the `Remove` method: it both causes a side effect (changing the list) *and* returns a value (whether an element was actually removed). If a method returns a value, that value is normally used, as in an `if` condition. `Remove` shows this is not always required: if you just want the side effect — try to remove an element, whether or not it is present — there is no need to check the return value. Once `someList` and `element` are defined, the call itself can stand alone (the following is a syntax fragment):

```csharp
someList.Remove(element);
```

Think carefully before *defining* a method that both has a side effect and returns a value. Most methods that return a value do not also have a side effect; when a method is used as an ordinary expression, it is easy to forget that it might also be doing something.

## Interactive List Example



As in contrast to `array`, `lists` are handy when you do not know how much
data there will be. A simple example would be reading in lines from the user
interactively:



````csharp
using System;
using System.Collections.Generic;
/// Return a List of lines entered by the user in response
/// to the prompt.  Lines in the List will be nonempty, since an
/// empty line terminates the input.
List<string> ReadLines(string prompt)
{
   List<string> lines = new List<string>();
   Console.WriteLine(prompt);
   Console.WriteLine("An empty line terminates input.");
   string? line = Console.ReadLine();
   while (line is not null && line.Length > 0) {
      lines.Add(line);
      line = Console.ReadLine();
   }
   return lines;
}

List<string> entered = ReadLines("Enter requested items:");
Console.WriteLine($"Saved {entered.Count}: {string.Join(", ", entered)}");
````

For a repeatable check, enter `PEN`, then `BOOK`, then an empty line. The final line is `Saved 2: PEN, BOOK`. The terminating blank line is not stored. End-of-input also stops the loop: `Console.ReadLine` can return null, so the condition checks for null before reading `Length`. Try an immediate blank line, one item, and repeated items. A list is useful because the number of entered lines is unknown.
