<fieldset><legend>1. What is a variable?</legend>
<label><input type="radio" name="cscs-variables-types-preview-1" value="0"> A named storage location for a value</label><br>
<label><input type="radio" name="cscs-variables-types-preview-1" value="1"> A command that always prints text</label><br>
<label><input type="radio" name="cscs-variables-types-preview-1" value="2"> A folder containing source files</label><br>
<label><input type="radio" name="cscs-variables-types-preview-1" value="3"> A fixed number of decimal places</label><br>
</fieldset>


```{dropdown} Answer
Answer: A named storage location for a value. A variable has a type and can be assigned a value.
```


<fieldset><legend>2. What is initialization?</legend>
<label><input type="radio" name="cscs-variables-types-preview-2" value="0"> Printing a value</label><br>
<label><input type="radio" name="cscs-variables-types-preview-2" value="1"> Giving a variable its first value</label><br>
<label><input type="radio" name="cscs-variables-types-preview-2" value="2"> Comparing two values</label><br>
<label><input type="radio" name="cscs-variables-types-preview-2" value="3"> Converting text into a number</label><br>
</fieldset>


```{dropdown} Answer
Answer: Giving a variable its first value. Initialization can accompany a declaration.
```


<fieldset><legend>3. What is a literal?</legend>
<label><input type="radio" name="cscs-variables-types-preview-3" value="0"> The name of a variable</label><br>
<label><input type="radio" name="cscs-variables-types-preview-3" value="1"> A keyboard input prompt</label><br>
<label><input type="radio" name="cscs-variables-types-preview-3" value="2"> A value written directly in source code</label><br>
<label><input type="radio" name="cscs-variables-types-preview-3" value="3"> The result of every method</label><br>
</fieldset>


```{dropdown} Answer
Answer: A value written directly in source code. Numbers, quoted text, and true or false can be literals.
```


<fieldset><legend>4. Which type stores a true or false value?</legend>
<label><input type="radio" name="cscs-variables-types-preview-4" value="0"> int</label><br>
<label><input type="radio" name="cscs-variables-types-preview-4" value="1"> string</label><br>
<label><input type="radio" name="cscs-variables-types-preview-4" value="2"> decimal</label><br>
<label><input type="radio" name="cscs-variables-types-preview-4" value="3"> bool</label><br>
</fieldset>


```{dropdown} Answer
Answer: bool. Boolean values represent conditions.
```


<fieldset><legend>5. What does an expression produce?</legend>
<label><input type="radio" name="cscs-variables-types-preview-5" value="0"> A value</label><br>
<label><input type="radio" name="cscs-variables-types-preview-5" value="1"> A new project folder</label><br>
<label><input type="radio" name="cscs-variables-types-preview-5" value="2"> Only a printed line</label><br>
<label><input type="radio" name="cscs-variables-types-preview-5" value="3"> Always a whole number</label><br>
</fieldset>


```{dropdown} Answer
Answer: A value. An expression can combine variables, operators, and method calls.
```


<fieldset><legend>6. What is an argument in a method call?</legend>
<label><input type="radio" name="cscs-variables-types-preview-6" value="0"> A result returned to the caller</label><br>
<label><input type="radio" name="cscs-variables-types-preview-6" value="1"> An input value supplied to the method</label><br>
<label><input type="radio" name="cscs-variables-types-preview-6" value="2"> A compiler warning</label><br>
<label><input type="radio" name="cscs-variables-types-preview-6" value="3"> A variable declaration keyword</label><br>
</fieldset>


```{dropdown} Answer
Answer: An input value supplied to the method. Arguments appear inside the call parentheses.
```


<fieldset><legend>7. What is a return value?</legend>
<label><input type="radio" name="cscs-variables-types-preview-7" value="0"> Text typed into the console</label><br>
<label><input type="radio" name="cscs-variables-types-preview-7" value="1"> A name reserved by C#</label><br>
<label><input type="radio" name="cscs-variables-types-preview-7" value="2"> A result handed back by a method</label><br>
<label><input type="radio" name="cscs-variables-types-preview-7" value="3"> The number of source files</label><br>
</fieldset>


```{dropdown} Answer
Answer: A result handed back by a method. Store the result or use it in another expression; void methods return no value.
```


<fieldset><legend>8. What is parsing?</legend>
<label><input type="radio" name="cscs-variables-types-preview-8" value="0"> Rounding a number</label><br>
<label><input type="radio" name="cscs-variables-types-preview-8" value="1"> Joining text</label><br>
<label><input type="radio" name="cscs-variables-types-preview-8" value="2"> Naming a variable</label><br>
<label><input type="radio" name="cscs-variables-types-preview-8" value="3"> Converting text into a typed value</label><br>
</fieldset>


```{dropdown} Answer
Answer: Converting text into a typed value. Numeric Parse methods convert valid numeric text.
```


<fieldset><legend>9. What is a cast?</legend>
<label><input type="radio" name="cscs-variables-types-preview-9" value="0"> An explicit conversion to a specified type</label><br>
<label><input type="radio" name="cscs-variables-types-preview-9" value="1"> A variable that changes its own type</label><br>
<label><input type="radio" name="cscs-variables-types-preview-9" value="2"> A console prompt</label><br>
<label><input type="radio" name="cscs-variables-types-preview-9" value="3"> A method that prints a report</label><br>
</fieldset>


```{dropdown} Answer
Answer: An explicit conversion to a specified type. A cast requests a conversion; some conversions lose a fractional part.
```


<fieldset><legend>10. What is formatted output?</legend>
<label><input type="radio" name="cscs-variables-types-preview-10" value="0"> A replacement for stored numeric values</label><br>
<label><input type="radio" name="cscs-variables-types-preview-10" value="1"> Text and values displayed with a controlled layout</label><br>
<label><input type="radio" name="cscs-variables-types-preview-10" value="2"> A numeric input parser</label><br>
<label><input type="radio" name="cscs-variables-types-preview-10" value="3"> An instruction to change every variable</label><br>
</fieldset>


```{dropdown} Answer
Answer: Text and values displayed with a controlled layout. Formatting controls the display, including decimal places and alignment.
```
