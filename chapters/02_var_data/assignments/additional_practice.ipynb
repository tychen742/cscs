## Unix Timestamp and Operator Practice



The **Unix epoch** begins at `1970-01-01 00:00:00 UTC`. A Unix timestamp counts seconds relative to that origin; it is a count, not a local time-zone label. `DateTimeOffset.UtcNow` captures a current instant, and `ToUnixTimeSeconds()` returns its timestamp. Local clock display and UTC display can describe the same instant differently.

This demonstration captures the instant once so all three representations refer to the same moment. Your output will differ from the recorded example.

````csharp
using System;
DateTimeOffset instant = DateTimeOffset.UtcNow;
Console.WriteLine($"Local: {instant.ToLocalTime():O}");
Console.WriteLine($"UTC: {instant:O}");
Console.WriteLine($"Unix seconds: {instant.ToUnixTimeSeconds()}");
````

For reproducible practice use `1_767_225_601`, the timestamp for `2026-01-01 00:00:01 UTC`. The underscores only separate digits for readability.

3. Store and print whether elapsed days exceed 20,000.
4. Copy the day count, use `+= 365`, and print the before/after values. This adds 365 days; it is not a general calendar-year calculation.
5. Use `timeStamp & 1` to test the low bit. Result 1 means an odd nonnegative timestamp; result 0 means even.

Assume a nonnegative timestamp and count days as 86,400 seconds for this arithmetic exercise. No decisions or loops are needed. Try one second earlier and predict which answers change.

````csharp
// Exercise: Use the operators to answer all five questions.
using System;
long timeStamp = 1_767_225_601;
// Calculate minutes, remaining seconds, hours, days, comparison,
// the adjusted day count, and the low bit.
Console.WriteLine(timeStamp); // replace with the requested calculations
````

````csharp
using System;
long timeStamp = 1_767_225_601;
long minutes = timeStamp / 60;
long remainderSeconds = timeStamp % 60;
long hours = minutes / 60;
long days = hours / 24;
bool overTwentyThousand = days > 20_000;
long adjustedDays = days;
adjustedDays += 365;
long lowBit = timeStamp & 1;
Console.WriteLine($"Minutes: {minutes}; remaining seconds: {remainderSeconds}");
Console.WriteLine($"Hours: {hours}; days: {days}");
Console.WriteLine($"Days > 20000: {overTwentyThousand}");
Console.WriteLine($"Before: {days}; after adding 365 days: {adjustedDays}");
Console.WriteLine($"Low bit (1 means odd): {lowBit}");
````
