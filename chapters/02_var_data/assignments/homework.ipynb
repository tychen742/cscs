````csharp
// Exercise: Summarize a team target
using System;

decimal target = 100m;
decimal sales = 80m;
bool trainingComplete = true;

// Your code starts here.
// TODO: complete the requested calculations and output.

// Your code ends here.
````

````csharp
// Solution: Summarize a team target
using System;

decimal target = 100m;
decimal sales = 80m;
bool trainingComplete = true;

decimal shortfall = Math.Max(target - sales, 0m);
bool targetMet = sales >= target;
bool bonusEligible = targetMet && trainingComplete;
Console.WriteLine($"Shortfall: {shortfall:F2}");
Console.WriteLine($"Target met: {targetMet}");
Console.WriteLine($"Bonus eligible: {bonusEligible}");
````

**Question 10: Report a budget difference**

An expense forecast is 143.567 and the budget is 150.00. Store the return value of Math.Round(forecast, 2) as roundedForecast. Compute the signed difference as budget - roundedForecast, then use Math.Abs for its magnitude. Print the rounded forecast, signed difference, and magnitude with F2. Retest with a forecast of 153.567; the signed difference should become negative while the magnitude stays nonnegative.


````csharp
// Exercise: Report a budget difference
using System;

double forecast = 143.567;
double budget = 150.00;

// Your code starts here.
// TODO: complete the requested calculations and output.

// Your code ends here.
````

````csharp
// Solution: Report a budget difference
using System;

double forecast = 143.567;
double budget = 150.00;

double roundedForecast = Math.Round(forecast, 2);
double difference = budget - roundedForecast;
double magnitude = Math.Abs(difference);
Console.WriteLine($"Forecast: {roundedForecast:F2}");
Console.WriteLine($"Budget difference: {difference:F2}");
Console.WriteLine($"Magnitude: {magnitude:F2}");
````
