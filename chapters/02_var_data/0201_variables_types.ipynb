These names associated with a data storage location are called
*variables*. In other words, variables represent the locations of
objects in memory. Some general rules for C# variables are:

- Every variable has a **type** that determines what values can be stored in the variable.
- The C# compiler guarantees that values stored in variables are always of the appropriate type.
- The value of a variable can be changed through **assignment** or through use of the ++ and -- operators.
- Variables are either initially assigned or initially unassigned.

Variables can have local scope or global scope:

- Local variables are declared within a method and can only be accessed
  within that method.
- Global variables typically refer to static fields that are accessible
  throughout the class.

## Declaring Local Variables




A declaration statement declares a new local variable, local constant, or local
reference variable. To declare a local variable, specify a type and provide
a variable name. You can declare multiple variables of the same type in one
statement:

> `type variableName;`

In this syntax,

- **type**
  is a C# type (such as int or string),
- **variableName**
  is the name of the variable (such as x or name)
- the semicolon `;`
  is the statement terminator.

For example, you may declare the following variables



````csharp
string greeting;     // declaring one string variable
int a, b, c;         // declaring multiple int variables
List<double> xs;     // a reference type (double list) variable
````


You can also provide initial values when declaring a variable.
The syntax is:



```text
type variableName = value;
```


For example,



````csharp
string greeting = "Hello";
int a = 3, b = 2, c = a + b;
List<double> xs = new();
````


In this syntax,

- the equal sign `=`
  is the assignment operator (note that it is not the equal sign in mathematics)

## The `var` Keyword




Local variables can be declared without giving an explicit type. The complier
will then infer the type of a variable from its initialization expression.
The keyword `var` is used in case of this. For example:



````csharp
var greeting = "Hello";
Console.WriteLine(greeting.GetType());  // output: System.String

var a = 100;
Console.WriteLine(a.GetType());  // output: System.Int32

var xs = new List<double>();
Console.WriteLine(xs.GetType());  // output: System.Collections.Generic.List`1[System.Double]
````

## Assignment Statements




If you need to change
the value to a variable, you need to use the **assignment statement**.
Note that an assignment statement is read from right to left. For example, if we
create a variable "area" and assign the value of "width * height" to it like:



````csharp
int area = 10;
int height = 10;
int width = 10;
area = width * height;
````


The assignment statement starts by evaluating the **expression** on the right-hand side: `width * height`. Since width * height is an
expression, the expression will be evaluated to a value like when evaluating an
expression in math, so the values are substituted as

```none
5 * 7
```

which is then evaluates to 35 and finally assigned to the variable, `area`, on the left.
The whole block of code could look something like follows. You may notice that we
are operating with three variables here.

````csharp
int width = 5, height, area;
height = 7;
area = width * height;
Console.WriteLine(area);
````

In the example above,
`width * height`, as an expression, evaluates to a value; while
`int width = 5, height, area;` is a declaration statement and `height = 7;`
an assignment statement.

You should also notice that the value of the most *recent* assignment is
remembered until reassigned. Also, it is common to use the same variable name
when we intend to change the value to the variable without keeping the previous
value, and after the initial declaration you don't need to declare type anymore:

````csharp
int a = 10;
a = 100;       // reassignment
````


To further illustrate how expression and statement work in C#, or in programming
languages in general, we can take a look at the example as follows. In this case,
we are making two assignment statements: one with a literal value and one with
an expression:

````csharp
int n = 7;
n = n + 1;
Console.WriteLine(n);
````

In short piece of code above, we have an initial assignment statement (`n = 7`),
a variable reassignment (`n = n + 1`) involving an expression `n + 1`, and
finally the *new* value of `n` is 8, replacing the old 7. In a normal C# program, use `Console.WriteLine(n);` when you want to display the value.

### Trace Reassignment: Playing Computer



To understand assignment, follow each statement using the most recently stored values. This is sometimes called **playing computer**. Predict the output of this small program before running it:

```csharp
int x = 3;
int y = x + 2;
y = 2 * y;
x = y - x;
Console.WriteLine(x + " " + y);
```

A state table records the values **after** each statement. A dash means that a local variable has not yet been declared; it is not a default zero.

| Statement | `x` afterward | `y` afterward | Reason |
| --- | ---: | ---: | --- |
| `int x = 3;` | 3 | — | Initialize `x`. |
| `int y = x + 2;` | 3 | 5 | Use the current `x`: `3 + 2`. |
| `y = 2 * y;` | 3 | 10 | Read the old `y`, calculate 10, then replace `y`. |
| `x = y - x;` | 7 | 10 | Read the current values: `10 - 3`. |
| `Console.WriteLine(x + " " + y);` | 7 | 10 | Print `7 10`; output does not change the variables. |

Unlike an equation asserting a permanent relationship, an assignment changes one stored value at that moment. Assigning `y = x + 2` does not make `y` automatically change whenever `x` changes. In this sequential program, the trace follows source order; later decisions and loops will make the order itself part of the reasoning.

````csharp
using System;
int x = 3;
int y = x + 2;
y = 2 * y;
x = y - x;
Console.WriteLine(x + " " + y);
````

Which statement changes only `reserved`? Does changing it automatically recompute `stock`? An assignment evaluates its right side once using the values available at that moment.


````csharp
using System;
int stock = 12;
int reserved = stock - 8;
stock = stock - reserved;
reserved = reserved + 1;
// TODO: print stock and reserved. Compare with your completed table.
````

````csharp
using System;
// Solution: the successive states are (12,4), (8,4), (8,5).
int stock = 12;
int reserved = stock - 8;
stock = stock - reserved;
reserved = reserved + 1;
Console.WriteLine($"stock={stock}, reserved={reserved}");
````

If you instead write `reserved = stock + 1` on the last line, its value becomes 9 because the current `stock` is 8. This is different from a spreadsheet relationship that recalculates dependents automatically.

### Guided Practice: Swap Two Prices


Two prices were entered in the wrong variables. Swap them using a third variable so the output is `regular=120, sale=95`. Predict what happens if you write `regular = sale;` first without saving the old value.


````csharp
using System;

int regular = 95;
int sale = 120;

// Your code starts here.
// Swap the two values using a temporary variable.
// Your code ends here.

Console.WriteLine($"regular={regular}, sale={sale}");
````

````csharp
using System;

// Solution
int regular = 95;
int sale = 120;

int saved = regular;   // keep the old value before it is overwritten
regular = sale;
sale = saved;

Console.WriteLine($"regular={regular}, sale={sale}");
````

Without the temporary variable, `regular = sale;` destroys the only copy of 95, and both variables end up holding 120. An assignment copies a value into a box; it does not remember what the box held before.

## Variable, Identifiers, and Literals




In programming languages, an **identifier** is defined as the name assigned to a
entity (class, interface, struct, delegate, or enum), member, *variable*, function,
method, or namespace, etc. A variable is an identifier/name assigned to a
memory location that stores a value or object reference.

**Literals**: Expressions with straight values such as `7` or `1.23` or `"hello"`
are called *literals* because they *literally* represent what they are.
For example:

- A `bool` variable has two literal values: true or false.
- A string literal is a series of characters delimited by double quotes in one line.
- An integral numeric type can have three kinds of literals: decimal, hexadecimal, and binary.

## Naming Convention of Identifiers




Some of the common conventions [^0201_variables_types-footnote-1] to follow when naming C# identifiers are:

- The characters must all be letters, digits, or underscores `_`,
  and must start with a letter or underscore.

- C# is case sensitive: The identifiers `name`, `NAME`, and `NaMe`
  are different.
- By convention, C# programs use **PascalCase** for namespace,
  class, and method names and **camelCase** for **variable** and
  method parameters. [^0201_variables_types-footnote-2]
- Use meaningful and descriptive names for variables, methods, and classes.
- Avoid keywords when naming identifiers (see {ref}`Section 1.3 <csharp-reserved-keywords>`).

## Optional: Escaped Identifiers and Contextual Keywords



Review {ref}`Names, Keywords, and Case in Section 1.3 <csharp-reserved-keywords>` for reserved keywords and the collapsed reference list. Prefer meaningful variable names that avoid those words.

Two exceptions help you read code from libraries or generated tools:

- **Escaped identifiers** prefix a reserved keyword with `@`, as in `@class`. The prefix tells the compiler to treat the word as a name; it is not part of the name itself. This can help when an external data field is named `class`.
- **Contextual keywords**, such as `var` and `value`, have special meanings only in particular contexts. Here, `var` asks the compiler to infer a local variable's type, while `value` is an ordinary variable name. Its special use in property setters comes later with classes.

The next example represents a shipment's service class, item count, and status. Predict the output before running. These are recognition examples; you do not need these naming exceptions for the core assignments.


````csharp
using System;
// An external shipment field is named "class".
string @class = "Standard";

// var infers int; itemCount remains a meaningful identifier.
var itemCount = 3;
// value is an ordinary identifier here, outside a property setter.
string value = "Ready";

Console.WriteLine($"{@class}: {itemCount} items, {value}.");

````

For a shipment model you design yourself, prefer names such as `serviceClass` and `status` to `@class` and `value`. The exceptions make existing code readable; they do not replace the naming conventions above.
