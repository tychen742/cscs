## Input: Reading from the Keyboard




At console or terminal, input is done in two parts: Prompt and read.
Because the lack of better visual cues, we use a *prompt* to instruct the user
how to respond. For example:



````csharp
Console.Write("Enter your name: ");
````


The user input from keyboard is automatically shown
in the terminal or console window. For a program to take in the characters typed,
another method in the `Console` class is used:
`Console.ReadLine`.

The read method reads the data from a line typed at the keyboard by the user.
When the user presses the Enter (Return) key, the sequence of characters,
form the string *value* of the read method.

With any method producing a value, the value is *lost* unless it
is *immediately* used. We therefore assign the value from the read method
to a variable like in



````csharp
string name;                                // declare variable

Console.WriteLine("Enter your name: ");     // prompt for input
name = Console.ReadLine();                  // save value to variable
````

Or, declare your variable and read in the input one line (e.g., "string name = Console.ReadLine();")
You can then print out the variable.


````csharp
Console.WriteLine("Enter your name: ");
string name = Console.ReadLine();

Console.WriteLine("Your name: " + name);                    // print to test
````

Now we have a variable containing the value for us to operate on.

4. Replace Program.cs with the complete client below. The int variable is declared before the returned value is used.
5. Save both files, build, and run from the project folder. Test a valid integer, a malformed line followed by a valid retry, and negative input. The helper is source code compiled into this same project, so there is no separate package-install step.

```csharp
{

```

The client and helper are also available in the {download}`complete demonstration project <../../demos/02/ui_helper.zip>`. Extract it, open the folder in VS Code, and run `dotnet run`.

Do not leave the template's top-level Console.WriteLine beside a second explicit Main. If you accidentally add two entry points, remove the unintended one rather than deleting the helper.

Keep the current ReadLine/Parse/TryParse explanations as well: a helper is useful only when its input and failure policy suit the program. The supplied version retries malformed integer text and reports exhausted input. In a regular C# program, including the VS Code projects used in this book, values are not displayed just because an expression exists in the source code. You must explicitly give the program an instruction to print to the console or terminal.

This printing is accomplished through a method with a long name: `Console.WriteLine`.
Like with math, you can pass a method a value to work on, by placing it in
parentheses after the name of the method.

`Console` is a C# class maintained by the system, that
interacts with the terminal or console window where text output
appears for the program. A method defined in that class is `WriteLine`.
To refer to a method like `WriteLine` in a different class, you must indicate
the location of the method with the "dot" notation shown:
class name, then `.`, then the method. This
gives the more elaborate name needed in the program.

The `Console.WriteLine` method automatically makes the printing
position advance to the next line, as when you press the Enter or Return key.
A variant, `Console.Write`, prints the parameter exactly, and nothing else.

## Composite Formatting




Instead of using the `+` operator, **composite format** (the
"fill-in-the-braces") gives us better control over output using `Console.WriteLine`
by using positional **string format index** to separate the string and the data variables.

In composite formatting, you use two arguments: 1) a composite format **string** with
**parameter specifier** (`format item` index) and 2) an **object list**. When printing,
the objects (variables or literal values) substitute the parameter specifiers one by one.

Observe the last two statements:



````csharp
string firstName;

Console.Write("Enter you first name: ");    // input will start here
firstName = Console.ReadLine();

Console.WriteLine("Hiya, " + firstName + "!");
Console.WriteLine("Hiya, {0}!", firstName);
````


You can imagine that there would be \{1}, \{2}... like this:



````csharp
string firstName = "Alex";
string lastName = "Kim";
Console.WriteLine("Hiya, {0} {1}!", firstName, lastName);
````


With composite formatting, we have the flexibility of writing the output string
and place the variables anywhere we prefer in the string.

A more elaborate silly examples that you could test in csharp would be:



````csharp
string first = "Peter";
string last = "Piper";
string what = "pick";
Console.WriteLine("{0} {1}, {0} {1}, {2}.", first, last, what);
````


It would print:

```
Peter Piper, Peter Piper, pick.
```

where parameter 0 is `first` (value `"Peter"`),
parameter 1 is `last` ( value `"Piper"`), and
parameter 2 is `what` (value `"pick"`).

As an example, try the following with `csharprepl`, `Run C#`, or a small VS Code project:



````csharp
int x = 7;
int y = 5;
Console.WriteLine("{0} plus {1} is {2}; {0} times {1} is {3}.", x, y, x+y, x*y);
````


to see it prints:



```text
7 plus 5 is 12; 7 times 5 is 35.
```


Note the following features of the parameters after the first string:

- These parameters can be any expression,
  and the expressions get evaluated before printing.
- These parameters to be substituted can be of any type.
- These parameters are automatically converted to a string form, just as in the
  use of the string `+` operation.

In fact the simple use of format strings shown so far can be completed replaced by
long expressions with `+`, if that is your taste.

### Format Specifiers




In addition to printing out strings using composite formatting as seen above, a common
way to format strings is using the **string.Format()** method to save the value of the
variable into another string. For example, we may save a format string to a variable:

````csharp
var msg = string.Format("There are {0} balls", 3);
````


Also, C# has **format specifiers** [^0203_input_output-format-specifiers-1] for different data types when
formatting data.
For example, `D` (decimal digits) formats an integer as whole-number digits:



````csharp
int value = 6324;
string output = string.Format("{{{0:D}}}", value);

Console.WriteLine(output);
// The example displays the following output:
//       {6324}
````


You may format two strings together. In the example below, the `{1,6:D}`
format item takes the second item, format it also as a decimal and the
string length will be 6 characters (right-aligned and padded with empty strings;
the 6 here is the `alignment component`):



````csharp
string.Format("{0:D}  {1,6:D}", 634, 868); // result: 634     868
````


`C` for currency:



````csharp
int myNumber = 100;
Console.WriteLine("{0:C}", myNumber);

// The example displays the following output
// if en-US is the current culture:
//        $100.00
````

```{index} format specifiers; standard numeric
```

The most commonly used standard numeric format specifiers are listed below. The
number after the letter is the **precision specifier**: for most specifiers it sets the
number of decimal places, and for `D` and `X` it sets the minimum number of digits. The
results shown are for the en-US culture.

| Specifier | Name | Use with | Format item | Value | Result |
| --------- | ---- | -------- | ----------- | ----- | ------ |
| `C` | Currency | any number | `{0:C}` | `1234.5m` | `$1,234.50` |
| `C0` | Currency, no decimals | any number | `{0:C0}` | `1234.5m` | `$1,235` |
| `D` | Decimal digits | integers only | `{0:D6}` | `42` | `000042` |
| `F` | Fixed-point | any number | `{0:F2}` | `3.14159` | `3.14` |
| `N` | Number with thousands separators | any number | `{0:N0}` | `1234567` | `1,234,567` |
| `P` | Percent (multiplies by 100) | any number | `{0:P1}` | `0.125` | `12.5%` |
| `E` | Scientific (exponential) | any number | `{0:E2}` | `1234.5` | `1.23E+003` |
| `X` | Hexadecimal | integers only | `{0:X4}` | `255` | `00FF` |

Specifiers that set decimal places round the value; they do not change the variable
itself. A business report typically uses `C` for money, `N0` for counts, and `P` for
rates:


````csharp
decimal revenue = 1234567.891m;
int orderId = 42;
double margin = 0.125;

Console.WriteLine("Revenue:    {0:C}", revenue);
Console.WriteLine("Rounded:    {0:N0}", revenue);
Console.WriteLine("Two places: {0:F2}", revenue);
Console.WriteLine("Margin:     {0:P1}", margin);
Console.WriteLine("Order ID:   {0:D6}", orderId);
````

### Guided Practice: Format an Order Line


Print one order line with composite formatting so the output is exactly `Stapler x3 @ $4.25 = $12.75`. Use `{2:F2}` style specifiers for the two money values and compute the total in the argument list.


````csharp
using System;

string item = "Stapler";
int quantity = 3;
decimal price = 4.25m;

// Your code starts here.
Console.WriteLine("{0} x{1}", item, quantity);
// Your code ends here.
````

````csharp
using System;

// Solution
string item = "Stapler";
int quantity = 3;
decimal price = 4.25m;

Console.WriteLine("{0} x{1} @ ${2:F2} = ${3:F2}", item, quantity, price, quantity * price);
````

The placeholders are numbered by position in the argument list, so `{3:F2}` formats the fourth argument, the computed total, with two decimals. Rewrite the same line with string interpolation after reading the next section and compare which is easier to read.

## String Interpolation




The `$` character identifies a string literal as an **interpolated string**. An interpolated
string is a string literal that might contain interpolation expressions. When an
interpolated string is resolved to a result string, the compiler replaces items with
interpolation expressions by the string representations of the expression results.
String interpolation provides a more readable, convenient syntax to format strings.
To identify a string literal as an interpolated string, prepend it with the `$` symbol.

Compare **composite formatting** and **string interpolation**:



````csharp
var name = "Mark";
var date = DateTime.Now;

// Composite formatting:
Console.WriteLine("Hello, {0}! Today is {1}, it's {2:HH:mm} now.", name, date.DayOfWeek, date);
// String interpolation:
Console.WriteLine($"Hello, {name}! Today is {date.DayOfWeek}, it's {date:HH:mm} now.");
// Both calls produce the same output that is similar to:
// Hello, Mark! Today is Wednesday, it's 19:40 now.
````

## Guided Practice: Read and Report an Invoice Line


An invoice line combines three separate jobs: prompt for text, interpret the numeric input, and format the calculated result. Keep those jobs visible so an output error does not get mistaken for an arithmetic error.

General forms (schematic):

```text
Console.Write(promptText);               // leave the cursor on this line
string text = Console.ReadLine()!;        // this exercise assumes a line is supplied
int quantity = int.Parse(text);          // this exercise assumes valid integer text
Console.WriteLine($"{label,-width} {amount,alignment:F2}");
```

`!` here states the exercise's supplied-input assumption; it is not a runtime check. Real missing input needs handling, developed later. A negative alignment left-aligns text, a positive alignment right-aligns it, and `F2` formats a number with two decimal places. Literal widths such as `-12` replace the schematic names. Formatting changes display, not the value used in a calculation.

For the worked line, use `Pens` at 2.50 per unit and type `4`. The quantity prompt stays beside the entered text. The unit-price times quantity calculation gives 10.00; the label takes a minimum of 12 columns and the amount eight. All examples here assume valid, nonnegative quantities and use an invariant culture so the decimal point is reproducible.


````csharp
using System;
using System.Globalization;
CultureInfo.CurrentCulture = CultureInfo.InvariantCulture;
Console.Write("Quantity: ");
string input = Console.ReadLine()!;
int quantity = int.Parse(input);
decimal total = 2.50m * quantity;
Console.WriteLine($"{"Pens",-12}{total,8:F2}");
````

In a terminal, the interaction is:

```text
Quantity: 4
Pens           10.00
```

The terminal echoes the typed `4`; it is not a second program print. A noninteractive runner may show only the prompt and report in captured stdout. There are eleven spaces between `Pens` and `10.00`: the two fields have combined width 20. With quantity zero, the amount is `0.00`; display alignment remains the same.
