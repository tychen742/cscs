A recursive method needs two promises. First, it must know when to stop. Second, every recursive call must move closer to that stopping point.

Recursion is not magic; it is ordinary method calls plus the call stack. The hard part is choosing a smaller version of the problem that has the same shape as the original problem.


## Base Case and Recursive Case

```{index} Base Case and Recursive Case
```

A **base case** returns an answer directly. A **recursive case** calls the same method on a smaller input and uses that result to build the current answer.

````csharp
using System;
using System.Linq;

foreach(int n in new[]{0,4})Console.WriteLine($"sum({n})={SumDown(n)}");
try{SumDown(-1);}catch(ArgumentOutOfRangeException){Console.WriteLine("negative input rejected");}
long SumDown(int n)
{
 if(n<0 || n>100)throw new ArgumentOutOfRangeException(nameof(n),"Teaching demo accepts 0 through 100.");
 return n==0?0:n+SumDown(n-1);
}
````

The contract accepts integers 0 through 100; zero returns zero and negative input is rejected. The cap keeps this linear-depth teaching demo small, rather than promising an unlimited safe recursion depth. For n>0, the nonnegative integer measure n decreases strictly. Assume the smaller call returns the correct sum of 1 through n−1; adding n gives the required sum through n. This is an induction argument: direct base case, smaller correct result, correct combination.

## Tracing the Call Stack

```{index} Tracing the Call Stack
```

The call stack stores each active method call until that call returns. A trace can make the order of calls and returns visible.

````csharp
using System;
using System.Linq;

Console.WriteLine($"Result: {Factorial(4)}");
Console.WriteLine($"0!={Factorial(0)}");
try{Factorial(-1);}catch(ArgumentOutOfRangeException){Console.WriteLine("negative factorial rejected");}
long Factorial(int n)
{
 if(n<0 || n>20)throw new ArgumentOutOfRangeException(nameof(n),"long factorial accepts 0 through 20.");
 Console.WriteLine($"call Factorial({n})");
 if(n<=1){Console.WriteLine($"return 1 from Factorial({n})");return 1;}
 long smaller=Factorial(n-1);
 long result=checked(n*smaller);
 Console.WriteLine($"return {result} from Factorial({n})");
 return result;
}
````

Each frame retains its own n and resumes after the smaller call returns. For 4!, calls enter with 4,3,2,1; returns deliver 1,2,6,24. The suspended multiplication is work done during unwinding. Zero factorial is one. A long holds 20! but not 21!, so the public contract rejects larger inputs; checked arithmetic makes accidental overflow visible. A mathematically terminating recursion can still exhaust the runtime stack.

## Progress Measures

```{index} Progress Measures
```

A progress measure proves that recursion is moving in the right direction. The measure does not have to be a single number, but it must get closer to a base case.

````csharp
using System;
using System.Linq;

int[] values={2,4,6,8,10};
Console.WriteLine(Contains(values,6));Console.WriteLine(Contains(values,7));
Console.WriteLine(Contains(Array.Empty<int>(),6));
bool Contains(int[] values,int target)
{
 ArgumentNullException.ThrowIfNull(values);
 return Scan(0);
 bool Scan(int index)
 {
  if(index==values.Length)return false;
  if(values[index]==target)return true;
  return Scan(index+1);
 }
}
````

The wrapper hides the helper index so clients cannot supply a negative or out-of-range start. Scan(index) promises whether the target occurs in the suffix [index,n). An empty suffix returns false; a matching first item returns true; otherwise the answer comes from the smaller suffix. The measure n−index decreases on every recursive call. Empty input needs no array access. Worst-case time and stack are Θ(n); a first-item match needs only one helper frame.

## Recursion and Iteration

```{index} Recursion and Iteration
```

Many recursive methods can be written with loops. Use recursion when the problem naturally breaks into smaller problems of the same shape, or when recursion makes the structure easier to see.

````csharp
using System;
using System.Linq;

int[] values={10,-3,8};
Console.WriteLine($"recursive={Sum(values)}, iterative={values.Sum(x=>(long)x)}");
long Sum(int[] values)
{
 ArgumentNullException.ThrowIfNull(values);
 return From(0);
 long From(int index)=>index==values.Length?0:values[index]+From(index+1);
}
````

Both sums visit every item, taking Θ(n) time under the integer cost model. The recursive version retains Θ(n) active frames; an iterative sum retains Θ(1) auxiliary state. Recursion changes expression of control flow, not the amount of data that must be processed. Prefer the loop for long flat reports; recursion is particularly useful for balanced splits and nested structures.

## Recursive Risks

```{index} Recursive Risks
```

A recursive method with a missing base case or no progress can call itself forever until the program runs out of stack space. In C#, recursive calls are not something to use casually for huge linear loops.

````csharp
using System;

Console.WriteLine(Countdown(3));
foreach(int n in new[]{-1,101})
{
 try{Countdown(n);}catch(ArgumentOutOfRangeException){Console.WriteLine($"reject {n}");}
}
int Countdown(int n)
{
 if(n<0||n>100)throw new ArgumentOutOfRangeException(nameof(n),"Small teaching input required.");
 return n==0?0:1+Countdown(n-1);
}
````

Check input domain, direct base answer, and a well-founded progress measure before running recursive code. Do not intentionally trigger stack exhaustion in the browser. Tail-call elimination is not guaranteed for these C# methods. A finite domain and arithmetic range are separate from the proof of termination; validate both when designing the interface.
