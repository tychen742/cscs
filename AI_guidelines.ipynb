# AI Guidelines


## Context

- this is a college level programming/computer science intro class
- this book is inspired by https://introcs.cs.luc.edu/ 
- the problem with introcs.cs.luc.edu is that they are not structured by the weeks. If you go to their [index page](https://introcs.cs.luc.edu/index.html) you see the book has many chapters and sections that can be better grouped.
- I want to create a better version of the luc book as a contribution to the open world. Andrew N. Harrington and George Thiruvathukal would love this idea.
- this book is not just about programming, it's about computer science and information technology, so the conceptual parts are important, not just syntax.
- There's a way to make covering 15 chapters possible: just keep the class going during midterm week, which is doable, except that we now have spring break, spring recession, and fall recession... sigh.


## Organization

- each chapter covers topics of one week-worth (time-wise) of student work and lecture. 
- the sections in each chapter are for lectures and they should in total cover topics for 100 to 120 minutes of lecturing at regular speed.
- For section topics coverage, 
  - The NN01-*.ipynb files (e.g., 0101-cs_ideas.ipynb) covers major and overall concepts of each chapter
  - NN02~N are sections each with its own focus
  - each section should have code examples where possible
- Assignments: in addition to content topics, each chapter should have a preview quiz, a lab, and a review (homework) section.
  - each preview quiz should cover 5 to 10 conceptual questions from the content sections.
  - each lab should have around 5 connected questions
  - each review (homework) should have 5 to 10 questions; the majority of them should be technical questions.


## Quality Considerations
- each chapter should cover essential topics
- each chapter and section should be structured sound pedagogically  
- consistency is key. 

## Formatting

- all NN00-*.ipynb files
  - should have # header, local TOC (only first and 2nd levels), and intro the the chapter according to the content of the whole chapter. The intro should be conceptually clear.
- - The NN00 files:
  - should have one # and no ## or ###. ## should be \<h2> and ### should be \<h3>.
  - should be formatted as: 
    - Title (#)
    - Why this Matters
    - Learning Goals
    - Chapter Flow
- each ## and ### headings should start with their own separate markdown cell with one blank line after the heading.
- each ## and ### header should have indexing. 
- Footnotes:
  - A footnotes should have an independent markdown cell.
  - Every lecture content sections (other than preview, lab, and review) should have Footnotes at the end of the section. 
  - The syntax is ```{rubric} Footnotes\n```. The syntax for the anchors and the footnote entries is [^N].


- when possible, use ASCII diagrams to explain concepts. Use the arrows and line components as seen below, they are probably UTF-8 symbols.
```
         ┌────────────────────────────────────────┐
         │           COMPUTER SYSTEM              │
┌──────┐ │  ┌─────────┐  ┌──────────┐  ┌────────┐ │  ┌─────────────┐
│ User │─┼─▶│  INPUT  │─▶│ PROCESS  │─▶│ OUTPUT │─┼─▶| System/User │
└──────┘ │  └─────────┘  └──────────┘  └────────┘ │  └────────────-┘
         │                    ▲                   │
         │              ┌─────┴─────┐             │
         │              │  STORAGE  │             │
         │              └───────────┘             │
         └────────────────────────────────────────┘

```
