## True or False

**1.** `x ?? y` returns `y` only when `x` is `null`.

```{dropdown} Answer
**True** — The null-coalescing operator returns the right-hand side only if the left is `null`.
```

**2.** Enabling nullable reference types makes `string` and `string?` behave identically.

```{dropdown} Answer
**False** — With `#nullable enable`, `string` is non-nullable; `string?` signals that null is allowed and the compiler warns on unsafe dereferences.
```

**3.** A generic type parameter `T` in `List<T>` can be *any* type — value type or reference type — unless a constraint limits it.

```{dropdown} Answer
**True** — Without a `where` constraint, `T` is unconstrained and accepts any type.
```

**4.** An `async` method that returns `Task` begins executing on a background thread as soon as it is called.

```{dropdown} Answer
**False** — `async` methods run synchronously until the first `await` that actually yields. No extra thread is created automatically.
```


## Multiple Choice

**1.** What does `order?.ShippingAddress?.City` return if `order` is `null`?

a) Throws `NullReferenceException`  
b) Returns `""`  
c) Returns `null`  
d) Compile error

```{dropdown} Answer
**c)** The null-conditional `?.` short-circuits and returns `null` without throwing.
```

**2.** What does `input ??= "default"` do?

a) Always assigns `"default"` to `input`  
b) Assigns `"default"` to `input` only if `input` is currently `null`  
c) Returns `"default"` without modifying `input`  
d) Compile error

```{dropdown} Answer
**b)** `??=` is null-coalescing assignment — assigns only when the variable is null.
```

**3.** What is the primary benefit of writing `static T Max<T>(T a, T b) where T : IComparable<T>`?

a) It avoids boxing/unboxing  
b) It works for any comparable type without writing separate overloads  
c) It prevents null arguments  
d) It compiles faster

```{dropdown} Answer
**b)** A single generic method replaces multiple type-specific overloads while remaining type-safe.
```

**4.** What return type should an `async` method use when it produces a value of type `string`?

a) `string`  
b) `void`  
c) `Task<string>`  
d) `async<string>`

```{dropdown} Answer
**c)** `Task<string>` lets callers `await` the method and receive the `string` result.
```
