## Short Answer

1. What does this switch expression return for input `'e'`?
   ```csharp
   char c = 'e';
   string kind = c switch {
       'a' or 'e' or 'i' or 'o' or 'u' => "vowel",
       >= 'a' and <= 'z'               => "consonant",
       _                               => "other"
   };
   ```
